```python
# =====================================================================
# SEED-IV — ΟΛΟΚΛΗΡΩΜΕΝΟ PIPELINE (Εκπαίδευση + Αξιολόγηση + XAI)
# =====================================================================

# --- ΒΗΜΑ 0: Εγκατάσταση απαραίτητων βιβλιοθηκών ---
try:
    import shap, lime, captum, mne
except ImportError:
    print("Εγκατάσταση απαραίτητων βιβλιοθηκών...")
    import subprocess, sys
    subprocess.run([sys.executable, "-m", "pip", "install", "-q",
                    "mne", "shap", "lime", "captum", "scikit-learn", "statsmodels"], check=True)

import os, re, math, json, zipfile, itertools, warnings, copy
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import scipy.io as sio
from scipy import signal as sps
from scipy.stats import spearmanr, kendalltau, wilcoxon

import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, TensorDataset

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import GroupKFold, LeaveOneGroupOut, train_test_split
from sklearn.metrics import (balanced_accuracy_score, f1_score, precision_recall_fscore_support,
                             cohen_kappa_score, matthews_corrcoef, roc_auc_score,
                             confusion_matrix, brier_score_loss)
from sklearn.manifold import TSNE
from statsmodels.stats.multitest import multipletests

from captum.attr import (IntegratedGradients, Saliency, Occlusion, DeepLift,
                         FeatureAblation, LayerGradCam, LRP)
import shap
from lime import lime_tabular

try:
    from google.colab import drive
    IN_COLAB = True
except ImportError:
    IN_COLAB = False

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Device: {device}")

try:
    display
except NameError:
    def display(x):
        print(x)

# --- ΡΥΘΜΙΣΕΙΣ ---
N_SEEDS = 1                  # 1 για γρήγορη εκτέλεση / αύξησέ το σε 3 για τελικό
SEEDS = [0]

RUN_FULL_LOSO = False
LOSO_MAX_SUBJECTS = 6        # Με n=5 η ελάχιστη εφικτή τιμή p του Wilcoxon είναι 0,0625

N_EPOCHS = 15                # Εποχές ανά fold
BATCH_SIZE = 64
LR = 5e-4

N_EXPLAIN_SAMPLES = 4
N_BOOTSTRAP = 1000
CI_ALPHA = 0.05

FS = 200                     # Sampling rate SEED-IV
WINDOW_SIZE = FS * 4          # 4 δευτερόλεπτα (800 δείγματα)
STEP = FS * 2                 # 2 δευτερόλεπτα (50% overlap)
SECONDS_LIMIT = 120           # Cap ανά trial. Με 30 δευτ. προέκυπταν μόλις ~13 παράθυρα
                              # ανά δοκιμασία και 936 συνολικά -- πολύ λίγα για να μάθει
                              # οτιδήποτε το δίκτυο, γι' αυτό κάθε split έμενε στο ~50%.

MAX_MAT_FILES = None         # None = όλα τα αρχεία (15 εθελοντές x 3 sessions).
                             # Με 6 αρχεία τα δεδομένα ήταν ανεπαρκή για εκπαίδευση.
N_CHANNELS = 62

LABELS_PER_SESSION = {
    1: [1, 2, 3, 0, 2, 0, 0, 1, 0, 1, 2, 3, 2, 3, 1, 1, 3, 0, 1, 3, 0, 2, 2, 3],
    2: [2, 1, 3, 0, 0, 2, 0, 2, 3, 1, 2, 3, 1, 1, 3, 0, 1, 3, 0, 1, 2, 2, 3, 0],
    3: [1, 2, 2, 3, 0, 0, 3, 1, 1, 2, 3, 0, 0, 2, 1, 3, 2, 3, 1, 0, 1, 3, 0, 3],
}

SEEDIV_CHANNELS = [
    'FP1', 'FPZ', 'FP2', 'AF3', 'AF4', 'F7', 'F5', 'F3', 'F1', 'FZ', 'F2', 'F4', 'F6', 'F8',
    'FT7', 'FC5', 'FC3', 'FC1', 'FCZ', 'FC2', 'FC4', 'FC6', 'FT8', 'T7', 'C5', 'C3', 'C1',
    'CZ', 'C2', 'C4', 'C6', 'T8', 'TP7', 'CP5', 'CP3', 'CP1', 'CPZ', 'CP2', 'CP4', 'CP6',
    'TP8', 'P7', 'P5', 'P3', 'P1', 'PZ', 'P2', 'P4', 'P6', 'P8', 'PO7', 'PO5', 'PO3', 'POZ',
    'PO4', 'PO6', 'PO8', 'CB1', 'O1', 'OZ', 'O2', 'CB2',
]

BANDS = {
    'delta': (1, 4), 'theta': (4, 8), 'alpha': (8, 13), 'beta': (13, 30), 'gamma': (30, 50),
}

os.makedirs('outputs', exist_ok=True)

# =====================================================================
# ΧΑΡΤΟΓΡΑΦΗΣΗ ΚΑΝΑΛΙΩΝ & TOPO-MAPS
# =====================================================================
def classify_channel(ch):
    name = ch.upper()
    if name.endswith('Z'):
        prefix, hemi = name[:-1], 'midline'
    else:
        m = re.match(r'^([A-Z]+)(\d+)$', name)
        if m:
            prefix, num = m.group(1), int(m.group(2))
            hemi = 'left' if num % 2 == 1 else 'right'
        else:
            prefix, hemi = name, 'midline'

    region_map = {
        'FP': ('frontal', 'anterior'), 'AF': ('frontal', 'anterior'), 'F': ('frontal', 'anterior'),
        'FC': ('central', 'anterior'), 'FT': ('temporal', 'anterior'),
        'C': ('central', 'midline'), 'T': ('temporal', 'midline'),
        'TP': ('temporal', 'posterior'), 'CP': ('parietal', 'anterior'),
        'P': ('parietal', 'posterior'), 'PO': ('occipital', 'posterior'),
        'O': ('occipital', 'posterior'), 'CB': ('occipital', 'posterior'),
    }
    region, axis = region_map.get(prefix, ('unknown', 'unknown'))
    return region, hemi, axis

CHANNEL_REGION = {ch: classify_channel(ch) for ch in SEEDIV_CHANNELS}

import mne
def build_montage_info(channels=SEEDIV_CHANNELS):
    montage = mne.channels.make_standard_montage('standard_1005')
    positions = montage.get_positions()['ch_pos']
    plottable = [ch for ch in channels if ch in positions]
    info = mne.create_info(ch_names=plottable, sfreq=FS, ch_types='eeg')
    info.set_montage(montage)
    return info, plottable

MNE_INFO, PLOTTABLE_CHANNELS = build_montage_info()
PLOTTABLE_IDX = [SEEDIV_CHANNELS.index(ch) for ch in PLOTTABLE_CHANNELS]

def plot_scalp_map(values_full, title, ax=None, cmap='RdBu_r'):
    values = np.asarray(values_full)[PLOTTABLE_IDX]
    fig = None
    if ax is None:
        fig, ax = plt.subplots(figsize=(4, 4))
    mne.viz.plot_topomap(values, MNE_INFO, axes=ax, show=False, cmap=cmap, contours=4)
    ax.set_title(title, fontsize=10)
    if fig is not None:
        plt.show()
    return ax

def channel_to_region_table(values, channels=SEEDIV_CHANNELS):
    rows = []
    for ch, v in zip(channels, values):
        region, hemi, axis = CHANNEL_REGION[ch]
        rows.append({'channel': ch, 'value': float(v), 'region': region, 'hemisphere': hemi, 'axis': axis})
    df = pd.DataFrame(rows)
    region_table = df.groupby('region')['value'].sum().sort_values(ascending=False)
    return df, region_table

def frontal_asymmetry(values, channels=SEEDIV_CHANNELS):
    pairs = [('AF3', 'AF4'), ('F3', 'F4'), ('F5', 'F6'), ('F7', 'F8'), ('FC3', 'FC4')]
    idx = {ch: i for i, ch in enumerate(channels)}
    out = {}
    for l, r in pairs:
        if l in idx and r in idx:
            out[f'{l}-{r}'] = float(values[idx[r]] - values[idx[l]])
    return out

# =====================================================================
# ΦΟΡΤΩΣΗ ΔΕΔΟΜΕΝΩΝ SEED-IV
# =====================================================================
# --- Εντοπισμός δεδομένων SEED-IV: λειτουργεί σε Kaggle ΚΑΙ σε Colab ---
# ΠΡΟΣΟΧΗ στη δομή του SEED-IV: τα .mat είναι οργανωμένα σε υποφακέλους 1/, 2/, 3/
# (ένας ανά session). Ο loader παρακάτω περιμένει τον ΓΟΝΙΚΟ φάκελο που τους περιέχει.
# Αν επιστρεφόταν απευθείας ο πρώτος φάκελος με .mat (π.χ. eeg_raw_data/1/), θα
# διαβαζόταν μόνο το session 1 και θα χάνονταν τα 2/3 των δεδομένων.
raw_data_dir = None


def _has_session_subdirs(d):
    return all(os.path.isdir(os.path.join(d, s)) for s in ['1', '2', '3'])


def _is_eeg_raw(d):
    # ΔΙΟΡΘΩΣΗ: το SEED-IV περιέχει ΠΟΛΛΟΥΣ φακέλους με υποφακέλους 1/2/3
    # (eeg_raw_data, eeg_feature_smooth, eye_feature_smooth, eye_raw_data...).
    # Θέλουμε ΜΟΝΟ το ακατέργαστο ΗΕΓ -- όχι eye-tracking, όχι έτοιμα χαρακτηριστικά.
    name = os.path.basename(d.rstrip('/')).lower()
    return 'eeg' in name and 'raw' in name and 'eye' not in name and 'feature' not in name


if os.path.exists('/kaggle/input'):
    candidates = [root for root, _, _ in os.walk('/kaggle/input')]

    # 1η προτεραιότητα: φάκελος ακατέργαστου ΗΕΓ με υποφακέλους sessions
    for root in sorted(candidates):
        if _is_eeg_raw(root) and _has_session_subdirs(root):
            raw_data_dir = root
            break

    # 2η: φάκελος ακατέργαστου ΗΕΓ χωρίς υποφακέλους
    if raw_data_dir is None:
        for root in sorted(candidates):
            if _is_eeg_raw(root):
                raw_data_dir = root
                break

    # 3η: οποιοσδήποτε φάκελος με .mat, ΕΚΤΟΣ από eye-tracking / έτοιμα χαρακτηριστικά
    if raw_data_dir is None:
        for root, _, files in sorted(os.walk('/kaggle/input')):
            low = root.lower()
            if 'eye' in low or 'feature' in low:
                continue
            if any(f.endswith('.mat') and not f.startswith('label') for f in files):
                parent = os.path.dirname(root)
                raw_data_dir = parent if _has_session_subdirs(parent) else root
                break

# Colab / τοπικά: αποσυμπίεση από .zip
if raw_data_dir is None:
    if IN_COLAB:
        if not os.path.exists('/content/drive'):
            drive.mount('/content/drive')
        zip_paths = ['/content/archive.zip', '/content/drive/MyDrive/archive.zip',
                     '/content/SEED_IV.zip']
    else:
        zip_paths = ['./archive.zip', './SEED_IV.zip']

    target_zip = next((p for p in zip_paths if os.path.exists(p)), None)
    if target_zip is None:
        raise FileNotFoundError(
            "Δεν βρέθηκαν δεδομένα SEED-IV -- ούτε .mat σε /kaggle/input, ούτε .zip σε Colab. "
            "Αν τρέχεις σε Kaggle, πρόσθεσε το dataset στο notebook (Add Input)."
        )

    extract_path = '/content/SEEDIV_extracted' if IN_COLAB else './SEEDIV_extracted'
    if not os.path.exists(extract_path):
        print("Αποσυμπίεση...")
        with zipfile.ZipFile(target_zip, 'r') as zf:
            zf.extractall(extract_path)

    raw_data_dir = os.path.join(extract_path, 'eeg_raw_data')
    if not os.path.exists(raw_data_dir):
        raw_data_dir = extract_path

print(f"Φάκελος δεδομένων: {raw_data_dir}")

# Kaggle: το /kaggle/working διατηρείται αυτόματα ως Output μετά το Save & Run All
OUT_DIR = '/kaggle/working/outputs' if os.path.exists('/kaggle/working') else 'outputs'
os.makedirs(OUT_DIR, exist_ok=True)
print(f"Φάκελος εξόδου: {OUT_DIR}")

def _trial_sort_key(mat_key):
    m = re.search(r'(\d+)$', mat_key)
    return int(m.group(1)) if m else -1

def load_seediv_with_manifest(raw_data_dir, max_files=MAX_MAT_FILES):
    X_list, y_list, rows = [], [], []
    wid = 0

    session_dirs = [d for d in ['1', '2', '3'] if os.path.isdir(os.path.join(raw_data_dir, d))]
    mat_entries = []
    if not session_dirs:
        for f in os.listdir(raw_data_dir):
            if f.endswith('.mat') and not f.startswith('label'):
                mat_entries.append((os.path.join(raw_data_dir, f), 1))
    else:
        for sdir in session_dirs:
            full_sdir = os.path.join(raw_data_dir, sdir)
            for f in os.listdir(full_sdir):
                if f.endswith('.mat') and not f.startswith('label'):
                    mat_entries.append((os.path.join(full_sdir, f), int(sdir)))

    if max_files is not None:
        mat_entries = mat_entries[:max_files]

    print(f"Επεξεργασία {len(mat_entries)} αρχείων .mat...")
    for mat_path, session_id in mat_entries:
        file_name = os.path.basename(mat_path)
        subject_id = file_name.split('_')[0]
        session_labels = LABELS_PER_SESSION.get(session_id, LABELS_PER_SESSION[1])

        try:
            raw = sio.loadmat(mat_path)
        except Exception:
            continue

        eeg_keys = [k for k in raw.keys() if not k.startswith('__') and 'label' not in k.lower()]
        eeg_keys = sorted(eeg_keys, key=_trial_sort_key)

        for trial_idx, key in enumerate(eeg_keys):
            if trial_idx >= len(session_labels):
                break
            signal = raw[key]
            if signal.ndim != 2:
                continue
            if signal.shape[0] != N_CHANNELS and signal.shape[1] == N_CHANNELS:
                signal = signal.T
            if signal.shape[0] != N_CHANNELS:
                continue

            raw_label = session_labels[trial_idx]
            if raw_label == 3:
                label = 1   # Happy
            elif raw_label == 1:
                label = 0   # Sad
            else:
                continue

            limit = min(signal.shape[1], FS * SECONDS_LIMIT)
            for start in range(0, limit - WINDOW_SIZE, STEP):
                seg = signal[:, start:start + WINDOW_SIZE]
                seg = (seg - np.mean(seg)) / (np.std(seg) + 1e-6)
                X_list.append(seg.T)
                y_list.append(label)
                rows.append({
                    'window_id': wid, 'subject_id': subject_id, 'trial_id': trial_idx,
                    'session_id': session_id, 'dataset': 'SEED-IV', 'label': label,
                    'start_sample': start,
                })
                wid += 1

    if len(X_list) == 0:
        raise RuntimeError(
            f"Δεν εξήχθη κανένα παράθυρο ΗΕΓ από τον φάκελο {raw_data_dir}. "
            "Πιθανότατα επιλέχθηκε φάκελος χωρίς ακατέργαστο ΗΕΓ (π.χ. eye-tracking). "
            "Όρισε χειροκίνητα raw_data_dir στον φάκελο eeg_raw_data."
        )
    X = np.array(X_list, dtype=np.float32)[:, np.newaxis, :, :]
    y = np.array(y_list, dtype=np.int64)
    manifest = pd.DataFrame(rows)
    return torch.tensor(X), torch.tensor(y), manifest

X_data, y_data, manifest = load_seediv_with_manifest(raw_data_dir)
manifest.to_csv(f'{OUT_DIR}/manifest.csv', index=False)
print(f"Σύνολο παραθύρων: {len(y_data)} | Happy: {y_data.sum().item()} ({y_data.float().mean().item():.1%})")

# =====================================================================
# SPLITS & ΜΟΝΤΕΛΟ ACTNN
# =====================================================================
def s1_window_split(manifest, seed):
    idx = manifest.index.values
    train_idx, test_idx = train_test_split(idx, test_size=0.2, random_state=seed, stratify=manifest['label'].values)
    yield train_idx, test_idx, 'S1_full'

def s2_trial_split(manifest, seed, n_splits=5):
    groups = manifest['subject_id'].astype(str) + '_' + manifest['trial_id'].astype(str)
    gkf = GroupKFold(n_splits=n_splits)
    idx = manifest.index.values
    for fold, (tr, te) in enumerate(gkf.split(idx, manifest['label'].values, groups=groups)):
        yield idx[tr], idx[te], f'S2_fold{fold}'

def s3_subject_split(manifest, seed, run_full=RUN_FULL_LOSO, max_subjects=LOSO_MAX_SUBJECTS):
    logo = LeaveOneGroupOut()
    idx = manifest.index.values
    groups = manifest['subject_id'].values
    subjects = sorted(manifest['subject_id'].unique())
    if not run_full:
        rng = np.random.RandomState(seed)
        subjects = list(rng.choice(subjects, size=min(max_subjects, len(subjects)), replace=False))
    for tr, te in logo.split(idx, manifest['label'].values, groups=groups):
        held_out_subject = manifest.iloc[te]['subject_id'].iloc[0]
        if held_out_subject not in subjects:
            continue
        yield idx[tr], idx[te], f'S3_subject{held_out_subject}'

class ACTNN(nn.Module):
    def __init__(self, n_channels=N_CHANNELS, n_classes=2):
        super().__init__()
        self.cnn = nn.Sequential(
            nn.Conv2d(1, 24, (1, n_channels)), nn.BatchNorm2d(24), nn.ELU(),
            nn.Conv2d(24, 48, (15, 1), stride=3), nn.BatchNorm2d(48), nn.ELU(), nn.Dropout(0.5),
        )
        self.transformer = nn.TransformerEncoder(
            nn.TransformerEncoderLayer(d_model=48, nhead=4, batch_first=True), num_layers=2)
        self.fc = nn.Linear(48, n_classes)
        self.attn = nn.Linear(48, 1)

    def forward(self, x):
        x = self.cnn(x).squeeze(3).permute(0, 2, 1)
        x = self.transformer(x)
        w = torch.softmax(self.attn(x), dim=1)
        return self.fc((x * w).sum(dim=1))

    def embed(self, x):
        feat = self.cnn(x).squeeze(3).permute(0, 2, 1)
        feat = self.transformer(feat)
        return feat.mean(dim=1)

def majority_class_baseline(y_test):
    _, counts = np.unique(y_test, return_counts=True)
    return counts.max() / counts.sum()

def full_metrics(y_true, y_pred, y_prob_pos):
    prec, rec, f1c, _ = precision_recall_fscore_support(y_true, y_pred, labels=[0, 1], zero_division=0)
    out = {
        'majority_baseline': majority_class_baseline(y_true),
        'balanced_accuracy': balanced_accuracy_score(y_true, y_pred),
        'macro_f1': f1_score(y_true, y_pred, average='macro', zero_division=0),
        'precision_class0': prec[0], 'recall_class0': rec[0],
        'precision_class1': prec[1], 'recall_class1': rec[1],
        'cohen_kappa': cohen_kappa_score(y_true, y_pred),
        'mcc': matthews_corrcoef(y_true, y_pred),
    }
    try:
        out['roc_auc'] = roc_auc_score(y_true, y_prob_pos)
    except ValueError:
        out['roc_auc'] = float('nan')
    return out

def train_one_split(train_idx, test_idx, seed, n_epochs=N_EPOCHS, verbose=True):
    torch.manual_seed(seed); np.random.seed(seed)
    train_loader = DataLoader(TensorDataset(X_data[train_idx], y_data[train_idx]), batch_size=BATCH_SIZE, shuffle=True)
    test_loader = DataLoader(TensorDataset(X_data[test_idx], y_data[test_idx]), batch_size=BATCH_SIZE, shuffle=False)

    model = ACTNN().to(device)
    optimizer = optim.Adam(model.parameters(), lr=LR)
    criterion = nn.CrossEntropyLoss()

    for epoch in range(n_epochs):
        model.train()
        train_correct, train_n, run_loss = 0, 0, 0.0
        for xb, yb in train_loader:
            xb, yb = xb.to(device), yb.to(device)
            optimizer.zero_grad()
            out = model(xb)
            loss = criterion(out, yb)
            loss.backward()
            optimizer.step()
            # Μέτρηση train accuracy: δείχνει αν το δίκτυο όντως μαθαίνει ή αν
            # έχει κολλήσει στο ποσοστό της πλειοψηφικής κλάσης.
            run_loss += loss.item() * len(yb)
            train_correct += (out.argmax(1) == yb).sum().item()
            train_n += len(yb)
        train_acc = train_correct / max(train_n, 1)
        train_loss = run_loss / max(train_n, 1)

        model.eval()
        val_correct, val_n = 0, 0
        with torch.no_grad():
            for xb, yb in test_loader:
                xb, yb = xb.to(device), yb.to(device)
                val_correct += (model(xb).argmax(1) == yb).sum().item()
                val_n += len(yb)
        val_acc = val_correct / max(val_n, 1)
        if verbose:
            print(f"  epoch {epoch+1:02d} | loss {train_loss:.4f} "
                  f"| train_acc {train_acc:.2%} | val_acc {val_acc:.2%}")

    model.eval()
    all_true, all_pred, all_prob = [], [], []
    with torch.no_grad():
        for xb, yb in test_loader:
            xb = xb.to(device)
            out = torch.softmax(model(xb), dim=1).cpu().numpy()
            all_true.append(yb.numpy())
            all_pred.append(out.argmax(1))
            all_prob.append(out[:, 1])
    y_true, y_pred, y_prob = np.concatenate(all_true), np.concatenate(all_pred), np.concatenate(all_prob)
    return model, y_true, y_pred, y_prob, full_metrics(y_true, y_pred, y_prob)

# =====================================================================
# ΕΚΤΕΛΕΣΗ ΕΚΠΑΙΔΕΥΣΗΣ S1 / S2 / S3
# =====================================================================
results = []
saved_models = {}

for seed in SEEDS:
    print(f"\n=== Seed {seed} ===")
    for train_idx, test_idx, fold_name in s1_window_split(manifest, seed):
        print(f"Training {fold_name}...")
        model, y_true, y_pred, y_prob, m = train_one_split(train_idx, test_idx, seed)
        m.update({'model': 'ACTNN', 'dataset': 'SEED-IV', 'split_level': 'S1', 'fold': fold_name, 'seed': seed})
        results.append(m)
        saved_models[('S1', seed)] = (model, test_idx, y_true, y_pred, y_prob)

    for train_idx, test_idx, fold_name in s2_trial_split(manifest, seed, n_splits=3):
        print(f"Training {fold_name}...")
        model, y_true, y_pred, y_prob, m = train_one_split(train_idx, test_idx, seed)
        m.update({'model': 'ACTNN', 'dataset': 'SEED-IV', 'split_level': 'S2', 'fold': fold_name, 'seed': seed})
        results.append(m)
        saved_models[('S2', seed, fold_name)] = (model, test_idx, y_true, y_pred, y_prob)

    for train_idx, test_idx, fold_name in s3_subject_split(manifest, seed):
        print(f"Training {fold_name}...")
        model, y_true, y_pred, y_prob, m = train_one_split(train_idx, test_idx, seed)
        m.update({'model': 'ACTNN', 'dataset': 'SEED-IV', 'split_level': 'S3', 'fold': fold_name, 'seed': seed})
        results.append(m)
        saved_models[('S3', seed, fold_name)] = (model, test_idx, y_true, y_pred, y_prob)

results_df = pd.DataFrame(results)
results_df.to_csv(f'{OUT_DIR}/results_S1_S2_S3_actnn_seediv.csv', index=False)
display(results_df[['split_level', 'fold', 'balanced_accuracy', 'macro_f1', 'majority_baseline']])

summary = (results_df.groupby('split_level')
           .agg(balanced_accuracy_mean=('balanced_accuracy', 'mean'),
                balanced_accuracy_std=('balanced_accuracy', 'std'),
                macro_f1_mean=('macro_f1', 'mean'),
                majority_baseline_mean=('majority_baseline', 'mean'))
           .reset_index())
display(summary)
s1_acc = summary.loc[summary.split_level == 'S1', 'balanced_accuracy_mean'].values
s3_acc = summary.loc[summary.split_level == 'S3', 'balanced_accuracy_mean'].values
if len(s1_acc) and len(s3_acc):
    print(f"Πτώση S1→S3 (balanced accuracy): {s1_acc[0] - s3_acc[0]:.2%}")

# --- Bootstrap CI για το S3 (όλα τα S3 folds μαζί) ---
s3_keys = [k for k in saved_models if k[0] == 'S3']
y_true_s3 = np.concatenate([saved_models[k][2] for k in s3_keys])
y_pred_s3 = np.concatenate([saved_models[k][3] for k in s3_keys])
y_prob_s3 = np.concatenate([saved_models[k][4] for k in s3_keys])
rng_b = np.random.RandomState(0)
boot = []
for _ in range(1000):
    ii = rng_b.randint(0, len(y_true_s3), len(y_true_s3))
    boot.append(balanced_accuracy_score(y_true_s3[ii], y_pred_s3[ii]))
lo, hi = np.percentile(boot, [2.5, 97.5])
print(f"S3 balanced accuracy: {np.mean(boot):.2%} (95% CI [{lo:.2%}, {hi:.2%}])")

# --- Wilcoxon: ακρίβεια ανά εθελοντή στο S1 έναντι του S3 ---
s1_key = next(k for k in saved_models if k[0] == 'S1')
_, s1_test_idx, s1_true, s1_pred, _ = saved_models[s1_key]
s1_df = pd.DataFrame({'y': s1_true, 'p': s1_pred,
                      'subject': manifest.loc[s1_test_idx, 'subject_id'].values})
s1_per_subj = s1_df.groupby('subject').apply(lambda g: (g['y'] == g['p']).mean())
s3_per_subj = {}
for k in s3_keys:
    _, t_idx, yt, yp, _ = saved_models[k]
    s3_per_subj[manifest.loc[t_idx, 'subject_id'].iloc[0]] = (yt == yp).mean()
s3_per_subj = pd.Series(s3_per_subj)
common = s1_per_subj.index.intersection(s3_per_subj.index)
if len(common) >= 5:
    stat, p = wilcoxon(s1_per_subj.loc[common], s3_per_subj.loc[common])
    print(f"Wilcoxon S1 vs S3 (n={len(common)} εθελοντές): stat={stat:.3f}, p={p:.4f}")
    if len(common) <= 5:
        print("  Σημείωση: με n=5 η ελάχιστη εφικτή τιμή p είναι 0,0625.")
else:
    print("Πολύ λίγοι κοινοί εθελοντές για Wilcoxon.")

# --- Βαθμονόμηση (ECE / Brier) για το S3 ---
bins = np.linspace(0, 1, 11)
ece = 0.0
for b_i in range(10):
    msk = (y_prob_s3 >= bins[b_i]) & ((y_prob_s3 < bins[b_i + 1]) if b_i < 9 else (y_prob_s3 <= 1))
    if msk.sum():
        ece += msk.mean() * abs((y_true_s3[msk] == 1).mean() - y_prob_s3[msk].mean())
print(f"ECE = {ece:.4f} | Brier = {brier_score_loss(y_true_s3, y_prob_s3):.4f}")

# --- Confusion matrices ---
fig, axes = plt.subplots(1, 3, figsize=(10.5, 3))
for ax, level in zip(axes, ['S1', 'S2', 'S3']):
    key = next((k for k in saved_models if k[0] == level), None)
    if key is None:
        continue
    _, _, yt, yp, _ = saved_models[key]
    sns.heatmap(confusion_matrix(yt, yp, labels=[0, 1]), annot=True, fmt='d', cmap='Blues',
                ax=ax, xticklabels=['Sad', 'Happy'], yticklabels=['Sad', 'Happy'])
    ax.set_title(level); ax.set_xlabel('Predicted'); ax.set_ylabel('True')
plt.tight_layout(); plt.show()

# Αποδέσμευση μνήμης GPU πριν την ανάλυση ερμηνευσιμότητας
import gc
for k in list(saved_models):
    if k[0] == 'S2':
        saved_models[k] = (None,) + tuple(saved_models[k][1:])
gc.collect()
if torch.cuda.is_available():
    torch.cuda.empty_cache()

# =====================================================================
# ΠΡΟΕΤΟΙΜΑΣΙΑ XAI EXPLAINERS
# =====================================================================
any_s3_key = next(k for k in saved_models if k[0] == 'S3')
xai_model, xai_test_idx, _, _, _ = saved_models[any_s3_key]
xai_model.eval()

xai_loader = DataLoader(TensorDataset(X_data[xai_test_idx], y_data[xai_test_idx]), batch_size=32, shuffle=False)
xai_batch_x, xai_batch_y = next(iter(xai_loader))
explain_samples = xai_batch_x[:N_EXPLAIN_SAMPLES].to(device)
explain_labels = xai_batch_y[:N_EXPLAIN_SAMPLES]
background = xai_batch_x[:min(20, len(xai_batch_x))].to(device)   # για SHAP/LIME

def _to_map(attr_tensor):
    return attr_tensor.squeeze().detach().cpu().numpy()

def explain_integrated_gradients(sample, target=1):
    return _to_map(IntegratedGradients(xai_model).attribute(sample, target=target, n_steps=20))

def explain_saliency(sample, target=1):
    return _to_map(Saliency(xai_model).attribute(sample.clone().requires_grad_(True), target=target))

def explain_occlusion(sample, target=1):
    return _to_map(Occlusion(xai_model).attribute(sample, target=target, sliding_window_shapes=(1, 20, 1), strides=(1, 10, 1)))

def explain_deeplift(sample, target=1):
    return _to_map(DeepLift(xai_model).attribute(sample, baselines=torch.zeros_like(sample), target=target))

def explain_feature_ablation(sample, target=1):
    """Ablation ανά ολόκληρο κανάλι (62 ηλεκτρόδια). Η είσοδος έχει σχήμα
    (1, 1, WINDOW_SIZE, N_CHANNELS), οπότε η μάσκα ομαδοποιεί κατά τον άξονα των καναλιών."""
    ablator = FeatureAblation(xai_model)
    feature_mask = (torch.arange(N_CHANNELS)
                    .reshape(1, 1, 1, N_CHANNELS)
                    .expand(1, 1, WINDOW_SIZE, N_CHANNELS).to(device))
    return _to_map(ablator.attribute(sample, target=target, feature_mask=feature_mask))


def explain_gradcam(sample, target=1):
    """Grad-CAM στο τελευταίο συνελικτικό επίπεδο. Προσοχή: cnn[-2] είναι ELU
    (τελευταίο στο Sequential είναι το Dropout) -- το πραγματικό τελευταίο Conv2d
    είναι το cnn[3]."""
    lgc = LayerGradCam(xai_model, xai_model.cnn[3])
    attr = lgc.attribute(sample, target=target)
    attr = torch.nn.functional.interpolate(attr, size=(WINDOW_SIZE, N_CHANNELS),
                                            mode='bilinear', align_corners=False)
    return _to_map(attr)


def explain_lrp(sample, target=1):
    """Το captum LRP δεν έχει κανόνα για ELU/TransformerEncoder· αν αποτύχει
    επιστρέφεται μηδενικός χάρτης αντί να διακοπεί η ανάλυση."""
    try:
        return _to_map(LRP(xai_model).attribute(sample.clone().requires_grad_(True), target=target))
    except Exception as e:
        print(f"  LRP απέτυχε ({e}); επιστρέφεται μηδενικός χάρτης.")
        return np.zeros((WINDOW_SIZE, N_CHANNELS))


def explain_shap_gradient(sample, target=1):
    """Gradient SHAP. Νεότερες εκδόσεις του shap επιστρέφουν επιπλέον άξονα κλάσεων,
    οπότε επιλέγεται ρητά η κλάση-στόχος."""
    explainer = shap.GradientExplainer(xai_model, background)
    sv = explainer.shap_values(sample)
    arr = np.array(sv[target] if isinstance(sv, list) and len(sv) > target
                   else (sv[0] if isinstance(sv, list) else sv))
    arr = np.squeeze(arr)
    if arr.ndim == 3:
        if arr.shape[-1] == 2:
            arr = arr[..., target]
        elif arr.shape[0] == 2:
            arr = arr[target]
        else:
            arr = arr[0]
    return arr


def _predict_flat(x_flat, batch=32):
    # ΔΙΟΡΘΩΣΗ: το LIME στέλνει χιλιάδες διαταραγμένα δείγματα ταυτόχρονα. Περνώντας τα
    # όλα μαζί από τον Transformer ζητούνταν ~10 GB μνήμης GPU (OutOfMemoryError).
    # Τώρα η πρόβλεψη γίνεται σε μικρές παρτίδες.
    x_flat = np.asarray(x_flat, dtype=np.float32)
    xai_model.eval()
    outs = []
    with torch.no_grad():
        for i in range(0, len(x_flat), batch):
            x_t = (torch.tensor(x_flat[i:i + batch])
                   .reshape(-1, 1, WINDOW_SIZE, N_CHANNELS).to(device))
            outs.append(torch.softmax(xai_model(x_t), dim=1).cpu().numpy())
    return np.concatenate(outs, axis=0)


def explain_lime(sample, target=1, num_features=60):
    """LIME πάνω στο επίπεδο διάνυσμα εισόδου."""
    sample_flat = sample.detach().reshape(1, -1).cpu().numpy()[0]
    bg_flat = background.detach().reshape(background.shape[0], -1).cpu().numpy()
    expl = lime_tabular.LimeTabularExplainer(
        bg_flat, mode='classification', class_names=['Sad', 'Happy'])
    exp = expl.explain_instance(sample_flat, _predict_flat, num_features=num_features,
                                 labels=[target], num_samples=500)
    out = np.zeros(WINDOW_SIZE * N_CHANNELS)
    for feat_idx, w in exp.as_map()[target]:
        out[feat_idx] = w
    return out.reshape(WINDOW_SIZE, N_CHANNELS)


def explain_kernelshap(sample, target=1, nsamples=60):
    """KernelSHAP σε επίπεδο ΚΑΝΑΛΙΟΥ (62 χαρακτηριστικά αντί για 800x62=49.600),
    ώστε το πλήθος συνδυασμών να παραμένει υπολογιστικά εφικτό."""
    base_np = sample.detach().cpu().numpy()[0, 0]     # (WINDOW_SIZE, N_CHANNELS)

    def predict_channel_masks(masks):
        out = []
        for mask in masks:
            tmp = base_np.copy()
            tmp[:, mask < 0.5] = 0.0
            x_t = torch.tensor(tmp[None, None, :, :], dtype=torch.float32).to(device)
            xai_model.eval()
            with torch.no_grad():
                out.append(torch.softmax(xai_model(x_t), dim=1).cpu().numpy()[0, target])
        return np.array(out)

    explainer = shap.KernelExplainer(predict_channel_masks, np.zeros((1, N_CHANNELS)))
    sv = explainer.shap_values(np.ones((1, N_CHANNELS)), nsamples=nsamples, silent=True)
    ch_scores = np.array(sv).reshape(-1)[:N_CHANNELS]
    return np.tile(ch_scores[None, :], (WINDOW_SIZE, 1))


EXPLAINERS = {
    'IntegratedGradients': explain_integrated_gradients,
    'Saliency': explain_saliency,
    'Occlusion': explain_occlusion,
    'DeepLift': explain_deeplift,
    'FeatureAblation': explain_feature_ablation,
    'GradCAM': explain_gradcam,
    'LRP': explain_lrp,
    'SHAP(Gradient)': explain_shap_gradient,
    'LIME': explain_lime,
    'KernelSHAP': explain_kernelshap,
}

# =====================================================================
# ΜΠΛΟΚ 3 — ΧΑΡΤΕΣ & ΜΕΤΩΠΙΑΙΑ ΑΣΥΜΜΕΤΡΙΑ
# =====================================================================
sample0 = explain_samples[0:1]
target0 = int(explain_labels[0].item())

ig_map = explain_integrated_gradients(sample0, target=target0)
channel_importance = np.abs(ig_map).sum(axis=0)

region_df, region_table = channel_to_region_table(channel_importance)
print("\n--- Relevance ανά Ανατομική Περιοχή ---")
display(region_table)

plot_scalp_map(channel_importance, title=f"SEED-IV (62ch): IG Relevance (Class {target0})")

asym = frontal_asymmetry(channel_importance)
print("\nΜετωπιαία Ασυμμετρία:")
for k, v in asym.items():
    print(f"  {k}: {v:+.4f}")


# =====================================================================
# ΞΕΧΩΡΙΣΤΟΣ ΤΟΠΟΓΡΑΦΙΚΟΣ ΧΑΡΤΗΣ ΚΑΙ TOP-5 ΓΙΑ ΚΑΘΕ ΜΕΘΟΔΟ XAI
# =====================================================================
all_maps = {}
for name, fn in EXPLAINERS.items():
    try:
        print(f"Υπολογισμός {name}...")
        all_maps[name] = fn(sample0, target=target0)
    except Exception as e:
        print(f"  Το {name} απέτυχε ({e}); παραλείπεται.")

# Έλεγχος σχήματος: ο χάρτης πρέπει να είναι (WINDOW_SIZE, N_CHANNELS)
clean_maps = {}
for name, m in all_maps.items():
    arr = np.asarray(m)
    if arr.ndim != 2:
        print(f"  Το {name} έδωσε χάρτη σχήματος {arr.shape}· παραλείπεται.")
        continue
    if arr.shape[1] != N_CHANNELS and arr.shape[0] == N_CHANNELS:
        print(f"  Το {name} έδωσε ανάστροφο χάρτη {arr.shape}· διορθώνεται.")
        arr = arr.T
    if arr.shape[1] != N_CHANNELS:
        print(f"  Το {name} έδωσε ασύμβατο χάρτη {arr.shape}· παραλείπεται.")
        continue
    clean_maps[name] = arr
all_maps = clean_maps
print(f"\nΜέθοδοι με έγκυρο χάρτη: {list(all_maps.keys())}")

print("\n===== Τοπογραφικοί χάρτες ανά μέθοδο XAI =====")
per_method_importance = {}
for method_name, attr_map in all_maps.items():
    imp = np.abs(attr_map).sum(axis=0)          # (62,)
    per_method_importance[method_name] = imp
    plot_scalp_map(imp, title=f"{method_name} — class {target0}")

    top5_idx = np.argsort(imp)[::-1][:5]
    print(f"\nTop-5 ηλεκτρόδια ({method_name}):")
    for i in top5_idx:
        ch = SEEDIV_CHANNELS[i]
        info_ch = CHANNEL_REGION.get(ch, ('unknown', 'unknown'))
        print(f"  {ch} ({info_ch[0]}-{info_ch[1]}): {imp[i]:.6f}")

# --- Συγκεντρωτικός πίνακας σημαντικότητας ανά ηλεκτρόδιο και μέθοδο ---
channel_scores_df = pd.DataFrame(per_method_importance, index=SEEDIV_CHANNELS)
channel_scores_df.index.name = 'channel'
print("\n===== Σημαντικότητα ανά ηλεκτρόδιο και μέθοδο (ACTNN / SEED-IV) =====")
display(channel_scores_df.round(6))
channel_scores_df.to_csv(f'{OUT_DIR}/xai_channel_scores_actnn_seediv.csv')

# --- Μετωπιαία ασυμμετρία ανά μέθοδο ---
print("\n===== Μετωπιαία ασυμμετρία ανά μέθοδο =====")
asym_rows = []
for method_name, imp in per_method_importance.items():
    a = frontal_asymmetry(imp)
    a['method'] = method_name
    asym_rows.append(a)
display(pd.DataFrame(asym_rows).set_index('method').round(6))


# =====================================================================
# t-SNE + ΠΟΣΟΤΙΚΑ ΜΕΤΡΑ ΔΙΑΧΩΡΙΣΙΜΟΤΗΤΑΣ
# Το t-SNE είναι μόνο οπτική ένδειξη και μπορεί να "δείξει" ομάδες που δεν
# αντιστοιχούν σε πραγματική δομή. Τα μέτρα αυτά το τεκμηριώνουν αριθμητικά.
# =====================================================================
from sklearn.metrics import silhouette_score, davies_bouldin_score

tsne_loader = DataLoader(TensorDataset(X_data[xai_test_idx], y_data[xai_test_idx]),
                          batch_size=64, shuffle=False)
xai_model.eval()
f_list, l_list = [], []
with torch.no_grad():
    for xb, yb in tsne_loader:
        f_list.append(xai_model.embed(xb.to(device)).cpu().numpy())
        l_list.append(yb.numpy())
features_s3 = np.concatenate(f_list)
labels_s3 = np.concatenate(l_list)

X_tsne = TSNE(n_components=2, init='pca', random_state=0,
              perplexity=min(30, max(5, len(features_s3) // 4))).fit_transform(features_s3)
plt.figure(figsize=(7, 5))
sns.scatterplot(x=X_tsne[:, 0], y=X_tsne[:, 1], hue=labels_s3, palette='coolwarm')
plt.title("t-SNE Clusters (S3 / LOSO model) — SEED-IV")
plt.show()


def fisher_ratio(feats, labels):
    """Λόγος διασποράς μεταξύ κλάσεων προς εντός κλάσεων (μεγαλύτερο = πιο διαχωρίσιμο)."""
    classes = np.unique(labels)
    overall_mean = feats.mean(axis=0)
    between, within = 0.0, 0.0
    for c in classes:
        fc = feats[labels == c]
        between += len(fc) * np.sum((fc.mean(axis=0) - overall_mean) ** 2)
        within += np.sum((fc - fc.mean(axis=0)) ** 2)
    return float(between / within) if within > 0 else float('nan')


def class_distances(feats, labels):
    """Μέση ενδο-κλασική και δια-κλασική απόσταση."""
    classes = np.unique(labels)
    centroids = {c: feats[labels == c].mean(axis=0) for c in classes}
    intra = np.mean([np.linalg.norm(feats[labels == c] - centroids[c], axis=1).mean()
                     for c in classes])
    inter = np.mean([np.linalg.norm(centroids[a] - centroids[b])
                     for i, a in enumerate(classes) for b in classes[i + 1:]])
    return float(intra), float(inter)


sil = silhouette_score(features_s3, labels_s3)
db = davies_bouldin_score(features_s3, labels_s3)
fr = fisher_ratio(features_s3, labels_s3)
intra, inter = class_distances(features_s3, labels_s3)

print("\n===== Ποσοτικά μέτρα διαχωρισιμότητας (ACTNN / SEED-IV, S3) =====")
print(f"Αναπαραστάσεις: {features_s3.shape} | Κατανομή κλάσεων: {np.bincount(labels_s3)}")
print(f"Silhouette score      : {sil:+.4f}   (→ +1 άριστος διαχωρισμός, ~0 επικάλυψη)")
print(f"Davies-Bouldin index  : {db:.4f}    (μικρότερο = καλύτερο)")
print(f"Fisher discriminant   : {fr:.4f}    (μεγαλύτερο = καλύτερο)")
print(f"Μέση ενδο-κλασική απόσταση : {intra:.4f}")
print(f"Μέση δια-κλασική απόσταση  : {inter:.4f}")
print(f"Λόγος inter/intra          : {inter/intra:.4f}  (>1 υποδηλώνει διαχωρίσιμες κλάσεις)")

separability_df = pd.DataFrame([{
    'model': 'ACTNN', 'dataset': 'SEED-IV', 'split_level': 'S3',
    'silhouette': sil, 'davies_bouldin': db, 'fisher_ratio': fr,
    'intra_class_dist': intra, 'inter_class_dist': inter, 'inter_over_intra': inter / intra,
}])
separability_df.to_csv(f'{OUT_DIR}/separability_metrics_actnn_seediv.csv', index=False)
display(separability_df.round(4))

# =====================================================================
# ΜΠΛΟΚ 4 — FAITHFULNESS & AOPC
# =====================================================================
def predicted_prob(sample, target):
    with torch.no_grad():
        return torch.softmax(xai_model(sample), dim=1)[0, target].item()

def deletion_insertion_curve(sample, target, attr_map, mode='deletion', k_values=[10, 20, 30]):
    flat = attr_map.flatten()
    order = np.argsort(-flat)
    n_total = order.size
    base = sample.clone()
    probs = []
    for k in k_values:
        idx_to_touch = order[:int(n_total * k / 100)]
        flat_base = base.view(1, 1, -1).clone()
        if mode == 'deletion':
            flat_base[0, 0, idx_to_touch] = 0.0
        else:
            new_flat = torch.zeros_like(flat_base)
            new_flat[0, 0, idx_to_touch] = base.view(1, 1, -1)[0, 0, idx_to_touch]
            flat_base = new_flat
        probs.append(predicted_prob(flat_base.view_as(base), target))
    return np.array(probs)

def aopc(base_prob, probs):
    probs = np.array(probs)
    return float(np.mean(base_prob - probs)) if len(probs) else float('nan')

faithfulness_rows = []
for name, fn in EXPLAINERS.items():
    for i in range(min(N_EXPLAIN_SAMPLES, len(explain_samples))):
        s = explain_samples[i:i + 1]
        t = int(explain_labels[i].item())
        base_p = predicted_prob(s, t)
        try:
            attr = np.asarray(fn(s, target=t))
        except Exception as e:
            print(f"  Faithfulness: το {name} απέτυχε ({e}); παραλείπεται.")
            break
        if attr.shape != (WINDOW_SIZE, N_CHANNELS):
            print(f"  Faithfulness: το {name} έδωσε σχήμα {attr.shape}; παραλείπεται.")
            break

        del_p = deletion_insertion_curve(s, t, attr, 'deletion')
        ins_p = deletion_insertion_curve(s, t, attr, 'insertion')
        rand_map = np.random.RandomState(i).rand(*attr.shape)
        del_p_r = deletion_insertion_curve(s, t, rand_map, 'deletion')

        faithfulness_rows.append({
            'explainer': name, 'sample': i,
            'AOPC_deletion': aopc(base_p, del_p),
            'AOPC_insertion': aopc(base_p, ins_p) * -1,
            'AOPC_deletion_random': aopc(base_p, del_p_r)
        })

faith_df = pd.DataFrame(faithfulness_rows).groupby('explainer').mean(numeric_only=True)
faith_df['deletion_beats_random'] = faith_df['AOPC_deletion'] > faith_df['AOPC_deletion_random']
print("\n--- Faithfulness Summary (AOPC) ---")
display(faith_df)
faith_df.to_csv(f'{OUT_DIR}/faithfulness_AOPC_actnn_seediv.csv')

# =====================================================================
# ΜΠΛΟΚ 5 — SANITY CHECKS
# =====================================================================
def cascading_randomization_test(model, sample, target):
    def explain(m):
        return Saliency(m).attribute(sample.clone().requires_grad_(True), target=target).squeeze().detach().cpu().numpy().flatten()
    orig = explain(model)
    res = {'none_randomized': 1.0}
    m = copy.deepcopy(model)
    for l_name in ['fc', 'attn', 'transformer', 'cnn']:
        layer = getattr(m, l_name)
        for p in layer.parameters():
            nn.init.normal_(p, mean=0.0, std=0.1)
        res[f'after_{l_name}'] = spearmanr(orig, explain(m))[0]
    return res

rand_res = cascading_randomization_test(xai_model, sample0, target0)
print("\n--- Model Randomization Test (Spearman ρ) ---")
for k, v in rand_res.items():
    print(f"  {k}: {v:+.3f}")

# =====================================================================
# ΜΠΛΟΚ 6 — ΣΥΜΦΩΝΙΑ EXPLAINERS
# =====================================================================
# Χρησιμοποιούνται οι χάρτες που υπολογίστηκαν ήδη (ΜΠΛΟΚ 3), αντί να ξανατρέξουν
# όλες οι μέθοδοι. Ο μηδενικός χάρτης του LRP εξαιρείται, αφού δεν ορίζεται συσχέτιση.
ch_imps = {n: v for n, v in per_method_importance.items() if np.abs(v).sum() > 0}
names = list(ch_imps.keys())
agree_spearman = pd.DataFrame(index=names, columns=names, dtype=float)
agree_jaccard5 = pd.DataFrame(index=names, columns=names, dtype=float)
for a, b in itertools.product(names, names):
    agree_spearman.loc[a, b] = spearmanr(ch_imps[a], ch_imps[b])[0]
    top_a = set(np.argsort(-ch_imps[a])[:5]); top_b = set(np.argsort(-ch_imps[b])[:5])
    agree_jaccard5.loc[a, b] = len(top_a & top_b) / len(top_a | top_b)
print("\n--- Jaccard Top-5 ηλεκτροδίων ---")
display(agree_jaccard5.round(2))
agree_spearman.to_csv(f'{OUT_DIR}/interexplainer_spearman_actnn_seediv.csv')

plt.figure(figsize=(7, 6))
sns.heatmap(agree_spearman.astype(float), annot=True, fmt='.2f', cmap='coolwarm', vmin=-1, vmax=1)
plt.title('Συμφωνία Explainers — Spearman ρ')
plt.tight_layout(); plt.show()

# =====================================================================
# ΜΠΛΟΚ 7 — ΦΑΣΜΑΤΙΚΗ ΕΞΗΓΗΣΙΜΟΤΗΤΑ
# =====================================================================
def bandpass_filter(x, low, high, fs=FS, order=4):
    nyq = fs / 2.0
    b, a = sps.butter(order, [max(low / nyq, 1e-3), min(high / nyq, 0.999)], btype='bandpass')
    return sps.filtfilt(b, a, x, axis=0).copy()

channels_x_bands = np.zeros((N_CHANNELS, len(BANDS)))
sample_np = sample0.squeeze().detach().cpu().numpy()

for b_idx, (band_name, (lo, hi)) in enumerate(BANDS.items()):
    band_signal = bandpass_filter(sample_np, lo, hi)
    band_tensor = torch.tensor(np.ascontiguousarray(band_signal[None, None, :, :]), dtype=torch.float32, device=device)
    ig_band = explain_integrated_gradients(band_tensor, target=target0)
    channels_x_bands[:, b_idx] = np.abs(ig_band).sum(axis=0)

cxb_df = pd.DataFrame(channels_x_bands, index=SEEDIV_CHANNELS, columns=list(BANDS.keys()))
fig, axes = plt.subplots(1, len(BANDS), figsize=(4 * len(BANDS), 4))
for ax, band_name in zip(axes, BANDS.keys()):
    plot_scalp_map(cxb_df[band_name].values, title=f"Band: {band_name}", ax=ax)
plt.tight_layout(); plt.show()

fa_alpha = frontal_asymmetry(cxb_df['alpha'].values)
print("\nFrontal Alpha Asymmetry (SEED-IV):")
for k, v in fa_alpha.items():
    print(f"  {k}: {v:+.4f}")

######################################################################
# ΑΝΑΛΥΣΗ ΟΜΟΙΟΤΗΤΑΣ ΑΝΑΠΑΡΑΣΤΑΣΕΩΝ (Linear CKA)
#
# Το CKA (Centered Kernel Alignment) μετρά πόσο όμοιες είναι δύο αναπαραστάσεις,
# σε κλίμακα 0 (εντελώς ανόμοιες) έως 1 (ταυτόσημες). Λειτουργεί ακόμα κι όταν
# οι δύο αναπαραστάσεις έχουν διαφορετικό πλήθος διαστάσεων, γι' αυτό επιτρέπει
# σύγκριση τόσο μεταξύ επιπέδων του ίδιου δικτύου όσο και μεταξύ διαφορετικών
# μοντέλων. Εδώ απαντώνται δύο ερωτήματα:
#   (α) σε ποιο στάδιο του δικτύου αλλάζει ουσιαστικά η πληροφορία, και
#   (β) μαθαίνουν τα μοντέλα S1 και S3 παρόμοιες αναπαραστάσεις;
######################################################################

def _center_gram(K):
    n = K.shape[0]
    H = np.eye(n) - np.ones((n, n)) / n
    return H @ K @ H


def linear_cka(X, Y):
    """Linear CKA μεταξύ δύο πινάκων αναπαραστάσεων (δείγματα x διαστάσεις).
    Τα δείγματα πρέπει να είναι τα ΙΔΙΑ και στη ΙΔΙΑ σειρά στους δύο πίνακες."""
    X = np.asarray(X, dtype=np.float64)
    Y = np.asarray(Y, dtype=np.float64)
    X = X - X.mean(axis=0, keepdims=True)
    Y = Y - Y.mean(axis=0, keepdims=True)
    Kx = _center_gram(X @ X.T)
    Ky = _center_gram(Y @ Y.T)
    hsic = float((Kx * Ky).sum())
    norm = float(np.sqrt((Kx * Kx).sum()) * np.sqrt((Ky * Ky).sum()))
    return hsic / norm if norm > 0 else float('nan')


def get_layer_reps(model, X_tensor, batch=64):
    """Εξάγει τις αναπαραστάσεις τριών σταδίων μέσω forward hooks, ώστε να μη
    χρειάζεται να αναπαραχθεί η εσωτερική λογική διαστάσεων του μοντέλου."""
    store, reps = {}, {'cnn': [], 'transformer': [], 'pooled': []}

    def mk_hook(name):
        def hook(_m, _i, out):
            store[name] = out.detach().cpu().numpy()
        return hook

    handles = [model.cnn.register_forward_hook(mk_hook('cnn')),
               model.transformer.register_forward_hook(mk_hook('transformer'))]
    model.eval()
    try:
        with torch.no_grad():
            for i in range(0, len(X_tensor), batch):
                xb = X_tensor[i:i + batch].to(device)
                pooled = model.embed(xb).cpu().numpy()
                reps['cnn'].append(store['cnn'].reshape(len(xb), -1))
                tr = store['transformer']
                reps['transformer'].append(tr.reshape(len(xb), -1))
                reps['pooled'].append(pooled)
    finally:
        for h in handles:
            h.remove()
    return {k: np.concatenate(v) for k, v in reps.items()}


# --- Κοινό υποσύνολο δειγμάτων: ίδια παράθυρα για κάθε μοντέλο που συγκρίνεται ---
rng_cka = np.random.RandomState(0)
cka_idx = rng_cka.choice(xai_test_idx, size=min(500, len(xai_test_idx)), replace=False)
X_cka = X_data[cka_idx]

reps_s3 = get_layer_reps(xai_model, X_cka)
for k, v in reps_s3.items():
    print(f"Αναπαράσταση '{k}': {v.shape}")

# --- (α) Ομοιότητα μεταξύ επιπέδων του μοντέλου S3 ---
layer_names = ['cnn', 'transformer', 'pooled']
cka_layers = pd.DataFrame(index=layer_names, columns=layer_names, dtype=float)
for a, b in itertools.product(layer_names, layer_names):
    cka_layers.loc[a, b] = linear_cka(reps_s3[a], reps_s3[b])

print("\n===== CKA μεταξύ επιπέδων (μοντέλο S3) =====")
display(cka_layers.round(3))

fig, ax = plt.subplots(figsize=(4.5, 3.8))
sns.heatmap(cka_layers.astype(float), annot=True, fmt='.3f', cmap='viridis',
            vmin=0, vmax=1, ax=ax)
ax.set_title('Linear CKA μεταξύ επιπέδων (S3)')
plt.tight_layout(); plt.show()

# --- (β) Ομοιότητα ίδιου επιπέδου μεταξύ μοντέλων S1 και S3 ---
s1_key = next((k for k in saved_models if k[0] == 'S1'), None)
model_s1 = saved_models[s1_key][0] if s1_key else None

if model_s1 is None:
    print("\nΤο μοντέλο S1 δεν είναι διαθέσιμο· η σύγκριση S1 vs S3 παραλείπεται.")
else:
    reps_s1 = get_layer_reps(model_s1, X_cka)
    rows = [{'layer': ln, 'CKA_S1_vs_S3': linear_cka(reps_s1[ln], reps_s3[ln])}
            for ln in layer_names]
    cka_cross = pd.DataFrame(rows).set_index('layer')
    print("\n===== CKA ίδιου επιπέδου: μοντέλο S1 έναντι μοντέλου S3 =====")
    display(cka_cross.round(3))
    print("Χαμηλές τιμές σημαίνουν ότι τα δύο πρωτόκολλα οδηγούν σε ποιοτικά")
    print("διαφορετικές αναπαραστάσεις, παρότι η αρχιτεκτονική είναι ίδια.")

    fig, ax = plt.subplots(figsize=(5, 3.2))
    cka_cross['CKA_S1_vs_S3'].plot(kind='bar', ax=ax, color='#4C72B0')
    ax.set_ylim(0, 1); ax.set_ylabel('Linear CKA')
    ax.set_title('Ομοιότητα αναπαραστάσεων S1 vs S3')
    plt.xticks(rotation=0); plt.tight_layout(); plt.show()

    cka_cross.to_csv(f'{OUT_DIR}/cka_s1_vs_s3.csv')

cka_layers.to_csv(f'{OUT_DIR}/cka_between_layers.csv')

# =====================================================================
# LAYER-WISE PROBING
# Γραμμικοί ταξινομητές πάνω στις παγωμένες αναπαραστάσεις ΚΑΘΕ σταδίου,
# ώστε να φανεί σε ποιο σημείο του δικτύου η κάθε πληροφορία γίνεται
# γραμμικά αποκωδικοποιήσιμη. Συγκρίνεται η συναισθηματική ετικέτα με
# μεταβλητές που το μοντέλο ΔΕΝ θα έπρεπε να κωδικοποιεί (ταυτότητα
# εθελοντή, δοκιμασία), οπότε υψηλή ακρίβεια σε αυτές αποτελεί ένδειξη
# συντόμευσης μέσω Διαρροής Ταυτότητας Υποκειμένου.
# =====================================================================
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split as _tts

META = manifest            # <-- στα δύο CCNN notebooks άλλαξέ το σε: META = info

probe_targets = {'label (συναίσθημα)': 'label',
                 'subject_id (ταυτότητα)': 'subject_id',
                 'trial_id (δοκιμασία)': 'trial_id'}
if 'session_id' in META.columns and META['session_id'].nunique() > 1:
    probe_targets['session_id (συνεδρία)'] = 'session_id'

probe_rows = []
for stage, feats in reps_s3.items():
    if np.allclose(feats.std(axis=0), 0):
        print(f"  Το στάδιο '{stage}' έχει μηδενική διασπορά· παραλείπεται.")
        continue
    for label_name, col in probe_targets.items():
        y = META.loc[cka_idx, col].values
        classes, counts = np.unique(y, return_counts=True)
        if len(classes) < 2 or counts.min() < 2:
            continue
        strat = y if counts.min() >= 2 else None
        Xtr, Xte, ytr, yte = _tts(feats, y, test_size=0.3, random_state=0, stratify=strat)
        clf = LogisticRegression(max_iter=1000).fit(Xtr, ytr)
        acc = clf.score(Xte, yte)
        chance = counts.max() / counts.sum()      # πλειοψηφική κλάση ως γραμμή βάσης
        probe_rows.append({'stage': stage, 'target': label_name,
                           'probe_accuracy': acc, 'chance_level': chance,
                           'ratio': acc / chance if chance > 0 else float('nan')})

probe_df = pd.DataFrame(probe_rows)
if not probe_df.empty:
    table = probe_df.pivot(index='stage', columns='target', values='probe_accuracy')
    print("\n===== Layer-wise probing: ακρίβεια γραμμικού ταξινομητή ανά στάδιο =====")
    print(table.round(4))
    print("\nΛόγος προς τη γραμμή βάσης (τιμές >>1 υποδηλώνουν αποκωδικοποιήσιμη πληροφορία):")
    print(probe_df.pivot(index='stage', columns='target', values='ratio').round(2))
    probe_df.to_csv('layerwise_probing.csv', index=False)

    fig, ax = plt.subplots(figsize=(1.6 * len(table) + 3, 3.4))
    table.plot(kind='bar', ax=ax)
    ax.set_ylabel('Ακρίβεια probe'); ax.set_ylim(0, 1)
    ax.set_title('Layer-wise probing ανά στάδιο')
    plt.xticks(rotation=0); plt.legend(fontsize=8); plt.tight_layout(); plt.show()
```
