```python
import os
import glob

# 1. Βρίσκουμε πόσοι εθελοντές επεξεργάστηκαν επιτυχώς
processed_files = glob.glob('/content/DEAP-Emotion-Recognition/Data/PREPROCESSED_DEAP_DATA/Participant_*.npy')
num_participants = len(processed_files)
print(f" Βρέθηκαν {num_participants} έτοιμοι εθελοντές. Διορθώνω τον κώδικα...")

# 2. Διορθώνουμε το main.py για να σταματήσει να ζητάει 32 εθελοντές
file_path = '/content/DEAP-Emotion-Recognition/main.py'
with open(file_path, 'r') as file:
    code = file.read()

# Αντικαθιστούμε το hardcoded 33 με τον πραγματικό αριθμό (συν 1)
code = code.replace("range(1, 33)", f"range(1, {num_participants + 1})")

with open(file_path, 'w') as file:
    file.write(code)
print(" Ο κώδικας διορθώθηκε!")

# 3. Ξεκινάμε κατευθείαν την εκπαίδευση!
# ΠΡΟΣΟΧΗ: Βάλαμε 'False' για να μην ξανακάνει το FFT των 24 λεπτών!
!python main.py Valence NONE False 0
```
