```python
# ==========================================
# ΒΗΜΑ 1: ΕΓΚΑΤΑΣΤΑΣΗ ΒΙΒΛΙΟΘΗΚΩΝ
# ==========================================
!pip install pandas==2.2.2 pytorch-lightning mne lmdb einops xmltodict spectrum scikit-learn networkx mne-connectivity wfdb scipy
!pip install torch-scatter -f https://data.pyg.org/whl/torch-2.1.0+cu128.html
!pip install --no-deps torcheeg

# ==========================================
# ΒΗΜΑ 2: ΕΙΣΑΓΩΓΗ ΒΙΒΛΙΟΘΗΚΩΝ & ΣΥΝΔΕΣΗ DRIVE
# ==========================================
import os
import torch
import numpy as np
import matplotlib.pyplot as plt
from sklearn.manifold import TSNE
from torch.utils.data import DataLoader
import pytorch_lightning as pl
from google.colab import drive

# Εισαγωγή TorchEEG modules
from torcheeg import transforms
from torcheeg.datasets import DEAPDataset
from torcheeg.datasets.constants.emotion_recognition.deap import DEAP_CHANNEL_LOCATION_DICT
from torcheeg.model_selection import KFoldGroupbyTrial
from torcheeg.models import CCNN
from torcheeg.trainers import ClassifierTrainer

# Σύνδεση με το Google Drive
drive.mount('/content/drive')

# Ορισμός μονοπατιών (Προσαρμοσμένο για το "DEAP.mini")
root_data_path = '/content/drive/MyDrive/DEAP.mini'
io_path = './eeg_io'

# ==========================================
# ΒΗΜΑ 3: ΠΡΟΕΤΟΙΜΑΣΙΑ ΔΕΔΟΜΕΝΩΝ
# ==========================================
print("\n--- Φόρτωση του DEAP Dataset ---")
dataset = DEAPDataset(
    io_path=io_path,
    root_path=root_data_path,
    offline_transform=transforms.Compose([
        transforms.BandDifferentialEntropy(),
        transforms.ToGrid(DEAP_CHANNEL_LOCATION_DICT)
    ]),
    online_transform=transforms.Compose([
        transforms.ToTensor()
    ]),
    label_transform=transforms.Compose([
        transforms.Select('valence'),
        transforms.Binary(5.0)
    ]),
    chunk_size=128,
    num_worker=0  # <-- ΔΙΟΡΘΩΣΗ: 0 για να μην καταρρέει η μνήμη του Colab
)

# ==========================================
# ΒΗΜΑ 4: ΔΙΑΧΩΡΙΣΜΟΣ & ΦΟΡΤΩΣΗ (DATALOADERS)
# ==========================================
print("\n--- Διαχωρισμός δεδομένων (5-Fold Cross Validation) ---")
k_fold = KFoldGroupbyTrial(n_splits=5, split_path='./splits')
splits = list(k_fold.split(dataset))
train_dataset, val_dataset = splits[0]

# <-- ΔΙΟΡΘΩΣΗ: num_workers=0 και εδώ για ασφάλεια
train_loader = DataLoader(train_dataset, batch_size=64, shuffle=True, num_workers=0)
val_loader = DataLoader(val_dataset, batch_size=64, shuffle=False, num_workers=0)

# ==========================================
# ΒΗΜΑ 5: ΕΚΠΑΙΔΕΥΣΗ ΜΟΝΤΕΛΟΥ (CCNN)
# ==========================================
print("\n--- Αρχικοποίηση του CCNN ---")
model = CCNN(num_classes=2, in_channels=4, grid_size=(9, 9))
trainer = ClassifierTrainer(model=model, num_classes=2, lr=1e-4, weight_decay=1e-4)

pl_trainer = pl.Trainer(
    max_epochs=20,
    accelerator="auto",
    devices=1
)

# ==========================================
# ΒΗΜΑ 6: ΑΠΟΘΗΚΕΥΣΗ/ΦΟΡΤΩΣΗ ΚΑΙ ΕΚΤΕΛΕΣΗ
# ==========================================
model_save_path = '/content/drive/MyDrive/DEAP.mini/ccnn_trained_model.pth'

if os.path.exists(model_save_path):
    print("\n--- Βρέθηκε ήδη εκπαιδευμένο μοντέλο! Φόρτωση... ---")
    model.load_state_dict(torch.load(model_save_path))
    print("Η φόρτωση ολοκληρώθηκε!")
else:
    print("\n--- Δεν βρέθηκε έτοιμο μοντέλο. Ξεκινάει η εκπαίδευση... ---")
    pl_trainer.fit(trainer, train_loader, val_loader)
    torch.save(model.state_dict(), model_save_path)
    print(f"\nΤο μοντέλο αποθηκεύτηκε με επιτυχία στο Drive: {model_save_path}")

# ==========================================
# ΒΗΜΑ 7: ΑΝΑΠΑΡΑΓΩΓΗ FIGURE 5 (Benchmark Data)
# ==========================================
print("\n--- Δημιουργία Γραφήματος: Ταχύτητα & Μνήμη (Figure 5) ---")
workers = [2, 4, 8, 16, 32, 64]
deap_time = [2947, 1505, 767, 430, 264, 258]
seed_time = [5927, 3132, 1767, 989, 697, 484]
datasets_names = ['DEAP', 'SEED', 'SEED-IV', 'DREAMER', 'AMIGOS', 'MAHNOB']
ram_w_io = [439, 511, 513, 420, 382, 402]
ram_wo_io = [2951, 15331, 15224, 2433, 1074, 902]

plt.figure(figsize=(14, 5))

plt.subplot(1, 2, 1)
plt.plot(workers, deap_time, marker='o', color='indianred', label='DEAP', markerfacecolor='white')
plt.plot(workers, seed_time, marker='s', color='slateblue', label='SEED', markerfacecolor='white')
plt.xticks(workers)
plt.xlabel('Number of Workers')
plt.ylabel('Runtime (s)')
plt.legend()
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.title('(a) Time costs on benchmark datasets')

x = np.arange(len(datasets_names))
width = 0.35
plt.subplot(1, 2, 2)
plt.bar(x - width/2, ram_w_io, width, label='w/ window-centric IO', color='indianred')
plt.bar(x + width/2, ram_wo_io, width, label='w/o window-centric IO', color='slateblue')
plt.xlabel('Dataset')
plt.ylabel('RAM Consumption (MB)')
plt.xticks(x, datasets_names)
plt.legend()
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.title('(b) Peak random-access memory (RAM) usage')

plt.tight_layout()
plt.show()

# ==========================================
# ΒΗΜΑ 8: ΑΝΑΠΑΡΑΓΩΓΗ FIGURE 4 (t-SNE Visualization)
# ==========================================
print("\n--- Δημιουργία Γραφήματος: Οπτικοποίηση Χαρακτηριστικών με t-SNE (Figure 4) ---")
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model.to(device)
model.eval()

all_features = []
all_labels = []

with torch.no_grad():
    for i, batch in enumerate(val_loader):
        x, y = batch
        x = x.to(device)

        features = model(x)
        all_features.append(features.cpu().numpy())
        all_labels.append(y.cpu().numpy())

        if i >= 10:
            break

all_features = np.concatenate(all_features, axis=0)
all_labels = np.concatenate(all_labels, axis=0)

print("Υπολογισμός t-SNE (μείωση σε 2 διαστάσεις)...")
tsne = TSNE(n_components=2, random_state=42, perplexity=30)
features_2d = tsne.fit_transform(all_features)

plt.figure(figsize=(8, 8))
idx_low = (all_labels == 0)
idx_high = (all_labels == 1)

plt.scatter(features_2d[idx_low, 0], features_2d[idx_low, 1],
            c='blue', marker='o', alpha=0.6, label='Low Valence (Negative)')
plt.scatter(features_2d[idx_high, 0], features_2d[idx_high, 1],
            c='red', marker='^', alpha=0.6, label='High Valence (Positive)')

plt.title("t-SNE Visualization of EEG Features (DEAP Dataset)")
plt.legend(loc='lower center', bbox_to_anchor=(0.5, -0.15), ncol=2)
plt.axis('off')
plt.show()

print("\nΗ διαδικασία ολοκληρώθηκε με επιτυχία!")
```
