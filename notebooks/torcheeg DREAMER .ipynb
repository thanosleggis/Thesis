```python
# Εγκατάσταση -- ΚΑΘΕ πακέτο σε ξεχωριστή εντολή. Το torcheeg.datasets κάνει eager
# import ΟΛΩΝ των datasets submodules (ακόμα κι αν δεν τα χρησιμοποιούμε, π.χ.
# MAHNOB, SEED), άρα χρειάζονται όλα τα δηλωμένα dependencies εγκατεστημένα.
import sys, types
if 'torch_scatter' not in sys.modules:
    _stub = types.ModuleType('torch_scatter')
    def _not_implemented(*args, **kwargs):
        raise NotImplementedError("torch_scatter stub -- δεν χρησιμοποιείται (μόνο CCNN)")
    _stub.scatter_add = _not_implemented
    sys.modules['torch_scatter'] = _stub

try:
    import torcheeg
    from torcheeg.datasets import DREAMERDataset
    from torcheeg.models import CCNN
    print("Το torcheeg είναι ήδη εγκατεστημένο.")
except ImportError:
    print("Εγκατάσταση torcheeg + dependencies...")
    import subprocess, sys
    for pkg in ["torcheeg --no-deps", "mne", "mne-connectivity", "captum", "lmdb",
                "scikit-learn", "seaborn", "pandas", "scipy", "einops", "xmltodict",
                "wfdb", "xlrd", "torchmetrics", "spectrum", "pytorch-lightning"]:
        subprocess.run([sys.executable, "-m", "pip", "install", "-q"] + pkg.split(), check=True)
    print("Η εγκατάσταση ολοκληρώθηκε.")


"""
CCNN (TorchEEG) / DREAMER -- ΔΙΟΡΘΩΜΕΝΟ, ΕΝΙΑΙΟ SCRIPT

Διορθώνει τα δύο επιβεβαιωμένα προβλήματα:
1. "Dead network": έλειπε το MeanStdNormalize μετά το BandDifferentialEntropy --
   τα ακατέργαστα χαρακτηριστικά DE είναι εκτός [-1,1] και κορεσμένα, οδηγώντας
   σε μηδενικά gradients (Balanced Accuracy = 50.00% ± 0.00% σε ΟΛΑ τα fold,
   ακόμα και το train accuracy κολλημένο στο majority baseline).
2. Fine-tuning leakage: το target subject επιλεγόταν αλφαβητικά/αριθμητικά πρώτο
   (sorted(...)[0]) αντί να προέρχεται από τα πραγματικά held-out υποκείμενα του
   καλύτερου fold -- τώρα επιλέγεται ρητά από αυτά.
"""

import os
import zipfile
import glob
import copy
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, Subset

# ΔΙΟΡΘΩΣΗ: το torcheeg.models κάνει eager import ΟΛΩΝ των αρχιτεκτονικών του (ακόμα
# κι αν θέλουμε μόνο το CCNN) -- και το RGNN (graph-based μοντέλο, άσχετο με εμάς)
# απαιτεί το πακέτο torch_scatter, που δεν εγκαθίσταται αξιόπιστα εδώ. Το "στήνουμε"
# ως ψεύτικο (stub) module πριν από οποιοδήποτε import του torcheeg.models, ώστε το
# RGNN να "βρει" κάτι να εισάγει και να μην σκάει -- δεν το χρησιμοποιούμε ποτέ.
import sys, types
if 'torch_scatter' not in sys.modules:
    _stub = types.ModuleType('torch_scatter')
    def _not_implemented(*args, **kwargs):
        raise NotImplementedError("torch_scatter stub -- δεν χρησιμοποιείται (μόνο CCNN)")
    _stub.scatter_add = _not_implemented
    sys.modules['torch_scatter'] = _stub

from torcheeg.datasets import DREAMERDataset
from torcheeg.datasets.constants.emotion_recognition.dreamer import DREAMER_CHANNEL_LOCATION_DICT
from torcheeg.models import CCNN
from torcheeg import transforms

from sklearn.model_selection import GroupKFold
from sklearn.metrics import (accuracy_score, balanced_accuracy_score, f1_score,
                              precision_recall_fscore_support, matthews_corrcoef,
                              roc_auc_score, cohen_kappa_score, confusion_matrix)

import matplotlib.pyplot as plt
import seaborn as sns
import mne

try:
    from google.colab import drive
    IN_COLAB = True
except ImportError:
    IN_COLAB = False

from captum.attr import Occlusion, IntegratedGradients

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Εκτέλεση σε: {device}")

# ================== ΡΥΘΜΙΣΕΙΣ ==================
BATCH_SIZE = 64
EPOCHS = 10
LR = 1e-4
N_SPLITS = 5

FINE_TUNE_EPOCHS = 5
FINE_TUNE_LR = 1e-5
FINE_TUNE_TRIAL_SPLIT = 0.7   # 70% trials για fine-tuning, 30% για τελικό έλεγχο

VALENCE_THRESHOLD = 3.0

DREAMER_CHANNELS = ["AF3", "F7", "F3", "FC5", "T7", "P7", "O1", "O2",
                    "P8", "T8", "FC6", "F4", "F8", "AF4"]
REGION_MAP = {
    "AF3": "frontal-left", "AF4": "frontal-right",
    "F7": "frontal-left", "F3": "frontal-left",
    "F4": "frontal-right", "F8": "frontal-right",
    "FC5": "frontal-central-left", "FC6": "frontal-central-right",
    "T7": "temporal-left", "T8": "temporal-right",
    "P7": "parietal-left", "P8": "parietal-right",
    "O1": "occipital-left", "O2": "occipital-right",
}


# ================== ΒΗΜΑ 1: Εντοπισμός DREAMER.mat (Colab ή Kaggle) ==================
MAT_FILE_PATH = None

# --- Kaggle: το αρχείο έρχεται συνήθως ήδη έτοιμο μέσα σε /kaggle/input/... ---
for root, _, files in os.walk('/kaggle/input'):
    for f in files:
        if f.endswith('.mat') and 'DREAMER' in f.upper():
            MAT_FILE_PATH = os.path.join(root, f)
            break
    if MAT_FILE_PATH:
        break

# --- Colab: DREAMER.zip στο Drive, χρειάζεται αποσυμπίεση ---
if MAT_FILE_PATH is None and IN_COLAB:
    if not os.path.exists('/content/drive'):
        drive.mount('/content/drive')
    ZIP_PATH = '/content/drive/MyDrive/DREAMER.zip'
    EXTRACT_TO = './dreamer_raw'
    if not os.path.exists(EXTRACT_TO):
        if not os.path.exists(ZIP_PATH):
            raise FileNotFoundError(f"Δεν βρέθηκε το {ZIP_PATH}.")
        print("Αποσυμπίεση...")
        with zipfile.ZipFile(ZIP_PATH, 'r') as zf:
            zf.extractall(EXTRACT_TO)
    mat_files = glob.glob(f"{EXTRACT_TO}/**/*.mat", recursive=True)
    if mat_files:
        MAT_FILE_PATH = mat_files[0]

if MAT_FILE_PATH is None:
    raise FileNotFoundError(
        "Δεν βρέθηκε το DREAMER.mat -- ούτε σε /kaggle/input, ούτε σε Colab Drive. "
        "Αν τρέχεις σε Kaggle, βεβαιώσου ότι έχεις προσθέσει το dataset στο notebook (Add Input)."
    )
print(f"mat_path = {MAT_FILE_PATH}")


# ================== ΒΗΜΑ 2: Φόρτωση dataset -- ΜΕ MeanStdNormalize (η διόρθωση) ==================
IO_PATH = './dreamer_processed_normalized'   # νέο io_path -- ώστε να ΜΗΝ διαβάσει τυχόν παλιά,
                                              # μη-κανονικοποιημένη cache από πριν
dataset = DREAMERDataset(
    io_path=IO_PATH,
    mat_path=MAT_FILE_PATH,
    offline_transform=transforms.Compose([
        transforms.BandDifferentialEntropy(),   # 4 ζώνες: Theta, Alpha, Beta, Gamma
        transforms.MeanStdNormalize(),          # ΔΙΟΡΘΩΣΗ: z-score στα DE χαρακτηριστικά --
                                                 # χωρίς αυτό το δίκτυο μένει "νεκρό" (μηδενικά
                                                 # gradients, Balanced Accuracy=50.00%±0.00% παντού)
        transforms.ToGrid(DREAMER_CHANNEL_LOCATION_DICT),
    ]),
    online_transform=transforms.Compose([
        transforms.ToTensor()
    ]),
    label_transform=transforms.Compose([
        transforms.Select('valence'),
        transforms.Binary(VALENCE_THRESHOLD)
    ]),
    num_worker=2
)
print(f"Dataset έτοιμο: {len(dataset)} παράθυρα")

info = dataset.info
print("Στήλες dataset.info:", list(info.columns))
print(f"Μοναδικά υποκείμενα: {info['subject_id'].nunique()}")

# ΔΙΟΡΘΩΣΗ: το dataset.info κρατάει το ΩΜΟ valence score, όχι την ήδη-binarized ετικέτα
# (το Binary(threshold) τρέχει μόνο μέσα στο __getitem__, όχι στο .info). Υπολογίζουμε
# εδώ ρητά την ίδια δυαδικοποίηση για χρήση σε manifest/ισορροπία/ομαδοποίηση.
info = info.copy()
info['label'] = (info['valence'] >= VALENCE_THRESHOLD).astype(int)

balance = info.groupby('subject_id')['label'].mean() * 100
plt.figure(figsize=(10, 4))
balance.sort_values().plot(kind='bar', color='#4C72B0')
plt.axhline(50, color='red', linestyle='--', linewidth=1, label='Ισορροπημένο (50%)')
plt.ylabel('% High Valence'); plt.xlabel('subject_id')
plt.title('Ισορροπία κλάσης Valence ανά υποκείμενο (DREAMER)')
plt.legend(); plt.tight_layout(); plt.show()



# ================== ΒΗΜΑ 3: Μετρικές ==================
def full_metrics_report(y_true, y_pred, y_prob_pos, title=""):
    prec, rec, f1c, _ = precision_recall_fscore_support(y_true, y_pred, labels=[0, 1], zero_division=0)
    vals, counts = np.unique(y_true, return_counts=True)
    majority_baseline = counts.max() / counts.sum()
    acc = accuracy_score(y_true, y_pred)
    bal_acc = balanced_accuracy_score(y_true, y_pred)
    macro_f1 = f1_score(y_true, y_pred, average='macro', zero_division=0)
    kappa = cohen_kappa_score(y_true, y_pred)
    mcc = matthews_corrcoef(y_true, y_pred)
    try:
        auc = roc_auc_score(y_true, y_prob_pos)
    except ValueError:
        auc = float('nan')

    print(f"\n===== {title} =====")
    print(f"Accuracy           : {acc:.2%}")
    print(f"Majority baseline  : {majority_baseline:.2%}  <-- αν το Accuracy δεν το ξεπερνά καθαρά, το μοντέλο δεν έμαθε")
    print(f"Balanced Accuracy  : {bal_acc:.2%}")
    print(f"Macro F1           : {macro_f1:.2%}")
    print(f"Cohen's kappa      : {kappa:.4f}")
    print(f"MCC                : {mcc:.4f}")
    print(f"ROC-AUC            : {auc:.4f}")
    print(f"Precision (0/1)    : {prec[0]:.3f} / {prec[1]:.3f}")
    print(f"Recall    (0/1)    : {rec[0]:.3f} / {rec[1]:.3f}")
    print(f"F1        (0/1)    : {f1c[0]:.3f} / {f1c[1]:.3f}")
    return {
        'accuracy': acc, 'majority_baseline': majority_baseline, 'balanced_accuracy': bal_acc,
        'macro_f1': macro_f1, 'cohen_kappa': kappa, 'mcc': mcc, 'roc_auc': auc,
    }


def plot_confusion(y_true, y_pred, title):
    cm = confusion_matrix(y_true, y_pred, labels=[0, 1])
    plt.figure(figsize=(3.5, 3))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=['Low', 'High'], yticklabels=['Low', 'High'])
    plt.xlabel('Predicted'); plt.ylabel('True'); plt.title(title, fontsize=10)
    plt.show()
    return cm


# ================== ΒΗΜΑ 4: Εκπαίδευση/αξιολόγηση ενός fold ==================
def train_one_fold(train_idx, test_idx, n_epochs=EPOCHS, verbose=True):
    model = CCNN(num_classes=2, in_channels=4, grid_size=(9, 9)).to(device)
    optimizer = optim.Adam(model.parameters(), lr=LR)
    criterion = nn.CrossEntropyLoss()

    train_loader = DataLoader(Subset(dataset, train_idx.tolist()), batch_size=BATCH_SIZE, shuffle=True)
    test_loader = DataLoader(Subset(dataset, test_idx.tolist()), batch_size=BATCH_SIZE, shuffle=False)

    for epoch in range(n_epochs):
        model.train()
        running_loss, correct, total = 0.0, 0, 0
        for X, y in train_loader:
            X, y = X.to(device), y.to(device)
            optimizer.zero_grad()
            out = model(X)
            loss = criterion(out, y)
            loss.backward()
            optimizer.step()
            running_loss += loss.item() * X.size(0)
            correct += (out.argmax(1) == y).sum().item()
            total += y.size(0)
        train_loss, train_acc = running_loss / total, 100.0 * correct / total

        model.eval()
        val_true, val_pred = [], []
        with torch.no_grad():
            for X, y in test_loader:
                X = X.to(device)
                out = model(X)
                val_pred.append(out.argmax(1).cpu().numpy())
                val_true.append(y.numpy())
        val_true = np.concatenate(val_true); val_pred = np.concatenate(val_pred)
        val_bal_acc = balanced_accuracy_score(val_true, val_pred)
        if verbose:
            print(f"Epoch {epoch+1:02d}/{n_epochs} | Train Loss {train_loss:.4f} Acc {train_acc:.2f}% "
                  f"| Val Bal.Acc {val_bal_acc:.2%}")

    model.eval()
    all_true, all_pred, all_prob = [], [], []
    with torch.no_grad():
        for X, y in test_loader:
            X = X.to(device)
            out = torch.softmax(model(X), dim=1).cpu().numpy()
            all_true.append(y.numpy())
            all_pred.append(out.argmax(1))
            all_prob.append(out[:, 1])
    y_true = np.concatenate(all_true); y_pred = np.concatenate(all_pred); y_prob = np.concatenate(all_prob)
    return model, y_true, y_pred, y_prob


# ================== ΒΗΜΑ 5: 5-Fold Subject-Independent Cross-Validation ==================
indices = info.index.values
groups = info['subject_id'].values

gkf = GroupKFold(n_splits=N_SPLITS)
fold_results = []
best_val_bal_acc = -1.0
best_model_state = None
best_fold_test_subjects = None

# Αποθηκεύονται για τις αναλύσεις αναπαραστάσεων και τον έλεγχο Wilcoxon
fold_states = {}
per_subject_acc = {'S1': {}, 'S3': {}}
s3_true_all, s3_pred_all = [], []

for fold, (train_idx, test_idx) in enumerate(gkf.split(indices, groups=groups)):
    train_subjects = sorted(info.iloc[train_idx]['subject_id'].unique().tolist())
    test_subjects = sorted(info.iloc[test_idx]['subject_id'].unique().tolist())
    print(f"\n########## FOLD {fold+1}/{N_SPLITS} ##########")
    print(f"Υποκείμενα στο train: {train_subjects}")
    print(f"Υποκείμενα στο test : {test_subjects}")

    model, y_true, y_pred, y_prob = train_one_fold(indices[train_idx], indices[test_idx])
    report = full_metrics_report(y_true, y_pred, y_prob, title=f"CCNN/DREAMER - Fold {fold+1} (subject-independent)")
    report['fold'] = fold + 1
    fold_results.append(report)

    # ΔΙΟΡΘΩΣΗ (μέρος 1/2): κρατάμε ΠΟΙΑ υποκείμενα ήταν πραγματικά held-out στο
    # καλύτερο fold, ώστε το fine-tuning παρακάτω να διαλέξει έγκυρο υποκείμενο.
    # Τα βάρη ΚΑΘΕ fold (όχι μόνο του καλύτερου) χρειάζονται για τη σύγκριση CKA
    fold_states[fold + 1] = copy.deepcopy(model.state_dict())

    # Ακρίβεια ανά υποκείμενο, για τον έλεγχο Wilcoxon S1 vs S3
    s3_true_all.append(y_true)
    s3_pred_all.append(y_pred)
    subj_ids = info.iloc[test_idx]['subject_id'].values
    for subj in np.unique(subj_ids):
        m = subj_ids == subj
        per_subject_acc['S3'][subj] = float((y_true[m] == y_pred[m]).mean())

    if report['balanced_accuracy'] > best_val_bal_acc:
        best_val_bal_acc = report['balanced_accuracy']
        best_model_state = copy.deepcopy(model.state_dict())
        best_fold_test_subjects = test_subjects

results_df = pd.DataFrame(fold_results)
print(f"\nΜέσος όρος Balanced Accuracy: {results_df['balanced_accuracy'].mean():.2%} "
      f"± {results_df['balanced_accuracy'].std():.2%}")
print(f"Μέσος όρος Macro F1: {results_df['macro_f1'].mean():.2%} ± {results_df['macro_f1'].std():.2%}")


# ================== ΒΗΜΑ 5β: Πρωτόκολλο τριών επιπέδων (S1 / S2) ==================
# Το 5-fold GroupKFold ανά υποκείμενο παραπάνω αντιστοιχεί λειτουργικά στο S3.
# Εδώ προστίθενται τα δύο ηπιότερα επίπεδα, ώστε να μετρηθεί η πτώση απόδοσης που
# οφείλεται στη Διαρροή Ταυτότητας Υποκειμένου:
#   S1 -- τυχαίος διαχωρισμός σε επίπεδο παραθύρου (80/20)
#   S2 -- GroupKFold ομαδοποιημένο ανά υποκείμενο ΚΑΙ δοκιμασία
from sklearn.model_selection import train_test_split
from scipy.stats import wilcoxon

protocol_results = []

print("\n" + "=" * 60)
print("ΕΠΙΠΕΔΟ S1 -- τυχαίος διαχωρισμός παραθύρων 80/20")
print("=" * 60)
s1_train, s1_test = train_test_split(
    indices, test_size=0.2, random_state=42, stratify=info['label'].values)

model_s1, y1_true, y1_pred, y1_prob = train_one_fold(s1_train, s1_test)
rep = full_metrics_report(y1_true, y1_pred, y1_prob, title="CCNN/DREAMER - S1 (window-level)")
rep.update({'split_level': 'S1', 'fold': 'S1_full'})
protocol_results.append(rep)
s1_state = copy.deepcopy(model_s1.state_dict())

s1_subjects = info.loc[s1_test, 'subject_id'].values
for subj in np.unique(s1_subjects):
    m = s1_subjects == subj
    per_subject_acc['S1'][subj] = float((y1_true[m] == y1_pred[m]).mean())

print("\n" + "=" * 60)
print("ΕΠΙΠΕΔΟ S2 -- GroupKFold ανά υποκείμενο και δοκιμασία")
print("=" * 60)
trial_groups = (info['subject_id'].astype(str) + '_' + info['trial_id'].astype(str)).values

for fold, (tr, te) in enumerate(GroupKFold(n_splits=N_SPLITS).split(indices, groups=trial_groups)):
    print(f"\n--- S2 fold {fold + 1}/{N_SPLITS} ---")
    _m, yt, yp, ypr = train_one_fold(indices[tr], indices[te])
    rep = full_metrics_report(yt, yp, ypr, title=f"CCNN/DREAMER - S2 fold {fold + 1}")
    rep.update({'split_level': 'S2', 'fold': f'S2_fold{fold}'})
    protocol_results.append(rep)

# Το S3 επαναχρησιμοποιείται από τον βρόχο του ΒΗΜΑΤΟΣ 5 -- δεν ξαναεκπαιδεύεται
for r in fold_results:
    rep = dict(r)
    rep.update({'split_level': 'S3', 'fold': f"S3_fold{r['fold']}"})
    protocol_results.append(rep)

s3_true_all = np.concatenate(s3_true_all)
s3_pred_all = np.concatenate(s3_pred_all)

protocol_df = pd.DataFrame(protocol_results)
protocol_summary = (protocol_df.groupby('split_level')
                    .agg(balanced_accuracy_mean=('balanced_accuracy', 'mean'),
                         balanced_accuracy_std=('balanced_accuracy', 'std'),
                         macro_f1_mean=('macro_f1', 'mean'),
                         majority_baseline_mean=('majority_baseline', 'mean'))
                    .reset_index())
print("\n===== Συγκεντρωτικά αποτελέσματα ανά επίπεδο =====")
print(protocol_summary.round(4))

s1_ba = protocol_summary.loc[protocol_summary.split_level == 'S1', 'balanced_accuracy_mean'].values
s3_ba = protocol_summary.loc[protocol_summary.split_level == 'S3', 'balanced_accuracy_mean'].values
if len(s1_ba) and len(s3_ba):
    print(f"\nΠτώση S1→S3 (balanced accuracy): {(s1_ba[0] - s3_ba[0]):.2%}")

rng_b = np.random.RandomState(0)
boot = []
for _ in range(1000):
    ii = rng_b.randint(0, len(s3_true_all), len(s3_true_all))
    boot.append(balanced_accuracy_score(s3_true_all[ii], s3_pred_all[ii]))
lo, hi = np.percentile(boot, [2.5, 97.5])
print(f"S3 balanced accuracy: {np.mean(boot):.2%} (95% CI [{lo:.2%}, {hi:.2%}])")

common = sorted(set(per_subject_acc['S1']) & set(per_subject_acc['S3']))
if len(common) >= 5:
    stat, p = wilcoxon([per_subject_acc['S1'][s] for s in common],
                       [per_subject_acc['S3'][s] for s in common])
    print(f"\nWilcoxon S1 vs S3 (n={len(common)} υποκείμενα): stat={stat:.3f}, p={p:.4f}")
else:
    print(f"\nΠολύ λίγα κοινά υποκείμενα ({len(common)}) για Wilcoxon.")

fig, axes = plt.subplots(1, 2, figsize=(7.5, 3.2))
for ax, (lvl, yt, yp) in zip(axes, [('S1', y1_true, y1_pred), ('S3', s3_true_all, s3_pred_all)]):
    sns.heatmap(confusion_matrix(yt, yp, labels=[0, 1]), annot=True, fmt='d', cmap='Blues',
                ax=ax, xticklabels=['Low', 'High'], yticklabels=['Low', 'High'])
    ax.set_title(lvl); ax.set_xlabel('Predicted'); ax.set_ylabel('True')
plt.tight_layout(); plt.show()



# ================== ΒΗΜΑ 6: Subject-Specific Fine-Tuning (ΔΙΟΡΘΩΜΕΝΟ) ==================
# ΔΙΟΡΘΩΣΗ (μέρος 2/2): το target subject επιλέγεται ΤΩΡΑ από τα πραγματικά held-out
# υποκείμενα του καλύτερου fold (best_fold_test_subjects) -- όχι sorted(...)[0], που
# μπορούσε να διαλέξει υποκείμενο ήδη μέσα στο training set του καλύτερου μοντέλου.
TARGET_SUBJECT = best_fold_test_subjects[0]
print(f"\nΣτοχευμένο υποκείμενο για fine-tuning (πραγματικά held-out από το καλύτερο fold): {TARGET_SUBJECT}")

subj_mask = info['subject_id'] == TARGET_SUBJECT
subj_idx = info.index[subj_mask].values
subj_trials = info.loc[subj_mask, 'trial_id'].values
unique_trials = np.unique(subj_trials)

rng = np.random.RandomState(42)
rng.shuffle(unique_trials)
split_point = max(1, int(FINE_TUNE_TRIAL_SPLIT * len(unique_trials)))
ft_trials, test_trials = unique_trials[:split_point], unique_trials[split_point:]
ft_idx = subj_idx[np.isin(subj_trials, ft_trials)]
subj_test_idx = subj_idx[np.isin(subj_trials, test_trials)]
print(f"Trials fine-tune: {len(ft_trials)} ({len(ft_idx)} παράθυρα) | "
      f"Trials test: {len(test_trials)} ({len(subj_test_idx)} παράθυρα)")

ft_model = CCNN(num_classes=2, in_channels=4, grid_size=(9, 9)).to(device)
ft_model.load_state_dict(best_model_state)

ft_loader = DataLoader(Subset(dataset, ft_idx.tolist()), batch_size=BATCH_SIZE, shuffle=True)
subj_test_loader = DataLoader(Subset(dataset, subj_test_idx.tolist()), batch_size=BATCH_SIZE, shuffle=False)

ft_model.eval()
all_true, all_pred, all_prob = [], [], []
with torch.no_grad():
    for X, y in subj_test_loader:
        X = X.to(device)
        out = torch.softmax(ft_model(X), dim=1).cpu().numpy()
        all_true.append(y.numpy()); all_pred.append(out.argmax(1)); all_prob.append(out[:, 1])
y_true_before = np.concatenate(all_true); y_pred_before = np.concatenate(all_pred); y_prob_before = np.concatenate(all_prob)
report_before = full_metrics_report(y_true_before, y_pred_before, y_prob_before,
                                     title=f"Πριν το fine-tuning (subject {TARGET_SUBJECT})")

optimizer_ft = optim.Adam(ft_model.parameters(), lr=FINE_TUNE_LR)
criterion = nn.CrossEntropyLoss()
for epoch in range(FINE_TUNE_EPOCHS):
    ft_model.train()
    running_loss, correct, total = 0.0, 0, 0
    for X, y in ft_loader:
        X, y = X.to(device), y.to(device)
        optimizer_ft.zero_grad()
        out = ft_model(X)
        loss = criterion(out, y)
        loss.backward()
        optimizer_ft.step()
        running_loss += loss.item() * X.size(0)
        correct += (out.argmax(1) == y).sum().item()
        total += y.size(0)
    print(f"Fine-tune epoch {epoch+1}/{FINE_TUNE_EPOCHS} | Loss {running_loss/total:.4f} "
          f"| Train Acc {100.0*correct/total:.2f}%")

ft_model.eval()
all_true, all_pred, all_prob = [], [], []
with torch.no_grad():
    for X, y in subj_test_loader:
        X = X.to(device)
        out = torch.softmax(ft_model(X), dim=1).cpu().numpy()
        all_true.append(y.numpy()); all_pred.append(out.argmax(1)); all_prob.append(out[:, 1])
y_true_after = np.concatenate(all_true); y_pred_after = np.concatenate(all_pred); y_prob_after = np.concatenate(all_prob)
report_after = full_metrics_report(y_true_after, y_pred_after, y_prob_after,
                                    title=f"Μετά το fine-tuning (subject {TARGET_SUBJECT})")

print(f"\nBalanced Accuracy πριν: {report_before['balanced_accuracy']:.2%} "
      f"-> μετά: {report_after['balanced_accuracy']:.2%}")
if len(np.unique(y_pred_after)) == 1:
    print("ΠΡΟΣΟΧΗ: το μοντέλο προβλέπει ΜΟΝΟ 1 κλάση μετά το fine-tuning (Majority Class Collapse).")



# ================== ΒΗΜΑ 7: XAI -- Occlusion + Integrated Gradients ==================
def attribution_to_electrode_scores(attr_grid, channel_location_dict, channels=DREAMER_CHANNELS):
    """Απομονώνει τα σκορ σημαντικότητας ΜΟΝΟ στις πραγματικές θέσεις ηλεκτροδίων
    μέσα στο πλέγμα 9x9 -- αγνοεί τα κενά (μη-ηλεκτρόδιο) κελιά του grid, ώστε να μην
    γίνεται αφηρημένη ερμηνεία πάνω σε μη υπαρκτά σημεία."""
    scores = {}
    for ch in channels:
        if ch in channel_location_dict:
            r, c = channel_location_dict[ch]
            # attr_grid: (bands, H, W) -- αθροίζουμε |attribution| σε όλες τις 4 ζώνες
            scores[ch] = float(np.abs(attr_grid[:, r, c]).sum())
    return scores


def build_montage_info(channels=DREAMER_CHANNELS):
    montage = mne.channels.make_standard_montage('standard_1020')
    info_mne = mne.create_info(ch_names=channels, sfreq=128, ch_types='eeg')
    info_mne.set_montage(montage)
    return info_mne


def plot_scalp_map(scores_dict, title, channels=DREAMER_CHANNELS):
    values = np.array([scores_dict.get(ch, 0.0) for ch in channels])
    fig, ax = plt.subplots(figsize=(4, 4))
    mne.viz.plot_topomap(values, MNE_INFO, axes=ax, show=False, cmap='RdBu_r', contours=4)
    ax.set_title(title, fontsize=10)
    plt.show()


def top5_by_region(scores_dict):
    df = pd.DataFrame([{'channel': ch, 'value': v, 'region': REGION_MAP.get(ch, 'unknown')}
                        for ch, v in scores_dict.items()])
    return df.sort_values('value', ascending=False)


MNE_INFO = build_montage_info()

# Πραγματικό δείγμα ελέγχου (από το test set του καλύτερου fold) -- στοχεύουμε High Valence
xai_model = CCNN(num_classes=2, in_channels=4, grid_size=(9, 9)).to(device)
xai_model.load_state_dict(best_model_state)
xai_model.eval()

high_valence_idx = None
for X, y in subj_test_loader:
    pos = (y == 1).nonzero(as_tuple=True)[0]
    if len(pos) > 0:
        high_valence_idx = X[pos[0]:pos[0] + 1].to(device)
        break
if high_valence_idx is None:
    # fallback: πάρε απλά το πρώτο δείγμα του test set αν δεν υπάρχει High Valence εκεί
    X, y = next(iter(subj_test_loader))
    high_valence_idx = X[0:1].to(device)

sample = high_valence_idx
target_class = 1  # High Valence

print("Υπολογισμός Occlusion attribution σε πραγματικό δείγμα...")
occlusion = Occlusion(xai_model)
occ_attr = occlusion.attribute(sample, target=target_class,
                                sliding_window_shapes=(4, 3, 3), strides=(1, 1, 1))
occ_grid = occ_attr.squeeze(0).detach().cpu().numpy()  # (4, 9, 9)

print("Υπολογισμός Integrated Gradients attribution...")
ig = IntegratedGradients(xai_model)
ig_attr = ig.attribute(sample, target=target_class, n_steps=50)
ig_grid = ig_attr.squeeze(0).detach().cpu().numpy()  # (4, 9, 9)

occ_scores = attribution_to_electrode_scores(occ_grid, DREAMER_CHANNEL_LOCATION_DICT)
ig_scores = attribution_to_electrode_scores(ig_grid, DREAMER_CHANNEL_LOCATION_DICT)

plot_scalp_map(occ_scores, title="Occlusion — High Valence")
plot_scalp_map(ig_scores, title="Integrated Gradients — High Valence")

print("\nTop-5 ηλεκτρόδια (Occlusion):")
for _, row in top5_by_region(occ_scores).head(5).iterrows():
    print(f"  {row['channel']} ({row['region']}): {row['value']:.4f}")

print("\nTop-5 ηλεκτρόδια (Integrated Gradients):")
for _, row in top5_by_region(ig_scores).head(5).iterrows():
    print(f"  {row['channel']} ({row['region']}): {row['value']:.4f}")

# ================== ΒΗΜΑ 8: Αποθήκευση αποτελεσμάτων ==================
# Kaggle: το /kaggle/working/ διατηρείται αυτόματα ως Output μετά από Save & Run All.
# Colab: αποθηκεύουμε στο Google Drive, ώστε να επιβιώσει και μετά τη λήξη του runtime.
if os.path.exists('/kaggle/working'):
    SAVE_DIR = '/kaggle/working/CCNN_DREAMER_fixed_results'
elif IN_COLAB:
    SAVE_DIR = '/content/drive/MyDrive/CCNN_DREAMER_fixed_results'
else:
    SAVE_DIR = './CCNN_DREAMER_fixed_results'

os.makedirs(SAVE_DIR, exist_ok=True)
results_df.to_csv(os.path.join(SAVE_DIR, 'fold_results.csv'), index=False)
torch.save(best_model_state, os.path.join(SAVE_DIR, 'best_model_state.pth'))
print(f"\nΤο καλύτερο μοντέλο και τα αποτελέσματα αποθηκεύτηκαν στο: {SAVE_DIR}")


# ================== ΒΗΜΑ 8: Ανάλυση εσωτερικών αναπαραστάσεων ==================
# (α) Διαχωρισιμότητα ανά επίπεδο, (β) ομοιότητα μέσω Linear CKA,
# (γ) layer-wise probing, (δ) μεταβολή παραμέτρων κατά τη μικρορύθμιση.
import itertools
from sklearn.metrics import silhouette_score, davies_bouldin_score
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split as _tts

CKA_LAYERS = ['conv1', 'conv2', 'conv3', 'conv4', 'lin1']


def _center_gram(K):
    n = K.shape[0]
    H = np.eye(n) - np.ones((n, n)) / n
    return H @ K @ H


def linear_cka(X, Y):
    """Linear CKA μεταξύ δύο πινάκων (δείγματα x διαστάσεις). Τα δείγματα πρέπει
    να είναι τα ΙΔΙΑ και στην ΙΔΙΑ σειρά στους δύο πίνακες."""
    X = np.asarray(X, dtype=np.float64) ; Y = np.asarray(Y, dtype=np.float64)
    X = X - X.mean(axis=0, keepdims=True) ; Y = Y - Y.mean(axis=0, keepdims=True)
    Kx, Ky = _center_gram(X @ X.T), _center_gram(Y @ Y.T)
    hsic = float((Kx * Ky).sum())
    norm = float(np.sqrt((Kx * Kx).sum()) * np.sqrt((Ky * Ky).sum()))
    return hsic / norm if norm > 0 else float('nan')


def fisher_ratio(feats, labels):
    """Λόγος διασποράς μεταξύ κλάσεων προς εντός κλάσεων (μεγαλύτερο = πιο διαχωρίσιμο)."""
    classes = np.unique(labels)
    overall = feats.mean(axis=0)
    between = within = 0.0
    for c in classes:
        fc = feats[labels == c]
        between += len(fc) * np.sum((fc.mean(axis=0) - overall) ** 2)
        within += np.sum((fc - fc.mean(axis=0)) ** 2)
    return float(between / within) if within > 0 else float('nan')


def class_distances(feats, labels):
    """Μέση ενδο-κλασική και δια-κλασική απόσταση."""
    classes = np.unique(labels)
    cent = {c: feats[labels == c].mean(axis=0) for c in classes}
    intra = np.mean([np.linalg.norm(feats[labels == c] - cent[c], axis=1).mean() for c in classes])
    inter = np.mean([np.linalg.norm(cent[a] - cent[b])
                     for i, a in enumerate(classes) for b in classes[i + 1:]])
    return float(intra), float(inter)


def get_layer_reps(model, loader):
    """Ενεργοποιήσεις των επιπέδων του CCNN μέσω forward hooks."""
    store, reps = {}, {n: [] for n in CKA_LAYERS}

    def mk(name):
        def hook(_m, _i, out):
            store[name] = out.detach().cpu().numpy()
        return hook

    handles = [getattr(model, n).register_forward_hook(mk(n))
               for n in CKA_LAYERS if hasattr(model, n)]
    model.eval()
    try:
        with torch.no_grad():
            for X, _y in loader:
                _ = model(X.to(device))
                for n in CKA_LAYERS:
                    if n in store:
                        reps[n].append(store[n].reshape(len(X), -1))
    finally:
        for h in handles:
            h.remove()
    return {n: np.concatenate(v) for n, v in reps.items() if v}


def load_ccnn(state):
    m = CCNN(num_classes=2, in_channels=4, grid_size=(9, 9)).to(device)
    m.load_state_dict(state)
    return m


# --- Δείγμα από ΟΛΟΚΛΗΡΟ το dataset, ώστε να υπάρχουν όλα τα υποκείμενα ---
rng_r = np.random.RandomState(0)
rep_idx = rng_r.choice(info.index.values, size=min(3000, len(info)), replace=False)
rep_loader = DataLoader(Subset(dataset, rep_idx.tolist()), batch_size=BATCH_SIZE, shuffle=False)
rep_labels = info.loc[rep_idx, 'label'].values

reps_s3 = get_layer_reps(load_ccnn(best_model_state), rep_loader)
for k, v in reps_s3.items():
    print(f"Αναπαράσταση '{k}': {v.shape}")

# --- (α) Μέτρα διαχωρισιμότητας ανά επίπεδο ---
sep_rows = []
for name, feats in reps_s3.items():
    if np.allclose(feats.std(axis=0), 0):
        continue
    intra, inter = class_distances(feats, rep_labels)
    sep_rows.append({'stage': name,
                     'silhouette': silhouette_score(feats, rep_labels),
                     'davies_bouldin': davies_bouldin_score(feats, rep_labels),
                     'fisher_ratio': fisher_ratio(feats, rep_labels),
                     'intra_class_dist': intra, 'inter_class_dist': inter,
                     'inter_over_intra': inter / intra if intra > 0 else float('nan')})
separability_df = pd.DataFrame(sep_rows).set_index('stage')
print("\n===== Μέτρα διαχωρισιμότητας ανά επίπεδο =====")
print(separability_df.round(4))

# --- (β) CKA: μεταξύ επιπέδων και μεταξύ S1/S3 ---
names = list(reps_s3.keys())
cka_layers = pd.DataFrame(index=names, columns=names, dtype=float)
for a, b in itertools.product(names, names):
    cka_layers.loc[a, b] = linear_cka(reps_s3[a], reps_s3[b])
print("\n===== CKA μεταξύ επιπέδων =====")
print(cka_layers.round(3))

fig, ax = plt.subplots(figsize=(5.5, 4.5))
sns.heatmap(cka_layers.astype(float), annot=True, fmt='.2f', cmap='viridis', vmin=0, vmax=1, ax=ax)
ax.set_title('Linear CKA μεταξύ επιπέδων')
plt.tight_layout(); plt.show()

reps_s1 = get_layer_reps(load_ccnn(s1_state), rep_loader)
cka_cross = pd.DataFrame([{'stage': n, 'CKA_S1_vs_S3': linear_cka(reps_s1[n], reps_s3[n])}
                          for n in names if n in reps_s1]).set_index('stage')
print("\n===== CKA ίδιου επιπέδου: μοντέλο S1 έναντι S3 =====")
print(cka_cross.round(3))
print("Χαμηλές τιμές σημαίνουν ότι τα δύο πρωτόκολλα οδηγούν σε ποιοτικά")
print("διαφορετικές αναπαραστάσεις, παρότι η αρχιτεκτονική είναι ίδια.")

fig, ax = plt.subplots(figsize=(5.5, 3.2))
cka_cross['CKA_S1_vs_S3'].plot(kind='bar', ax=ax, color='#4C72B0')
ax.set_ylim(0, 1); ax.set_ylabel('Linear CKA'); ax.set_title('Ομοιότητα αναπαραστάσεων S1 vs S3')
plt.xticks(rotation=0); plt.tight_layout(); plt.show()

# --- (γ) Layer-wise probing ---
probe_targets = {'label (συναίσθημα)': 'label',
                 'subject_id (ταυτότητα)': 'subject_id',
                 'trial_id (δοκιμασία)': 'trial_id'}
probe_rows = []
for stage, feats in reps_s3.items():
    if np.allclose(feats.std(axis=0), 0):
        continue
    for lname, col in probe_targets.items():
        if col not in info.columns:
            continue
        y = info.loc[rep_idx, col].values
        classes, counts = np.unique(y, return_counts=True)
        if len(classes) < 2 or counts.min() < 2:
            continue
        Xtr, Xte, ytr, yte = _tts(feats, y, test_size=0.3, random_state=0, stratify=y)
        clf = LogisticRegression(max_iter=1000).fit(Xtr, ytr)
        acc = clf.score(Xte, yte)
        chance = counts.max() / counts.sum()
        probe_rows.append({'stage': stage, 'target': lname, 'probe_accuracy': acc,
                           'chance_level': chance, 'ratio': acc / chance})

probe_df = pd.DataFrame(probe_rows)
if not probe_df.empty:
    acc_tab = probe_df.pivot(index='stage', columns='target', values='probe_accuracy')
    acc_tab = acc_tab.reindex([s for s in CKA_LAYERS if s in acc_tab.index])
    print("\n===== Layer-wise probing: ακρίβεια ανά επίπεδο =====")
    print(acc_tab.round(4))
    print("\nΛόγος προς τη γραμμή βάσης (τιμές >> 1 = αποκωδικοποιήσιμη πληροφορία):")
    print(probe_df.pivot(index='stage', columns='target', values='ratio')
          .reindex(acc_tab.index).round(2))

    fig, ax = plt.subplots(figsize=(1.6 * len(acc_tab) + 3, 3.4))
    acc_tab.plot(kind='bar', ax=ax)
    ax.set_ylabel('Ακρίβεια probe'); ax.set_ylim(0, 1)
    ax.set_title('Linear probing ανά επίπεδο')
    plt.xticks(rotation=0); plt.legend(fontsize=8); plt.tight_layout(); plt.show()

# --- (δ) Μεταβολή παραμέτρων κατά τη μικρορύθμιση ---
# Το CKA συγκρίνει τι ΠΑΡΑΓΕΙ το δίκτυο· εδώ συγκρίνεται τι ΑΛΛΑΞΕ μέσα του.
tuned = globals().get('ft_model', globals().get('finetune_model'))
if tuned is not None:
    after = tuned.state_dict()
    wrows = []
    for pname, w_before in best_model_state.items():
        if not torch.is_floating_point(w_before):
            continue
        delta = (after[pname] - w_before).float()
        n_before = float(w_before.float().norm())
        wrows.append({'parameter': pname, 'shape': tuple(w_before.shape),
                      'abs_change': float(delta.norm()),
                      'rel_change': float(delta.norm()) / n_before if n_before > 0 else float('nan')})
    weight_df = pd.DataFrame(wrows).sort_values('rel_change', ascending=False)
    print("\n===== Μεταβολή παραμέτρων κατά τη μικρορύθμιση =====")
    print(weight_df.round(6).to_string(index=False))

    weight_df['layer'] = weight_df['parameter'].str.split('.').str[0]
    layer_change = weight_df.groupby('layer')['rel_change'].mean().sort_values(ascending=False)
    print("\nΜέση σχετική μεταβολή ανά επίπεδο:")
    print(layer_change.round(6))

    fig, ax = plt.subplots(figsize=(6, 3.4))
    layer_change.plot(kind='bar', ax=ax, color='#C44E52')
    ax.set_ylabel('Μέση σχετική μεταβολή')
    ax.set_title('Ποια επίπεδα άλλαξαν περισσότερο στη μικρορύθμιση')
    plt.xticks(rotation=0); plt.tight_layout(); plt.show()
else:
    print("\nΔεν βρέθηκε μικρορυθμισμένο μοντέλο· η ανάλυση παραμέτρων παραλείπεται.")
```
