```python
# =====================================================================
# MozartussNet / DEAP -- ΜΕ ΠΡΑΓΜΑΤΙΚΗ ΧΡΗΣΗ ΤΟΥ ΑΠΟΘΕΤΗΡΙΟΥ
#
# Η εξαγωγή χαρακτηριστικών γίνεται καλώντας ΑΠΕΥΘΕΙΑΣ τη συνάρτηση fft_processing
# του αποθετηρίου github.com/Mozartuss/DEAP-Emotion-Recognition, με τις ίδιες
# ακριβώς παραμέτρους που χρησιμοποιεί το main.py του αποθετηρίου.
#
# ΣΗΜΕΙΩΣΗ: δεν εκτελείται το main.py αυτούσιο, επειδή εκτός από την εξαγωγή
# χαρακτηριστικών εκπαιδεύει και το δικό του μοντέλο LSTM για 200 εποχές και
# παράγει γραφήματα -- βήματα άσχετα με την παρούσα εργασία. Καλούνται μόνο οι
# συναρτήσεις προεπεξεργασίας (fft_processing), ώστε τα παραγόμενα χαρακτηριστικά
# να είναι πανομοιότυπα με εκείνα του αποθετηρίου.
# =====================================================================

import os
import re
import sys
import glob
import copy
import shutil
import subprocess
import itertools
import time

import numpy as np
import pandas as pd

# ---------- Εγκατάσταση εξαρτήσεων του αποθετηρίου ----------
def pip_install(*pkgs):
    subprocess.run([sys.executable, '-m', 'pip', 'install', '-q', *pkgs], check=True)

try:
    import pyeeg
except ImportError:
    print("Εγκατάσταση pyeeg (εξάρτηση του αποθετηρίου)...")
    pip_install('git+https://github.com/forrestbao/pyeeg.git')

for mod, pkg in [('tqdm', 'tqdm'), ('typer', 'typer'), ('shap', 'shap'),
                 ('lime', 'lime'), ('mne', 'mne'), ('statsmodels', 'statsmodels'),
                 ('seaborn', 'seaborn')]:
    try:
        __import__(mod)
    except ImportError:
        print(f"Εγκατάσταση {pkg}...")
        pip_install(pkg)

import tensorflow as tf
from tensorflow import keras
from sklearn.model_selection import GroupKFold, LeaveOneGroupOut, train_test_split
from sklearn.metrics import (balanced_accuracy_score, f1_score, precision_recall_fscore_support,
                             cohen_kappa_score, matthews_corrcoef, roc_auc_score,
                             confusion_matrix, brier_score_loss,
                             silhouette_score, davies_bouldin_score)
from sklearn.manifold import TSNE
from scipy.stats import spearmanr
import matplotlib.pyplot as plt
import seaborn as sns
import shap
from lime import lime_tabular
import mne

try:
    display
except NameError:
    def display(x):
        print(x)

gpu_devices = tf.config.list_physical_devices('GPU')
print(f"GPU διαθέσιμη: {gpu_devices if gpu_devices else 'ΟΧΙ -- τρέχει σε CPU!'}")

# ================== ΡΥΘΜΙΣΕΙΣ ==================
BATCH_SIZE = 512
N_EPOCHS = 15
SEEDS = [0]
N_BOOTSTRAP = 1000
CI_ALPHA = 0.05
K_VALUES = [5, 10, 20, 30, 50]

# Πόσοι εθελοντές θα επεξεργαστούν. Η εξαγωγή χαρακτηριστικών του αποθετηρίου είναι
# αργή (488 παράθυρα ανά δοκιμασία x 40 δοκιμασίες x 32 κανάλια ανά εθελοντή), οπότε
# για γρήγορη δοκιμή ολόκληρου του pipeline βάλε π.χ. 5. Για την τελική εκτέλεση: None.
MAX_PARTICIPANTS = None

REPO_URL = 'https://github.com/Mozartuss/DEAP-Emotion-Recognition.git'
REPO_DIR = '/kaggle/working/DEAP-Emotion-Recognition' if os.path.exists('/kaggle/working') \
           else '/content/DEAP-Emotion-Recognition'
OUT_DIR = '/kaggle/working/outputs' if os.path.exists('/kaggle/working') else '/content/outputs'
os.makedirs(OUT_DIR, exist_ok=True)

# ---------- Εντοπισμός των ακατέργαστων .dat ----------
RAW_DIR = None
for root, _, files in os.walk('/kaggle/input'):
    if any(f.endswith('.dat') for f in files):
        RAW_DIR = root
        break
if RAW_DIR is None:
    for p in ['/content/drive/MyDrive/DEAP.mini',
              '/content/drive/MyDrive/data_preprocessed_python']:
        if os.path.exists(p) and glob.glob(f'{p}/*.dat'):
            RAW_DIR = p
            break
if RAW_DIR is None:
    raise FileNotFoundError(
        "Δεν βρέθηκαν αρχεία .dat του DEAP. Αν τρέχεις σε Kaggle, πρόσθεσε το dataset "
        "στο notebook (Add Input)."
    )
print(f"Ακατέργαστα δεδομένα : {RAW_DIR}")
print(f"Φάκελος εξόδου       : {OUT_DIR}")


# ================== ΒΗΜΑ 1: Κλωνοποίηση και προσαρμογή του αποθετηρίου ==================
if not os.path.exists(REPO_DIR):
    print(f"Κλωνοποίηση {REPO_URL} ...")
    subprocess.run(['git', 'clone', '--depth', '1', REPO_URL, REPO_DIR], check=True)
else:
    print("Το αποθετήριο υπάρχει ήδη τοπικά.")

# Προσαρμογή των διαδρομών του αποθετηρίου: τα ακατέργαστα δεδομένα βρίσκονται στο
# read-only /kaggle/input, ενώ τα παραγόμενα .npy πρέπει να γραφτούν στο /kaggle/working.
PREPROC_DIR = os.path.join(REPO_DIR, 'Data', 'PREPROCESSED_DEAP_DATA')
os.makedirs(PREPROC_DIR, exist_ok=True)

# ΚΡΙΣΙΜΗ ΔΙΟΡΘΩΣΗ ΣΥΜΒΑΤΟΤΗΤΑΣ: το αποθετήριο καλεί np.save(..., fix_imports=True).
# Το όρισμα fix_imports καταργήθηκε στο NumPy 2.0, οπότε η εξαγωγή χαρακτηριστικών
# τερματίζει με TypeError σε κάθε σύγχρονο περιβάλλον. Αφαιρείται εδώ προγραμματιστικά.
import re as _re
for _f in ['PreProcessing/FFT.py', 'PreProcessing/BuildDataset.py']:
    _p = os.path.join(REPO_DIR, _f)
    if os.path.exists(_p):
        with open(_p) as _fh:
            _src = _fh.read()
        if 'fix_imports' in _src:
            with open(_p, 'w') as _fh:
                _fh.write(_re.sub(r',\s*fix_imports=True', '', _src))
            print(f"Διορθώθηκε ασυμβατότητα NumPy 2.x στο {_f}")

constants_path = os.path.join(REPO_DIR, 'Utils', 'Constants.py')
with open(constants_path) as f:
    const_src = f.read()

if 'PATCHED_BY_THESIS' not in const_src:
    const_src = const_src.replace(
        'RAW_DATA_PATH = Path(ROOT_PATH, "Data", "RAW_DEAP_DATASET")',
        f'RAW_DATA_PATH = Path(r"{RAW_DIR}")  # PATCHED_BY_THESIS'
    )
    with open(constants_path, 'w') as f:
        f.write(const_src)
    print("Οι διαδρομές του αποθετηρίου προσαρμόστηκαν.")

# Το αποθετήριο χρησιμοποιεί απόλυτα imports (from Utils... / from PreProcessing...),
# οπότε πρέπει να μπει στο sys.path και να γίνει cwd.
if REPO_DIR not in sys.path:
    sys.path.insert(0, REPO_DIR)
os.chdir(REPO_DIR)

from Utils.Constants import DEAP_ELECTRODES, FREQUENCIES        # noqa: E402
from Utils.DataHandler import LoadData                           # noqa: E402
from PreProcessing.FFT import fft_processing                     # noqa: E402

print(f"Ηλεκτρόδια αποθετηρίου : {len(DEAP_ELECTRODES)}")
print(f"Φασματικές ζώνες       : {FREQUENCIES}")

DEAP_CHANNELS = list(DEAP_ELECTRODES)
BAND_NAMES = list(FREQUENCIES)
N_BANDS = len(BAND_NAMES)

CHANNEL_REGION = {
    'Fp1': ('frontal', 'left'), 'Fp2': ('frontal', 'right'),
    'AF3': ('frontal', 'left'), 'AF4': ('frontal', 'right'),
    'F3': ('frontal', 'left'), 'F4': ('frontal', 'right'),
    'F7': ('frontal', 'left'), 'F8': ('frontal', 'right'), 'Fz': ('frontal', 'mid'),
    'FC5': ('central', 'left'), 'FC1': ('central', 'left'),
    'FC6': ('central', 'right'), 'FC2': ('central', 'right'),
    'C3': ('central', 'left'), 'C4': ('central', 'right'), 'Cz': ('central', 'mid'),
    'T7': ('temporal', 'left'), 'T8': ('temporal', 'right'),
    'CP5': ('parietal', 'left'), 'CP1': ('parietal', 'left'),
    'CP6': ('parietal', 'right'), 'CP2': ('parietal', 'right'),
    'P3': ('parietal', 'left'), 'P4': ('parietal', 'right'),
    'P7': ('parietal', 'left'), 'P8': ('parietal', 'right'), 'Pz': ('parietal', 'mid'),
    'PO3': ('occipital', 'left'), 'PO4': ('occipital', 'right'),
    'O1': ('occipital', 'left'), 'O2': ('occipital', 'right'), 'Oz': ('occipital', 'mid'),
}


# ================== ΒΗΜΑ 2: Εξαγωγή χαρακτηριστικών μέσω του αποθετηρίου ==================
# Οι παράμετροι είναι ΑΚΡΙΒΩΣ αυτές που χρησιμοποιεί το main.py του αποθετηρίου:
#   band=[4, 8, 12, 16, 25, 45], window_size=256 (2 δευτ.), step_size=16 (0,125 δευτ.),
#   sample_rate=128, και τα 32 κανάλια. Παράγονται 32 x 5 = 160 χαρακτηριστικά ισχύος
#   ανά παράθυρο, μέσω pyeeg.bin_power.
existing = sorted(glob.glob(f'{PREPROC_DIR}/Participant_*.npy'))
print(f"\nΉδη επεξεργασμένοι εθελοντές: {len(existing)}")

t_start = time.time()
load_data = LoadData(RAW_DIR)
processed_count = 0

for filename, data in load_data.yield_raw_data():
    if MAX_PARTICIPANTS is not None and processed_count >= MAX_PARTICIPANTS:
        break
    fft_processing(subject=data,
                   filename=filename,
                   channels=range(len(DEAP_ELECTRODES)),
                   band=[4, 8, 12, 16, 25, 45],
                   window_size=256,
                   step_size=16,
                   sample_rate=128,
                   overwrite=False,
                   fs=False)
    processed_count += 1
    print(f"  {processed_count} εθελοντές | {time.time() - t_start:.0f} δευτ.")

processed_files = sorted(glob.glob(f'{PREPROC_DIR}/Participant_*.npy'),
                         key=lambda p: int(re.search(r'Participant_(\d+)', p).group(1)))
if MAX_PARTICIPANTS is not None:
    processed_files = processed_files[:MAX_PARTICIPANTS]
if not processed_files:
    raise FileNotFoundError(f"Δεν παρήχθησαν αρχεία στο {PREPROC_DIR}.")
print(f"\nΣυνολικά αρχεία προς φόρτωση: {len(processed_files)}")


# ================== ΒΗΜΑ 3: Φόρτωση των παραγόμενων χαρακτηριστικών ==================
# Κάθε γραμμή του .npy είναι [διάνυσμα 160 χαρακτηριστικών, δυαδικές ετικέτες [Valence, Arousal]].
# Κρατάμε το Valence (πρώτο στοιχείο), όπως ορίζει η εργασία.
X_list, y_list, rows = [], [], []
wid = 0
WINDOWS_PER_TRIAL = 488   # (8064 - 256) / 16 = 488 παράθυρα ανά δοκιμασία

for path in processed_files:
    sub_id = re.search(r'Participant_(\d+)', os.path.basename(path)).group(1)
    raw = np.load(path, allow_pickle=True)
    sub_X = np.stack(raw[:, 0]).astype(np.float32)
    sub_y = np.array([int(lbl[0]) for lbl in raw[:, 1]], dtype=np.int64)

    for i in range(len(sub_y)):
        rows.append({'window_id': wid, 'subject_id': sub_id,
                     'trial_id': i // WINDOWS_PER_TRIAL, 'label': int(sub_y[i])})
        wid += 1
    X_list.append(sub_X)
    y_list.append(sub_y)

X_data = np.concatenate(X_list, axis=0)
y_data = np.concatenate(y_list, axis=0)
manifest = pd.DataFrame(rows)
N_FEATURES = X_data.shape[1]

print(f"\nX_data: {X_data.shape} | y_data: {y_data.shape}")
print(f"Θετικά (High Valence): {y_data.mean():.1%} | Εθελοντές: {manifest['subject_id'].nunique()}")
manifest.to_csv(f'{OUT_DIR}/manifest_mozartuss_deap.csv', index=False)

FEATURES_MATCH_LAYOUT = (N_FEATURES == len(DEAP_CHANNELS) * N_BANDS)
if not FEATURES_MATCH_LAYOUT:
    print(f"ΠΡΟΣΟΧΗ: {N_FEATURES} χαρακτηριστικά -- δεν ταιριάζουν με "
          f"{len(DEAP_CHANNELS)}x{N_BANDS}. Οι τοπογραφικοί χάρτες θα παραλειφθούν.")


# ================== ΒΗΜΑ 4: Μοντέλο, μετρικές, splits ==================
def build_substitute_model(n_features):
    model = keras.Sequential([
        keras.layers.Input(shape=(n_features,)),
        keras.layers.Dense(64, activation='relu', name='dense_64'),
        keras.layers.Dropout(0.3),
        keras.layers.Dense(32, activation='relu', name='dense_32'),
        keras.layers.Dropout(0.3),
        keras.layers.Dense(1, activation='sigmoid', name='output'),
    ])
    model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
    return model


def majority_class_baseline(y_test):
    vals, counts = np.unique(y_test, return_counts=True)
    return counts.max() / counts.sum()


def full_metrics(y_true, y_pred, y_prob_pos):
    prec, rec, f1c, _ = precision_recall_fscore_support(y_true, y_pred, labels=[0, 1], zero_division=0)
    out = {
        'majority_baseline': majority_class_baseline(y_true),
        'balanced_accuracy': balanced_accuracy_score(y_true, y_pred),
        'macro_f1': f1_score(y_true, y_pred, average='macro', zero_division=0),
        'precision_class0': prec[0], 'recall_class0': rec[0],
        'precision_class1': prec[1], 'recall_class1': rec[1],
        'cohen_kappa': cohen_kappa_score(y_true, y_pred),
        'mcc': matthews_corrcoef(y_true, y_pred),
    }
    try:
        out['roc_auc'] = roc_auc_score(y_true, y_prob_pos)
    except ValueError:
        out['roc_auc'] = float('nan')
    out['learned_anything'] = out['balanced_accuracy'] > out['majority_baseline'] + 0.02
    return out


def s1_window_split(manifest, seed):
    idx = manifest.index.values
    tr, te = train_test_split(idx, test_size=0.2, random_state=seed,
                              stratify=manifest['label'].values)
    yield tr, te, 'S1_full'


def s2_trial_split(manifest, seed, n_splits=3):
    groups = manifest['subject_id'].astype(str) + '_' + manifest['trial_id'].astype(str)
    gkf = GroupKFold(n_splits=n_splits)
    idx = manifest.index.values
    for fold, (tr, te) in enumerate(gkf.split(idx, manifest['label'].values, groups=groups)):
        yield idx[tr], idx[te], f'S2_fold{fold}'


def s3_subject_split(manifest, seed, max_subjects=3):
    logo = LeaveOneGroupOut()
    idx = manifest.index.values
    groups = manifest['subject_id'].values
    subjects = sorted(manifest['subject_id'].unique())
    rng = np.random.RandomState(seed)
    subjects = list(rng.choice(subjects, size=min(max_subjects, len(subjects)), replace=False))
    for tr, te in logo.split(idx, manifest['label'].values, groups=groups):
        held_out = manifest.iloc[te]['subject_id'].iloc[0]
        if held_out not in subjects:
            continue
        yield idx[tr], idx[te], f'S3_subject{held_out}'


def train_one_split(train_idx, test_idx, seed, n_epochs=N_EPOCHS, fold_name=None):
    tf.random.set_seed(seed); np.random.seed(seed)
    if fold_name:
        print(f"\n-- {fold_name} -- ({len(train_idx)} train / {len(test_idx)} test δείγματα)")

    Xtr, ytr = X_data[train_idx], y_data[train_idx]
    Xte, yte = X_data[test_idx], y_data[test_idx]
    mu, sd = Xtr.mean(0, keepdims=True), Xtr.std(0, keepdims=True) + 1e-6
    Xtr, Xte = (Xtr - mu) / sd, (Xte - mu) / sd

    model = build_substitute_model(Xtr.shape[1])
    es = keras.callbacks.EarlyStopping(monitor='val_accuracy', patience=3, restore_best_weights=True)

    t0 = time.time()
    # verbose=1: εμφανίζει ανά εποχή loss/accuracy σε train ΚΑΙ validation, ώστε να
    # φαίνεται αν το δίκτυο όντως μαθαίνει ή αν έχει κολλήσει στην πλειοψηφική κλάση.
    history = model.fit(Xtr, ytr, validation_data=(Xte, yte), epochs=n_epochs,
                        batch_size=BATCH_SIZE, callbacks=[es], verbose=1)

    tr_acc, va_acc = history.history['accuracy'], history.history['val_accuracy']
    print(f"   Train accuracy: {tr_acc[0]:.2%} (1η εποχή) -> {tr_acc[-1]:.2%} (τελευταία) "
          f"| Val accuracy: {va_acc[0]:.2%} -> {va_acc[-1]:.2%}")
    print(f"   Διάρκεια: {time.time() - t0:.1f} δευτ.")

    y_prob = model.predict(Xte, batch_size=BATCH_SIZE, verbose=0).ravel()
    y_pred = (y_prob >= 0.5).astype(int)
    return model, yte, y_pred, y_prob, full_metrics(yte, y_pred, y_prob), (mu, sd)


# ================== ΒΗΜΑ 5: Εκπαίδευση S1 / S2 / S3 ==================
results, saved_models = [], {}

for seed in SEEDS:
    print(f"\n{'='*60}\n=== Seed {seed} ===\n{'='*60}")

    for level, splitter in [('S1', lambda: s1_window_split(manifest, seed)),
                            ('S2', lambda: s2_trial_split(manifest, seed, n_splits=3)),
                            ('S3', lambda: s3_subject_split(manifest, seed))]:
        print(f"\n### {level} ###")
        for train_idx, test_idx, fold_name in splitter():
            model, yte, ypred, yprob, m, norm = train_one_split(
                train_idx, test_idx, seed, fold_name=fold_name)
            m.update({'model': 'SubstituteMLP', 'dataset': 'DEAP',
                      'split_level': level, 'fold': fold_name, 'seed': seed})
            results.append(m)
            key = (level, seed) if level == 'S1' else (level, seed, fold_name)
            saved_models[key] = (model, test_idx, yte, ypred, yprob, norm)
            print(f"   -> Balanced Acc: {m['balanced_accuracy']:.2%} | Majority baseline: "
                  f"{m['majority_baseline']:.2%} | Macro F1: {m['macro_f1']:.2%}")

results_df = pd.DataFrame(results)
results_df.to_csv(f'{OUT_DIR}/results_S1_S2_S3_mozartuss_deap.csv', index=False)

summary = results_df.groupby('split_level').agg(
    balanced_accuracy_mean=('balanced_accuracy', 'mean'),
    balanced_accuracy_std=('balanced_accuracy', 'std'),
    macro_f1_mean=('macro_f1', 'mean'),
    majority_baseline_mean=('majority_baseline', 'mean')).reset_index()

s1_acc = summary.loc[summary.split_level == 'S1', 'balanced_accuracy_mean'].values
s3_acc = summary.loc[summary.split_level == 'S3', 'balanced_accuracy_mean'].values
drop = (s1_acc[0] - s3_acc[0]) if len(s1_acc) and len(s3_acc) else float('nan')

print("\n--- Συγκεντρωτικά Αποτελέσματα ---")
display(summary)
print(f"Πτώση S1->S3 (balanced accuracy): {drop:.2%}")


# ================== ΒΗΜΑ 6: Bootstrap CI & confusion matrices ==================
def bootstrap_ci(y_true, y_pred, metric_fn, n_boot=N_BOOTSTRAP, alpha=CI_ALPHA, seed=0):
    rng = np.random.RandomState(seed)
    n = len(y_true)
    stats = [metric_fn(y_true[rng.randint(0, n, n)], y_pred[rng.randint(0, n, n)])
             for _ in range(n_boot)]
    lo, hi = np.percentile(stats, [100 * alpha / 2, 100 * (1 - alpha / 2)])
    return float(np.mean(stats)), float(lo), float(hi)


any_s3_key = next(k for k in saved_models if k[0] == 'S3')
model_s3, test_idx_s3, yte_s3, ypred_s3, yprob_s3, (mu_s3, sd_s3) = saved_models[any_s3_key]

mean_ba, lo, hi = bootstrap_ci(yte_s3, ypred_s3, balanced_accuracy_score)
print(f"\nS3 balanced accuracy: {mean_ba:.2%} (95% CI [{lo:.2%}, {hi:.2%}])")

fig, axes = plt.subplots(1, 3, figsize=(11, 3.2))
for ax, level in zip(axes, ['S1', 'S2', 'S3']):
    key = next((k for k in saved_models if k[0] == level), None)
    if key is None:
        continue
    _, _, yt, yp, _, _ = saved_models[key]
    cm = confusion_matrix(yt, yp, labels=[0, 1])
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', ax=ax,
                xticklabels=['Low', 'High'], yticklabels=['Low', 'High'])
    ax.set_title(level, fontsize=10); ax.set_xlabel('Predicted'); ax.set_ylabel('True')
plt.tight_layout(); plt.show()

ece_bins = np.linspace(0, 1, 11)
ece = sum((((yprob_s3 >= ece_bins[i]) & (yprob_s3 < ece_bins[i + 1])).sum() / len(yprob_s3)) *
          abs((yte_s3[(yprob_s3 >= ece_bins[i]) & (yprob_s3 < ece_bins[i + 1])] == 1).mean() -
              yprob_s3[(yprob_s3 >= ece_bins[i]) & (yprob_s3 < ece_bins[i + 1])].mean())
          for i in range(10)
          if ((yprob_s3 >= ece_bins[i]) & (yprob_s3 < ece_bins[i + 1])).sum() > 0)
print(f"ECE = {ece:.4f} | Brier = {brier_score_loss(yte_s3, yprob_s3):.4f}")


# ================== ΒΗΜΑ 7: t-SNE & ποσοτικά μέτρα διαχωρισιμότητας ==================
feature_extractor = keras.Model(inputs=model_s3.inputs,
                                outputs=model_s3.get_layer('dense_32').output)
rng_t = np.random.RandomState(0)
tsne_idx = rng_t.choice(test_idx_s3, size=min(2000, len(test_idx_s3)), replace=False)
X_tsne_raw = (X_data[tsne_idx] - mu_s3) / sd_s3
features_s3 = feature_extractor.predict(X_tsne_raw, batch_size=BATCH_SIZE, verbose=0)
labels_s3 = y_data[tsne_idx]

tsne_xy = TSNE(n_components=2, random_state=0, perplexity=30, init='pca').fit_transform(features_s3)
plt.figure(figsize=(7, 6))
for c, name, color in [(0, 'Low Valence', '#C44E52'), (1, 'High Valence', '#4C72B0')]:
    m_ = labels_s3 == c
    plt.scatter(tsne_xy[m_, 0], tsne_xy[m_, 1], s=8, alpha=0.6, c=color, label=name)
plt.legend(); plt.xlabel('t-SNE 1'); plt.ylabel('t-SNE 2')
plt.title('t-SNE αναπαραστάσεων -- S3 (LOSO) μοντέλο')
plt.tight_layout(); plt.show()


def fisher_ratio(feats, labels):
    """Λόγος διασποράς μεταξύ κλάσεων προς εντός κλάσεων (μεγαλύτερο = πιο διαχωρίσιμο)."""
    classes = np.unique(labels)
    overall_mean = feats.mean(axis=0)
    between, within = 0.0, 0.0
    for c in classes:
        fc = feats[labels == c]
        between += len(fc) * np.sum((fc.mean(axis=0) - overall_mean) ** 2)
        within += np.sum((fc - fc.mean(axis=0)) ** 2)
    return float(between / within) if within > 0 else float('nan')


def class_distances(feats, labels):
    """Μέση ενδο-κλασική και δια-κλασική απόσταση."""
    classes = np.unique(labels)
    centroids = {c: feats[labels == c].mean(axis=0) for c in classes}
    intra = np.mean([np.linalg.norm(feats[labels == c] - centroids[c], axis=1).mean()
                     for c in classes])
    inter = np.mean([np.linalg.norm(centroids[a] - centroids[b])
                     for i, a in enumerate(classes) for b in classes[i + 1:]])
    return float(intra), float(inter)


sil = silhouette_score(features_s3, labels_s3)
db = davies_bouldin_score(features_s3, labels_s3)
fr = fisher_ratio(features_s3, labels_s3)
intra, inter = class_distances(features_s3, labels_s3)

print("\n===== Ποσοτικά μέτρα διαχωρισιμότητας (MozartussNet / DEAP, S3) =====")
print(f"Αναπαραστάσεις: {features_s3.shape} | Κατανομή κλάσεων: {np.bincount(labels_s3)}")
print(f"Silhouette score      : {sil:+.4f}   (→ +1 άριστος διαχωρισμός, ~0 επικάλυψη)")
print(f"Davies-Bouldin index  : {db:.4f}    (μικρότερο = καλύτερο)")
print(f"Fisher discriminant   : {fr:.4f}    (μεγαλύτερο = καλύτερο)")
print(f"Μέση ενδο-κλασική απόσταση : {intra:.4f}")
print(f"Μέση δια-κλασική απόσταση  : {inter:.4f}")
print(f"Λόγος inter/intra          : {inter/intra:.4f}  (>1 υποδηλώνει διαχωρίσιμες κλάσεις)")

pd.DataFrame([{
    'model': 'MozartussNet', 'dataset': 'DEAP', 'split_level': 'S3',
    'silhouette': sil, 'davies_bouldin': db, 'fisher_ratio': fr,
    'intra_class_dist': intra, 'inter_class_dist': inter, 'inter_over_intra': inter / intra,
}]).to_csv(f'{OUT_DIR}/separability_metrics_mozartuss_deap.csv', index=False)


# ================== ΒΗΜΑ 8: XAI ==================
# ΣΗΜΕΙΩΣΗ ΓΙΑ ΤΗΝ ΕΠΙΛΟΓΗ ΜΕΘΟΔΩΝ:
# Το μοντέλο δέχεται 160 πινακοποιημένα χαρακτηριστικά, όχι σήμα σε μορφή
# (κανάλια x χρόνος). Συνεπώς:
#   - Grad-CAM ΔΕΝ εφαρμόζεται: απαιτεί συνελικτικό επίπεδο, ενώ εδώ το δίκτυο
#     είναι καθαρά πυκνό (Dense). Παραλείπεται συνειδητά και όχι λόγω σφάλματος.
#   - Occlusion σε χρονικά παράθυρα ΔΕΝ εφαρμόζεται (δεν υπάρχει χρονικός άξονας).
#     Το αντίστοιχό του εδώ είναι το Feature Ablation, δηλαδή ο μηδενισμός
#     μεμονωμένων χαρακτηριστικών.
# Εφαρμόζονται: Permutation Importance, Feature Ablation, Gradient SHAP,
# KernelSHAP και LIME -- πέντε μέθοδοι συνολικά.

def predict_proba_s3(X_raw):
    Xn = (X_raw - mu_s3) / sd_s3
    p1 = model_s3.predict(Xn, batch_size=BATCH_SIZE, verbose=0).ravel()
    return np.stack([1 - p1, p1], axis=1)


rng = np.random.RandomState(0)
perm_idx = rng.choice(test_idx_s3, size=min(300, len(test_idx_s3)), replace=False)
X_eval, y_eval = X_data[perm_idx], y_data[perm_idx]

print("\nΥπολογισμός Permutation Importance...")
base_pred = predict_proba_s3(X_eval)[:, 1]
base_acc = ((base_pred >= 0.5).astype(int) == y_eval).mean()
perm_importance = np.zeros(N_FEATURES)
for f in range(N_FEATURES):
    X_perm = X_eval.copy()
    rng.shuffle(X_perm[:, f])
    perm_acc = ((predict_proba_s3(X_perm)[:, 1] >= 0.5).astype(int) == y_eval).mean()
    perm_importance[f] = base_acc - perm_acc

print("Υπολογισμός Feature Ablation...")
# Μηδενίζει ένα-ένα κάθε χαρακτηριστικό (στην κανονικοποιημένη κλίμακα το 0 αντιστοιχεί
# στον μέσο όρο του συνόλου εκπαίδευσης) και μετρά τη μεταβολή της μέσης πιθανότητας.
ablation_importance = np.zeros(N_FEATURES)
base_prob_mean = base_pred.mean()
for f in range(N_FEATURES):
    X_abl = X_eval.copy()
    X_abl[:, f] = mu_s3[0, f]
    ablation_importance[f] = base_prob_mean - predict_proba_s3(X_abl)[:, 1].mean()

print("Υπολογισμός Gradient SHAP...")
try:
    X_eval_norm = (X_eval - mu_s3) / sd_s3
    bg_norm = X_eval_norm[rng.choice(len(X_eval_norm), size=min(50, len(X_eval_norm)), replace=False)]
    grad_explainer = shap.GradientExplainer(model_s3, bg_norm)
    gshap_vals = grad_explainer.shap_values(X_eval_norm[:50])
    arr = gshap_vals[0] if isinstance(gshap_vals, list) else gshap_vals
    gradient_shap_importance = np.abs(np.array(arr)).reshape(-1, N_FEATURES).mean(axis=0)
except Exception as e:
    print(f"  Το Gradient SHAP απέτυχε ({e}); παραλείπεται.")
    gradient_shap_importance = None

print("Υπολογισμός KernelSHAP...")
background = shap.sample(X_eval, 30)
kernel_explainer = shap.KernelExplainer(lambda x: predict_proba_s3(x)[:, 1], background)
shap_values = kernel_explainer.shap_values(X_eval[:20], nsamples=50, silent=True)
shap_importance = np.abs(np.array(shap_values)).reshape(-1, N_FEATURES).mean(axis=0)

print("Υπολογισμός LIME...")
lime_explainer = lime_tabular.LimeTabularExplainer(X_eval, mode='classification',
                                                    class_names=['Low', 'High'])
lime_importance = np.zeros(N_FEATURES)
n_lime = min(10, len(X_eval))
for i in range(n_lime):
    exp = lime_explainer.explain_instance(X_eval[i], predict_proba_s3, num_features=N_FEATURES)
    for feat_idx, w in exp.as_map()[1]:
        lime_importance[feat_idx] += abs(w)
lime_importance /= n_lime

methods = {
    'PermutationImportance': perm_importance,
    'FeatureAblation': ablation_importance,
    'KernelSHAP': shap_importance,
    'LIME': lime_importance,
}
if gradient_shap_importance is not None:
    methods['GradientSHAP'] = gradient_shap_importance
print(f"\nΜέθοδοι με έγκυρο αποτέλεσμα: {list(methods.keys())}")


# ================== ΒΗΜΑ 9: Τοπογραφικοί χάρτες ανά μέθοδο ==================
# Τα 160 χαρακτηριστικά ανάγονται πίσω σε 32 ηλεκτρόδια (άθροισμα των 5 ζωνών
# κάθε καναλιού), ώστε τα ευρήματα να αποδίδονται σε ανατομικές θέσεις και όχι
# σε ανώνυμους αριθμούς χαρακτηριστικών.
def features_to_channels(importance, n_channels=len(DEAP_CHANNELS), n_bands=N_BANDS):
    return np.abs(np.asarray(importance)).reshape(n_channels, n_bands).sum(axis=1)


def features_to_bands(importance, n_channels=len(DEAP_CHANNELS), n_bands=N_BANDS):
    return np.abs(np.asarray(importance)).reshape(n_channels, n_bands).sum(axis=0)


def build_montage_info(channels=DEAP_CHANNELS):
    montage = mne.channels.make_standard_montage('standard_1020')
    positions = montage.get_positions()['ch_pos']
    plottable = [ch for ch in channels if ch in positions]
    info = mne.create_info(ch_names=plottable, sfreq=128, ch_types='eeg')
    info.set_montage(montage)
    return info, plottable


if FEATURES_MATCH_LAYOUT:
    MNE_INFO, PLOTTABLE = build_montage_info()
    PLOT_IDX = [DEAP_CHANNELS.index(ch) for ch in PLOTTABLE]

    def plot_scalp_map(ch_values, title):
        fig, ax = plt.subplots(figsize=(4, 4))
        mne.viz.plot_topomap(np.asarray(ch_values)[PLOT_IDX], MNE_INFO,
                             axes=ax, show=False, cmap='RdBu_r', contours=4)
        ax.set_title(title, fontsize=10)
        plt.show()

    print("\n===== Τοπογραφικοί χάρτες ανά μέθοδο XAI =====")
    channel_importances = {}
    for name, imp in methods.items():
        ch_imp = features_to_channels(imp)
        channel_importances[name] = ch_imp
        plot_scalp_map(ch_imp, title=f"{name} — High Valence")

        top5 = np.argsort(ch_imp)[::-1][:5]
        print(f"\nTop-5 ηλεκτρόδια ({name}):")
        for i in top5:
            ch = DEAP_CHANNELS[i]
            region, hemi = CHANNEL_REGION[ch]
            print(f"  {ch} ({region}-{hemi}): {ch_imp[i]:.6f}")

    channel_scores_df = pd.DataFrame(channel_importances, index=DEAP_CHANNELS)
    channel_scores_df.index.name = 'channel'
    print("\n===== Σημαντικότητα ανά ηλεκτρόδιο και μέθοδο =====")
    display(channel_scores_df.round(6))
    channel_scores_df.to_csv(f'{OUT_DIR}/xai_channel_scores_mozartuss_deap.csv')

    # --- Σημαντικότητα ανά φασματική ζώνη ---
    band_df = pd.DataFrame({name: features_to_bands(imp) for name, imp in methods.items()},
                           index=BAND_NAMES)
    band_df.index.name = 'band'
    print("\n===== Σημαντικότητα ανά φασματική ζώνη και μέθοδο =====")
    display(band_df.round(6))
    band_df.to_csv(f'{OUT_DIR}/xai_band_scores_mozartuss_deap.csv')

    fig, ax = plt.subplots(figsize=(7, 4))
    band_df.plot(kind='bar', ax=ax)
    ax.set_ylabel('Συνολική σημαντικότητα')
    ax.set_title('Σημαντικότητα ανά φασματική ζώνη')
    plt.tight_layout(); plt.show()

    # --- Heatmap Καναλιών x Ζωνών (μέσος όρος όλων των μεθόδων, κανονικοποιημένος) ---
    norm_maps = []
    for imp in methods.values():
        grid = np.abs(np.asarray(imp)).reshape(len(DEAP_CHANNELS), N_BANDS)
        rng_ = grid.max() - grid.min()
        norm_maps.append((grid - grid.min()) / rng_ if rng_ > 0 else grid)
    cxb = pd.DataFrame(np.mean(norm_maps, axis=0), index=DEAP_CHANNELS, columns=BAND_NAMES)
    plt.figure(figsize=(6, 9))
    sns.heatmap(cxb, annot=True, fmt='.2f', cmap='viridis')
    plt.title('Channels × Bands relevance (μέσος όρος μεθόδων)')
    plt.tight_layout(); plt.show()

    # --- Μετωπιαία ασυμμετρία ανά μέθοδο ---
    def frontal_asymmetry(ch_values):
        pairs = [('Fp1', 'Fp2'), ('AF3', 'AF4'), ('F3', 'F4'), ('F7', 'F8'), ('FC5', 'FC6')]
        idx_map = {ch: i for i, ch in enumerate(DEAP_CHANNELS)}
        return {f'{l}-{r}': float(ch_values[idx_map[r]] - ch_values[idx_map[l]])
                for l, r in pairs if l in idx_map and r in idx_map}

    asym_rows = []
    for name, ch_imp in channel_importances.items():
        a = frontal_asymmetry(ch_imp); a['method'] = name
        asym_rows.append(a)
    print("\n===== Μετωπιαία ασυμμετρία ανά μέθοδο (δεξιά - αριστερή) =====")
    display(pd.DataFrame(asym_rows).set_index('method').round(6))


# ================== ΒΗΜΑ 10: Faithfulness (Deletion/Insertion + AOPC) ==================
def aopc(probs, base_prob):
    """Μέση πτώση πιθανότητας σε σχέση με την αρχική πρόβλεψη."""
    probs = np.asarray(probs)
    return float(np.mean(base_prob - probs)) if probs.size else float('nan')


def deletion_insertion_curve(sample_row, importance, mode='deletion', k_values=K_VALUES):
    order = np.argsort(-np.abs(importance))
    probs = []
    for k in k_values:
        n_k = int(N_FEATURES * k / 100)
        touched = order[:n_k]
        x = sample_row.copy()
        if mode == 'deletion':
            x[touched] = mu_s3[0, touched]          # αντικατάσταση με τη μέση τιμή
        else:
            keep = x[touched].copy()
            x = mu_s3[0].copy()
            x[touched] = keep
        probs.append(predict_proba_s3(x[None, :])[0, 1])
    return np.array(probs)


faithfulness_rows = []
n_faith = min(5, len(X_eval))
for name, imp in methods.items():
    del_all, ins_all, del_r_all, ins_r_all, base_all = [], [], [], [], []
    for i in range(n_faith):
        row = X_eval[i]
        base_p = predict_proba_s3(row[None, :])[0, 1]
        rand_imp = np.random.RandomState(i).rand(N_FEATURES)
        del_all.append(deletion_insertion_curve(row, imp, 'deletion'))
        ins_all.append(deletion_insertion_curve(row, imp, 'insertion'))
        del_r_all.append(deletion_insertion_curve(row, rand_imp, 'deletion'))
        ins_r_all.append(deletion_insertion_curve(row, rand_imp, 'insertion'))
        base_all.append(base_p)
    base_m = float(np.mean(base_all))
    faithfulness_rows.append({
        'explainer': name,
        'AOPC_deletion': aopc(np.mean(del_all, axis=0), base_m),
        'AOPC_insertion': aopc(np.mean(ins_all, axis=0), base_m) * -1,
        'AOPC_deletion_random': aopc(np.mean(del_r_all, axis=0), base_m),
        'AOPC_insertion_random': aopc(np.mean(ins_r_all, axis=0), base_m) * -1,
    })

faithfulness_df = pd.DataFrame(faithfulness_rows)
faithfulness_df['deletion_beats_random'] = (faithfulness_df['AOPC_deletion'] >
                                             faithfulness_df['AOPC_deletion_random'])
print("\n===== Faithfulness (AOPC) =====")
display(faithfulness_df.round(5))
faithfulness_df.to_csv(f'{OUT_DIR}/faithfulness_AOPC_mozartuss_deap.csv', index=False)


# ================== ΒΗΜΑ 11: Sanity checks ==================
print("\n===== Sanity check: τυχαιοποίηση βαρών μοντέλου =====")
orig_imp = methods['PermutationImportance']
rand_model = build_substitute_model(N_FEATURES)
rand_model.set_weights([np.random.normal(0, 0.1, w.shape) for w in model_s3.get_weights()])


def predict_proba_rand(X_raw):
    Xn = (X_raw - mu_s3) / sd_s3
    p1 = rand_model.predict(Xn, batch_size=BATCH_SIZE, verbose=0).ravel()
    return np.stack([1 - p1, p1], axis=1)


rand_base = ((predict_proba_rand(X_eval)[:, 1] >= 0.5).astype(int) == y_eval).mean()
rand_imp = np.zeros(N_FEATURES)
rng2 = np.random.RandomState(1)
for f in range(N_FEATURES):
    Xp = X_eval.copy()
    rng2.shuffle(Xp[:, f])
    rand_imp[f] = rand_base - ((predict_proba_rand(Xp)[:, 1] >= 0.5).astype(int) == y_eval).mean()

rho_model, _ = spearmanr(orig_imp, rand_imp)
print(f"Spearman ρ (εκπαιδευμένο vs τυχαιοποιημένο μοντέλο): {rho_model:+.4f}")
print("Αναμενόμενο: κοντά στο μηδέν -- αλλιώς οι εξηγήσεις δεν εξαρτώνται από τα μαθημένα βάρη.")


# ================== ΒΗΜΑ 12: Συμφωνία μεταξύ explainers ==================
def jaccard_topk(v1, v2, k=10):
    top1 = set(np.argsort(-np.abs(v1))[:k])
    top2 = set(np.argsort(-np.abs(v2))[:k])
    return len(top1 & top2) / len(top1 | top2)


names = list(methods.keys())
agree = pd.DataFrame(index=names, columns=names, dtype=float)
agree_jaccard = pd.DataFrame(index=names, columns=names, dtype=float)
for a, b in itertools.product(names, names):
    agree.loc[a, b] = spearmanr(methods[a], methods[b])[0]
    agree_jaccard.loc[a, b] = jaccard_topk(methods[a], methods[b], 10)

print("\n--- Spearman ρ (συμφωνία explainers) ---")
display(agree.round(2))
print("\n--- Jaccard Top-10 χαρακτηριστικών ---")
display(agree_jaccard.round(2))

fig, ax = plt.subplots(figsize=(6, 5))
sns.heatmap(agree.astype(float), annot=True, fmt='.2f', cmap='coolwarm', vmin=-1, vmax=1, ax=ax)
ax.set_title('Συμφωνία explainers — Spearman ρ')
plt.tight_layout(); plt.show()

agree.to_csv(f'{OUT_DIR}/interexplainer_spearman_mozartuss_deap.csv')
agree_jaccard.to_csv(f'{OUT_DIR}/interexplainer_jaccard_mozartuss_deap.csv')
print(f"\nΌλα τα αποτελέσματα αποθηκεύτηκαν στο: {OUT_DIR}")

# =====================================================================
# ΑΝΑΛΥΣΗ ΟΜΟΙΟΤΗΤΑΣ ΑΝΑΠΑΡΑΣΤΑΣΕΩΝ (Linear CKA)
#
# Το CKA (Centered Kernel Alignment) μετρά πόσο όμοιες είναι δύο αναπαραστάσεις,
# σε κλίμακα 0 (εντελώς ανόμοιες) έως 1 (ταυτόσημες). Λειτουργεί ακόμα κι όταν
# οι δύο αναπαραστάσεις έχουν διαφορετικό πλήθος διαστάσεων, γι' αυτό επιτρέπει
# σύγκριση τόσο μεταξύ επιπέδων του ίδιου δικτύου όσο και μεταξύ μοντέλων που
# εκπαιδεύτηκαν με διαφορετικό πρωτόκολλο διαχωρισμού.
# =====================================================================

def _center_gram(K):
    n = K.shape[0]
    H = np.eye(n) - np.ones((n, n)) / n
    return H @ K @ H


def linear_cka(X, Y):
    """Linear CKA μεταξύ δύο πινάκων (δείγματα x διαστάσεις). Τα δείγματα πρέπει
    να είναι τα ΙΔΙΑ και στη ΙΔΙΑ σειρά στους δύο πίνακες."""
    X = np.asarray(X, dtype=np.float64)
    Y = np.asarray(Y, dtype=np.float64)
    X = X - X.mean(axis=0, keepdims=True)
    Y = Y - Y.mean(axis=0, keepdims=True)
    Kx = _center_gram(X @ X.T)
    Ky = _center_gram(Y @ Y.T)
    hsic = float((Kx * Ky).sum())
    norm = float(np.sqrt((Kx * Kx).sum()) * np.sqrt((Ky * Ky).sum()))
    return hsic / norm if norm > 0 else float('nan')


LAYER_NAMES = ['dense_64', 'dense_32']


def get_layer_reps(model, X_raw, norm_stats):
    """Επιστρέφει τις ενεργοποιήσεις των δύο πυκνών επιπέδων.
    ΣΗΜΑΝΤΙΚΟ: κάθε μοντέλο τροφοδοτείται με τα ΔΙΚΑ ΤΟΥ στατιστικά κανονικοποίησης,
    αφού με αυτά εκπαιδεύτηκε· διαφορετικά η σύγκριση θα ήταν άκυρη."""
    mu, sd = norm_stats
    Xn = (X_raw - mu) / sd
    extractor = keras.Model(inputs=model.inputs,
                            outputs=[model.get_layer(n).output for n in LAYER_NAMES])
    outs = extractor.predict(Xn, batch_size=BATCH_SIZE, verbose=0)
    return {name: np.asarray(o) for name, o in zip(LAYER_NAMES, outs)}


# --- Κοινό υποσύνολο δειγμάτων: ίδια ακριβώς παράθυρα για κάθε μοντέλο ---
rng_cka = np.random.RandomState(0)
cka_idx = rng_cka.choice(test_idx_s3, size=min(500, len(test_idx_s3)), replace=False)
X_cka = X_data[cka_idx]

reps_s3 = get_layer_reps(model_s3, X_cka, (mu_s3, sd_s3))
for k, v in reps_s3.items():
    print(f"Αναπαράσταση '{k}': {v.shape}")

# --- (α) Ομοιότητα μεταξύ των δύο επιπέδων του μοντέλου S3 ---
cka_layers = pd.DataFrame(index=LAYER_NAMES, columns=LAYER_NAMES, dtype=float)
for a, b in itertools.product(LAYER_NAMES, LAYER_NAMES):
    cka_layers.loc[a, b] = linear_cka(reps_s3[a], reps_s3[b])

print("\n===== CKA μεταξύ επιπέδων (μοντέλο S3) =====")
display(cka_layers.round(3))

fig, ax = plt.subplots(figsize=(4, 3.4))
sns.heatmap(cka_layers.astype(float), annot=True, fmt='.3f', cmap='viridis',
            vmin=0, vmax=1, ax=ax)
ax.set_title('Linear CKA μεταξύ επιπέδων (S3)')
plt.tight_layout(); plt.show()
cka_layers.to_csv(f'{OUT_DIR}/cka_between_layers_mozartuss_deap.csv')

# --- (β) Ομοιότητα ίδιου επιπέδου μεταξύ των μοντέλων S1 και S3 ---
s1_key = next((k for k in saved_models if k[0] == 'S1'), None)

if s1_key is None:
    print("\nΤο μοντέλο S1 δεν είναι διαθέσιμο· η σύγκριση S1 vs S3 παραλείπεται.")
else:
    model_s1, _, _, _, _, norm_s1 = saved_models[s1_key]
    reps_s1 = get_layer_reps(model_s1, X_cka, norm_s1)
    cka_cross = pd.DataFrame(
        [{'layer': ln, 'CKA_S1_vs_S3': linear_cka(reps_s1[ln], reps_s3[ln])}
         for ln in LAYER_NAMES]).set_index('layer')

    print("\n===== CKA ίδιου επιπέδου: μοντέλο S1 έναντι μοντέλου S3 =====")
    display(cka_cross.round(3))
    print("Χαμηλές τιμές σημαίνουν ότι τα δύο πρωτόκολλα οδηγούν σε ποιοτικά")
    print("διαφορετικές αναπαραστάσεις, παρότι η αρχιτεκτονική είναι ίδια.")

    fig, ax = plt.subplots(figsize=(4.5, 3.2))
    cka_cross['CKA_S1_vs_S3'].plot(kind='bar', ax=ax, color='#4C72B0')
    ax.set_ylim(0, 1); ax.set_ylabel('Linear CKA')
    ax.set_title('Ομοιότητα αναπαραστάσεων S1 vs S3')
    plt.xticks(rotation=0); plt.tight_layout(); plt.show()
    cka_cross.to_csv(f'{OUT_DIR}/cka_s1_vs_s3_mozartuss_deap.csv')

# =====================================================================
# LAYER-WISE PROBING
# Γραμμικοί ταξινομητές πάνω στις παγωμένες αναπαραστάσεις ΚΑΘΕ σταδίου,
# ώστε να φανεί σε ποιο σημείο του δικτύου η κάθε πληροφορία γίνεται
# γραμμικά αποκωδικοποιήσιμη. Συγκρίνεται η συναισθηματική ετικέτα με
# μεταβλητές που το μοντέλο ΔΕΝ θα έπρεπε να κωδικοποιεί (ταυτότητα
# εθελοντή, δοκιμασία), οπότε υψηλή ακρίβεια σε αυτές αποτελεί ένδειξη
# συντόμευσης μέσω Διαρροής Ταυτότητας Υποκειμένου.
# =====================================================================
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split as _tts

META = manifest            # <-- στα δύο CCNN notebooks άλλαξέ το σε: META = info

probe_targets = {'label (συναίσθημα)': 'label',
                 'subject_id (ταυτότητα)': 'subject_id',
                 'trial_id (δοκιμασία)': 'trial_id'}
if 'session_id' in META.columns and META['session_id'].nunique() > 1:
    probe_targets['session_id (συνεδρία)'] = 'session_id'

probe_rows = []
for stage, feats in reps_s3.items():
    if np.allclose(feats.std(axis=0), 0):
        print(f"  Το στάδιο '{stage}' έχει μηδενική διασπορά· παραλείπεται.")
        continue
    for label_name, col in probe_targets.items():
        y = META.loc[cka_idx, col].values
        classes, counts = np.unique(y, return_counts=True)
        if len(classes) < 2 or counts.min() < 2:
            continue
        strat = y if counts.min() >= 2 else None
        Xtr, Xte, ytr, yte = _tts(feats, y, test_size=0.3, random_state=0, stratify=strat)
        clf = LogisticRegression(max_iter=1000).fit(Xtr, ytr)
        acc = clf.score(Xte, yte)
        chance = counts.max() / counts.sum()      # πλειοψηφική κλάση ως γραμμή βάσης
        probe_rows.append({'stage': stage, 'target': label_name,
                           'probe_accuracy': acc, 'chance_level': chance,
                           'ratio': acc / chance if chance > 0 else float('nan')})

probe_df = pd.DataFrame(probe_rows)
if not probe_df.empty:
    table = probe_df.pivot(index='stage', columns='target', values='probe_accuracy')
    print("\n===== Layer-wise probing: ακρίβεια γραμμικού ταξινομητή ανά στάδιο =====")
    print(table.round(4))
    print("\nΛόγος προς τη γραμμή βάσης (τιμές >>1 υποδηλώνουν αποκωδικοποιήσιμη πληροφορία):")
    print(probe_df.pivot(index='stage', columns='target', values='ratio').round(2))
    probe_df.to_csv('layerwise_probing.csv', index=False)

    fig, ax = plt.subplots(figsize=(1.6 * len(table) + 3, 3.4))
    table.plot(kind='bar', ax=ax)
    ax.set_ylabel('Ακρίβεια probe'); ax.set_ylim(0, 1)
    ax.set_title('Layer-wise probing ανά στάδιο')
    plt.xticks(rotation=0); plt.legend(fontsize=8); plt.tight_layout(); plt.show()
```
