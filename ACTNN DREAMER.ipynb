```python

######################################################################
# ACTNN / DREAMER — Πλήρες pipeline (εκπαίδευση + αξιολόγηση + XAI)
######################################################################

# --- ΒΗΜΑ 0: Εγκατάσταση βιβλιοθηκών (ΜΠΛΟΚ 0: ένα μόνο, καθαρό install cell) ---
try:
    import braindecode, shap, lime, captum, mne
except ImportError:
    print("Εγκατάσταση απαραίτητων βιβλιοθηκών...")
    import subprocess, sys
    subprocess.run([sys.executable, "-m", "pip", "install", "-q",
                     "braindecode", "mne", "shap", "lime", "captum",
                     "scikit-learn", "statsmodels"], check=True)

import os, math, json, zipfile, itertools, warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import scipy.io as sio
from scipy import signal as sps
from scipy.stats import spearmanr, kendalltau, wilcoxon

import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, TensorDataset, Subset

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import GroupKFold, LeaveOneGroupOut, train_test_split
from sklearn.metrics import (balanced_accuracy_score, f1_score, precision_recall_fscore_support,
                              cohen_kappa_score, matthews_corrcoef, roc_auc_score,
                              confusion_matrix, brier_score_loss)
from sklearn.manifold import TSNE
from sklearn.linear_model import LogisticRegression
from statsmodels.stats.multitest import multipletests

from captum.attr import (IntegratedGradients, Saliency, Occlusion, DeepLift,
                          FeatureAblation, LayerGradCam, LRP)
import shap
from lime import lime_tabular

try:
    from google.colab import drive
    IN_COLAB = True
except ImportError:
    IN_COLAB = False

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Device: {device}")

try:
    display  # ορίζεται αυτόματα σε Jupyter/Colab
except NameError:
    def display(x):  # fallback αν τρέχει σαν απλό .py script
        print(x)

# --- ΡΥΘΜΙΣΕΙΣ ---
N_SEEDS = 1
SEEDS = [0]

RUN_FULL_LOSO = False
LOSO_MAX_SUBJECTS = 5   # >=5 ώστε ο έλεγχος Wilcoxon να είναι εφικτός

N_EPOCHS = 15
BATCH_SIZE = 256
LR = 1e-3

N_EXPLAIN_SAMPLES = 2
SECONDS_LIMIT = 60   # όπως και στα υπόλοιπα μοντέλα της εργασίας, ώστε τα
                     # αποτελέσματα να είναι άμεσα συγκρίσιμα μεταξύ αρχιτεκτονικών

# FS/WINDOW_SIZE/STEP -- έλειπαν εντελώς, χρησιμοποιούνται στη φόρτωση δεδομένων/φίλτρα/MNE
FS = 128            # συχνότητα δειγματοληψίας DREAMER (Emotiv EPOC)
WINDOW_SIZE = 128   # 1 δευτερόλεπτο παράθυρο (128 δείγματα)
STEP = 64           # βήμα ολίσθησης 64 δειγμάτων -> επικάλυψη 50%

# N_BOOTSTRAP/CI_ALPHA -- έλειπαν επίσης, χρησιμοποιούνται στο bootstrap_ci (ΜΠΛΟΚ 9)
N_BOOTSTRAP = 1000   # αριθμός bootstrap resamples για το 95% CI
CI_ALPHA = 0.05      # 95% διάστημα εμπιστοσύνης

# ΜΠΛΟΚ 3: 14 κανάλια Emotiv EPOC (DREAMER), με τη σειρά που δίνει η τεκμηρίωση του dataset
DREAMER_CHANNELS = ['AF3', 'F7', 'F3', 'FC5', 'T7', 'P7', 'O1',
                     'O2', 'P8', 'T8', 'FC6', 'F4', 'F8', 'AF4']

# Ομαδοποίηση περιοχών + ημισφαίριο/άξονας (χοντρική, βασισμένη στο 10-20 σύστημα)
CHANNEL_REGION = {
    'AF3': ('frontal', 'left', 'anterior'),  'AF4': ('frontal', 'right', 'anterior'),
    'F7':  ('frontal', 'left', 'anterior'),  'F8':  ('frontal', 'right', 'anterior'),
    'F3':  ('frontal', 'left', 'anterior'),  'F4':  ('frontal', 'right', 'anterior'),
    'FC5': ('central', 'left', 'anterior'),  'FC6': ('central', 'right', 'anterior'),
    'T7':  ('temporal', 'left', 'midline'),  'T8':  ('temporal', 'right', 'midline'),
    'P7':  ('parietal', 'left', 'posterior'),'P8':  ('parietal', 'right', 'posterior'),
    'O1':  ('occipital', 'left', 'posterior'),'O2': ('occipital', 'right', 'posterior'),
}

# ΜΠΛΟΚ 7: ζώνες συχνοτήτων
BANDS = {
    'delta': (1, 4), 'theta': (4, 8), 'alpha': (8, 13), 'beta': (13, 30), 'gamma': (30, 50),
}

# Kaggle: το /kaggle/working διατηρείται αυτόματα ως Output μετά το Save & Run All
OUT_DIR = '/kaggle/working/outputs' if os.path.exists('/kaggle/working') else 'outputs'
os.makedirs(OUT_DIR, exist_ok=True)
print(f"Φάκελος εξόδου: {OUT_DIR}")


######################################################################
# ΜΠΛΟΚ 0 — Φόρτωση δεδομένων + manifest
######################################################################

# --- ΒΗΜΑ 1: Εντοπισμός DREAMER.mat (Kaggle ή Colab) ---
mat_path = None

# Kaggle: το αρχείο βρίσκεται ήδη έτοιμο μέσα στο /kaggle/input
for root, _, files in os.walk('/kaggle/input'):
    for f in files:
        if f.endswith('.mat') and 'DREAMER' in f.upper():
            mat_path = os.path.join(root, f)
            break
    if mat_path:
        break

# Colab / τοπικά: αποσυμπίεση από DREAMER.zip
if mat_path is None:
    if IN_COLAB:
        if not os.path.exists('/content/drive'):
            drive.mount('/content/drive')
        zip_paths = ['/content/DREAMER.zip', '/content/drive/MyDrive/DREAMER.zip']
    else:
        zip_paths = ['./DREAMER.zip']

    target_zip = next((p for p in zip_paths if os.path.exists(p)), None)
    if target_zip is None:
        raise FileNotFoundError(
            "Δεν βρέθηκε το DREAMER.mat ούτε σε /kaggle/input, ούτε DREAMER.zip σε Colab. "
            "Αν τρέχεις σε Kaggle, πρόσθεσε το dataset στο notebook (Add Input)."
        )

    extract_path = '/content/DREAMER_extracted' if IN_COLAB else './DREAMER_extracted'
    if not os.path.exists(extract_path):
        print("Αποσυμπίεση...")
        with zipfile.ZipFile(target_zip, 'r') as zf:
            zf.extractall(extract_path)

    for root, dirs, files in os.walk(extract_path):
        for f in files:
            if f.endswith('.mat') and 'DREAMER' in f:
                mat_path = os.path.join(root, f)
                break
        if mat_path:
            break

print(f"mat_path = {mat_path}")


def load_dreamer_with_manifest(path, window_size=WINDOW_SIZE, step=STEP, seconds_limit=SECONDS_LIMIT):
    """Ενιαίος loader. Sliding window + z-score normalization ανά παράθυρο (όπως στο αρχικό),
    ΑΛΛΑ τώρα κρατάει subject_id/trial_id/start_sample για κάθε παράθυρο ώστε να χτίζεται το manifest.
    """
    raw = sio.loadmat(path)
    data_struct = raw['DREAMER'][0, 0]['Data']

    X_list, y_list = [], []
    rows = []  # για το manifest
    wid = 0

    for sub in range(23):
        try:
            eeg_root = data_struct[0, sub]['EEG'][0, 0]
            stimuli = (eeg_root['stimuli'][0, 0] if 'stimuli' in eeg_root.dtype.names
                       else eeg_root['Signals'][0, 0])
            valence = data_struct[0, sub]['ScoreValence'][0, 0]
        except Exception:
            continue

        for trial in range(18):
            label = 1 if valence[trial, 0] >= 3 else 0
            signal = stimuli[trial, 0]
            limit = min(len(signal), FS * seconds_limit)  # ΜΠΛΟΚ 0: το cap αναφέρεται μία φορά

            for start in range(0, limit - window_size, step):
                seg = signal[start:start + window_size, :]
                seg = (seg - np.mean(seg)) / (np.std(seg) + 1e-6)
                X_list.append(seg.T)  # (14, 128)
                y_list.append(label)
                rows.append({
                    'window_id': wid, 'subject_id': sub, 'trial_id': trial,
                    'session_id': sub,  # DREAMER: ένα session ανά εθελοντή, δεν υπάρχει ξεχωριστό session index
                    'dataset': 'DREAMER', 'label': label, 'start_sample': start,
                })
                wid += 1

    X = np.array(X_list, dtype=np.float32)[:, np.newaxis, :, :]
    y = np.array(y_list, dtype=np.int64)
    manifest = pd.DataFrame(rows)
    return torch.tensor(X), torch.tensor(y), manifest


X_data, y_data, manifest = load_dreamer_with_manifest(mat_path)
manifest.to_csv(f'{OUT_DIR}/manifest.csv', index=False)
print(f"Σύνολο παραθύρων: {len(y_data)} | Θετικά (High Valence): {y_data.sum().item()} "
      f"({y_data.float().mean().item():.1%})")
display(manifest.head())


######################################################################
# ΜΠΛΟΚ 3 (μέρος 1) — Χαρτογράφηση καναλιών & βοηθητικά για scalp maps
######################################################################

import mne

def build_montage_info(channels=DREAMER_CHANNELS):
    montage = mne.channels.make_standard_montage('standard_1020')
    info = mne.create_info(ch_names=channels, sfreq=FS, ch_types='eeg')
    info.set_montage(montage, on_missing='warn')
    return info

MNE_INFO = build_montage_info()


def plot_scalp_map(values, title, ax=None, cmap='RdBu_r'):
    """values: array (14,) — μία τιμή relevance/importance ανά κανάλι."""
    fig = None
    if ax is None:
        fig, ax = plt.subplots(figsize=(4, 4))
    mne.viz.plot_topomap(values, MNE_INFO, axes=ax, show=False, cmap=cmap, contours=4)
    ax.set_title(title, fontsize=10)
    if fig is not None:
        plt.show()
    return ax


def channel_to_region_table(values, channels=DREAMER_CHANNELS):
    """ΜΠΛΟΚ 3: άθροισμα attribution ανά περιοχή (π.χ. |IG| αθροισμένο στον χρόνο ανά κανάλι)."""
    rows = []
    for ch, v in zip(channels, values):
        region, hemi, axis = CHANNEL_REGION[ch]
        rows.append({'channel': ch, 'value': float(v), 'region': region,
                     'hemisphere': hemi, 'axis': axis})
    df = pd.DataFrame(rows)
    region_table = df.groupby('region')['value'].sum().sort_values(ascending=False)
    return df, region_table


def frontal_asymmetry(values, channels=DREAMER_CHANNELS):
    """ΜΠΛΟΚ 3: ημισφαιρική ασυμμετρία σε μετωπιαία ζεύγη (π.χ. AF3/AF4, F3/F4, F7/F8)."""
    pairs = [('AF3', 'AF4'), ('F3', 'F4'), ('F7', 'F8'), ('FC5', 'FC6')]
    idx = {ch: i for i, ch in enumerate(channels)}
    out = {}
    for l, r in pairs:
        if l in idx and r in idx:
            out[f'{l}-{r}'] = float(values[idx[r]] - values[idx[l]])  # >0 => δεξιά > αριστερή
    return out


######################################################################
# ΜΠΛΟΚ 1 — Τρία επίπεδα διαχωρισμού (S1 / S2 / S3)
######################################################################

def s1_window_split(manifest, seed):
    idx = manifest.index.values
    train_idx, test_idx = train_test_split(idx, test_size=0.2, random_state=seed,
                                            stratify=manifest['label'].values)
    yield train_idx, test_idx, 'S1_full'


def s2_trial_split(manifest, seed, n_splits=5):
    groups = manifest['subject_id'].astype(str) + '_' + manifest['trial_id'].astype(str)
    gkf = GroupKFold(n_splits=n_splits)
    idx = manifest.index.values
    for fold, (tr, te) in enumerate(gkf.split(idx, manifest['label'].values, groups=groups)):
        yield idx[tr], idx[te], f'S2_fold{fold}'


def s3_subject_split(manifest, seed, run_full=RUN_FULL_LOSO, max_subjects=LOSO_MAX_SUBJECTS):
    logo = LeaveOneGroupOut()
    idx = manifest.index.values
    groups = manifest['subject_id'].values
    subjects = sorted(manifest['subject_id'].unique())
    if not run_full:
        rng = np.random.RandomState(seed)
        subjects = list(rng.choice(subjects, size=min(max_subjects, len(subjects)), replace=False))
    for tr, te in logo.split(idx, manifest['label'].values, groups=groups):
        held_out_subject = manifest.iloc[te]['subject_id'].iloc[0]
        if held_out_subject not in subjects:
            continue
        yield idx[tr], idx[te], f'S3_subject{held_out_subject}'


######################################################################
# Μοντέλο — ACTNN (ενιαίος ορισμός, ΜΠΛΟΚ 0)
######################################################################

class ACTNN(nn.Module):
    def __init__(self, n_channels=14, n_classes=2):
        super().__init__()
        self.cnn = nn.Sequential(
            nn.Conv2d(1, 16, (n_channels, 1)), nn.BatchNorm2d(16), nn.ELU(),
            nn.Conv2d(16, 32, (1, 10), stride=2), nn.BatchNorm2d(32), nn.ELU(), nn.Dropout(0.5),
        )
        self.transformer = nn.TransformerEncoder(
            nn.TransformerEncoderLayer(d_model=32, nhead=4, batch_first=True), num_layers=2)
        self.fc = nn.Linear(32, n_classes)
        self.attn = nn.Linear(32, 1)

    def forward(self, x):
        x = self.cnn(x).squeeze(2).permute(0, 2, 1)
        x = self.transformer(x)
        w = torch.softmax(self.attn(x), dim=1)
        return self.fc((x * w).sum(dim=1))

    def embed(self, x):
        """Χρήσιμο για t-SNE (ΜΠΛΟΚ 0) και identity probing (ΜΠΛΟΚ 8)."""
        feat = self.cnn(x).squeeze(2).permute(0, 2, 1)
        feat = self.transformer(feat)
        return feat.mean(dim=1)


######################################################################
# ΜΠΛΟΚ 2 — Μετρικές, majority-class baseline, calibration
######################################################################

def majority_class_baseline(y_test):
    vals, counts = np.unique(y_test, return_counts=True)
    return counts.max() / counts.sum()


def full_metrics(y_true, y_pred, y_prob_pos):
    """y_prob_pos: predicted probability of class 1 (High Valence)."""
    prec, rec, f1c, _ = precision_recall_fscore_support(y_true, y_pred, labels=[0, 1], zero_division=0)
    out = {
        'majority_baseline': majority_class_baseline(y_true),
        'balanced_accuracy': balanced_accuracy_score(y_true, y_pred),
        'macro_f1': f1_score(y_true, y_pred, average='macro', zero_division=0),
        'precision_class0': prec[0], 'recall_class0': rec[0],
        'precision_class1': prec[1], 'recall_class1': rec[1],
        'cohen_kappa': cohen_kappa_score(y_true, y_pred),
        'mcc': matthews_corrcoef(y_true, y_pred),
    }
    try:
        out['roc_auc'] = roc_auc_score(y_true, y_prob_pos)
    except ValueError:
        out['roc_auc'] = float('nan')  # μία μόνο κλάση παρόν στο test fold
    out['learned_anything'] = out['balanced_accuracy'] > out['majority_baseline'] + 0.02
    return out


def plot_confusion(y_true, y_pred, title, ax=None):
    cm = confusion_matrix(y_true, y_pred, labels=[0, 1])
    fig = None
    if ax is None:
        fig, ax = plt.subplots(figsize=(3.5, 3))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', ax=ax,
                xticklabels=['Low', 'High'], yticklabels=['Low', 'High'])
    ax.set_xlabel('Predicted'); ax.set_ylabel('True'); ax.set_title(title, fontsize=10)
    if fig is not None:
        plt.show()
    return cm


def expected_calibration_error(y_true, y_prob_pos, n_bins=10):
    bins = np.linspace(0, 1, n_bins + 1)
    ece = 0.0
    n = len(y_true)
    for i in range(n_bins):
        mask = (y_prob_pos >= bins[i]) & (y_prob_pos < bins[i + 1] if i < n_bins - 1
                                            else y_prob_pos <= bins[i + 1])
        if mask.sum() == 0:
            continue
        conf = y_prob_pos[mask].mean()
        acc = (y_true[mask] == 1).mean()
        ece += (mask.sum() / n) * abs(acc - conf)
    return ece


def reliability_diagram(y_true, y_prob_pos, n_bins=10, title='Reliability diagram'):
    bins = np.linspace(0, 1, n_bins + 1)
    centers, accs, confs = [], [], []
    for i in range(n_bins):
        lo, hi = bins[i], bins[i + 1]
        mask = (y_prob_pos >= lo) & (y_prob_pos < hi if i < n_bins - 1 else y_prob_pos <= hi)
        if mask.sum() == 0:
            continue
        centers.append((lo + hi) / 2)
        accs.append((y_true[mask] == 1).mean())
        confs.append(y_prob_pos[mask].mean())
    plt.figure(figsize=(4, 4))
    plt.plot([0, 1], [0, 1], 'k--', label='ideal')
    plt.plot(confs, accs, 'o-', label='model')
    plt.xlabel('Confidence'); plt.ylabel('Empirical accuracy'); plt.title(title); plt.legend()
    plt.show()
    ece = expected_calibration_error(y_true, y_prob_pos, n_bins)
    brier = brier_score_loss((y_true == 1).astype(int), y_prob_pos)
    print(f"ECE = {ece:.4f} | Brier = {brier:.4f}")
    return ece, brier


def per_subject_accuracy(y_true, y_pred, subject_ids):
    df = pd.DataFrame({'y': y_true, 'p': y_pred, 'subject': subject_ids})
    return df.groupby('subject').apply(lambda g: (g['y'] == g['p']).mean())


######################################################################
# Εκπαίδευση ενός split (χρησιμοποιείται από S1/S2/S3, ΜΠΛΟΚ 1 & 9)
######################################################################

def train_one_split(train_idx, test_idx, seed, n_epochs=N_EPOCHS, verbose=True, fold_name=None):
    torch.manual_seed(seed); np.random.seed(seed)

    train_ds = TensorDataset(X_data[train_idx], y_data[train_idx])
    test_ds = TensorDataset(X_data[test_idx], y_data[test_idx])
    train_loader = DataLoader(train_ds, batch_size=BATCH_SIZE, shuffle=True)
    test_loader = DataLoader(test_ds, batch_size=BATCH_SIZE, shuffle=False)

    model = ACTNN().to(device)
    optimizer = optim.Adam(model.parameters(), lr=LR)
    scheduler = optim.lr_scheduler.ReduceLROnPlateau(optimizer, 'max', patience=3, factor=0.5)
    criterion = nn.CrossEntropyLoss()

    if verbose and fold_name is not None:
        print(f"-- {fold_name} --")   # header ώστε να ξεχωρίζει κάθε fold στην έξοδο

    for epoch in range(n_epochs):
        model.train()
        train_correct, train_n = 0, 0                     # μέτρηση train accuracy
        for xb, yb in train_loader:
            xb, yb = xb.to(device), yb.to(device)
            optimizer.zero_grad()
            out = model(xb)
            loss = criterion(out, yb)
            loss.backward()
            optimizer.step()
            train_correct += (out.argmax(1) == yb).sum().item()
            train_n += len(yb)
        train_acc = train_correct / max(train_n, 1)

        model.eval()
        val_correct, val_n = 0, 0
        with torch.no_grad():
            for xb, yb in test_loader:
                xb, yb = xb.to(device), yb.to(device)
                val_correct += (model(xb).argmax(1) == yb).sum().item()
                val_n += len(yb)
        val_acc = val_correct / max(val_n, 1)
        scheduler.step(val_acc)
        if verbose:
            print(f"  epoch {epoch+1:02d} | train_acc {train_acc:.2%} | val_acc {val_acc:.2%}")

    # Τελικές προβλέψεις για μετρικές
    model.eval()
    all_true, all_pred, all_prob = [], [], []
    with torch.no_grad():
        for xb, yb in test_loader:
            xb = xb.to(device)
            out = torch.softmax(model(xb), dim=1).cpu().numpy()
            all_true.append(yb.numpy())
            all_pred.append(out.argmax(1))
            all_prob.append(out[:, 1])
    y_true = np.concatenate(all_true)
    y_pred = np.concatenate(all_pred)
    y_prob = np.concatenate(all_prob)
    metrics = full_metrics(y_true, y_pred, y_prob)
    return model, y_true, y_pred, y_prob, metrics


######################################################################
# ΜΠΛΟΚ 1 + ΜΠΛΟΚ 9 — Εκτέλεση S1/S2/S3 × πολλά seeds, κεντρικός πίνακας, στατιστική επικύρωση
######################################################################

results = []          # μία γραμμή ανά (split_level, fold, seed)
saved_models = {}      # κρατάμε ένα μοντέλο ανά split-level για τα επόμενα blocks (XAI κ.λπ.)

for seed in SEEDS:
    print(f"\n=== Seed {seed} ===")

    # S1 — window-level (αντιπαράδειγμα)
    for train_idx, test_idx, fold_name in s1_window_split(manifest, seed):
        model, y_true, y_pred, y_prob, m = train_one_split(train_idx, test_idx, seed, fold_name=fold_name)
        m.update({'model': 'ACTNN', 'dataset': 'DREAMER', 'split_level': 'S1', 'fold': fold_name, 'seed': seed})
        results.append(m)
        saved_models[('S1', seed)] = (model, test_idx, y_true, y_pred, y_prob)

    # S2 — trial-level
    for train_idx, test_idx, fold_name in s2_trial_split(manifest, seed, n_splits=5):
        model, y_true, y_pred, y_prob, m = train_one_split(train_idx, test_idx, seed, fold_name=fold_name)
        m.update({'model': 'ACTNN', 'dataset': 'DREAMER', 'split_level': 'S2', 'fold': fold_name, 'seed': seed})
        results.append(m)
        saved_models[('S2', seed, fold_name)] = (model, test_idx, y_true, y_pred, y_prob)

    # S3 — subject-level (LOSO)
    for train_idx, test_idx, fold_name in s3_subject_split(manifest, seed):
        model, y_true, y_pred, y_prob, m = train_one_split(train_idx, test_idx, seed, fold_name=fold_name)
        m.update({'model': 'ACTNN', 'dataset': 'DREAMER', 'split_level': 'S3', 'fold': fold_name, 'seed': seed})
        results.append(m)
        saved_models[('S3', seed, fold_name)] = (model, test_idx, y_true, y_pred, y_prob)

results_df = pd.DataFrame(results)
results_df.to_csv(f'{OUT_DIR}/results_S1_S2_S3.csv', index=False)
display(results_df.head())

# --- Κεντρικός πίνακας: Model × Dataset × Acc S1/S2/S3 × Πτώση S1→S3 (ΜΠΛΟΚ 1) ---
summary = (results_df.groupby('split_level')
           .agg(balanced_accuracy_mean=('balanced_accuracy', 'mean'),
                balanced_accuracy_std=('balanced_accuracy', 'std'),
                macro_f1_mean=('macro_f1', 'mean'),
                macro_f1_std=('macro_f1', 'std'),
                majority_baseline_mean=('majority_baseline', 'mean'))
           .reset_index())

s1_acc = summary.loc[summary.split_level == 'S1', 'balanced_accuracy_mean'].values
s3_acc = summary.loc[summary.split_level == 'S3', 'balanced_accuracy_mean'].values
drop = (s1_acc[0] - s3_acc[0]) if len(s1_acc) and len(s3_acc) else float('nan')
print(f"Πτώση S1→S3 (balanced accuracy): {drop:.2%}")
display(summary)

# --- ΜΠΛΟΚ 9: Bootstrap CI στο test set (για το καλύτερο/S3 μοντέλο, ενδεικτικά seed 0) ---
def bootstrap_ci(y_true, y_pred, metric_fn, n_boot=N_BOOTSTRAP, alpha=CI_ALPHA, seed=0):
    rng = np.random.RandomState(seed)
    n = len(y_true)
    stats = []
    for _ in range(n_boot):
        idx = rng.randint(0, n, n)
        stats.append(metric_fn(y_true[idx], y_pred[idx]))
    lo, hi = np.percentile(stats, [100 * alpha / 2, 100 * (1 - alpha / 2)])
    return np.mean(stats), lo, hi


any_s3_key = next(k for k in saved_models if k[0] == 'S3')
_, _, y_true_s3, y_pred_s3, _ = saved_models[any_s3_key]
mean_ba, lo, hi = bootstrap_ci(y_true_s3, y_pred_s3, balanced_accuracy_score)
print(f"S3 balanced accuracy: {mean_ba:.2%} (95% CI [{lo:.2%}, {hi:.2%}])")

# --- ΜΠΛΟΚ 9: Wilcoxon signed-rank σε επίπεδο εθελοντή (S1 vs S3), + Holm-Bonferroni ---
# Χρειάζεται per-subject accuracy και για τα δύο splits, πάνω στο ίδιο σύνολο subjects.
s1_key = next(k for k in saved_models if k[0] == 'S1')
_, s1_test_idx, s1_true, s1_pred, _ = saved_models[s1_key]
s1_subjects = manifest.loc[s1_test_idx, 'subject_id'].values
s1_per_subj = per_subject_accuracy(s1_true, s1_pred, s1_subjects)

s3_keys = [k for k in saved_models if k[0] == 'S3']
s3_per_subj = {}
for k in s3_keys:
    _, test_idx, y_true, y_pred, _ = saved_models[k]
    subj = manifest.loc[test_idx, 'subject_id'].iloc[0]
    s3_per_subj[subj] = (y_true == y_pred).mean()
s3_per_subj = pd.Series(s3_per_subj)

common = s1_per_subj.index.intersection(s3_per_subj.index)
if len(common) >= 5:
    stat, p = wilcoxon(s1_per_subj.loc[common], s3_per_subj.loc[common])
    reject, p_adj, _, _ = multipletests([p], method='holm')
    print(f"Wilcoxon S1 vs S3 (n={len(common)} εθελοντές): stat={stat:.3f}, p={p:.4f}, "
          f"p_holm={p_adj[0]:.4f}, σημαντικό={reject[0]}")
else:
    print("Πολύ λίγοι κοινοί εθελοντές μεταξύ S1-test-set και S3-folds για Wilcoxon "
          "(αύξησε LOSO_MAX_SUBJECTS ή RUN_FULL_LOSO).")


######################################################################
# ΜΠΛΟΚ 2 (συνέχεια) — Confusion matrices, calibration, ισορροπία κλάσεων ανά subject
######################################################################

# Confusion matrix + calibration για κάθε split-level (seed 0, ένα ενδεικτικό fold)
fig, axes = plt.subplots(1, 3, figsize=(11, 3.5))
for ax, level in zip(axes, ['S1', 'S2', 'S3']):
    key = next((k for k in saved_models if k[0] == level), None)
    if key is None:
        continue
    _, _, y_true, y_pred, _ = saved_models[key]
    plot_confusion(y_true, y_pred, title=f"{level}", ax=ax)
plt.tight_layout(); plt.show()

# Reliability diagram για S3 (το πιο αξιόπιστο επίπεδο)
_, _, y_true_s3, y_pred_s3, y_prob_s3 = saved_models[any_s3_key]
reliability_diagram(y_true_s3, y_prob_s3, title='Reliability — S3 (LOSO)')

# ΜΠΛΟΚ 2: ισορροπία κλάσεων ανά subject (ο s13-τύπου εθελοντής θα ξεχωρίσει)
subj_balance = manifest.groupby('subject_id')['label'].mean().sort_values()
plt.figure(figsize=(9, 3))
sns.barplot(x=subj_balance.index.astype(str), y=subj_balance.values, color='steelblue')
plt.axhline(0.5, color='red', linestyle='--', linewidth=1)
plt.ylabel('% High Valence'); plt.xlabel('subject_id')
plt.title('Ισορροπία κλάσεων ανά εθελοντή (DREAMER)')
plt.xticks(rotation=90)
plt.show()
outlier_subjects = subj_balance[(subj_balance < 0.15) | (subj_balance > 0.85)]
print("Πιθανοί degenerate-risk subjects (πολύ ανισόρροποι):")
print(outlier_subjects)


######################################################################
# t-SNE (από το αρχικό, κρατιέται ως ήταν) — χρησιμοποιεί το S3 μοντέλο
######################################################################

tsne_model, tsne_test_idx, _, _, _ = saved_models[any_s3_key]
tsne_loader = DataLoader(TensorDataset(X_data[tsne_test_idx], y_data[tsne_test_idx]),
                          batch_size=128, shuffle=False)
tsne_model.eval()
f_list, l_list = [], []
with torch.no_grad():
    for xb, yb in tsne_loader:
        f_list.append(tsne_model.embed(xb.to(device)).cpu().numpy())
        l_list.append(yb.numpy())
X_tsne = TSNE(n_components=2, init='pca', random_state=0).fit_transform(np.concatenate(f_list))
plt.figure(figsize=(7, 5))
sns.scatterplot(x=X_tsne[:, 0], y=X_tsne[:, 1], hue=np.concatenate(l_list), palette='coolwarm')
plt.title("t-SNE Clusters (S3 / LOSO model)")
plt.show()

######################################################################
# ΠΟΣΟΤΙΚΑ ΜΕΤΡΑ ΔΙΑΧΩΡΙΣΙΜΟΤΗΤΑΣ
# Το t-SNE είναι μόνο οπτική ένδειξη και μπορεί να "δείξει" ομάδες που δεν
# αντιστοιχούν σε πραγματική δομή, ανάλογα με το perplexity. Τα μέτρα αυτά
# τεκμηριώνουν αριθμητικά τον βαθμό διαχωρισμού των δύο κλάσεων.
######################################################################
from sklearn.metrics import silhouette_score, davies_bouldin_score

features_s3 = np.concatenate(f_list)
labels_s3 = np.concatenate(l_list)


def fisher_ratio(feats, labels):
    """Λόγος διασποράς μεταξύ κλάσεων προς εντός κλάσεων (μεγαλύτερο = πιο διαχωρίσιμο)."""
    classes = np.unique(labels)
    overall_mean = feats.mean(axis=0)
    between, within = 0.0, 0.0
    for c in classes:
        fc = feats[labels == c]
        between += len(fc) * np.sum((fc.mean(axis=0) - overall_mean) ** 2)
        within += np.sum((fc - fc.mean(axis=0)) ** 2)
    return float(between / within) if within > 0 else float('nan')


def class_distances(feats, labels):
    """Μέση ενδο-κλασική και δια-κλασική απόσταση."""
    classes = np.unique(labels)
    centroids = {c: feats[labels == c].mean(axis=0) for c in classes}
    intra = np.mean([np.linalg.norm(feats[labels == c] - centroids[c], axis=1).mean()
                     for c in classes])
    inter = np.mean([np.linalg.norm(centroids[a] - centroids[b])
                     for i, a in enumerate(classes) for b in classes[i + 1:]])
    return float(intra), float(inter)


sil = silhouette_score(features_s3, labels_s3)
db = davies_bouldin_score(features_s3, labels_s3)
fr = fisher_ratio(features_s3, labels_s3)
intra, inter = class_distances(features_s3, labels_s3)

print("\n===== Ποσοτικά μέτρα διαχωρισιμότητας (ACTNN / DREAMER, S3) =====")
print(f"Αναπαραστάσεις: {features_s3.shape} | Κατανομή κλάσεων: {np.bincount(labels_s3)}")
print(f"Silhouette score      : {sil:+.4f}   (→ +1 άριστος διαχωρισμός, ~0 επικάλυψη)")
print(f"Davies-Bouldin index  : {db:.4f}    (μικρότερο = καλύτερο)")
print(f"Fisher discriminant   : {fr:.4f}    (μεγαλύτερο = καλύτερο)")
print(f"Μέση ενδο-κλασική απόσταση : {intra:.4f}")
print(f"Μέση δια-κλασική απόσταση  : {inter:.4f}")
print(f"Λόγος inter/intra          : {inter/intra:.4f}  (>1 υποδηλώνει διαχωρίσιμες κλάσεις)")

separability_df = pd.DataFrame([{
    'model': 'ACTNN', 'dataset': 'DREAMER', 'split_level': 'S3',
    'silhouette': sil, 'davies_bouldin': db, 'fisher_ratio': fr,
    'intra_class_dist': intra, 'inter_class_dist': inter, 'inter_over_intra': inter / intra,
}])
separability_df.to_csv(f'{OUT_DIR}/separability_metrics_actnn_dreamer.csv', index=False)
display(separability_df.round(4))


######################################################################
# ΜΠΛΟΚ 4 προετοιμασία — XAI explainers
######################################################################

xai_model, xai_test_idx, _, _, _ = saved_models[any_s3_key]
xai_model.eval()

xai_loader = DataLoader(TensorDataset(X_data[xai_test_idx], y_data[xai_test_idx]),
                         batch_size=32, shuffle=False)
xai_batch_x, xai_batch_y = next(iter(xai_loader))
background = xai_batch_x[:min(30, len(xai_batch_x))].to(device)
explain_samples = xai_batch_x[:N_EXPLAIN_SAMPLES].to(device)
explain_labels = xai_batch_y[:N_EXPLAIN_SAMPLES]


def _to_map(attr_tensor):
    return attr_tensor.squeeze().detach().cpu().numpy()  # (14, 128)


def explain_integrated_gradients(sample, target=1):
    ig = IntegratedGradients(xai_model)
    attr = ig.attribute(sample, target=target, n_steps=32)
    return _to_map(attr)


def explain_saliency(sample, target=1):
    sal = Saliency(xai_model)
    attr = sal.attribute(sample.requires_grad_(True), target=target)
    return _to_map(attr)


def explain_occlusion(sample, target=1):
    occ = Occlusion(xai_model)
    attr = occ.attribute(sample, target=target, sliding_window_shapes=(1, 1, 10), strides=(1, 1, 5))
    return _to_map(attr)


def explain_deeplift(sample, target=1):
    dl = DeepLift(xai_model)
    baseline = torch.zeros_like(sample)
    attr = dl.attribute(sample, baselines=baseline, target=target)
    return _to_map(attr)


def explain_feature_ablation_channel(sample, target=1):
    """Ablation ανά ολόκληρο κανάλι (14 features), broadcast σε (14,128) στο τέλος."""
    ablator = FeatureAblation(xai_model)
    feature_mask = torch.arange(14).reshape(1, 1, 14, 1).expand(1, 1, 14, 128).to(device)
    attr = ablator.attribute(sample, target=target, feature_mask=feature_mask)
    return _to_map(attr)


def explain_gradcam(sample, target=1):
    # ΔΙΟΡΘΩΣΗ: το cnn[-2] είναι ELU (τελευταίο στο Sequential είναι το Dropout),
    # όχι Conv2d. Το Grad-CAM πρέπει να δείχνει στο πραγματικό τελευταίο συνελικτικό
    # επίπεδο, δηλαδή στο cnn[3] = Conv2d(16, 32, (1, 10), stride=2).
    target_layer = xai_model.cnn[3]
    lgc = LayerGradCam(xai_model, target_layer)
    attr = lgc.attribute(sample, target=target)
    attr = torch.nn.functional.interpolate(attr, size=(14, 128), mode='bilinear', align_corners=False)
    return _to_map(attr)


def explain_lrp(sample, target=1):
    try:
        lrp = LRP(xai_model)
        attr = lrp.attribute(sample.requires_grad_(True), target=target)
        return _to_map(attr)
    except Exception as e:
        print(f"LRP απέτυχε ({e}); επιστρέφεται 0-map. (Το TransformerEncoder δεν υποστηρίζεται πάντα.)")
        return np.zeros((14, 128))


def explain_shap_gradient(background, sample, target=1):
    explainer = shap.GradientExplainer(xai_model, background)
    sv = explainer.shap_values(sample)

    # ΔΙΟΡΘΩΣΗ: παλαιότερες εκδόσεις του shap επιστρέφουν λίστα (μία είσοδος ανά κλάση),
    # ενώ νεότερες έναν ενιαίο πίνακα με ΕΠΙΠΛΕΟΝ άξονα κλάσεων στο τέλος, π.χ.
    # (1, 1, 14, 128, 2). Το προηγούμενο `np.squeeze(arr)[0]` έπαιρνε το πρώτο ΚΑΝΑΛΙ
    # αντί για την κλάση-στόχο, δίνοντας σχήμα (128, 2) και τελικά 128 τιμές αντί για 14.
    if isinstance(sv, list):
        arr = np.array(sv[target] if len(sv) > target else sv[0])
    else:
        arr = np.array(sv)
    arr = np.squeeze(arr)

    if arr.ndim == 3:
        if arr.shape[-1] == 2:      # (14, 128, n_classes) -> επιλογή κλάσης
            arr = arr[..., target]
        elif arr.shape[0] == 2:     # (n_classes, 14, 128) -> επιλογή κλάσης
            arr = arr[target]
        else:
            arr = arr[0]
    return arr


def explain_lime(all_X_flat, sample_flat, target=1):
    def predict_p(x):
        x_t = torch.tensor(x, dtype=torch.float32).reshape(-1, 1, 14, 128).to(device)
        xai_model.eval()
        with torch.no_grad():
            return torch.softmax(xai_model(x_t), dim=1).cpu().numpy()
    expl = lime_tabular.LimeTabularExplainer(all_X_flat, mode='classification',
                                              class_names=['Low Valence', 'High Valence'])
    # ΔΙΟΡΘΩΣΗ: ζητάμε ρητά την κλάση-στόχο (labels=[target]) και διαβάζουμε τα βάρη
    # ΑΥΤΗΣ της κλάσης. Πριν, το as_map()[1] επέστρεφε πάντα την κλάση 1, οπότε αν το
    # target ήταν 0 οι εξηγήσεις αφορούσαν την αντίθετη κλάση από τις άλλες μεθόδους.
    exp = expl.explain_instance(sample_flat, predict_p, num_features=40, labels=[target])
    out = np.zeros(14 * 128)
    for feat_idx, w in exp.as_map()[target]:
        out[feat_idx] = w
    return out.reshape(14, 128)


def explain_kernelshap(all_X_flat, sample_flat, target=1, nsamples=100):
    def predict_p1(x):
        x_t = torch.tensor(x, dtype=torch.float32).reshape(-1, 1, 14, 128).to(device)
        xai_model.eval()
        with torch.no_grad():
            return torch.softmax(xai_model(x_t), dim=1).cpu().numpy()[:, target]
    bg = shap.sample(all_X_flat, 30)
    explainer = shap.KernelExplainer(predict_p1, bg)
    sv = explainer.shap_values(sample_flat, nsamples=nsamples, silent=True)
    return np.array(sv).reshape(14, 128)


EXPLAINERS_CHANNELTIME = {
    'IntegratedGradients': lambda s: explain_integrated_gradients(s),
    'Saliency': lambda s: explain_saliency(s),
    'Occlusion': lambda s: explain_occlusion(s),
    'DeepLift': lambda s: explain_deeplift(s),
    'FeatureAblation': lambda s: explain_feature_ablation_channel(s),
    'GradCAM': lambda s: explain_gradcam(s),
    'LRP': lambda s: explain_lrp(s),
}

print("Explainers έτοιμοι:", list(EXPLAINERS_CHANNELTIME.keys()) + ['SHAP(Gradient)', 'LIME', 'KernelSHAP'])


######################################################################
# ΜΠΛΟΚ 3 (μέρος 2) — Εφαρμογή: relevance ανά ηλεκτρόδιο/περιοχή + scalp maps
######################################################################

sample0 = explain_samples[0:1]
target0 = int(explain_labels[0].item())

ig_map = explain_integrated_gradients(sample0, target=target0)          # (14,128)
channel_importance = np.abs(ig_map).sum(axis=1)                          # (14,)

region_df, region_table = channel_to_region_table(channel_importance)
print(region_df.sort_values('value', ascending=False))
print("\nRelevance ανά περιοχή:\n", region_table)

plot_scalp_map(channel_importance, title=f"IG relevance — target class {target0}")

asym = frontal_asymmetry(channel_importance)
print("\nΜετωπιαία ασυμμετρία (δεξιά - αριστερή, >0 => δεξιά ισχυρότερη):")
for k, v in asym.items():
    print(f"  {k}: {v:+.4f}")


######################################################################
# ΞΕΧΩΡΙΣΤΟΣ ΤΟΠΟΓΡΑΦΙΚΟΣ ΧΑΡΤΗΣ ΚΑΙ TOP-5 ΓΙΑ ΚΑΘΕ ΜΕΘΟΔΟ XAI
# Μέχρι εδώ παραγόταν scalp map μόνο για το Integrated Gradients. Εδώ
# υπολογίζεται και απεικονίζεται ξεχωριστά κάθε μέθοδος, ώστε να είναι
# δυνατή η οπτική και ποσοτική σύγκρισή τους.
######################################################################
all_maps = {}
for name, fn in EXPLAINERS_CHANNELTIME.items():
    try:
        print(f"Υπολογισμός {name}...")
        all_maps[name] = fn(sample0)
    except Exception as e:
        print(f"  Το {name} απέτυχε ({e}); παραλείπεται.")

# Οι τρεις μέθοδοι που δέχονται διαφορετικά ορίσματα
sample0_flat = sample0.detach().reshape(1, -1).cpu().numpy()[0]
all_X_flat = background.detach().reshape(background.shape[0], -1).cpu().numpy()

for label, call in [
    ("SHAP(Gradient)", lambda: explain_shap_gradient(background, sample0, target=target0)),
    ("LIME", lambda: explain_lime(all_X_flat, sample0_flat, target=target0)),
    ("KernelSHAP", lambda: explain_kernelshap(all_X_flat, sample0_flat, target=target0)),
]:
    try:
        print(f"Υπολογισμός {label}...")
        all_maps[label] = call()
    except Exception as e:
        print(f"  Το {label} απέτυχε ({e}); παραλείπεται.")

# --- Έλεγχος και κανονικοποίηση σχήματος κάθε χάρτη σε (κανάλια, χρόνος) ---
# Ασφαλιστική δικλείδα: κάποιες βιβλιοθήκες επιστρέφουν τον χάρτη ανάστροφα ή με
# επιπλέον άξονα. Χωρίς αυτόν τον έλεγχο, το plot_topomap σκάει με λάθος πλήθος τιμών.
clean_maps = {}
for method_name, attr_map in all_maps.items():
    arr = np.asarray(attr_map)
    if arr.ndim != 2:
        print(f"  Το {method_name} έδωσε χάρτη σχήματος {arr.shape}· παραλείπεται.")
        continue
    if arr.shape[0] != len(DREAMER_CHANNELS) and arr.shape[1] == len(DREAMER_CHANNELS):
        print(f"  Το {method_name} έδωσε ανάστροφο χάρτη {arr.shape}· διορθώνεται.")
        arr = arr.T
    if arr.shape[0] != len(DREAMER_CHANNELS):
        print(f"  Το {method_name} έδωσε ασύμβατο χάρτη {arr.shape}· παραλείπεται.")
        continue
    clean_maps[method_name] = arr

all_maps = clean_maps
print(f"\nΜέθοδοι με έγκυρο χάρτη: {list(all_maps.keys())}")

print("\n===== Τοπογραφικοί χάρτες ανά μέθοδο XAI =====")
per_method_importance = {}
for method_name, attr_map in all_maps.items():
    imp = np.abs(attr_map).sum(axis=1)     # (14,)
    per_method_importance[method_name] = imp
    plot_scalp_map(imp, title=f"{method_name} — target class {target0}")

    top5_idx = np.argsort(imp)[::-1][:5]
    print(f"\nTop-5 ηλεκτρόδια ({method_name}):")
    for i in top5_idx:
        ch = DREAMER_CHANNELS[i]
        region, hemi = CHANNEL_REGION[ch][0], CHANNEL_REGION[ch][1]
        print(f"  {ch} ({region}-{hemi}): {imp[i]:.6f}")

# --- Συγκεντρωτικός πίνακας σημαντικότητας ανά ηλεκτρόδιο και μέθοδο ---
channel_scores_df = pd.DataFrame(per_method_importance, index=DREAMER_CHANNELS)
channel_scores_df.index.name = 'channel'
print("\n===== Σημαντικότητα ανά ηλεκτρόδιο και μέθοδο (ACTNN / DREAMER) =====")
display(channel_scores_df.round(6))
channel_scores_df.to_csv(f'{OUT_DIR}/xai_channel_scores_actnn_dreamer.csv')

# --- Μετωπιαία ασυμμετρία ανά μέθοδο ---
print("\n===== Μετωπιαία ασυμμετρία ανά μέθοδο (δεξιά - αριστερή) =====")
asym_rows = []
for method_name, imp in per_method_importance.items():
    a = frontal_asymmetry(imp)
    a['method'] = method_name
    asym_rows.append(a)
display(pd.DataFrame(asym_rows).set_index('method').round(6))

# --- Heatmaps Καναλιών x Χρόνου ανά μέθοδο ---
n_show = len(all_maps)
fig, axes = plt.subplots(n_show, 1, figsize=(12, 2.4 * n_show))
if n_show == 1:
    axes = [axes]
for ax, (name, m) in zip(axes, all_maps.items()):
    arr = np.asarray(m)
    vmax = np.abs(arr).max() or 1.0
    im = ax.imshow(arr, aspect='auto', cmap='RdBu_r', vmin=-vmax, vmax=vmax)
    ax.set_title(name, fontsize=10)
    ax.set_ylabel('Ηλεκτρόδια')
    ax.set_yticks(range(len(DREAMER_CHANNELS)))
    ax.set_yticklabels(DREAMER_CHANNELS, fontsize=6)
    fig.colorbar(im, ax=ax, fraction=0.02)
axes[-1].set_xlabel('Χρόνος (δείγματα)')
plt.tight_layout()
plt.show()


######################################################################
# ΜΠΛΟΚ 4 — Faithfulness: Deletion / Insertion tests + AOPC (+ τυχαία βάση)
######################################################################

K_VALUES = [5, 10, 20, 30, 50]


def predicted_prob(sample, target):
    with torch.no_grad():
        return torch.softmax(xai_model(sample), dim=1)[0, target].item()


def rank_flat_indices(attr_map, descending=True):
    flat = attr_map.flatten()
    order = np.argsort(-flat if descending else flat)
    return order  # indices into flattened (14*128) array, most-important first


def deletion_insertion_curve(sample, target, attr_map, mode='deletion', k_values=K_VALUES):
    order = rank_flat_indices(attr_map)
    n_total = order.size
    base = sample.clone()
    probs = []
    for k in k_values:
        n_k = int(n_total * k / 100)
        idx_to_touch = order[:n_k]
        perturbed = base.clone().flatten(2)  # (1,1,14*128) view via reshape below
        flat = perturbed.view(1, 1, -1).clone()
        if mode == 'deletion':
            flat[0, 0, idx_to_touch] = 0.0
        else:  # insertion: ξεκινάμε από μηδενικό σήμα, προσθέτουμε το top-k
            new_flat = torch.zeros_like(flat)
            new_flat[0, 0, idx_to_touch] = base.view(1, 1, -1)[0, 0, idx_to_touch]
            flat = new_flat
        perturbed_sample = flat.view_as(base)
        probs.append(predicted_prob(perturbed_sample, target))
    return np.array(probs)


def aopc(probs, base_prob):
    return float(np.mean(base_prob - probs)) if len(probs) else float('nan')


faithfulness_rows = []
for name, fn in EXPLAINERS_CHANNELTIME.items():
    for i in range(min(N_EXPLAIN_SAMPLES, len(explain_samples))):
        sample = explain_samples[i:i + 1]
        target = int(explain_labels[i].item())
        base_prob = predicted_prob(sample, target)
        attr_map = fn(sample)

        del_probs = deletion_insertion_curve(sample, target, attr_map, 'deletion')
        ins_probs = deletion_insertion_curve(sample, target, attr_map, 'insertion')

        rand_map = np.random.RandomState(i).rand(14, 128)
        del_probs_rand = deletion_insertion_curve(sample, target, rand_map, 'deletion')
        ins_probs_rand = deletion_insertion_curve(sample, target, rand_map, 'insertion')

        faithfulness_rows.append({
            'explainer': name, 'sample': i,
            'AOPC_deletion': aopc(del_probs, base_prob),
            # ΔΙΟΡΘΩΣΗ: η σειρά ορισμάτων ήταν ανάποδη -- aopc(probs, base_prob), όχι
            # aopc(base_prob, probs). Το base_prob είναι μονή τιμή (float), οπότε όταν
            # μπαίνει ΠΡΩΤΟ ως "probs", το len(probs) μέσα στο aopc() σκάει με
            # "TypeError: object of type 'float' has no len()".
            'AOPC_insertion': aopc(ins_probs, base_prob) * -1,  # insertion: υψηλότερο = καλύτερο
            'AOPC_deletion_random': aopc(del_probs_rand, base_prob),
            'AOPC_insertion_random': aopc(ins_probs_rand, base_prob) * -1,
        })

faithfulness_df = pd.DataFrame(faithfulness_rows)
faithfulness_summary = faithfulness_df.groupby('explainer').mean(numeric_only=True)
faithfulness_summary['deletion_beats_random'] = (
    faithfulness_summary['AOPC_deletion'] > faithfulness_summary['AOPC_deletion_random'])
faithfulness_summary.to_csv(f'{OUT_DIR}/faithfulness_AOPC.csv')
display(faithfulness_summary)


######################################################################
# ΜΠΛΟΚ 5 — Sanity checks (model & data randomization)
######################################################################

import copy

def cascading_randomization_test(model, sample, target, explain_fn_name='Saliency'):
    """Τυχαιοποιεί σταδιακά επίπεδα από την έξοδο προς την είσοδο, μετρά Spearman ρ
    του χάρτη με τον αρχικό (μη τυχαιοποιημένο) χάρτη."""
    def explain_with(m):
        sal = Saliency(m)
        attr = sal.attribute(sample.clone().requires_grad_(True), target=target)
        return attr.squeeze().detach().cpu().numpy().flatten()

    original_map = explain_with(model)

    layers_in_order = ['fc', 'attn', 'transformer', 'cnn']  # έξοδος -> είσοδος
    results = {'none_randomized': 1.0}
    m = copy.deepcopy(model)
    for layer_name in layers_in_order:
        layer = getattr(m, layer_name)
        for p in layer.parameters():
            nn.init.normal_(p, mean=0.0, std=0.1)
        new_map = explain_with(m)
        rho, _ = spearmanr(original_map, new_map)
        results[f'after_randomizing_{layer_name}'] = rho
    return results


rand_results = cascading_randomization_test(xai_model, explain_samples[0:1],
                                             int(explain_labels[0].item()))
print("Model-randomization test (Spearman ρ με τον αρχικό χάρτη):")
for k, v in rand_results.items():
    print(f"  {k}: {v:+.3f}")

plt.figure(figsize=(6, 3))
plt.plot(list(rand_results.keys()), list(rand_results.values()), 'o-')
plt.xticks(rotation=30, ha='right')
plt.ylabel('Spearman ρ με τον αρχικό χάρτη')
plt.title('Cascading model-randomization sanity check')
plt.axhline(0, color='gray', linestyle='--', linewidth=1)
plt.tight_layout(); plt.show()

# --- Data (label) randomization test ---
# Εκπαιδεύουμε ένα δεύτερο ACTNN με ανακατεμένες ετικέτες πάνω στο ίδιο train set, και
# συγκρίνουμε τους χάρτες saliency με το κανονικό μοντέλο.

shuffled_y = y_data.clone()
rng = np.random.RandomState(0)
perm = rng.permutation(len(shuffled_y))
shuffled_y = shuffled_y[perm]

_, s3_test_idx, _, _, _ = saved_models[any_s3_key]
s3_train_idx = np.setdiff1d(manifest.index.values, s3_test_idx)

shuffled_train_ds = TensorDataset(X_data[s3_train_idx], shuffled_y[s3_train_idx])
shuffled_loader = DataLoader(shuffled_train_ds, batch_size=BATCH_SIZE, shuffle=True)

shuffled_model = ACTNN().to(device)
opt = optim.Adam(shuffled_model.parameters(), lr=LR)
crit = nn.CrossEntropyLoss()
for epoch in range(N_EPOCHS):
    shuffled_model.train()
    for xb, yb in shuffled_loader:
        xb, yb = xb.to(device), yb.to(device)
        opt.zero_grad()
        loss = crit(shuffled_model(xb), yb)
        loss.backward()
        opt.step()

sal_normal = Saliency(xai_model)
sal_shuffled = Saliency(shuffled_model)
sample = explain_samples[0:1]
target = int(explain_labels[0].item())
map_normal = sal_normal.attribute(sample.clone().requires_grad_(True), target=target).squeeze().detach().cpu().numpy()
map_shuffled = sal_shuffled.attribute(sample.clone().requires_grad_(True), target=target).squeeze().detach().cpu().numpy()
rho_data, _ = spearmanr(map_normal.flatten(), map_shuffled.flatten())
print(f"Data-randomization test: Spearman ρ (κανονικό vs shuffled-labels μοντέλο) = {rho_data:+.3f}")
print("Αναμενόμενο: χαμηλό ρ. Αν παραμένει υψηλό, ο explainer αντικατοπτρίζει κυρίως τη δομή του σήματος.")


######################################################################
# ΜΠΛΟΚ 6 — Ποσοτική συμφωνία μεταξύ explainers
######################################################################

def rbo(list1, list2, p=0.9):
    """Rank-Biased Overlap (απλή, μη-εξωτερικευμένη υλοποίηση)."""
    s1, s2 = set(), set()
    score, weight = 0.0, 1.0
    for d in range(max(len(list1), len(list2))):
        if d < len(list1):
            s1.add(list1[d])
        if d < len(list2):
            s2.add(list2[d])
        overlap = len(s1 & s2)
        score += weight * (overlap / (d + 1))
        weight *= p
    return (1 - p) * score


def jaccard_topk(v1, v2, k):
    top1 = set(np.argsort(-np.abs(v1))[:k])
    top2 = set(np.argsort(-np.abs(v2))[:k])
    return len(top1 & top2) / len(top1 | top2)


# Υπολογισμός channel-importance (|attr| αθροισμένο στον χρόνο) ανά explainer, ίδιο δείγμα
channel_importances = {}
for name, fn in EXPLAINERS_CHANNELTIME.items():
    attr_map = fn(sample0)
    channel_importances[name] = np.abs(attr_map).sum(axis=1)

names = list(channel_importances.keys())
agree_spearman = pd.DataFrame(index=names, columns=names, dtype=float)
agree_kendall = pd.DataFrame(index=names, columns=names, dtype=float)
agree_jaccard5 = pd.DataFrame(index=names, columns=names, dtype=float)
agree_rbo = pd.DataFrame(index=names, columns=names, dtype=float)

for a, b in itertools.product(names, names):
    va, vb = channel_importances[a], channel_importances[b]
    agree_spearman.loc[a, b] = spearmanr(va, vb)[0]
    agree_kendall.loc[a, b] = kendalltau(va, vb)[0]
    agree_jaccard5.loc[a, b] = jaccard_topk(va, vb, 5)
    order_a = list(np.argsort(-np.abs(va)))
    order_b = list(np.argsort(-np.abs(vb)))
    agree_rbo.loc[a, b] = rbo(order_a, order_b)

print("Spearman ρ (channel importance):")
display(agree_spearman.round(2))
print("\nJaccard top-5 κανάλια:")
display(agree_jaccard5.round(2))

fig, ax = plt.subplots(figsize=(6, 5))
sns.heatmap(agree_spearman.astype(float), annot=True, fmt='.2f', cmap='coolwarm', vmin=-1, vmax=1, ax=ax)
ax.set_title('Συμφωνία explainers — Spearman ρ (channel importance)')
plt.tight_layout(); plt.show()


######################################################################
# ΜΠΛΟΚ 7 — Φασματική εξηγησιμότητα
######################################################################

def bandstop_filter(x, low, high, fs=FS, order=4):
    """x: (14, 128) numpy. Επιστρέφει το σήμα με αφαιρεμένη τη ζώνη [low, high]."""
    nyq = fs / 2
    low_n, high_n = max(low / nyq, 1e-3), min(high / nyq, 0.999)
    b, a = sps.butter(order, [low_n, high_n], btype='bandstop')
    return sps.filtfilt(b, a, x, axis=-1)


def bandpass_filter(x, low, high, fs=FS, order=4):
    nyq = fs / 2
    low_n, high_n = max(low / nyq, 1e-3), min(high / nyq, 0.999)
    b, a = sps.butter(order, [low_n, high_n], btype='bandpass')
    return sps.filtfilt(b, a, x, axis=-1)


# --- Band ablation: πτώση Macro F1 ανά ζώνη, στο test set του S3 μοντέλου ---
_, s3_test_idx_full, y_true_full, y_pred_full, _ = saved_models[any_s3_key]
X_test_np = X_data[s3_test_idx_full].numpy()  # (N,1,14,128)
y_test_np = y_data[s3_test_idx_full].numpy()

base_macro_f1 = f1_score(y_true_full, y_pred_full, average='macro', zero_division=0)
band_drop_rows = []
for band_name, (lo, hi) in BANDS.items():
    filtered = np.stack([bandstop_filter(x[0], lo, hi) for x in X_test_np]).astype(np.float32)
    filtered_t = torch.tensor(filtered[:, None, :, :]).to(device)
    with torch.no_grad():
        preds = xai_model(filtered_t).argmax(1).cpu().numpy()
    band_f1 = f1_score(y_test_np, preds, average='macro', zero_division=0)
    band_drop_rows.append({'band': band_name, 'range_hz': f"{lo}-{hi}",
                           'macro_f1_after_removal': band_f1, 'drop': base_macro_f1 - band_f1})

band_drop_df = pd.DataFrame(band_drop_rows).sort_values('drop', ascending=False)
print(f"Baseline Macro F1 (χωρίς αφαίρεση ζώνης): {base_macro_f1:.3f}")
display(band_drop_df)

# --- Band-specific Integrated Gradients + heatmap Channels × Bands ---
channels_x_bands = np.zeros((14, len(BANDS)))
sample_np = sample0.squeeze().detach().cpu().numpy()  # (14,128)

for b_idx, (band_name, (lo, hi)) in enumerate(BANDS.items()):
    band_signal = bandpass_filter(sample_np, lo, hi)
    band_tensor = torch.tensor(np.ascontiguousarray(band_signal[None, None, :, :]),
                               dtype=torch.float32, device=device)
    ig_band = explain_integrated_gradients(band_tensor, target=target0)
    channels_x_bands[:, b_idx] = np.abs(ig_band).sum(axis=1)

cxb_df = pd.DataFrame(channels_x_bands, index=DREAMER_CHANNELS, columns=list(BANDS.keys()))
plt.figure(figsize=(6, 5))
sns.heatmap(cxb_df, annot=True, fmt='.2f', cmap='viridis')
plt.title(f'Channels × Bands relevance (IG, target class {target0})')
plt.tight_layout(); plt.show()

fig, axes = plt.subplots(1, len(BANDS), figsize=(4 * len(BANDS), 4))
for ax, band_name in zip(axes, BANDS.keys()):
    plot_scalp_map(cxb_df[band_name].values, title=band_name, ax=ax)
plt.tight_layout(); plt.show()

fa_frontal = frontal_asymmetry(cxb_df['alpha'].values)
print("Frontal alpha asymmetry (δεξιά-αριστερή, ένδειξη valence στη βιβλιογραφία):")
for k, v in fa_frontal.items():
    print(f"  {k}: {v:+.4f}")


######################################################################
# ΜΠΛΟΚ 8 (προαιρετικό) — Έλεγχος για shortcuts (subject/session/trial identity probing)
######################################################################

def get_embeddings(model, idx):
    loader = DataLoader(TensorDataset(X_data[idx], y_data[idx]), batch_size=128, shuffle=False)
    model.eval()
    feats = []
    with torch.no_grad():
        for xb, _ in loader:
            feats.append(model.embed(xb.to(device)).cpu().numpy())
    return np.concatenate(feats)


probe_idx = manifest.index.values  # όλο το manifest, split ξανά με train_test_split παρακάτω
emb = get_embeddings(xai_model, probe_idx)

probe_targets = {
    'label (συναίσθημα)': manifest['label'].values,
    'subject_id (ταυτότητα)': manifest['subject_id'].values,
    'trial_id': manifest['trial_id'].values,
}

probe_results = {}
for target_name, target_vals in probe_targets.items():
    Xtr, Xte, ytr, yte = train_test_split(emb, target_vals, test_size=0.3, random_state=0,
                                           stratify=target_vals if target_name != 'trial_id' else None)
    clf = LogisticRegression(max_iter=1000)
    clf.fit(Xtr, ytr)
    acc = clf.score(Xte, yte)
    chance = 1.0 / len(np.unique(target_vals))
    probe_results[target_name] = {'probe_accuracy': acc, 'chance_level': chance}

probe_df = pd.DataFrame(probe_results).T
print("Probing accuracy (frozen S3-encoder embeddings):")
display(probe_df)

######################################################################
# ΑΝΑΛΥΣΗ ΟΜΟΙΟΤΗΤΑΣ ΑΝΑΠΑΡΑΣΤΑΣΕΩΝ (Linear CKA)
#
# Το CKA (Centered Kernel Alignment) μετρά πόσο όμοιες είναι δύο αναπαραστάσεις,
# σε κλίμακα 0 (εντελώς ανόμοιες) έως 1 (ταυτόσημες). Λειτουργεί ακόμα κι όταν
# οι δύο αναπαραστάσεις έχουν διαφορετικό πλήθος διαστάσεων, γι' αυτό επιτρέπει
# σύγκριση τόσο μεταξύ επιπέδων του ίδιου δικτύου όσο και μεταξύ διαφορετικών
# μοντέλων. Εδώ απαντώνται δύο ερωτήματα:
#   (α) σε ποιο στάδιο του δικτύου αλλάζει ουσιαστικά η πληροφορία, και
#   (β) μαθαίνουν τα μοντέλα S1 και S3 παρόμοιες αναπαραστάσεις;
######################################################################

def _center_gram(K):
    n = K.shape[0]
    H = np.eye(n) - np.ones((n, n)) / n
    return H @ K @ H


def linear_cka(X, Y):
    """Linear CKA μεταξύ δύο πινάκων αναπαραστάσεων (δείγματα x διαστάσεις).
    Τα δείγματα πρέπει να είναι τα ΙΔΙΑ και στη ΙΔΙΑ σειρά στους δύο πίνακες."""
    X = np.asarray(X, dtype=np.float64)
    Y = np.asarray(Y, dtype=np.float64)
    X = X - X.mean(axis=0, keepdims=True)
    Y = Y - Y.mean(axis=0, keepdims=True)
    Kx = _center_gram(X @ X.T)
    Ky = _center_gram(Y @ Y.T)
    hsic = float((Kx * Ky).sum())
    norm = float(np.sqrt((Kx * Kx).sum()) * np.sqrt((Ky * Ky).sum()))
    return hsic / norm if norm > 0 else float('nan')


def get_layer_reps(model, X_tensor, batch=64):
    """Εξάγει τις αναπαραστάσεις τριών σταδίων μέσω forward hooks, ώστε να μη
    χρειάζεται να αναπαραχθεί η εσωτερική λογική διαστάσεων του μοντέλου."""
    store, reps = {}, {'cnn': [], 'transformer': [], 'pooled': []}

    def mk_hook(name):
        def hook(_m, _i, out):
            store[name] = out.detach().cpu().numpy()
        return hook

    handles = [model.cnn.register_forward_hook(mk_hook('cnn')),
               model.transformer.register_forward_hook(mk_hook('transformer'))]
    model.eval()
    try:
        with torch.no_grad():
            for i in range(0, len(X_tensor), batch):
                xb = X_tensor[i:i + batch].to(device)
                pooled = model.embed(xb).cpu().numpy()
                reps['cnn'].append(store['cnn'].reshape(len(xb), -1))
                tr = store['transformer']
                reps['transformer'].append(tr.reshape(len(xb), -1))
                reps['pooled'].append(pooled)
    finally:
        for h in handles:
            h.remove()
    return {k: np.concatenate(v) for k, v in reps.items()}


# --- Κοινό υποσύνολο δειγμάτων: ίδια παράθυρα για κάθε μοντέλο που συγκρίνεται ---
rng_cka = np.random.RandomState(0)
cka_idx = rng_cka.choice(xai_test_idx, size=min(500, len(xai_test_idx)), replace=False)
X_cka = X_data[cka_idx]

reps_s3 = get_layer_reps(xai_model, X_cka)
for k, v in reps_s3.items():
    print(f"Αναπαράσταση '{k}': {v.shape}")

# --- (α) Ομοιότητα μεταξύ επιπέδων του μοντέλου S3 ---
layer_names = ['cnn', 'transformer', 'pooled']
cka_layers = pd.DataFrame(index=layer_names, columns=layer_names, dtype=float)
for a, b in itertools.product(layer_names, layer_names):
    cka_layers.loc[a, b] = linear_cka(reps_s3[a], reps_s3[b])

print("\n===== CKA μεταξύ επιπέδων (μοντέλο S3) =====")
display(cka_layers.round(3))

fig, ax = plt.subplots(figsize=(4.5, 3.8))
sns.heatmap(cka_layers.astype(float), annot=True, fmt='.3f', cmap='viridis',
            vmin=0, vmax=1, ax=ax)
ax.set_title('Linear CKA μεταξύ επιπέδων (S3)')
plt.tight_layout(); plt.show()

# --- (β) Ομοιότητα ίδιου επιπέδου μεταξύ μοντέλων S1 και S3 ---
s1_key = next((k for k in saved_models if k[0] == 'S1'), None)
model_s1 = saved_models[s1_key][0] if s1_key else None

if model_s1 is None:
    print("\nΤο μοντέλο S1 δεν είναι διαθέσιμο· η σύγκριση S1 vs S3 παραλείπεται.")
else:
    reps_s1 = get_layer_reps(model_s1, X_cka)
    rows = [{'layer': ln, 'CKA_S1_vs_S3': linear_cka(reps_s1[ln], reps_s3[ln])}
            for ln in layer_names]
    cka_cross = pd.DataFrame(rows).set_index('layer')
    print("\n===== CKA ίδιου επιπέδου: μοντέλο S1 έναντι μοντέλου S3 =====")
    display(cka_cross.round(3))
    print("Χαμηλές τιμές σημαίνουν ότι τα δύο πρωτόκολλα οδηγούν σε ποιοτικά")
    print("διαφορετικές αναπαραστάσεις, παρότι η αρχιτεκτονική είναι ίδια.")

    fig, ax = plt.subplots(figsize=(5, 3.2))
    cka_cross['CKA_S1_vs_S3'].plot(kind='bar', ax=ax, color='#4C72B0')
    ax.set_ylim(0, 1); ax.set_ylabel('Linear CKA')
    ax.set_title('Ομοιότητα αναπαραστάσεων S1 vs S3')
    plt.xticks(rotation=0); plt.tight_layout(); plt.show()

    cka_cross.to_csv(f'{OUT_DIR}/cka_s1_vs_s3.csv')

cka_layers.to_csv(f'{OUT_DIR}/cka_between_layers.csv')
# =====================================================================
# LAYER-WISE PROBING
# Γραμμικοί ταξινομητές πάνω στις παγωμένες αναπαραστάσεις ΚΑΘΕ σταδίου,
# ώστε να φανεί σε ποιο σημείο του δικτύου η κάθε πληροφορία γίνεται
# γραμμικά αποκωδικοποιήσιμη. Συγκρίνεται η συναισθηματική ετικέτα με
# μεταβλητές που το μοντέλο ΔΕΝ θα έπρεπε να κωδικοποιεί (ταυτότητα
# εθελοντή, δοκιμασία), οπότε υψηλή ακρίβεια σε αυτές αποτελεί ένδειξη
# συντόμευσης μέσω Διαρροής Ταυτότητας Υποκειμένου.
# =====================================================================
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split as _tts

META = manifest            # <-- στα δύο CCNN notebooks άλλαξέ το σε: META = info

probe_targets = {'label (συναίσθημα)': 'label',
                 'subject_id (ταυτότητα)': 'subject_id',
                 'trial_id (δοκιμασία)': 'trial_id'}
if 'session_id' in META.columns and META['session_id'].nunique() > 1:
    probe_targets['session_id (συνεδρία)'] = 'session_id'

probe_rows = []
for stage, feats in reps_s3.items():
    if np.allclose(feats.std(axis=0), 0):
        print(f"  Το στάδιο '{stage}' έχει μηδενική διασπορά· παραλείπεται.")
        continue
    for label_name, col in probe_targets.items():
        y = META.loc[cka_idx, col].values
        classes, counts = np.unique(y, return_counts=True)
        if len(classes) < 2 or counts.min() < 2:
            continue
        strat = y if counts.min() >= 2 else None
        Xtr, Xte, ytr, yte = _tts(feats, y, test_size=0.3, random_state=0, stratify=strat)
        clf = LogisticRegression(max_iter=1000).fit(Xtr, ytr)
        acc = clf.score(Xte, yte)
        chance = counts.max() / counts.sum()      # πλειοψηφική κλάση ως γραμμή βάσης
        probe_rows.append({'stage': stage, 'target': label_name,
                           'probe_accuracy': acc, 'chance_level': chance,
                           'ratio': acc / chance if chance > 0 else float('nan')})

probe_df = pd.DataFrame(probe_rows)
if not probe_df.empty:
    table = probe_df.pivot(index='stage', columns='target', values='probe_accuracy')
    print("\n===== Layer-wise probing: ακρίβεια γραμμικού ταξινομητή ανά στάδιο =====")
    print(table.round(4))
    print("\nΛόγος προς τη γραμμή βάσης (τιμές >>1 υποδηλώνουν αποκωδικοποιήσιμη πληροφορία):")
    print(probe_df.pivot(index='stage', columns='target', values='ratio').round(2))
    probe_df.to_csv('layerwise_probing.csv', index=False)

    fig, ax = plt.subplots(figsize=(1.6 * len(table) + 3, 3.4))
    table.plot(kind='bar', ax=ax)
    ax.set_ylabel('Ακρίβεια probe'); ax.set_ylim(0, 1)
    ax.set_title('Layer-wise probing ανά στάδιο')
    plt.xticks(rotation=0); plt.legend(fontsize=8); plt.tight_layout(); plt.show()
```
