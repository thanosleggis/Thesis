```python
# =====================================================================
# CNN-LSTM / DEAP -- ΕΝΙΑΙΟ, ΔΙΟΡΘΩΜΕΝΟ SCRIPT
# Κρατάει: πλήρες σετ μετρικών, calibration, 5 μεθόδους XAI (LRP/InputxGrad,
# SHAP, Grad-CAM, Occlusion, LIME), γράφημα ισορροπίας ανά subject.
# Προσθέτει: S1/S2/S3 πρωτόκολλο, ονόματα ηλεκτροδίων + scalp maps +
# frontal asymmetry, Faithfulness/AOPC, sanity checks, inter-explainer
# agreement, φασματική ανάλυση, bootstrap/Wilcoxon, ΚΑΙ διορθώνει το
# fine-tuning leakage (target subject τώρα επιλέγεται από πραγματικά
# held-out S3 υποκείμενο, όχι τυχαία από όλο το dataset).
# ΝΕΟ σε αυτή την έκδοση: σωστό install ανά πακέτο (lime/mne δεν χάνονταν
# πια σιωπηλά), έλεγχος GPU στην αρχή, progress bar + χρονόμετρο ανά fold,
# άμεση εκτύπωση αποτελεσμάτων μετά από κάθε fold (όχι μόνο στο τέλος).
# =====================================================================

import os, random, copy, itertools, time
import pickle
import numpy as np
import pandas as pd
import tensorflow as tf
from tensorflow.keras.utils import to_categorical
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, Conv1D, MaxPooling1D, LSTM, Dense, Dropout, BatchNormalization
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.callbacks import ModelCheckpoint, EarlyStopping

from scipy import signal as sps
from scipy.stats import spearmanr, wilcoxon
from sklearn.model_selection import train_test_split, GroupKFold, LeaveOneGroupOut
from sklearn.manifold import TSNE
from sklearn.metrics import (classification_report, confusion_matrix, f1_score,
                              balanced_accuracy_score, matthews_corrcoef,
                              roc_auc_score, cohen_kappa_score, brier_score_loss,
                              precision_recall_fscore_support)
from sklearn.calibration import calibration_curve
from statsmodels.stats.multitest import multipletests
from collections import Counter

import matplotlib.pyplot as plt
import seaborn as sns

import subprocess, sys

def ensure_installed(pip_name, import_name=None):
    import_name = import_name or pip_name
    try:
        __import__(import_name)
    except ImportError:
        print(f" Εγκατάσταση {pip_name}...")
        subprocess.run([sys.executable, '-m', 'pip', 'install', '-q', pip_name], check=True)

for pip_name in ['shap', 'lime', 'mne']:
    ensure_installed(pip_name)

import shap
import lime
import lime.lime_tabular
import scipy.ndimage
import mne

gpu_devices = tf.config.list_physical_devices('GPU')
print(f" GPU διαθέσιμη: {gpu_devices if gpu_devices else 'ΟΧΙ -- τρέχει σε CPU!'}")

# --- Εντοπισμός φακέλου με τα .dat του DEAP: λειτουργεί σε Kaggle ΚΑΙ σε Colab ---
DATA_PATH = None
for root, _, files in os.walk('/kaggle/input'):
    if any(f.endswith('.dat') for f in files):
        DATA_PATH = root
        break

if DATA_PATH is None:
    try:
        from google.colab import drive as _colab_drive
        if not os.path.exists('/content/drive'):
            _colab_drive.mount('/content/drive', force_remount=True)
        for p in ['/content/drive/MyDrive/DEAP.mini', '/content/drive/MyDrive/DEAP_mini',
                  '/content/drive/MyDrive/deap_mini']:
            if os.path.exists(p) and any(f.endswith('.dat') for f in os.listdir(p)):
                DATA_PATH = p
                break
    except ImportError:
        pass

if DATA_PATH is None:
    raise FileNotFoundError(
        "Δεν βρέθηκε φάκελος με αρχεία .dat του DEAP -- ούτε σε /kaggle/input, ούτε σε Colab Drive. "
        "Αν τρέχεις σε Kaggle, πρόσθεσε το dataset στο notebook (Add Input)."
    )
print(f" Βρέθηκε ο φάκελος δεδομένων: {DATA_PATH}")

OUT_DIR = '/kaggle/working/outputs' if os.path.exists('/kaggle/working') else '/content/outputs'
os.makedirs(OUT_DIR, exist_ok=True)
print(f" Τα αποτελέσματα θα αποθηκευτούν στο: {OUT_DIR}")

# --- ΡΥΘΜΙΣΕΙΣ ---
WINDOW_SIZE = 128
STRIDE = 64
CHANNELS = 32
BASELINE_SAMPLES = 384   # 3 δευτερόλεπτα baseline στα 128Hz, αφαιρούνται πριν το sliding window
EPOCHS = 20
BATCH_SIZE = 128

RUN_FULL_LOSO = False
LOSO_MAX_SUBJECTS = 5
N_EXPLAIN_SAMPLES = 2
N_BOOTSTRAP = 1000
CI_ALPHA = 0.05
K_VALUES = [5, 10, 20, 30, 50]

DREAMER_CHANNELS = [   # όνομα κρατιέται ίδιο για να μην αλλάξει τίποτα άλλο στο script· είναι τα κανάλια DEAP
    'Fp1', 'AF3', 'F3', 'F7', 'FC5', 'FC1', 'C3', 'T7', 'CP5', 'CP1', 'P3', 'P7',
    'PO3', 'O1', 'Oz', 'Pz', 'Fp2', 'AF4', 'Fz', 'F4', 'F8', 'FC6', 'FC2', 'Cz',
    'C4', 'T8', 'CP6', 'CP2', 'P4', 'P8', 'PO4', 'O2',
]

CHANNEL_REGION = {
    'Fp1': ('frontal', 'left'), 'Fp2': ('frontal', 'right'),
    'AF3': ('frontal', 'left'), 'AF4': ('frontal', 'right'),
    'F3': ('frontal', 'left'), 'F4': ('frontal', 'right'),
    'F7': ('frontal', 'left'), 'F8': ('frontal', 'right'), 'Fz': ('frontal', 'mid'),
    'FC5': ('central', 'left'), 'FC1': ('central', 'left'),
    'FC6': ('central', 'right'), 'FC2': ('central', 'right'),
    'C3': ('central', 'left'), 'C4': ('central', 'right'), 'Cz': ('central', 'mid'),
    'T7': ('temporal', 'left'), 'T8': ('temporal', 'right'),
    'CP5': ('parietal', 'left'), 'CP1': ('parietal', 'left'),
    'CP6': ('parietal', 'right'), 'CP2': ('parietal', 'right'),
    'P3': ('parietal', 'left'), 'P4': ('parietal', 'right'),
    'P7': ('parietal', 'left'), 'P8': ('parietal', 'right'), 'Pz': ('parietal', 'mid'),
    'PO3': ('occipital', 'left'), 'PO4': ('occipital', 'right'),
    'O1': ('occipital', 'left'), 'O2': ('occipital', 'right'), 'Oz': ('occipital', 'mid'),
}
BANDS = {'delta': (1, 4), 'theta': (4, 8), 'alpha': (8, 13), 'beta': (13, 30), 'gamma': (30, 50)}
FS = 128


# =====================================================================
# ΒΗΜΑ 1 -- Φόρτωση δεδομένων + manifest
# =====================================================================
def load_deap_with_manifest(folder):
    print(f" Ξεκινάει η φόρτωση των εθελοντών DEAP από {folder}... (μερικά λεπτά)")
    files = sorted(f for f in os.listdir(folder) if f.endswith('.dat'))

    X_list, y_list, rows = [], [], []
    wid = 0
    for fi, filename in enumerate(files):
        subject_id = filename.replace('.dat', '')   # π.χ. "s01"
        try:
            with open(os.path.join(folder, filename), 'rb') as f:
                content = pickle.load(f, encoding='latin1')
            data, labels = content['data'], content['labels']
        except Exception as e:
            print(f" Σφάλμα στο {filename}: {e}")
            continue

        for trial in range(len(data)):
            trial_signal = data[trial, :32, :].T.astype('float32')      # (χρόνος, 32 κανάλια)
            clean_signal = trial_signal[BASELINE_SAMPLES:, :]           # αφαίρεση baseline
            label = 1 if labels[trial, 0] >= 5.0 else 0                 # Binary(5.0) -- κλίμακα DEAP

            for start in range(0, clean_signal.shape[0] - WINDOW_SIZE, STRIDE):
                window = clean_signal[start:start + WINDOW_SIZE, :]
                if window.shape[0] != WINDOW_SIZE:
                    continue
                window = (window - np.mean(window)) / (np.std(window) + 1e-6)
                X_list.append(window)
                y_list.append(label)
                rows.append({'window_id': wid, 'subject_id': subject_id, 'trial_id': trial,
                             'label': label, 'start_sample': start})
                wid += 1

    X = np.array(X_list, dtype='float32')
    y = np.array(y_list, dtype='int64')
    manifest = pd.DataFrame(rows)
    print(" Η επεξεργασία (Sliding Window) ολοκληρώθηκε!")
    return X, y, manifest


X_data, y_data, manifest = load_deap_with_manifest(DATA_PATH)
manifest.to_csv(f'{OUT_DIR}/manifest_deap_cnnlstm.csv', index=False)
print(f"\n Σύνολο παραθύρων: {len(y_data)} | High Valence: {y_data.mean():.1%} "
      f"| Εθελοντές: {manifest['subject_id'].nunique()}")

# --- Γράφημα ισορροπίας κλάσεων ανά εθελοντή ---
subj_balance = manifest.groupby('subject_id')['label'].mean().sort_values() * 100
plt.figure(figsize=(12, 5))
plt.bar(range(len(subj_balance)), subj_balance.values,
        tick_label=list(subj_balance.index), color='steelblue')
plt.axhline(50, color='red', linestyle='--', label='Ισορροπία (50%)')
plt.ylabel('% High Valence'); plt.xlabel('Subject ID')
plt.title('Ισορροπία κλάσεων ανά εθελοντή (DEAP)')
plt.xticks(rotation=90); plt.legend(); plt.tight_layout(); plt.show()

skewed = subj_balance[(subj_balance < 30) | (subj_balance > 70)]
if len(skewed):
    print(" Προσοχή -- έντονη ανισορροπία κλάσεων:")
    print(skewed)


# =====================================================================
# ΒΗΜΑ 2 -- S1/S2/S3 splits
# =====================================================================
def s1_window_split(manifest, seed):
    idx = manifest.index.values
    train_idx, test_idx = train_test_split(idx, test_size=0.2, random_state=seed,
                                            stratify=manifest['label'].values)
    yield train_idx, test_idx, 'S1_full'


def s2_trial_split(manifest, seed, n_splits=5):
    groups = manifest['subject_id'].astype(str) + '_' + manifest['trial_id'].astype(str)
    gkf = GroupKFold(n_splits=n_splits)
    idx = manifest.index.values
    for fold, (tr, te) in enumerate(gkf.split(idx, manifest['label'].values, groups=groups)):
        yield idx[tr], idx[te], f'S2_fold{fold}'


def s3_subject_split(manifest, seed, run_full=RUN_FULL_LOSO, max_subjects=LOSO_MAX_SUBJECTS):
    logo = LeaveOneGroupOut()
    idx = manifest.index.values
    groups = manifest['subject_id'].values
    subjects = sorted(manifest['subject_id'].unique())
    if not run_full:
        rng = np.random.RandomState(seed)
        subjects = list(rng.choice(subjects, size=min(max_subjects, len(subjects)), replace=False))
    for tr, te in logo.split(idx, manifest['label'].values, groups=groups):
        held_out = manifest.iloc[te]['subject_id'].iloc[0]
        if held_out not in subjects:
            continue
        yield idx[tr], idx[te], f'S3_subject{held_out}'


# =====================================================================
# ΒΗΜΑ 3 -- Μοντέλο
# =====================================================================
def build_cnn_lstm(n_channels=CHANNELS, window_size=WINDOW_SIZE):
    inputs = Input(shape=(window_size, n_channels))
    x = Conv1D(64, 3, activation='relu', padding='same')(inputs)
    x = BatchNormalization()(x)
    x = MaxPooling1D(2)(x)
    x = Dropout(0.3)(x)
    x = Conv1D(128, 3, activation='relu', padding='same')(x)
    x = BatchNormalization()(x)
    x = MaxPooling1D(2)(x)
    x = Dropout(0.3)(x)
    x = LSTM(128)(x)
    x = Dropout(0.4)(x)
    x = Dense(64, activation='relu')(x)
    outputs = Dense(2, activation='softmax')(x)
    model = Model(inputs, outputs)
    model.compile(optimizer=Adam(0.001), loss='categorical_crossentropy', metrics=['accuracy'])
    return model


# =====================================================================
# ΒΗΜΑ 4 -- Μετρικές
# =====================================================================
def majority_class_baseline(y_true):
    vals, counts = np.unique(y_true, return_counts=True)
    return counts.max() / counts.sum()


def full_metrics(y_true, y_pred, y_prob_pos):
    prec, rec, f1c, _ = precision_recall_fscore_support(y_true, y_pred, labels=[0, 1], zero_division=0)
    out = {
        'majority_baseline': majority_class_baseline(y_true),
        'balanced_accuracy': balanced_accuracy_score(y_true, y_pred),
        'macro_f1': f1_score(y_true, y_pred, average='macro', zero_division=0),
        'precision_class0': prec[0], 'recall_class0': rec[0],
        'precision_class1': prec[1], 'recall_class1': rec[1],
        'cohen_kappa': cohen_kappa_score(y_true, y_pred),
        'mcc': matthews_corrcoef(y_true, y_pred),
    }
    try:
        out['roc_auc'] = roc_auc_score(y_true, y_prob_pos)
    except ValueError:
        out['roc_auc'] = float('nan')
    out['learned_anything'] = out['balanced_accuracy'] > out['majority_baseline'] + 0.02
    return out


def plot_confusion(y_true, y_pred, title, ax=None):
    cm = confusion_matrix(y_true, y_pred, labels=[0, 1])
    fig = None
    if ax is None:
        fig, ax = plt.subplots(figsize=(3.5, 3))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', ax=ax,
                xticklabels=['Low', 'High'], yticklabels=['Low', 'High'])
    ax.set_xlabel('Predicted'); ax.set_ylabel('True'); ax.set_title(title, fontsize=10)
    if fig is not None:
        plt.show()
    return cm


def expected_calibration_error(y_true, y_proba_pos, n_bins=10):
    bins = np.linspace(0.0, 1.0, n_bins + 1)
    ece = 0.0
    n = len(y_true)
    for i in range(n_bins):
        mask = (y_proba_pos >= bins[i]) & (y_proba_pos < bins[i + 1] if i < n_bins - 1 else y_proba_pos <= bins[i+1])
        if mask.sum() == 0:
            continue
        acc_bin = (y_true[mask] == (y_proba_pos[mask] >= 0.5).astype(int)).mean()
        conf_bin = y_proba_pos[mask].mean()
        ece += (mask.sum() / n) * abs(acc_bin - conf_bin)
    return ece


def reliability_diagram(y_true, y_proba_pos, title):
    prob_true, prob_pred = calibration_curve(y_true, y_proba_pos, n_bins=10)
    plt.figure(figsize=(4, 4))
    plt.plot([0, 1], [0, 1], 'k--', label='ideal')
    plt.plot(prob_pred, prob_true, 'o-', label='model')
    plt.xlabel('Confidence'); plt.ylabel('Empirical accuracy'); plt.title(title); plt.legend()
    plt.show()
    ece = expected_calibration_error(y_true, y_proba_pos)
    brier = brier_score_loss(y_true, y_proba_pos)
    print(f"ECE = {ece:.4f} | Brier = {brier:.4f}")
    return ece, brier


def per_subject_accuracy(y_true, y_pred, subject_ids):
    df = pd.DataFrame({'y': y_true, 'p': y_pred, 'subject': subject_ids})
    return df.groupby('subject').apply(lambda g: (g['y'] == g['p']).mean())


# =====================================================================
# ΒΗΜΑ 5 -- Εκπαίδευση ενός split
# =====================================================================
def train_one_split(train_idx, test_idx, seed, n_epochs=EPOCHS, fold_name=None):
    np.random.seed(seed); tf.random.set_seed(seed)
    if fold_name:
        print(f"-- {fold_name} -- ({len(train_idx)} train / {len(test_idx)} test παράθυρα)")

    Xtr, ytr = X_data[train_idx], to_categorical(y_data[train_idx], 2)
    Xte, yte = X_data[test_idx], to_categorical(y_data[test_idx], 2)

    model = build_cnn_lstm()
    callbacks = [EarlyStopping(monitor='val_accuracy', patience=5, restore_best_weights=True, mode='max')]

    t0 = time.time()
    history = model.fit(Xtr, ytr, epochs=n_epochs, batch_size=BATCH_SIZE,
                        validation_data=(Xte, yte), callbacks=callbacks, verbose=1)

    tr_acc = history.history['accuracy']
    va_acc = history.history['val_accuracy']
    print(f"   Train accuracy: {tr_acc[0]:.2%} (1η εποχή) -> {tr_acc[-1]:.2%} (τελευταία) "
          f"| Val accuracy: {va_acc[0]:.2%} -> {va_acc[-1]:.2%}")
    print(f"   (fold '{fold_name}' ολοκληρώθηκε σε {time.time() - t0:.1f} δευτ.)")

    y_prob = model.predict(Xte, verbose=0)
    y_pred = np.argmax(y_prob, axis=1)
    y_true = y_data[test_idx]
    metrics = full_metrics(y_true, y_pred, y_prob[:, 1])
    return model, y_true, y_pred, y_prob[:, 1], metrics


# =====================================================================
# ΒΗΜΑ 6 -- Εκτέλεση S1/S2/S3, κεντρικός πίνακας, bootstrap, Wilcoxon
# =====================================================================
SEED = 0
results = []
saved_models = {}

print("\n=== S1 (window-level, αντιπαράδειγμα) ===")
for train_idx, test_idx, fold_name in s1_window_split(manifest, SEED):
    model, y_true, y_pred, y_prob, m = train_one_split(train_idx, test_idx, SEED, fold_name=fold_name)
    m.update({'model': 'CNN-LSTM', 'dataset': 'DEAP', 'split_level': 'S1', 'fold': fold_name})
    results.append(m)
    saved_models[('S1',)] = (model, test_idx, y_true, y_pred, y_prob)
    print(f"   -> Balanced Acc: {m['balanced_accuracy']:.2%} | Majority baseline: {m['majority_baseline']:.2%} "
          f"| Macro F1: {m['macro_f1']:.2%}")

print("\n=== S2 (trial-level, 5-fold GroupKFold) ===")
for train_idx, test_idx, fold_name in s2_trial_split(manifest, SEED, n_splits=5):
    model, y_true, y_pred, y_prob, m = train_one_split(train_idx, test_idx, SEED, fold_name=fold_name)
    m.update({'model': 'CNN-LSTM', 'dataset': 'DEAP', 'split_level': 'S2', 'fold': fold_name})
    results.append(m)
    saved_models[('S2', fold_name)] = (model, test_idx, y_true, y_pred, y_prob)
    print(f"   -> Balanced Acc: {m['balanced_accuracy']:.2%} | Majority baseline: {m['majority_baseline']:.2%} "
          f"| Macro F1: {m['macro_f1']:.2%}")

print("\n=== S3 (subject-level, LOSO) ===")
for train_idx, test_idx, fold_name in s3_subject_split(manifest, SEED):
    model, y_true, y_pred, y_prob, m = train_one_split(train_idx, test_idx, SEED, fold_name=fold_name)
    m.update({'model': 'CNN-LSTM', 'dataset': 'DEAP', 'split_level': 'S3', 'fold': fold_name})
    results.append(m)
    saved_models[('S3', fold_name)] = (model, test_idx, y_true, y_pred, y_prob)
    print(f"   -> Balanced Acc: {m['balanced_accuracy']:.2%} | Majority baseline: {m['majority_baseline']:.2%} "
          f"| Macro F1: {m['macro_f1']:.2%}")

results_df = pd.DataFrame(results)
results_df.to_csv(f'{OUT_DIR}/results_S1_S2_S3_deap_cnnlstm.csv', index=False)
print(results_df)

summary = (results_df.groupby('split_level')
           .agg(balanced_accuracy_mean=('balanced_accuracy', 'mean'),
                balanced_accuracy_std=('balanced_accuracy', 'std'),
                macro_f1_mean=('macro_f1', 'mean'),
                majority_baseline_mean=('majority_baseline', 'mean'))
           .reset_index())
s1_acc = summary.loc[summary.split_level == 'S1', 'balanced_accuracy_mean'].values
s3_acc = summary.loc[summary.split_level == 'S3', 'balanced_accuracy_mean'].values
drop = (s1_acc[0] - s3_acc[0]) if len(s1_acc) and len(s3_acc) else float('nan')
print(f"\nΠτώση S1→S3 (balanced accuracy): {drop:.2%}")
print(summary)

any_s3_key = next(k for k in saved_models if k[0] == 'S3')
_, s3_test_idx, y_true_s3, y_pred_s3, y_prob_s3 = saved_models[any_s3_key]


def bootstrap_ci(y_true, y_pred, metric_fn, n_boot=N_BOOTSTRAP, alpha=CI_ALPHA, seed=0):
    rng = np.random.RandomState(seed)
    n = len(y_true)
    stats = [metric_fn(y_true[rng.randint(0, n, n)], y_pred[rng.randint(0, n, n)]) for _ in range(n_boot)]
    lo, hi = np.percentile(stats, [100 * alpha / 2, 100 * (1 - alpha / 2)])
    return np.mean(stats), lo, hi


mean_ba, lo, hi = bootstrap_ci(y_true_s3, y_pred_s3, balanced_accuracy_score)
print(f"S3 balanced accuracy: {mean_ba:.2%} (95% CI [{lo:.2%}, {hi:.2%}])")

s1_test_idx_full = list(s1_window_split(manifest, SEED))[0][1]
_, _, s1_true, s1_pred, _ = saved_models[('S1',)]
s1_subjects = manifest.loc[s1_test_idx_full, 'subject_id'].values
s1_per_subj = per_subject_accuracy(s1_true, s1_pred, s1_subjects)

s3_per_subj = {}
for k in [k for k in saved_models if k[0] == 'S3']:
    _, test_idx, yte, ypred, _ = saved_models[k]
    subj = manifest.loc[test_idx, 'subject_id'].iloc[0]
    s3_per_subj[subj] = (yte == ypred).mean()
s3_per_subj = pd.Series(s3_per_subj)

common = s1_per_subj.index.intersection(s3_per_subj.index)
if len(common) >= 5:
    stat, p = wilcoxon(s1_per_subj.loc[common], s3_per_subj.loc[common])
    reject, p_adj, _, _ = multipletests([p], method='holm')
    print(f"Wilcoxon S1 vs S3 (n={len(common)}): stat={stat:.3f}, p={p:.4f}, "
          f"p_holm={p_adj[0]:.4f}, σημαντικό={reject[0]}")
else:
    print("Πολύ λίγα κοινά υποκείμενα για Wilcoxon (αύξησε LOSO_MAX_SUBJECTS).")

fig, axes = plt.subplots(1, 3, figsize=(11, 3.5))
for ax, level in zip(axes, ['S1', 'S2', 'S3']):
    key = next((k for k in saved_models if k[0] == level), None)
    if key is None:
        continue
    _, _, yt, yp, _ = saved_models[key]
    plot_confusion(yt, yp, title=level, ax=ax)
plt.tight_layout(); plt.show()

reliability_diagram(y_true_s3, y_prob_s3, title='Reliability -- S3 (LOSO)')


# =====================================================================
# ΒΗΜΑ 7 -- t-SNE (πάνω στο S3 μοντέλο)
# =====================================================================
xai_model, xai_test_idx, _, _, _ = saved_models[any_s3_key]

feature_model = Model(inputs=xai_model.inputs, outputs=xai_model.layers[-2].output)
X_test_xai = X_data[xai_test_idx]
sample_size = min(1000, len(X_test_xai))
idx = np.random.choice(len(X_test_xai), sample_size, replace=False)
features = feature_model.predict(X_test_xai[idx], verbose=0)
tsne_results = TSNE(n_components=2, random_state=42, perplexity=30).fit_transform(features)

plt.figure(figsize=(8, 6))
scatter = plt.scatter(tsne_results[:, 0], tsne_results[:, 1], c=y_data[xai_test_idx][idx], cmap='bwr', alpha=0.6)
plt.legend(handles=scatter.legend_elements()[0], labels=['Low Valence', 'High Valence'])
plt.title('t-SNE -- S3 (LOSO) μοντέλο'); plt.show()

# =====================================================================
# ΠΟΣΟΤΙΚΑ ΜΕΤΡΑ ΔΙΑΧΩΡΙΣΙΜΟΤΗΤΑΣ
# Το t-SNE είναι μόνο οπτική ένδειξη και μπορεί να "δείξει" ομάδες που δεν
# αντιστοιχούν σε πραγματική δομή, ανάλογα με το perplexity. Τα μέτρα αυτά
# τεκμηριώνουν αριθμητικά τον βαθμό διαχωρισμού των δύο κλάσεων.
# =====================================================================
from sklearn.metrics import silhouette_score, davies_bouldin_score

labels_feat = y_data[xai_test_idx][idx]


def fisher_ratio(feats, labels):
    """Λόγος διασποράς μεταξύ κλάσεων προς εντός κλάσεων (μεγαλύτερο = πιο διαχωρίσιμο)."""
    classes = np.unique(labels)
    overall_mean = feats.mean(axis=0)
    between, within = 0.0, 0.0
    for c in classes:
        fc = feats[labels == c]
        between += len(fc) * np.sum((fc.mean(axis=0) - overall_mean) ** 2)
        within += np.sum((fc - fc.mean(axis=0)) ** 2)
    return float(between / within) if within > 0 else float('nan')


def class_distances(feats, labels):
    """Μέση ενδο-κλασική και δια-κλασική απόσταση."""
    classes = np.unique(labels)
    centroids = {c: feats[labels == c].mean(axis=0) for c in classes}
    intra = np.mean([np.linalg.norm(feats[labels == c] - centroids[c], axis=1).mean()
                      for c in classes])
    inter = np.mean([np.linalg.norm(centroids[a] - centroids[b])
                      for i, a in enumerate(classes) for b in classes[i + 1:]])
    return float(intra), float(inter)


sil = silhouette_score(features, labels_feat)
db = davies_bouldin_score(features, labels_feat)
fr = fisher_ratio(features, labels_feat)
intra, inter = class_distances(features, labels_feat)

print("\n===== Ποσοτικά μέτρα διαχωρισιμότητας (CNN-LSTM / DEAP, S3) =====")
print(f"Αναπαραστάσεις: {features.shape} | Κατανομή κλάσεων: {np.bincount(labels_feat)}")
print(f"Silhouette score      : {sil:+.4f}   (→ +1 άριστος διαχωρισμός, ~0 επικάλυψη)")
print(f"Davies-Bouldin index  : {db:.4f}    (μικρότερο = καλύτερο)")
print(f"Fisher discriminant   : {fr:.4f}    (μεγαλύτερο = καλύτερο)")
print(f"Μέση ενδο-κλασική απόσταση : {intra:.4f}")
print(f"Μέση δια-κλασική απόσταση  : {inter:.4f}")
print(f"Λόγος inter/intra          : {inter/intra:.4f}  (>1 υποδηλώνει διαχωρίσιμες κλάσεις)")

separability_df = pd.DataFrame([{
    'model': 'CNN-LSTM', 'dataset': 'DEAP', 'split_level': 'S3',
    'silhouette': sil, 'davies_bouldin': db, 'fisher_ratio': fr,
    'intra_class_dist': intra, 'inter_class_dist': inter, 'inter_over_intra': inter / intra,
}])
separability_df.to_csv(f'{OUT_DIR}/separability_metrics_deap_cnnlstm.csv', index=False)
print(separability_df.round(4))


# =====================================================================
# ΒΗΜΑ 8 -- XAI: 5 μέθοδοι (πάνω στο ΠΡΑΓΜΑΤΙΚΟ S3 μοντέλο)
# =====================================================================
sample_idx = 0
input_data = X_test_xai[sample_idx:sample_idx + 1]
input_tensor = tf.convert_to_tensor(input_data, dtype=tf.float32)
target_class = int(np.argmax(xai_model.predict(input_data, verbose=0)[0]))
class_name = "High Valence" if target_class == 1 else "Low Valence"
print(f"\nΔείγμα XAI (S3/LOSO μοντέλο): πρόβλεψη = {class_name}")

X_train_xai_idx = np.setdiff1d(manifest.index.values, xai_test_idx)
background = X_data[X_train_xai_idx][np.random.choice(len(X_train_xai_idx), 30, replace=False)]


def explain_inputxgrad(model, sample, target):
    t = tf.convert_to_tensor(sample, dtype=tf.float32)
    with tf.GradientTape() as tape:
        tape.watch(t)
        preds = model(t)
        score = preds[0, target]
    grads = tape.gradient(score, t)
    return (grads * t)[0].numpy()


def explain_shap(model, background, sample, target):
    explainer = shap.GradientExplainer(model, background)
    sv = explainer.shap_values(sample)
    if isinstance(sv, list):
        arr = np.array(sv[target])[0] if len(sv) > target else np.array(sv[0])[0]
    else:
        arr = np.array(sv)
        arr = arr[0, ..., target] if arr.ndim >= 4 and arr.shape[-1] == 2 else arr[0]
    return arr


def explain_gradcam(model, sample, target, window_size=WINDOW_SIZE):
    last_conv = [l for l in model.layers if isinstance(l, tf.keras.layers.Conv1D)][-1]
    grad_model = tf.keras.models.Model(model.inputs, [last_conv.output, model.output])
    t = tf.convert_to_tensor(sample, dtype=tf.float32)
    with tf.GradientTape() as tape:
        conv_out, preds = grad_model(t)
        loss = tf.math.log(preds[:, target] + 1e-10)
    grads = tape.gradient(loss, conv_out)
    pooled = tf.reduce_mean(grads, axis=(0, 1))
    cam = tf.reduce_sum(tf.multiply(pooled, conv_out[0]), axis=-1).numpy()
    cam = np.maximum(cam, 0)
    if np.max(cam) == 0:
        cam = cam + 1e-10
    cam = cam / np.max(cam)
    cam_resized = scipy.ndimage.zoom(cam, window_size / len(cam))
    return np.tile(cam_resized[:, None], (1, sample.shape[-1]))


def explain_occlusion(model, sample, target, window_size=WINDOW_SIZE, win=8):
    scores = np.zeros(window_size)
    base_pred = model.predict(sample, verbose=0)[0][target]
    for t0 in range(0, window_size, win):
        temp = sample.copy()
        temp[0, t0:t0 + win, :] = 0
        new_pred = model.predict(temp, verbose=0)[0][target]
        scores[t0:t0 + win] = base_pred - new_pred
    return np.tile(scores[:, None], (1, sample.shape[-1]))


def explain_lime(model, background, sample, target, window_size=WINDOW_SIZE, n_channels=CHANNELS):
    lime_explainer = lime.lime_tabular.LimeTabularExplainer(
        training_data=background.reshape(background.shape[0], -1), mode='classification')

    def predict_fn(x):
        return model.predict(x.reshape(-1, window_size, n_channels), verbose=0)

    exp = lime_explainer.explain_instance(sample.flatten(), predict_fn,
                                           num_features=window_size * n_channels, labels=[target])
    out = np.zeros(window_size * n_channels)
    for feat_idx, w in exp.local_exp[target]:
        out[feat_idx] = w
    return out.reshape(window_size, n_channels)


def explain_feature_ablation(model, sample, target, window_size=WINDOW_SIZE, n_channels=CHANNELS):
    """Μηδενίζει ένα-ένα ολόκληρο κανάλι και μετρά την πτώση της πιθανότητας πρόβλεψης.
    Σε αντίθεση με το Occlusion (που κρύβει χρονικά παράθυρα), εδώ η σημαντικότητα
    αποδίδεται απευθείας σε επίπεδο ηλεκτροδίου."""
    base_pred = model.predict(sample, verbose=0)[0][target]
    scores = np.zeros(n_channels)
    for ch in range(n_channels):
        temp = sample.copy()
        temp[0, :, ch] = 0.0
        scores[ch] = base_pred - model.predict(temp, verbose=0)[0][target]
    return np.tile(scores[None, :], (window_size, 1))


def explain_kernelshap(model, background, sample, target,
                       window_size=WINDOW_SIZE, n_channels=CHANNELS, nsamples=200):
    """KernelSHAP σε επίπεδο καναλιού: κάθε "χαρακτηριστικό" είναι ένα ολόκληρο
    ηλεκτρόδιο, ώστε ο αριθμός συνδυασμών να παραμένει υπολογιστικά διαχειρίσιμος
    (32 κανάλια αντί για 32x128=4096 μεμονωμένα σημεία)."""
    def predict_channels(masks):
        out = []
        for mask in masks:
            temp = sample.copy()
            for ch in range(n_channels):
                if mask[ch] < 0.5:
                    temp[0, :, ch] = 0.0
            out.append(model.predict(temp, verbose=0)[0][target])
        return np.array(out)

    explainer = shap.KernelExplainer(predict_channels, np.zeros((1, n_channels)))
    sv = explainer.shap_values(np.ones((1, n_channels)), nsamples=nsamples, silent=True)
    channel_scores = np.array(sv).reshape(-1)[:n_channels]
    return np.tile(channel_scores[None, :], (window_size, 1))


EXPLAINERS = {
    'InputxGrad': lambda s: explain_inputxgrad(xai_model, s, target_class),
    'SHAP': lambda s: explain_shap(xai_model, background, s, target_class),
    'GradCAM': lambda s: explain_gradcam(xai_model, s, target_class),
    'Occlusion': lambda s: explain_occlusion(xai_model, s, target_class),
    'LIME': lambda s: explain_lime(xai_model, background, s, target_class),
    'FeatureAblation': lambda s: explain_feature_ablation(xai_model, s, target_class),
    'KernelSHAP': lambda s: explain_kernelshap(xai_model, background, s, target_class),
}
print("Explainers έτοιμοι:", list(EXPLAINERS.keys()))

maps = {}
for name, fn in EXPLAINERS.items():
    try:
        print(f"Υπολογισμός {name}...")
        maps[name] = fn(input_data)
    except Exception as e:
        print(f"  Το {name} απέτυχε ({e}); παραλείπεται.")
print(f"\nΜέθοδοι με έγκυρο χάρτη: {list(maps.keys())}")

fig, axs = plt.subplots(len(maps), 1, figsize=(14, 4 * len(maps)))
for ax, (name, m) in zip(axs, maps.items()):
    im = ax.imshow(m.T, aspect='auto', cmap='coolwarm')
    ax.set_title(name); ax.set_ylabel('Ηλεκτρόδια')
    ax.set_yticks(range(len(DREAMER_CHANNELS))); ax.set_yticklabels(DREAMER_CHANNELS, fontsize=7)
    fig.colorbar(im, ax=ax)
axs[-1].set_xlabel('Χρόνος (Time Steps)')
plt.tight_layout(); plt.show()


# =====================================================================
# ΒΗΜΑ 9 -- Χαρτογράφηση ηλεκτροδίων + scalp maps (ΞΕΧΩΡΙΣΤΑ ΑΝΑ ΜΕΘΟΔΟ) + frontal asymmetry
# =====================================================================
def build_montage_info(channels=DREAMER_CHANNELS):
    montage = mne.channels.make_standard_montage('standard_1020')
    info = mne.create_info(ch_names=channels, sfreq=FS, ch_types='eeg')
    info.set_montage(montage, on_missing='warn')
    return info

MNE_INFO = build_montage_info()


def plot_scalp_map(values, title):
    fig, ax = plt.subplots(figsize=(4, 4))
    mne.viz.plot_topomap(values, MNE_INFO, axes=ax, show=False, cmap='RdBu_r', contours=4)
    ax.set_title(title, fontsize=10)
    plt.show()


def channel_to_region_table(values, channels=DREAMER_CHANNELS):
    rows = [{'channel': ch, 'value': float(v), 'region': CHANNEL_REGION[ch][0], 'hemisphere': CHANNEL_REGION[ch][1]}
            for ch, v in zip(channels, values)]
    df = pd.DataFrame(rows)
    return df, df.groupby('region')['value'].sum().sort_values(ascending=False)


def frontal_asymmetry(values, channels=DREAMER_CHANNELS):
    pairs = [('Fp1', 'Fp2'), ('AF3', 'AF4'), ('F3', 'F4'), ('F7', 'F8'), ('FC5', 'FC6')]
    idx = {ch: i for i, ch in enumerate(channels)}
    return {f'{l}-{r}': float(values[idx[r]] - values[idx[l]]) for l, r in pairs if l in idx and r in idx}


ig_map = maps['InputxGrad']
channel_importance = np.abs(ig_map).sum(axis=0)

region_df, region_table = channel_to_region_table(channel_importance)
print(region_df.sort_values('value', ascending=False))
print("\nRelevance ανά περιοχή:\n", region_table)
plot_scalp_map(channel_importance, title=f"InputxGrad relevance — target {class_name}")

asym = frontal_asymmetry(channel_importance)
print("\nΜετωπιαία ασυμμετρία (δεξιά - αριστερή):")
for k, v in asym.items():
    print(f"  {k}: {v:+.4f}")

print("\n===== Scalp maps ανά μέθοδο XAI =====")
per_method_channel_importance = {}
for method_name, attr_map in maps.items():
    imp = np.abs(attr_map).sum(axis=0)
    per_method_channel_importance[method_name] = imp
    plot_scalp_map(imp, title=f"{method_name} relevance — target {class_name}")
    top5_idx = np.argsort(imp)[::-1][:5]
    print(f"\nTop-5 ηλεκτρόδια ({method_name}):")
    for i in top5_idx:
        ch = DREAMER_CHANNELS[i]
        region, hemi = CHANNEL_REGION[ch]
        print(f"  {ch} ({region}-{hemi}): {imp[i]:.4f}")

# --- Συγκεντρωτικός πίνακας σημαντικότητας ανά ηλεκτρόδιο και μέθοδο ---
channel_scores_df = pd.DataFrame(per_method_channel_importance, index=DREAMER_CHANNELS)
channel_scores_df.index.name = 'channel'
print("\n===== Σημαντικότητα ανά ηλεκτρόδιο και μέθοδο (CNN-LSTM / DEAP) =====")
print(channel_scores_df.round(4))
channel_scores_df.to_csv(f'{OUT_DIR}/xai_channel_scores_deap_cnnlstm.csv')

# --- Μετωπιαία ασυμμετρία ανά μέθοδο ---
print("\n===== Μετωπιαία ασυμμετρία ανά μέθοδο (δεξιά - αριστερή) =====")
asym_rows = []
for method_name, imp in per_method_channel_importance.items():
    a = frontal_asymmetry(imp)
    a['method'] = method_name
    asym_rows.append(a)
print(pd.DataFrame(asym_rows).set_index('method').round(4))


# =====================================================================
# ΒΗΜΑ 10 -- Faithfulness: Deletion/Insertion + AOPC
# =====================================================================
def predicted_prob(model, sample, target):
    return model.predict(sample, verbose=0)[0][target]


def rank_flat_indices(attr_map, descending=True):
    flat = attr_map.flatten()
    return np.argsort(-flat if descending else flat)


def deletion_insertion_curve(model, sample, target, attr_map, mode='deletion', k_values=K_VALUES):
    order = rank_flat_indices(attr_map)
    n_total = order.size
    base = sample.copy()
    probs = []
    for k in k_values:
        n_k = int(n_total * k / 100)
        idx_touch = order[:n_k]
        flat = base.reshape(1, -1).copy()
        if mode == 'deletion':
            flat[0, idx_touch] = 0.0
        else:
            new_flat = np.zeros_like(flat)
            new_flat[0, idx_touch] = base.reshape(1, -1)[0, idx_touch]
            flat = new_flat
        perturbed = flat.reshape(base.shape)
        probs.append(predicted_prob(model, perturbed, target))
    return np.array(probs)


def aopc(probs, base_prob):
    return float(np.mean(base_prob - probs)) if len(probs) else float('nan')


faithfulness_rows = []
base_prob = predicted_prob(xai_model, input_data, target_class)
for name, m in maps.items():
    del_p = deletion_insertion_curve(xai_model, input_data, target_class, m, 'deletion')
    ins_p = deletion_insertion_curve(xai_model, input_data, target_class, m, 'insertion')
    rand_map = np.random.RandomState(0).rand(*m.shape)
    del_r = deletion_insertion_curve(xai_model, input_data, target_class, rand_map, 'deletion')
    ins_r = deletion_insertion_curve(xai_model, input_data, target_class, rand_map, 'insertion')
    faithfulness_rows.append({
        'explainer': name,
        'AOPC_deletion': aopc(del_p, base_prob),
        # ΔΙΟΡΘΩΣΗ: η σειρά ορισμάτων ήταν ανάποδη -- aopc(probs, base_prob), όχι
        # aopc(base_prob, probs). Το base_prob είναι μονή τιμή, οπότε όταν μπαίνει
        # ΠΡΩΤΟ ως "probs", το len() μέσα στο aopc() σκάει.
        'AOPC_insertion': aopc(ins_p, base_prob) * -1,
        'AOPC_deletion_random': aopc(del_r, base_prob),
        'AOPC_insertion_random': aopc(ins_r, base_prob) * -1,
    })
faithfulness_df = pd.DataFrame(faithfulness_rows)
faithfulness_df['deletion_beats_random'] = faithfulness_df['AOPC_deletion'] > faithfulness_df['AOPC_deletion_random']
faithfulness_df.to_csv(f'{OUT_DIR}/faithfulness_AOPC_deap_cnnlstm.csv', index=False)
print(faithfulness_df)


# =====================================================================
# ΒΗΜΑ 11 -- Sanity checks: model & data randomization
# =====================================================================
def cascading_randomization_test(model, sample, target):
    def explain_with(m):
        return explain_inputxgrad(m, sample, target).flatten()

    original_map = explain_with(model)
    m2 = tf.keras.models.clone_model(model)
    m2.set_weights(model.get_weights())

    results = {'none_randomized': 1.0}
    randomizable_layers = [l for l in m2.layers if l.get_weights()]
    for layer in reversed(randomizable_layers):
        new_weights = [np.random.normal(0, 0.1, w.shape).astype(w.dtype) for w in layer.get_weights()]
        layer.set_weights(new_weights)
        rho, _ = spearmanr(original_map, explain_with(m2))
        results[f'after_randomizing_{layer.name}'] = rho
    return results


rand_results = cascading_randomization_test(xai_model, input_data, target_class)
print("Model-randomization test (Spearman ρ με τον αρχικό χάρτη):")
for k, v in rand_results.items():
    print(f"  {k}: {v:+.3f}")
plt.figure(figsize=(7, 3))
plt.plot(list(rand_results.keys()), list(rand_results.values()), 'o-')
plt.xticks(rotation=45, ha='right'); plt.ylabel('Spearman ρ')
plt.title('Cascading model-randomization sanity check')
plt.axhline(0, color='gray', linestyle='--'); plt.tight_layout(); plt.show()

_, s3_test_idx_check, _, _, _ = saved_models[any_s3_key]
s3_train_idx_check = np.setdiff1d(manifest.index.values, s3_test_idx_check)
shuffled_y = y_data.copy()
rng = np.random.RandomState(0)
shuffled_y[s3_train_idx_check] = rng.permutation(shuffled_y[s3_train_idx_check])

shuffled_model = build_cnn_lstm()
shuffled_model.fit(X_data[s3_train_idx_check], to_categorical(shuffled_y[s3_train_idx_check], 2),
                    epochs=min(EPOCHS, 15), batch_size=BATCH_SIZE, verbose=0)

map_normal = explain_inputxgrad(xai_model, input_data, target_class).flatten()
map_shuffled = explain_inputxgrad(shuffled_model, input_data, target_class).flatten()
rho_data, _ = spearmanr(map_normal, map_shuffled)
print(f"\nData-randomization test: Spearman ρ (κανονικό vs shuffled-labels) = {rho_data:+.3f}")
print("Αναμενόμενο: χαμηλό ρ.")


# =====================================================================
# ΒΗΜΑ 12 -- Ποσοτική συμφωνία μεταξύ των 5 explainers
# =====================================================================
def jaccard_topk(v1, v2, k):
    top1 = set(np.argsort(-np.abs(v1))[:k])
    top2 = set(np.argsort(-np.abs(v2))[:k])
    return len(top1 & top2) / len(top1 | top2)


channel_importances = {name: np.abs(m).sum(axis=0) for name, m in maps.items()}
names = list(channel_importances.keys())
agree_spearman = pd.DataFrame(index=names, columns=names, dtype=float)
agree_jaccard5 = pd.DataFrame(index=names, columns=names, dtype=float)
for a, b in itertools.product(names, names):
    va, vb = channel_importances[a], channel_importances[b]
    agree_spearman.loc[a, b] = spearmanr(va, vb)[0]
    agree_jaccard5.loc[a, b] = jaccard_topk(va, vb, 5)

print("Spearman ρ (channel importance):"); print(agree_spearman.round(2))
print("\nJaccard top-5:"); print(agree_jaccard5.round(2))
fig, ax = plt.subplots(figsize=(6, 5))
sns.heatmap(agree_spearman.astype(float), annot=True, fmt='.2f', cmap='coolwarm', vmin=-1, vmax=1, ax=ax)
ax.set_title('Συμφωνία explainers — Spearman ρ'); plt.tight_layout(); plt.show()


# =====================================================================
# ΒΗΜΑ 13 -- Φασματική εξηγησιμότητα
# =====================================================================
def bandstop_filter(x, low, high, fs=FS, order=4):
    nyq = fs / 2
    low_n, high_n = max(low / nyq, 1e-3), min(high / nyq, 0.999)
    b, a = sps.butter(order, [low_n, high_n], btype='bandstop')
    return sps.filtfilt(b, a, x, axis=0)


def bandpass_filter(x, low, high, fs=FS, order=4):
    nyq = fs / 2
    low_n, high_n = max(low / nyq, 1e-3), min(high / nyq, 0.999)
    b, a = sps.butter(order, [low_n, high_n], btype='bandpass')
    return sps.filtfilt(b, a, x, axis=0)


X_test_np = X_data[xai_test_idx]
y_test_np = y_data[xai_test_idx]
base_preds = np.argmax(xai_model.predict(X_test_np, verbose=0), axis=1)
base_macro_f1 = f1_score(y_test_np, base_preds, average='macro', zero_division=0)

band_drop_rows = []
for band_name, (lo, hi) in BANDS.items():
    filtered = np.stack([bandstop_filter(x, lo, hi) for x in X_test_np]).astype('float32')
    preds = np.argmax(xai_model.predict(filtered, verbose=0), axis=1)
    band_f1 = f1_score(y_test_np, preds, average='macro', zero_division=0)
    band_drop_rows.append({'band': band_name, 'range_hz': f"{lo}-{hi}",
                            'macro_f1_after_removal': band_f1, 'drop': base_macro_f1 - band_f1})
band_drop_df = pd.DataFrame(band_drop_rows).sort_values('drop', ascending=False)
print(f"Baseline Macro F1: {base_macro_f1:.3f}")
print(band_drop_df)

channels_x_bands = np.zeros((len(DREAMER_CHANNELS), len(BANDS)))
sample_np = input_data[0]
for b_idx, (band_name, (lo, hi)) in enumerate(BANDS.items()):
    band_signal = bandpass_filter(sample_np, lo, hi)
    band_map = explain_inputxgrad(xai_model, band_signal[None, ...].astype('float32'), target_class)
    channels_x_bands[:, b_idx] = np.abs(band_map).sum(axis=0)

cxb_df = pd.DataFrame(channels_x_bands, index=DREAMER_CHANNELS, columns=list(BANDS.keys()))
plt.figure(figsize=(6, 5))
sns.heatmap(cxb_df, annot=True, fmt='.2f', cmap='viridis')
plt.title(f'Channels × Bands relevance (InputxGrad, {class_name})')
plt.tight_layout(); plt.show()

fig, axes = plt.subplots(1, len(BANDS), figsize=(4 * len(BANDS), 4))
for ax, band_name in zip(axes, BANDS.keys()):
    mne.viz.plot_topomap(cxb_df[band_name].values, MNE_INFO, axes=ax, show=False, cmap='RdBu_r')
    ax.set_title(band_name)
plt.tight_layout(); plt.show()

fa_alpha = frontal_asymmetry(cxb_df['alpha'].values)
print("Frontal alpha asymmetry:")
for k, v in fa_alpha.items():
    print(f"  {k}: {v:+.4f}")


# =====================================================================
# ΒΗΜΑ 14 -- Subject-Specific Fine-Tuning (ΔΙΟΡΘΩΜΕΝΟ: γνήσιο held-out υποκείμενο)
# =====================================================================
target_sub_id = any_s3_key[1].replace('S3_subject', '')
print(f"\nΣτοχευμένο υποκείμενο για fine-tuning (πραγματικά held-out): {target_sub_id}")

subj_mask = manifest['subject_id'] == target_sub_id
subj_idx = manifest.index[subj_mask].values
subj_trials = manifest.loc[subj_mask, 'trial_id'].values
unique_trials = np.unique(subj_trials)
rng = np.random.RandomState(42)
rng.shuffle(unique_trials)
split_point = max(1, int(0.7 * len(unique_trials)))
ft_trials, test_trials = unique_trials[:split_point], unique_trials[split_point:]
ft_idx = subj_idx[np.isin(subj_trials, ft_trials)]
subj_test_idx = subj_idx[np.isin(subj_trials, test_trials)]

X_ft, y_ft = X_data[ft_idx], to_categorical(y_data[ft_idx], 2)
X_subj_test, y_subj_test = X_data[subj_test_idx], to_categorical(y_data[subj_test_idx], 2)
y_subj_test_int = y_data[subj_test_idx]

sub_baseline = majority_class_baseline(y_subj_test_int) * 100
print(f"Majority-class baseline (test του υποκειμένου): {sub_baseline:.2f}%")

proba_before = xai_model.predict(X_subj_test, verbose=0)
pred_before = np.argmax(proba_before, axis=1)
m_before = full_metrics(y_subj_test_int, pred_before, proba_before[:, 1])
print(f"Πριν το fine-tuning -- Balanced Accuracy: {m_before['balanced_accuracy']:.2%}")

ft_model = tf.keras.models.clone_model(xai_model)
ft_model.set_weights(xai_model.get_weights())
for layer in ft_model.layers[:-4]:
    layer.trainable = False
ft_model.compile(optimizer=Adam(learning_rate=0.0001), loss='categorical_crossentropy', metrics=['accuracy'])
ft_model.fit(X_ft, y_ft, epochs=10, batch_size=32, validation_data=(X_subj_test, y_subj_test), verbose=0)

proba_after = ft_model.predict(X_subj_test, verbose=0)
pred_after = np.argmax(proba_after, axis=1)
m_after = full_metrics(y_subj_test_int, pred_after, proba_after[:, 1])
print(f"Μετά το fine-tuning -- Balanced Accuracy: {m_after['balanced_accuracy']:.2%}")

if len(np.unique(pred_after)) == 1:
    print("ΠΡΟΣΟΧΗ: το μοντέλο προβλέπει ΜΟΝΟ 1 κλάση μετά το fine-tuning (Majority Class Collapse).")
elif m_after['balanced_accuracy'] * 100 <= sub_baseline + 1.0:
    print("ΠΡΟΣΟΧΗ: η ακρίβεια δεν ξεπερνά ουσιαστικά το majority baseline του υποκειμένου.")

######################################################################
# ΑΝΑΛΥΣΗ ΟΜΟΙΟΤΗΤΑΣ ΑΝΑΠΑΡΑΣΤΑΣΕΩΝ (Linear CKA)
#
# Το CKA (Centered Kernel Alignment) μετρά πόσο όμοιες είναι δύο αναπαραστάσεις,
# σε κλίμακα 0 (εντελώς ανόμοιες) έως 1 (ταυτόσημες), ακόμα κι όταν έχουν
# διαφορετικό πλήθος διαστάσεων. Εξετάζονται δύο ερωτήματα:
#   (α) πώς μετασχηματίζεται η πληροφορία από στάδιο σε στάδιο του δικτύου, και
#   (β) μαθαίνουν τα μοντέλα S1 και S3 παρόμοιες αναπαραστάσεις;
######################################################################

def _center_gram(K):
    n = K.shape[0]
    H = np.eye(n) - np.ones((n, n)) / n
    return H @ K @ H


def linear_cka(X, Y):
    """Linear CKA μεταξύ δύο πινάκων (δείγματα x διαστάσεις). Τα δείγματα πρέπει
    να είναι τα ΙΔΙΑ και στην ΙΔΙΑ σειρά στους δύο πίνακες."""
    X = np.asarray(X, dtype=np.float64)
    Y = np.asarray(Y, dtype=np.float64)
    X = X - X.mean(axis=0, keepdims=True)
    Y = Y - Y.mean(axis=0, keepdims=True)
    Kx = _center_gram(X @ X.T)
    Ky = _center_gram(Y @ Y.T)
    hsic = float((Kx * Ky).sum())
    norm = float(np.sqrt((Kx * Kx).sum()) * np.sqrt((Ky * Ky).sum()))
    return hsic / norm if norm > 0 else float('nan')


def find_cka_layers(model):
    """Εντοπίζει δυναμικά τα στάδια προς σύγκριση. Το δίκτυο κατασκευάζεται με το
    functional API χωρίς ονόματα επιπέδων, οπότε τα επίπεδα αναγνωρίζονται από τον
    ΤΥΠΟ τους: τα δύο συνελικτικά μπλοκ, το αναδρομικό επίπεδο και το πυκνό επίπεδο
    πριν την έξοδο.

    ΔΙΟΡΘΩΣΗ: η ιδιότητα layer.output_shape καταργήθηκε στο Keras 3. Για τα πυκνά
    επίπεδα χρησιμοποιείται η layer.units, με εναλλακτική το layer.output.shape.
    """
    picked, conv_seen = [], 0
    for layer in model.layers:
        cls = type(layer).__name__
        if cls == 'Conv1D':
            conv_seen += 1
            picked.append((f'conv{conv_seen}', layer))
        elif cls == 'LSTM':
            picked.append(('lstm', layer))
        elif cls == 'Dense':
            units = getattr(layer, 'units', None)
            if units is None:
                try:
                    units = int(layer.output.shape[-1])
                except Exception:
                    units = None
            if units != 2:      # εξαιρείται το επίπεδο εξόδου
                picked.append(('dense', layer))
    return picked


def get_layer_reps(model, X_raw, batch=128):
    """Ενεργοποιήσεις κάθε σταδίου. Οι συνελικτικές έξοδοι είναι τρισδιάστατες
    (δείγματα x χρόνος x φίλτρα) και ισοπεδώνονται σε δύο διαστάσεις."""
    picked = find_cka_layers(model)
    extractor = Model(inputs=model.inputs, outputs=[l.output for _n, l in picked])
    outs = extractor.predict(X_raw, batch_size=batch, verbose=0)
    if not isinstance(outs, list):
        outs = [outs]
    return {name: np.asarray(o).reshape(len(X_raw), -1)
            for (name, _l), o in zip(picked, outs)}


# --- Κοινό υποσύνολο δειγμάτων: ίδια ακριβώς παράθυρα για κάθε μοντέλο ---
rng_cka = np.random.RandomState(0)
cka_idx = rng_cka.choice(xai_test_idx, size=min(500, len(xai_test_idx)), replace=False)
X_cka = X_data[cka_idx]

reps_s3 = get_layer_reps(xai_model, X_cka)
for k, v in reps_s3.items():
    print(f"Αναπαράσταση '{k}': {v.shape}")

# --- (α) Ομοιότητα μεταξύ σταδίων του μοντέλου S3 ---
names = list(reps_s3.keys())
cka_layers = pd.DataFrame(index=names, columns=names, dtype=float)
for a, b in itertools.product(names, names):
    cka_layers.loc[a, b] = linear_cka(reps_s3[a], reps_s3[b])

print("\n===== CKA μεταξύ σταδίων (μοντέλο S3) =====")
print(cka_layers.round(3))

fig, ax = plt.subplots(figsize=(5, 4))
sns.heatmap(cka_layers.astype(float), annot=True, fmt='.3f', cmap='viridis',
            vmin=0, vmax=1, ax=ax)
ax.set_title('Linear CKA μεταξύ σταδίων (S3)')
plt.tight_layout(); plt.show()
cka_layers.to_csv(f'{OUT_DIR}/cka_between_layers.csv')

# --- (β) Ομοιότητα ίδιου σταδίου μεταξύ των μοντέλων S1 και S3 ---
s1_key = next((k for k in saved_models if k[0] == 'S1'), None)

if s1_key is None:
    print("\nΤο μοντέλο S1 δεν είναι διαθέσιμο· η σύγκριση S1 vs S3 παραλείπεται.")
else:
    model_s1 = saved_models[s1_key][0]
    reps_s1 = get_layer_reps(model_s1, X_cka)
    cka_cross = pd.DataFrame(
        [{'stage': n, 'CKA_S1_vs_S3': linear_cka(reps_s1[n], reps_s3[n])}
         for n in names if n in reps_s1]).set_index('stage')

    print("\n===== CKA ίδιου σταδίου: μοντέλο S1 έναντι μοντέλου S3 =====")
    print(cka_cross.round(3))
    print("Χαμηλές τιμές σημαίνουν ότι τα δύο πρωτόκολλα οδηγούν σε ποιοτικά")
    print("διαφορετικές αναπαραστάσεις, παρότι η αρχιτεκτονική είναι ίδια.")

    fig, ax = plt.subplots(figsize=(5, 3.2))
    cka_cross['CKA_S1_vs_S3'].plot(kind='bar', ax=ax, color='#4C72B0')
    ax.set_ylim(0, 1); ax.set_ylabel('Linear CKA')
    ax.set_title('Ομοιότητα αναπαραστάσεων S1 vs S3')
    plt.xticks(rotation=0); plt.tight_layout(); plt.show()
    cka_cross.to_csv(f'{OUT_DIR}/cka_s1_vs_s3.csv')

# =====================================================================
# LAYER-WISE PROBING
# Γραμμικοί ταξινομητές πάνω στις παγωμένες αναπαραστάσεις ΚΑΘΕ σταδίου,
# ώστε να φανεί σε ποιο σημείο του δικτύου η κάθε πληροφορία γίνεται
# γραμμικά αποκωδικοποιήσιμη. Συγκρίνεται η συναισθηματική ετικέτα με
# μεταβλητές που το μοντέλο ΔΕΝ θα έπρεπε να κωδικοποιεί (ταυτότητα
# εθελοντή, δοκιμασία), οπότε υψηλή ακρίβεια σε αυτές αποτελεί ένδειξη
# συντόμευσης μέσω Διαρροής Ταυτότητας Υποκειμένου.
# =====================================================================
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split as _tts

META = manifest            # <-- στα δύο CCNN notebooks άλλαξέ το σε: META = info

probe_targets = {'label (συναίσθημα)': 'label',
                 'subject_id (ταυτότητα)': 'subject_id',
                 'trial_id (δοκιμασία)': 'trial_id'}
if 'session_id' in META.columns and META['session_id'].nunique() > 1:
    probe_targets['session_id (συνεδρία)'] = 'session_id'

probe_rows = []
for stage, feats in reps_s3.items():
    if np.allclose(feats.std(axis=0), 0):
        print(f"  Το στάδιο '{stage}' έχει μηδενική διασπορά· παραλείπεται.")
        continue
    for label_name, col in probe_targets.items():
        y = META.loc[cka_idx, col].values
        classes, counts = np.unique(y, return_counts=True)
        if len(classes) < 2 or counts.min() < 2:
            continue
        strat = y if counts.min() >= 2 else None
        Xtr, Xte, ytr, yte = _tts(feats, y, test_size=0.3, random_state=0, stratify=strat)
        clf = LogisticRegression(max_iter=1000).fit(Xtr, ytr)
        acc = clf.score(Xte, yte)
        chance = counts.max() / counts.sum()      # πλειοψηφική κλάση ως γραμμή βάσης
        probe_rows.append({'stage': stage, 'target': label_name,
                           'probe_accuracy': acc, 'chance_level': chance,
                           'ratio': acc / chance if chance > 0 else float('nan')})

probe_df = pd.DataFrame(probe_rows)
if not probe_df.empty:
    table = probe_df.pivot(index='stage', columns='target', values='probe_accuracy')
    print("\n===== Layer-wise probing: ακρίβεια γραμμικού ταξινομητή ανά στάδιο =====")
    print(table.round(4))
    print("\nΛόγος προς τη γραμμή βάσης (τιμές >>1 υποδηλώνουν αποκωδικοποιήσιμη πληροφορία):")
    print(probe_df.pivot(index='stage', columns='target', values='ratio').round(2))
    probe_df.to_csv('layerwise_probing.csv', index=False)

    fig, ax = plt.subplots(figsize=(1.6 * len(table) + 3, 3.4))
    table.plot(kind='bar', ax=ax)
    ax.set_ylabel('Ακρίβεια probe'); ax.set_ylim(0, 1)
    ax.set_title('Layer-wise probing ανά στάδιο')
    plt.xticks(rotation=0); plt.legend(fontsize=8); plt.tight_layout(); plt.show()
```
