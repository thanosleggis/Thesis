```python
!pip install -q torcheeg --no-deps
!pip install -q mne
!pip install -q mne-connectivity
!pip install -q captum
!pip install -q lmdb
!pip install -q scikit-learn
!pip install -q seaborn
!pip install -q pandas
!pip install -q scipy
!pip install -q einops
!pip install -q xmltodict
!pip install -q wfdb
!pip install -q xlrd
!pip install -q torchmetrics
!pip install -q spectrum
!pip install -q pytorch-lightning
!pip install -q lime
!pip install -q shap
print("Η εγκατάσταση ολοκληρώθηκε.")

import sys, types
if 'torch_scatter' not in sys.modules:
    _stub = types.ModuleType('torch_scatter')
    def _not_implemented(*args, **kwargs):
        raise NotImplementedError("Το torch_scatter είναι ψεύτικο (stub) εδώ -- δεν χρησιμοποιούμε RGNN.")
    _stub.scatter_add = _not_implemented
    sys.modules['torch_scatter'] = _stub

from torcheeg.datasets import DEAPDataset
from torcheeg import transforms
from torcheeg.datasets.constants import DEAP_CHANNEL_LOCATION_DICT
from torcheeg.models import CCNN
print("Η εισαγωγή του torcheeg.datasets και torcheeg.models.CCNN έγινε επιτυχώς.")
import os, re, glob, shutil

pattern = re.compile(r'^s\d+\.dat$', re.IGNORECASE)
candidate_files = [f for f in glob.glob('/kaggle/input/**/*.dat', recursive=True)
                    if pattern.match(os.path.basename(f))]

if not candidate_files:
    raise FileNotFoundError(
        "Δεν βρέθηκαν αρχεία s##.dat κάτω από /kaggle/input. Πρόσθεσε το DEAP.mini "
        "dataset στο notebook ('+ Add Input') πριν συνεχίσεις."
    )

LOCAL_DATA_DIR = '/kaggle/working/deap_clean'
if os.path.exists(LOCAL_DATA_DIR):
    shutil.rmtree(LOCAL_DATA_DIR)
os.makedirs(LOCAL_DATA_DIR, exist_ok=True)

for f in candidate_files:
    shutil.copy(f, os.path.join(LOCAL_DATA_DIR, os.path.basename(f)))

print(f"Βρέθηκαν και αντιγράφηκαν {len(candidate_files)} αρχεία .dat")
import shutil, os
from torcheeg.datasets import DEAPDataset
from torcheeg import transforms
from torcheeg.datasets.constants import DEAP_CHANNEL_LOCATION_DICT

IO_PATH = '/kaggle/working/deap_processed'
if os.path.exists(IO_PATH):
    shutil.rmtree(IO_PATH)

dataset = DEAPDataset(
    io_path=IO_PATH,
    root_path=LOCAL_DATA_DIR,
    offline_transform=transforms.Compose([
        transforms.BandDifferentialEntropy(),
        transforms.MeanStdNormalize(),
        transforms.ToGrid(DEAP_CHANNEL_LOCATION_DICT)
    ]),
    online_transform=transforms.Compose([transforms.ToTensor()]),
    label_transform=transforms.Compose([
        transforms.Select('valence'),
        transforms.Binary(5.0)
    ]),
    num_worker=4
)
print(f"Το DEAP dataset είναι έτοιμο. Συνολικά παράθυρα: {len(dataset)}")
print(dataset.info.head())
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

DEAP_CHANNELS = [
    "Fp1", "AF3", "F3", "F7", "FC5", "FC1", "C3", "T7",
    "CP5", "CP1", "P3", "P7", "PO3", "O1", "Oz", "Pz",
    "Fp2", "AF4", "F4", "F8", "FC6", "FC2", "C4", "T8",
    "CP6", "CP2", "P4", "P8", "PO4", "O2", "Cz", "Fz"
]
REGION_MAP = {
    "Fp1": "frontal-left", "Fp2": "frontal-right", "AF3": "frontal-left", "AF4": "frontal-right",
    "F3": "frontal-left", "F4": "frontal-right", "F7": "frontal-left", "F8": "frontal-right",
    "Fz": "frontal-mid",
    "FC5": "frontal-central-left", "FC1": "frontal-central-left",
    "FC6": "frontal-central-right", "FC2": "frontal-central-right",
    "C3": "central-left", "C4": "central-right", "Cz": "central-mid",
    "T7": "temporal-left", "T8": "temporal-right",
    "CP5": "centro-parietal-left", "CP1": "centro-parietal-left",
    "CP6": "centro-parietal-right", "CP2": "centro-parietal-right",
    "P3": "parietal-left", "P4": "parietal-right", "P7": "parietal-left",
    "P8": "parietal-right", "Pz": "parietal-mid",
    "PO3": "parieto-occipital-left", "PO4": "parieto-occipital-right",
    "O1": "occipital-left", "O2": "occipital-right", "Oz": "occipital-mid",
}

info = dataset.info
print("Στήλες dataset.info:", list(info.columns))
print(f"Μοναδικά υποκείμενα: {info['subject_id'].nunique()}")

VALENCE_THRESHOLD = 5.0
balance = info.groupby('subject_id')['valence'].apply(lambda s: (s >= VALENCE_THRESHOLD).mean() * 100)
plt.figure(figsize=(10, 4))
balance.sort_values().plot(kind='bar', color='#4C72B0')
plt.axhline(50, color='red', linestyle='--', linewidth=1, label='Ισορροπημένο (50%)')
plt.ylabel('% High Valence')
plt.xlabel('subject_id')
plt.title('Ισορροπία κλάσης Valence ανά υποκείμενο (DEAP)')
plt.legend()
plt.tight_layout()
plt.show()
import numpy as np
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, Subset
from torcheeg.models import CCNN
from sklearn.model_selection import GroupKFold
from sklearn.metrics import (
    accuracy_score, balanced_accuracy_score, f1_score,
    precision_recall_fscore_support, matthews_corrcoef,
    roc_auc_score, cohen_kappa_score, confusion_matrix
)
import seaborn as sns
import matplotlib.pyplot as plt

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
BATCH_SIZE = 64
EPOCHS = 10
LR = 1e-3
print(f"Εκτέλεση σε: {device}")


def train_one_epoch(model, loader, optimizer, criterion):
    model.train()
    running_loss, correct, total = 0.0, 0, 0
    for X, y in loader:
        X, y = X.to(device), y.to(device)
        optimizer.zero_grad()
        out = model(X)
        loss = criterion(out, y)
        loss.backward()
        optimizer.step()
        running_loss += loss.item() * X.size(0)
        correct += (out.argmax(1) == y).sum().item()
        total += y.size(0)
    return running_loss / total, 100.0 * correct / total


@torch.no_grad()
def evaluate(model, loader):
    model.eval()
    criterion = nn.CrossEntropyLoss()
    running_loss, total = 0.0, 0
    all_true, all_pred, all_prob = [], [], []
    for X, y in loader:
        X, y = X.to(device), y.to(device)
        out = model(X)
        loss = criterion(out, y)
        running_loss += loss.item() * X.size(0)
        total += y.size(0)
        prob = torch.softmax(out, dim=1)[:, 1]
        all_true.append(y.cpu().numpy())
        all_pred.append(out.argmax(1).cpu().numpy())
        all_prob.append(prob.cpu().numpy())
    y_true = np.concatenate(all_true)
    y_pred = np.concatenate(all_pred)
    y_prob = np.concatenate(all_prob)
    return running_loss / total, y_true, y_pred, y_prob


def majority_class_baseline(y_true):
    values, counts = np.unique(y_true, return_counts=True)
    return counts.max() / counts.sum() * 100.0


def full_report(y_true, y_pred, y_prob, title=""):
    acc = accuracy_score(y_true, y_pred) * 100
    bal_acc = balanced_accuracy_score(y_true, y_pred) * 100
    macro_f1 = f1_score(y_true, y_pred, average='macro') * 100
    prec, rec, f1_per_class, support = precision_recall_fscore_support(
        y_true, y_pred, labels=[0, 1], zero_division=0)
    mcc = matthews_corrcoef(y_true, y_pred)
    kappa = cohen_kappa_score(y_true, y_pred)
    try:
        auc = roc_auc_score(y_true, y_prob)
    except ValueError:
        auc = float('nan')
    baseline = majority_class_baseline(y_true)

    print(f"\n===== {title} =====")
    print(f"Accuracy           : {acc:.2f}%")
    print(f"Majority baseline  : {baseline:.2f}%")
    print(f"Balanced Accuracy  : {bal_acc:.2f}%")
    print(f"Macro F1           : {macro_f1:.2f}%")
    print(f"Cohen's kappa      : {kappa:.4f}")
    print(f"MCC                : {mcc:.4f}")
    print(f"ROC-AUC            : {auc:.4f}")
    print(f"Precision (0/1)    : {prec[0]:.3f} / {prec[1]:.3f}")
    print(f"Recall    (0/1)    : {rec[0]:.3f} / {rec[1]:.3f}")
    print(f"F1        (0/1)    : {f1_per_class[0]:.3f} / {f1_per_class[1]:.3f}")

    cm = confusion_matrix(y_true, y_pred, labels=[0, 1])
    plt.figure(figsize=(4, 3.5))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
                xticklabels=['Low V', 'High V'], yticklabels=['Low V', 'High V'])
    plt.xlabel('Predicted')
    plt.ylabel('True')
    plt.title(f'Confusion Matrix — {title}')
    plt.tight_layout()
    plt.show()

    return {
        'accuracy': acc, 'majority_baseline': baseline, 'balanced_accuracy': bal_acc,
        'macro_f1': macro_f1, 'kappa': kappa, 'mcc': mcc, 'roc_auc': auc,
        'precision': prec, 'recall': rec, 'f1_per_class': f1_per_class, 'confusion_matrix': cm
    }
    info = dataset.info

# ΔΙΟΡΘΩΣΗ: το dataset.info περιέχει την ΑΚΑΤΕΡΓΑΣΤΗ βαθμολογία valence, όχι τη
# δυαδική ετικέτα -- η κατωφλιοποίηση Binary(5.0) εφαρμόζεται μόνο στα δεδομένα που
# επιστρέφει ο DataLoader. Η στήλη 'label' δημιουργείται εδώ ρητά, ώστε να είναι
# διαθέσιμη για τη στρωματοποίηση του S1 και για το layer-wise probing.
info = info.copy()
info['label'] = (info['valence'] >= VALENCE_THRESHOLD).astype(int)
print("Κατανομή ετικετών:", info['label'].value_counts().to_dict())

# ================== ΒΗΜΑ 4β: Βοηθητικές συναρτήσεις ==================
# Η λογική εκπαίδευσης είναι inline μέσα στον βρόχο· εδώ τυποποιείται σε συνάρτηση,
# ώστε να μπορεί να κληθεί και για τα επίπεδα S1 και S2.
import copy, itertools
from sklearn.model_selection import train_test_split
from scipy.stats import wilcoxon


def train_one_fold(train_idx, test_idx, n_epochs=EPOCHS):
    train_loader = DataLoader(Subset(dataset, train_idx.tolist()),
                              batch_size=BATCH_SIZE, shuffle=True)
    test_loader = DataLoader(Subset(dataset, test_idx.tolist()),
                             batch_size=BATCH_SIZE, shuffle=False)
    model = CCNN(num_classes=2, in_channels=4, grid_size=(9, 9)).to(device)
    optimizer = optim.Adam(model.parameters(), lr=LR)
    criterion = nn.CrossEntropyLoss()

    for epoch in range(1, n_epochs + 1):
        train_loss, train_acc = train_one_epoch(model, train_loader, optimizer, criterion)
        _, yt, yp, _ = evaluate(model, test_loader)
        print(f"  Epoch {epoch:02d}/{n_epochs} | Train Loss {train_loss:.4f} Acc {train_acc:.2f}% "
              f"| Val Bal.Acc {balanced_accuracy_score(yt, yp) * 100:.2f}%")

    _, y_true, y_pred, y_prob = evaluate(model, test_loader)
    return model, y_true, y_pred, y_prob


full_metrics_report = full_report   # ενιαία ονομασία με το DREAMER notebook

indices = info.index.values
groups = info['subject_id'].values

N_SPLITS = 5
gkf = GroupKFold(n_splits=N_SPLITS)

fold_reports = []
best_val_bal_acc = -1
best_model_state = None

# Αποθηκεύονται για τις αναλύσεις αναπαραστάσεων και τον έλεγχο Wilcoxon
fold_states = {}
per_subject_acc = {'S1': {}, 'S3': {}}
s3_true_all, s3_pred_all = [], []

for fold, (train_idx, test_idx) in enumerate(gkf.split(indices, groups=groups)):
    print(f"\n########## FOLD {fold + 1}/{N_SPLITS} ##########")
    print(f"Υποκείμενα στο train: {sorted(set(groups[train_idx]))}")
    print(f"Υποκείμενα στο test : {sorted(set(groups[test_idx]))}")

    train_loader = DataLoader(Subset(dataset, train_idx.tolist()), batch_size=BATCH_SIZE, shuffle=True)
    test_loader = DataLoader(Subset(dataset, test_idx.tolist()), batch_size=BATCH_SIZE, shuffle=False)

    model = CCNN(num_classes=2, in_channels=4, grid_size=(9, 9)).to(device)
    optimizer = optim.Adam(model.parameters(), lr=LR)
    criterion = nn.CrossEntropyLoss()

    for epoch in range(1, EPOCHS + 1):
        train_loss, train_acc = train_one_epoch(model, train_loader, optimizer, criterion)
        val_loss, y_true, y_pred, y_prob = evaluate(model, test_loader)
        val_bal_acc = balanced_accuracy_score(y_true, y_pred) * 100
        print(f"Epoch {epoch:02d}/{EPOCHS} | Train Loss {train_loss:.4f} Acc {train_acc:.2f}% "
              f"| Val Loss {val_loss:.4f} Bal.Acc {val_bal_acc:.2f}%")

    _, y_true, y_pred, y_prob = evaluate(model, test_loader)
    report = full_report(y_true, y_pred, y_prob, title=f"CCNN/DEAP - Fold {fold + 1} (subject-independent)")
    report['fold'] = fold + 1
    report['test_subjects'] = sorted(set(groups[test_idx]))
    fold_reports.append(report)

    # Τα βάρη ΚΑΘΕ fold (όχι μόνο του καλύτερου) χρειάζονται για τη σύγκριση CKA
    fold_states[fold + 1] = copy.deepcopy(model.state_dict())

    # Ακρίβεια ανά υποκείμενο, για τον έλεγχο Wilcoxon S1 vs S3
    s3_true_all.append(y_true)
    s3_pred_all.append(y_pred)
    subj_ids = groups[test_idx]
    for subj in np.unique(subj_ids):
        m = subj_ids == subj
        per_subject_acc['S3'][subj] = float((y_true[m] == y_pred[m]).mean())

    if report['balanced_accuracy'] > best_val_bal_acc:
        best_val_bal_acc = report['balanced_accuracy']
        # ΔΙΟΡΘΩΣΗ: deepcopy -- χωρίς αυτό το state_dict παραμένει αναφορά στους
        # ζωντανούς τανυστές του μοντέλου και μπορεί να αλλοιωθεί.
        best_model_state = copy.deepcopy(model.state_dict())

summary = pd.DataFrame([{
    'fold': r['fold'], 'accuracy': r['accuracy'], 'majority_baseline': r['majority_baseline'],
    'balanced_accuracy': r['balanced_accuracy'], 'macro_f1': r['macro_f1'],
    'mcc': r['mcc'], 'roc_auc': r['roc_auc'],
} for r in fold_reports])
print("\n===== Συγκεντρωτικός πίνακας (5-fold, subject-independent) =====")
print(summary)
print(f"\nΜέσος όρος Balanced Accuracy: {summary['balanced_accuracy'].mean():.2f}% ± {summary['balanced_accuracy'].std():.2f}%")


# ================== ΒΗΜΑ 5β: Πρωτόκολλο τριών επιπέδων (S1 / S2) ==================
# Το 5-fold GroupKFold ανά υποκείμενο παραπάνω αντιστοιχεί λειτουργικά στο S3.
# Εδώ προστίθενται τα δύο ηπιότερα επίπεδα, ώστε να μετρηθεί η πτώση απόδοσης που
# οφείλεται στη Διαρροή Ταυτότητας Υποκειμένου:
#   S1 -- τυχαίος διαχωρισμός σε επίπεδο παραθύρου (80/20)
#   S2 -- GroupKFold ομαδοποιημένο ανά υποκείμενο ΚΑΙ δοκιμασία
# ΣΗΜΕΙΩΣΗ: στο notebook αυτό οι μετρικές είναι σε κλίμακα 0-100, όχι 0-1.
protocol_results = []

print("\n" + "=" * 60)
print("ΕΠΙΠΕΔΟ S1 -- τυχαίος διαχωρισμός παραθύρων 80/20")
print("=" * 60)
s1_train, s1_test = train_test_split(
    indices, test_size=0.2, random_state=42, stratify=info['label'].values)

model_s1, y1_true, y1_pred, y1_prob = train_one_fold(s1_train, s1_test)
rep = full_report(y1_true, y1_pred, y1_prob, title="CCNN/DEAP - S1 (window-level)")
rep.update({'split_level': 'S1', 'fold': 'S1_full'})
protocol_results.append(rep)
s1_state = copy.deepcopy(model_s1.state_dict())

s1_subjects = info.loc[s1_test, 'subject_id'].values
for subj in np.unique(s1_subjects):
    m = s1_subjects == subj
    per_subject_acc['S1'][subj] = float((y1_true[m] == y1_pred[m]).mean())

print("\n" + "=" * 60)
print("ΕΠΙΠΕΔΟ S2 -- GroupKFold ανά υποκείμενο και δοκιμασία")
print("=" * 60)
trial_groups = (info['subject_id'].astype(str) + '_' + info['trial_id'].astype(str)).values

for fold, (tr, te) in enumerate(GroupKFold(n_splits=N_SPLITS).split(indices, groups=trial_groups)):
    print(f"\n--- S2 fold {fold + 1}/{N_SPLITS} ---")
    _m, yt, yp, ypr = train_one_fold(indices[tr], indices[te])
    rep = full_report(yt, yp, ypr, title=f"CCNN/DEAP - S2 fold {fold + 1}")
    rep.update({'split_level': 'S2', 'fold': f'S2_fold{fold}'})
    protocol_results.append(rep)

# Το S3 επαναχρησιμοποιείται από τον βρόχο παραπάνω -- δεν ξαναεκπαιδεύεται
for r in fold_reports:
    rep = dict(r)
    rep.update({'split_level': 'S3', 'fold': f"S3_fold{r['fold']}"})
    protocol_results.append(rep)

s3_true_all = np.concatenate(s3_true_all)
s3_pred_all = np.concatenate(s3_pred_all)

protocol_df = pd.DataFrame([{k: v for k, v in r.items()
                             if k not in ('confusion_matrix', 'f1_per_class', 'test_subjects')}
                            for r in protocol_results])
protocol_summary = (protocol_df.groupby('split_level')
                    .agg(balanced_accuracy_mean=('balanced_accuracy', 'mean'),
                         balanced_accuracy_std=('balanced_accuracy', 'std'),
                         macro_f1_mean=('macro_f1', 'mean'),
                         majority_baseline_mean=('majority_baseline', 'mean'))
                    .reset_index())
print("\n===== Συγκεντρωτικά αποτελέσματα ανά επίπεδο (κλίμακα 0-100) =====")
print(protocol_summary.round(2))

s1_ba = protocol_summary.loc[protocol_summary.split_level == 'S1', 'balanced_accuracy_mean'].values
s3_ba = protocol_summary.loc[protocol_summary.split_level == 'S3', 'balanced_accuracy_mean'].values
if len(s1_ba) and len(s3_ba):
    print(f"\nΠτώση S1→S3 (balanced accuracy): {(s1_ba[0] - s3_ba[0]):.2f} ποσοστιαίες μονάδες")

rng_b = np.random.RandomState(0)
boot = []
for _ in range(1000):
    ii = rng_b.randint(0, len(s3_true_all), len(s3_true_all))
    boot.append(balanced_accuracy_score(s3_true_all[ii], s3_pred_all[ii]))
lo, hi = np.percentile(boot, [2.5, 97.5])
print(f"S3 balanced accuracy: {np.mean(boot):.2%} (95% CI [{lo:.2%}, {hi:.2%}])")

common = sorted(set(per_subject_acc['S1']) & set(per_subject_acc['S3']))
if len(common) >= 5:
    stat, p = wilcoxon([per_subject_acc['S1'][s] for s in common],
                       [per_subject_acc['S3'][s] for s in common])
    print(f"\nWilcoxon S1 vs S3 (n={len(common)} υποκείμενα): stat={stat:.3f}, p={p:.4f}")
else:
    print(f"\nΠολύ λίγα κοινά υποκείμενα ({len(common)}) για Wilcoxon.")

fig, axes = plt.subplots(1, 2, figsize=(7.5, 3.2))
for ax, (lvl, yt, yp) in zip(axes, [('S1', y1_true, y1_pred), ('S3', s3_true_all, s3_pred_all)]):
    sns.heatmap(confusion_matrix(yt, yp, labels=[0, 1]), annot=True, fmt='d', cmap='Blues',
                ax=ax, xticklabels=['Low', 'High'], yticklabels=['Low', 'High'])
    ax.set_title(lvl); ax.set_xlabel('Predicted'); ax.set_ylabel('True')
plt.tight_layout(); plt.show()
print(f"Μέσος όρος Macro F1: {summary['macro_f1'].mean():.2f}% ± {summary['macro_f1'].std():.2f}%")
FINE_TUNE_EPOCHS = 5
FINE_TUNE_LR = 1e-5
TARGET_SUBJECT = sorted(info['subject_id'].unique())[0]

print(f"Στοχευμένο υποκείμενο για fine-tuning: {TARGET_SUBJECT}")

subject_mask = info['subject_id'] == TARGET_SUBJECT
subject_idx = info.index[subject_mask].values
subject_trials = info.loc[subject_mask, 'trial_id'].values

unique_trials = np.unique(subject_trials)
rng = np.random.RandomState(42)
rng.shuffle(unique_trials)
split_point = max(1, int(0.7 * len(unique_trials)))
ft_trials, test_trials = unique_trials[:split_point], unique_trials[split_point:]

ft_idx = subject_idx[np.isin(subject_trials, ft_trials)]
subj_test_idx = subject_idx[np.isin(subject_trials, test_trials)]

ft_loader = DataLoader(Subset(dataset, ft_idx.tolist()), batch_size=BATCH_SIZE, shuffle=True)
subj_test_loader = DataLoader(Subset(dataset, subj_test_idx.tolist()), batch_size=BATCH_SIZE, shuffle=False)

finetune_model = CCNN(num_classes=2, in_channels=4, grid_size=(9, 9)).to(device)
finetune_model.load_state_dict(best_model_state)

_, y_true_before, y_pred_before, y_prob_before = evaluate(finetune_model, subj_test_loader)
report_before = full_report(y_true_before, y_pred_before, y_prob_before, title=f"Πριν το fine-tuning (subject {TARGET_SUBJECT})")

ft_optimizer = optim.Adam(finetune_model.parameters(), lr=FINE_TUNE_LR)
ft_criterion = nn.CrossEntropyLoss()
for epoch in range(1, FINE_TUNE_EPOCHS + 1):
    loss, acc = train_one_epoch(finetune_model, ft_loader, ft_optimizer, ft_criterion)
    print(f"Fine-tune epoch {epoch}/{FINE_TUNE_EPOCHS} | Loss {loss:.4f} | Train Acc {acc:.2f}%")

_, y_true_after, y_pred_after, y_prob_after = evaluate(finetune_model, subj_test_loader)
report_after = full_report(y_true_after, y_pred_after, y_prob_after, title=f"Μετά το fine-tuning (subject {TARGET_SUBJECT})")
print(f"\nΒalanced Accuracy πριν: {report_before['balanced_accuracy']:.2f}% -> μετά: {report_after['balanced_accuracy']:.2f}%")
FINE_TUNE_EPOCHS = 5
FINE_TUNE_LR = 1e-5
TARGET_SUBJECT = sorted(info['subject_id'].unique())[0]

print(f"Στοχευμένο υποκείμενο για fine-tuning: {TARGET_SUBJECT}")

subject_mask = info['subject_id'] == TARGET_SUBJECT
subject_idx = info.index[subject_mask].values
subject_trials = info.loc[subject_mask, 'trial_id'].values

unique_trials = np.unique(subject_trials)
rng = np.random.RandomState(42)
rng.shuffle(unique_trials)
split_point = max(1, int(0.7 * len(unique_trials)))
ft_trials, test_trials = unique_trials[:split_point], unique_trials[split_point:]

ft_idx = subject_idx[np.isin(subject_trials, ft_trials)]
subj_test_idx = subject_idx[np.isin(subject_trials, test_trials)]

ft_loader = DataLoader(Subset(dataset, ft_idx.tolist()), batch_size=BATCH_SIZE, shuffle=True)
subj_test_loader = DataLoader(Subset(dataset, subj_test_idx.tolist()), batch_size=BATCH_SIZE, shuffle=False)

finetune_model = CCNN(num_classes=2, in_channels=4, grid_size=(9, 9)).to(device)
finetune_model.load_state_dict(best_model_state)

_, y_true_before, y_pred_before, y_prob_before = evaluate(finetune_model, subj_test_loader)
report_before = full_report(y_true_before, y_pred_before, y_prob_before, title=f"Πριν το fine-tuning (subject {TARGET_SUBJECT})")

ft_optimizer = optim.Adam(finetune_model.parameters(), lr=FINE_TUNE_LR)
ft_criterion = nn.CrossEntropyLoss()
for epoch in range(1, FINE_TUNE_EPOCHS + 1):
    loss, acc = train_one_epoch(finetune_model, ft_loader, ft_optimizer, ft_criterion)
    print(f"Fine-tune epoch {epoch}/{FINE_TUNE_EPOCHS} | Loss {loss:.4f} | Train Acc {acc:.2f}%")

_, y_true_after, y_pred_after, y_prob_after = evaluate(finetune_model, subj_test_loader)
report_after = full_report(y_true_after, y_pred_after, y_prob_after, title=f"Μετά το fine-tuning (subject {TARGET_SUBJECT})")
print(f"\nΒalanced Accuracy πριν: {report_before['balanced_accuracy']:.2f}% -> μετά: {report_after['balanced_accuracy']:.2f}%")
import mne
from captum.attr import Occlusion, IntegratedGradients, LayerGradCam
from lime import lime_tabular
import shap

xai_model = CCNN(num_classes=2, in_channels=4, grid_size=(9, 9)).to(device)
xai_model.load_state_dict(best_model_state)
xai_model.eval()

grid_positions = DEAP_CHANNEL_LOCATION_DICT

def attribution_to_electrode_scores(attr_grid_2d):
    scores = np.zeros(len(DEAP_CHANNELS))
    for ch_idx, ch_name in enumerate(DEAP_CHANNELS):
        if ch_name in grid_positions:
            r, c = grid_positions[ch_name]
            scores[ch_idx] = attr_grid_2d[r, c]
    return scores

def plot_deap_topomap(scores, title):
    info_mne = mne.create_info(ch_names=DEAP_CHANNELS, sfreq=128, ch_types='eeg')
    montage = mne.channels.make_standard_montage('standard_1020')
    info_mne.set_montage(montage, match_case=False)
    fig, ax = plt.subplots(figsize=(6, 5))
    vmax = np.max(np.abs(scores))
    vmax = vmax if vmax > 1e-9 else 1e-9
    im, _ = mne.viz.plot_topomap(scores, info_mne, axes=ax, show=False, cmap='Reds',
                                  names=DEAP_CHANNELS, sphere=(0., 0., 0., 0.095), vlim=(0, vmax))
    plt.title(title, fontsize=12, pad=15)
    plt.colorbar(im, ax=ax, label="Attribution Score")
    plt.tight_layout()
    plt.show()

sample_inputs, sample_labels = next(iter(test_loader))
sample_inputs = sample_inputs.to(device)
sample_x = sample_inputs[:1]
target_class = 1

print("Occlusion...")
occlusion = Occlusion(xai_model)
attr_occ = occlusion.attribute(sample_x, target=target_class, sliding_window_shapes=(4, 3, 3))
occ_scores = attribution_to_electrode_scores(np.abs(attr_occ[0].detach().cpu().numpy()).sum(axis=0))
plot_deap_topomap(occ_scores, "Occlusion — High Valence")

print("Integrated Gradients...")
ig = IntegratedGradients(xai_model)
attr_ig = ig.attribute(sample_x, target=target_class, n_steps=50)
ig_scores = attribution_to_electrode_scores(np.abs(attr_ig[0].detach().cpu().numpy()).sum(axis=0))
plot_deap_topomap(ig_scores, "Integrated Gradients — High Valence")

print("Grad-CAM...")
gradcam = LayerGradCam(xai_model, xai_model.conv4)
attr_gc = gradcam.attribute(sample_x, target=target_class, relu_attributions=True)
gc_scores = attribution_to_electrode_scores(attr_gc[0, 0].detach().cpu().numpy())
plot_deap_topomap(gc_scores, "Grad-CAM — High Valence")

print("SHAP...")
background_inputs, _ = next(iter(train_loader))
background = background_inputs[:32].to(device)
explainer_shap = shap.GradientExplainer(xai_model, background)
shap_values = explainer_shap.shap_values(sample_x)
shap_target = shap_values[target_class][0] if isinstance(shap_values, list) else shap_values[0, ..., target_class]
shap_scores = attribution_to_electrode_scores(np.abs(shap_target).sum(axis=0))
plot_deap_topomap(shap_scores, "SHAP — High Valence")

print("LIME...")
def model_predict_proba(flat_batch):
    xai_model.eval()
    with torch.no_grad():
        t = torch.tensor(flat_batch, dtype=torch.float32).reshape(-1, 4, 9, 9).to(device)
        probs = torch.softmax(xai_model(t), dim=1)
    return probs.cpu().numpy()

lime_background = background_inputs.reshape(background_inputs.shape[0], -1).numpy()
sample_flat = sample_x.reshape(1, -1).cpu().numpy()[0]
feature_names = [f"B{b}_R{r}_C{c}" for b in range(4) for r in range(9) for c in range(9)]
explainer_lime = lime_tabular.LimeTabularExplainer(
    training_data=lime_background, mode='classification',
    feature_names=feature_names, class_names=['Low Valence', 'High Valence'])
exp = explainer_lime.explain_instance(sample_flat, model_predict_proba,
                                       num_features=len(feature_names), num_samples=500, top_labels=1)
lime_map = np.zeros(len(feature_names))
for feat_idx, weight in exp.local_exp[exp.top_labels[0]]:
    lime_map[feat_idx] = weight
lime_scores = attribution_to_electrode_scores(np.abs(lime_map.reshape(4, 9, 9)).sum(axis=0))
plot_deap_topomap(lime_scores, "LIME — High Valence")

def top5(scores, name):
    idx = np.argsort(scores)[::-1][:5]
    print(f"\nTop-5 ({name}):")
    for i in idx:
        ch = DEAP_CHANNELS[i]
        print(f"  {ch} ({REGION_MAP.get(ch, '?')}): {scores[i]:.4f}")

top5(occ_scores, "Occlusion"); top5(ig_scores, "Integrated Gradients")
top5(gc_scores, "Grad-CAM"); top5(shap_scores, "SHAP"); top5(lime_scores, "LIME")

# =====================================================================
# ΠΡΟΣΘΗΚΗ Α -- t-SNE + ΠΟΣΟΤΙΚΑ ΜΕΤΡΑ ΔΙΑΧΩΡΙΣΙΜΟΤΗΤΑΣ
# =====================================================================
from sklearn.manifold import TSNE
from sklearn.metrics import silhouette_score, davies_bouldin_score

def extract_features(model, loader, max_samples=2000):
    """Παίρνει τις αναπαραστάσεις του προτελευταίου επιπέδου (lin1) μέσω forward hook."""
    feats, labs = [], []
    store = {}

    def hook(module, inp, out):
        store['feat'] = out.detach().cpu().numpy()

    target_layer = getattr(model, 'lin1', None)
    if target_layer is None:
        target_layer = getattr(model, 'conv4')
    handle = target_layer.register_forward_hook(hook)

    model.eval()
    with torch.no_grad():
        for X, y in loader:
            _ = model(X.to(device))
            f = store['feat']
            f = f.reshape(f.shape[0], -1)
            feats.append(f)
            labs.append(y.numpy())
            if sum(len(a) for a in labs) >= max_samples:
                break
    handle.remove()
    return np.concatenate(feats)[:max_samples], np.concatenate(labs)[:max_samples]


def fisher_ratio(features, labels):
    classes = np.unique(labels)
    overall_mean = features.mean(axis=0)
    between, within = 0.0, 0.0
    for c in classes:
        fc = features[labels == c]
        between += len(fc) * np.sum((fc.mean(axis=0) - overall_mean) ** 2)
        within += np.sum((fc - fc.mean(axis=0)) ** 2)
    return float(between / within) if within > 0 else float('nan')


def class_distances(features, labels):
    classes = np.unique(labels)
    centroids = {c: features[labels == c].mean(axis=0) for c in classes}
    intra = np.mean([np.linalg.norm(features[labels == c] - centroids[c], axis=1).mean()
                      for c in classes])
    inter = np.mean([np.linalg.norm(centroids[a] - centroids[b])
                      for i, a in enumerate(classes) for b in classes[i + 1:]])
    return float(intra), float(inter)


tsne_indices = np.random.RandomState(0).choice(len(dataset), size=min(2000, len(dataset)), replace=False)
tsne_loader = DataLoader(Subset(dataset, tsne_indices.tolist()), batch_size=BATCH_SIZE, shuffle=False)

features, labels_feat = extract_features(xai_model, tsne_loader)
print(f"Αναπαραστάσεις: {features.shape} | Κατανομή κλάσεων: {np.bincount(labels_feat)}")

sil = silhouette_score(features, labels_feat)
db = davies_bouldin_score(features, labels_feat)
fr = fisher_ratio(features, labels_feat)
intra, inter = class_distances(features, labels_feat)

print("\n===== Ποσοτικά μέτρα διαχωρισιμότητας (CCNN / DEAP) =====")
print(f"Silhouette score      : {sil:+.4f}   (→ +1 άριστος διαχωρισμός, ~0 επικάλυψη, <0 λάθος ανάθεση)")
print(f"Davies-Bouldin index  : {db:.4f}    (μικρότερο = καλύτερο)")
print(f"Fisher discriminant   : {fr:.4f}    (μεγαλύτερο = καλύτερο)")
print(f"Μέση ενδο-κλασική απόσταση : {intra:.4f}")
print(f"Μέση δια-κλασική απόσταση  : {inter:.4f}")
print(f"Λόγος inter/intra          : {inter/intra:.4f}  (>1 υποδηλώνει διαχωρίσιμες κλάσεις)")

tsne_xy = TSNE(n_components=2, random_state=0, perplexity=30, init='pca').fit_transform(features)
plt.figure(figsize=(7, 6))
for c, name, color in [(0, 'Low Valence', '#C44E52'), (1, 'High Valence', '#4C72B0')]:
    m = labels_feat == c
    plt.scatter(tsne_xy[m, 0], tsne_xy[m, 1], s=8, alpha=0.6, c=color, label=name)
plt.legend()
plt.title(f"t-SNE αναπαραστάσεων CCNN / DEAP\nSilhouette = {sil:+.3f} | Davies-Bouldin = {db:.3f}")
plt.xlabel("t-SNE 1"); plt.ylabel("t-SNE 2")
plt.tight_layout(); plt.show()


# =====================================================================
# ΠΡΟΣΘΗΚΗ Β -- Feature Ablation με topomap
# (το notebook είχε ήδη Occlusion, Integrated Gradients, Grad-CAM, SHAP και LIME·
#  εδώ προστίθεται η μόνη μέθοδος που έλειπε)
# =====================================================================
from captum.attr import FeatureAblation

print("\nFeature Ablation...")
# Κάθε κελί (r,c) του πλέγματος αφαιρείται ΜΑΖΙ σε όλες τις 4 ζώνες, ώστε η
# σημαντικότητα να αντιστοιχεί σε ηλεκτρόδιο και όχι σε μεμονωμένη φασματική ζώνη.
feature_mask = torch.arange(9 * 9).reshape(1, 1, 9, 9).expand(1, 4, 9, 9).to(device)
ablator = FeatureAblation(xai_model)
attr_fa = ablator.attribute(sample_x, target=target_class, feature_mask=feature_mask)
fa_scores = attribution_to_electrode_scores(np.abs(attr_fa[0].detach().cpu().numpy()).sum(axis=0))
plot_deap_topomap(fa_scores, "Feature Ablation — High Valence")
top5(fa_scores, "Feature Ablation")

# --- Συγκεντρωτικός πίνακας σημαντικότητας ανά ηλεκτρόδιο και μέθοδο ---
all_methods = {"Occlusion": occ_scores, "IntegratedGradients": ig_scores,
               "GradCAM": gc_scores, "FeatureAblation": fa_scores,
               "SHAP": shap_scores, "LIME": lime_scores}
summary_df = pd.DataFrame(all_methods, index=DEAP_CHANNELS)
summary_df.index.name = 'channel'
print("\n===== Σημαντικότητα ανά ηλεκτρόδιο και μέθοδο (DEAP) =====")
print(summary_df.round(4))
summary_df.to_csv('/kaggle/working/xai_channel_scores_deap.csv')

import os
save_dir = '/kaggle/working/DEAP_Saved_Experiment'
os.makedirs(save_dir, exist_ok=True)
torch.save(best_model_state, os.path.join(save_dir, 'ccnn_deap_best_subject_independent.pth'))
summary.to_csv(os.path.join(save_dir, 'deap_subject_independent_fold_results.csv'), index=False)
print(f"Αποθηκεύτηκαν στο: {save_dir}")
print("Κάνε 'Save Version' στο Kaggle ώστε να διατηρηθούν μόνιμα ως Output.")


# ================== ΒΗΜΑ 8: Ανάλυση εσωτερικών αναπαραστάσεων ==================
# (α) Διαχωρισιμότητα ανά επίπεδο, (β) ομοιότητα μέσω Linear CKA,
# (γ) layer-wise probing, (δ) μεταβολή παραμέτρων κατά τη μικρορύθμιση.
import itertools
from sklearn.metrics import silhouette_score, davies_bouldin_score
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split as _tts

CKA_LAYERS = ['conv1', 'conv2', 'conv3', 'conv4', 'lin1']


def _center_gram(K):
    n = K.shape[0]
    H = np.eye(n) - np.ones((n, n)) / n
    return H @ K @ H


def linear_cka(X, Y):
    """Linear CKA μεταξύ δύο πινάκων (δείγματα x διαστάσεις). Τα δείγματα πρέπει
    να είναι τα ΙΔΙΑ και στην ΙΔΙΑ σειρά στους δύο πίνακες."""
    X = np.asarray(X, dtype=np.float64) ; Y = np.asarray(Y, dtype=np.float64)
    X = X - X.mean(axis=0, keepdims=True) ; Y = Y - Y.mean(axis=0, keepdims=True)
    Kx, Ky = _center_gram(X @ X.T), _center_gram(Y @ Y.T)
    hsic = float((Kx * Ky).sum())
    norm = float(np.sqrt((Kx * Kx).sum()) * np.sqrt((Ky * Ky).sum()))
    return hsic / norm if norm > 0 else float('nan')


def fisher_ratio(feats, labels):
    """Λόγος διασποράς μεταξύ κλάσεων προς εντός κλάσεων (μεγαλύτερο = πιο διαχωρίσιμο)."""
    classes = np.unique(labels)
    overall = feats.mean(axis=0)
    between = within = 0.0
    for c in classes:
        fc = feats[labels == c]
        between += len(fc) * np.sum((fc.mean(axis=0) - overall) ** 2)
        within += np.sum((fc - fc.mean(axis=0)) ** 2)
    return float(between / within) if within > 0 else float('nan')


def class_distances(feats, labels):
    """Μέση ενδο-κλασική και δια-κλασική απόσταση."""
    classes = np.unique(labels)
    cent = {c: feats[labels == c].mean(axis=0) for c in classes}
    intra = np.mean([np.linalg.norm(feats[labels == c] - cent[c], axis=1).mean() for c in classes])
    inter = np.mean([np.linalg.norm(cent[a] - cent[b])
                     for i, a in enumerate(classes) for b in classes[i + 1:]])
    return float(intra), float(inter)


def get_layer_reps(model, loader):
    """Ενεργοποιήσεις των επιπέδων του CCNN μέσω forward hooks."""
    store, reps = {}, {n: [] for n in CKA_LAYERS}

    def mk(name):
        def hook(_m, _i, out):
            store[name] = out.detach().cpu().numpy()
        return hook

    handles = [getattr(model, n).register_forward_hook(mk(n))
               for n in CKA_LAYERS if hasattr(model, n)]
    model.eval()
    try:
        with torch.no_grad():
            for X, _y in loader:
                _ = model(X.to(device))
                for n in CKA_LAYERS:
                    if n in store:
                        reps[n].append(store[n].reshape(len(X), -1))
    finally:
        for h in handles:
            h.remove()
    return {n: np.concatenate(v) for n, v in reps.items() if v}


def load_ccnn(state):
    m = CCNN(num_classes=2, in_channels=4, grid_size=(9, 9)).to(device)
    m.load_state_dict(state)
    return m


# --- Δείγμα από ΟΛΟΚΛΗΡΟ το dataset, ώστε να υπάρχουν όλα τα υποκείμενα ---
rng_r = np.random.RandomState(0)
rep_idx = rng_r.choice(info.index.values, size=min(3000, len(info)), replace=False)
rep_loader = DataLoader(Subset(dataset, rep_idx.tolist()), batch_size=BATCH_SIZE, shuffle=False)
rep_labels = info.loc[rep_idx, 'label'].values

reps_s3 = get_layer_reps(load_ccnn(best_model_state), rep_loader)
for k, v in reps_s3.items():
    print(f"Αναπαράσταση '{k}': {v.shape}")

# --- (α) Μέτρα διαχωρισιμότητας ανά επίπεδο ---
sep_rows = []
for name, feats in reps_s3.items():
    if np.allclose(feats.std(axis=0), 0):
        continue
    intra, inter = class_distances(feats, rep_labels)
    sep_rows.append({'stage': name,
                     'silhouette': silhouette_score(feats, rep_labels),
                     'davies_bouldin': davies_bouldin_score(feats, rep_labels),
                     'fisher_ratio': fisher_ratio(feats, rep_labels),
                     'intra_class_dist': intra, 'inter_class_dist': inter,
                     'inter_over_intra': inter / intra if intra > 0 else float('nan')})
separability_df = pd.DataFrame(sep_rows).set_index('stage')
print("\n===== Μέτρα διαχωρισιμότητας ανά επίπεδο =====")
print(separability_df.round(4))

# --- (β) CKA: μεταξύ επιπέδων και μεταξύ S1/S3 ---
names = list(reps_s3.keys())
cka_layers = pd.DataFrame(index=names, columns=names, dtype=float)
for a, b in itertools.product(names, names):
    cka_layers.loc[a, b] = linear_cka(reps_s3[a], reps_s3[b])
print("\n===== CKA μεταξύ επιπέδων =====")
print(cka_layers.round(3))

fig, ax = plt.subplots(figsize=(5.5, 4.5))
sns.heatmap(cka_layers.astype(float), annot=True, fmt='.2f', cmap='viridis', vmin=0, vmax=1, ax=ax)
ax.set_title('Linear CKA μεταξύ επιπέδων')
plt.tight_layout(); plt.show()

reps_s1 = get_layer_reps(load_ccnn(s1_state), rep_loader)
cka_cross = pd.DataFrame([{'stage': n, 'CKA_S1_vs_S3': linear_cka(reps_s1[n], reps_s3[n])}
                          for n in names if n in reps_s1]).set_index('stage')
print("\n===== CKA ίδιου επιπέδου: μοντέλο S1 έναντι S3 =====")
print(cka_cross.round(3))
print("Χαμηλές τιμές σημαίνουν ότι τα δύο πρωτόκολλα οδηγούν σε ποιοτικά")
print("διαφορετικές αναπαραστάσεις, παρότι η αρχιτεκτονική είναι ίδια.")

fig, ax = plt.subplots(figsize=(5.5, 3.2))
cka_cross['CKA_S1_vs_S3'].plot(kind='bar', ax=ax, color='#4C72B0')
ax.set_ylim(0, 1); ax.set_ylabel('Linear CKA'); ax.set_title('Ομοιότητα αναπαραστάσεων S1 vs S3')
plt.xticks(rotation=0); plt.tight_layout(); plt.show()

# --- (γ) Layer-wise probing ---
probe_targets = {'label (συναίσθημα)': 'label',
                 'subject_id (ταυτότητα)': 'subject_id',
                 'trial_id (δοκιμασία)': 'trial_id'}
probe_rows = []
for stage, feats in reps_s3.items():
    if np.allclose(feats.std(axis=0), 0):
        continue
    for lname, col in probe_targets.items():
        if col not in info.columns:
            continue
        y = info.loc[rep_idx, col].values
        classes, counts = np.unique(y, return_counts=True)
        if len(classes) < 2 or counts.min() < 2:
            continue
        Xtr, Xte, ytr, yte = _tts(feats, y, test_size=0.3, random_state=0, stratify=y)
        clf = LogisticRegression(max_iter=1000).fit(Xtr, ytr)
        acc = clf.score(Xte, yte)
        chance = counts.max() / counts.sum()
        probe_rows.append({'stage': stage, 'target': lname, 'probe_accuracy': acc,
                           'chance_level': chance, 'ratio': acc / chance})

probe_df = pd.DataFrame(probe_rows)
if not probe_df.empty:
    acc_tab = probe_df.pivot(index='stage', columns='target', values='probe_accuracy')
    acc_tab = acc_tab.reindex([s for s in CKA_LAYERS if s in acc_tab.index])
    print("\n===== Layer-wise probing: ακρίβεια ανά επίπεδο =====")
    print(acc_tab.round(4))
    print("\nΛόγος προς τη γραμμή βάσης (τιμές >> 1 = αποκωδικοποιήσιμη πληροφορία):")
    print(probe_df.pivot(index='stage', columns='target', values='ratio')
          .reindex(acc_tab.index).round(2))

    fig, ax = plt.subplots(figsize=(1.6 * len(acc_tab) + 3, 3.4))
    acc_tab.plot(kind='bar', ax=ax)
    ax.set_ylabel('Ακρίβεια probe'); ax.set_ylim(0, 1)
    ax.set_title('Linear probing ανά επίπεδο')
    plt.xticks(rotation=0); plt.legend(fontsize=8); plt.tight_layout(); plt.show()

# --- (δ) Μεταβολή παραμέτρων κατά τη μικρορύθμιση ---
# Το CKA συγκρίνει τι ΠΑΡΑΓΕΙ το δίκτυο· εδώ συγκρίνεται τι ΑΛΛΑΞΕ μέσα του.
tuned = globals().get('ft_model', globals().get('finetune_model'))
if tuned is not None:
    after = tuned.state_dict()
    wrows = []
    for pname, w_before in best_model_state.items():
        if not torch.is_floating_point(w_before):
            continue
        delta = (after[pname] - w_before).float()
        n_before = float(w_before.float().norm())
        wrows.append({'parameter': pname, 'shape': tuple(w_before.shape),
                      'abs_change': float(delta.norm()),
                      'rel_change': float(delta.norm()) / n_before if n_before > 0 else float('nan')})
    weight_df = pd.DataFrame(wrows).sort_values('rel_change', ascending=False)
    print("\n===== Μεταβολή παραμέτρων κατά τη μικρορύθμιση =====")
    print(weight_df.round(6).to_string(index=False))

    weight_df['layer'] = weight_df['parameter'].str.split('.').str[0]
    layer_change = weight_df.groupby('layer')['rel_change'].mean().sort_values(ascending=False)
    print("\nΜέση σχετική μεταβολή ανά επίπεδο:")
    print(layer_change.round(6))

    fig, ax = plt.subplots(figsize=(6, 3.4))
    layer_change.plot(kind='bar', ax=ax, color='#C44E52')
    ax.set_ylabel('Μέση σχετική μεταβολή')
    ax.set_title('Ποια επίπεδα άλλαξαν περισσότερο στη μικρορύθμιση')
    plt.xticks(rotation=0); plt.tight_layout(); plt.show()
else:
    print("\nΔεν βρέθηκε μικρορυθμισμένο μοντέλο· η ανάλυση παραμέτρων παραλείπεται.")
```
