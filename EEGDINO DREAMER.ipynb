```python
try:
    import braindecode, captum, lime, mne, shap
except ImportError:
    print("Εγκατάσταση απαραίτητων βιβλιοθηκών...")
    !pip install -q "braindecode[hub]" shap lime captum mne scikit-learn statsmodels

import copy
import itertools
import json
import math
import os
import re
import warnings
import zipfile

warnings.filterwarnings("ignore")

import matplotlib.pyplot as plt
import mne
import numpy as np
import pandas as pd
from scipy import signal as sps
from scipy.stats import kendalltau, spearmanr, wilcoxon
import seaborn as sns
from sklearn.manifold import TSNE
from sklearn.metrics import (
    balanced_accuracy_score,
    brier_score_loss,
    cohen_kappa_score,
    confusion_matrix,
    f1_score,
    matthews_corrcoef,
    precision_recall_fscore_support,
    roc_auc_score,
)
from sklearn.model_selection import GroupKFold, LeaveOneGroupOut, train_test_split
from statsmodels.stats.multitest import multipletests
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, TensorDataset

from braindecode.models import EEGDINO
from captum.attr import (
    DeepLift,
    FeatureAblation,
    IntegratedGradients,
    Occlusion,
    Saliency,
    LRP,
)
from lime import lime_tabular
import shap

# Απενεργοποίηση του Colab drive mount για εκτέλεση στο Kaggle
IN_COLAB = False

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Device: {device}")

try:
    display
except NameError:

  def display(x):
    print(x)


N_SEEDS = 2
SEEDS = [0, 1][:N_SEEDS]

RUN_FULL_LOSO = False
LOSO_MAX_SUBJECTS = 4

N_EPOCHS = (
    10  # fine-tuning ενός pretrained transformer· λίγες εποχές συνήθως αρκούν
)
BATCH_SIZE = 32
LR = 1e-4  # ίδιο με το αρχικό — χαμηλό, κατάλληλο για fine-tuning

N_EXPLAIN_SAMPLES = 5
N_BOOTSTRAP = 1000
CI_ALPHA = 0.05

DATASET_NAME = "DREAMER"
FS = 128
WINDOW_SIZE = 128
STEP = 64
SECONDS_LIMIT = 60

CHANNELS = [
    "AF3",
    "F7",
    "F3",
    "FC5",
    "T7",
    "P7",
    "O1",
    "O2",
    "P8",
    "T8",
    "FC6",
    "F4",
    "F8",
    "AF4",
]
N_CHANNELS = len(CHANNELS)

BANDS = {
    "delta": (1, 4),
    "theta": (4, 8),
    "alpha": (8, 13),
    "beta": (13, 30),
    "gamma": (30, 50),
}

os.makedirs("outputs", exist_ok=True)

# Χρήση της διαδρομής εισόδου στο Kaggle
DATA_PATH = "/kaggle/input/datasets/thanosleggitsas/dreamer-zip"

mat_path = None
extract_path = "/kaggle/working/DREAMER_extracted"

# Εύρεση του αρχείου είτε είναι απευθείας .mat είτε .zip
if os.path.exists(DATA_PATH):
  if os.path.isfile(DATA_PATH):
    if DATA_PATH.endswith(".mat"):
      mat_path = DATA_PATH
    elif DATA_PATH.endswith(".zip"):
      if not os.path.exists(extract_path):
        print("Αποσυμπίεση...")
        with zipfile.ZipFile(DATA_PATH, "r") as zf:
          zf.extractall(extract_path)
  else:
    for root, _, files in os.walk(DATA_PATH):
      for f in files:
        if f.endswith(".mat") and "DREAMER" in f.upper():
          mat_path = os.path.join(root, f)
          break
        elif f.endswith(".zip") and "DREAMER" in f.upper():
          if not os.path.exists(extract_path):
            print("Αποσυμπίεση...")
            with zipfile.ZipFile(os.path.join(root, f), "r") as zf:
              zf.extractall(extract_path)

if mat_path is None and os.path.exists(extract_path):
  for root, _, files in os.walk(extract_path):
    for f in files:
      if f.endswith(".mat") and "DREAMER" in f.upper():
        mat_path = os.path.join(root, f)
        break
    if mat_path:
      break

if mat_path is None:
  raise FileNotFoundError(
      f"Δεν βρέθηκε το DREAMER.mat στη διαδρομή {DATA_PATH}."
  )
print(f"mat_path = {mat_path}")


def load_dreamer_with_manifest(
    path, window_size=WINDOW_SIZE, step=STEP, seconds_limit=SECONDS_LIMIT
):
  raw = sio.loadmat(path)
  data_struct = raw["DREAMER"][0, 0]["Data"]

  X_list, y_list, rows = [], [], []
  wid = 0
  for sub in range(23):
    try:
      eeg_root = data_struct[0, sub]["EEG"][0, 0]
      stimuli = (
          eeg_root["stimuli"][0, 0]
          if "stimuli" in eeg_root.dtype.names
          else eeg_root["Signals"][0, 0]
      )
      valence = data_struct[0, sub]["ScoreValence"][0, 0]
    except Exception:
      continue

    for trial in range(18):
      label = 1 if valence[trial, 0] >= 3.0 else 0
      signal = stimuli[trial, 0]
      limit = min(len(signal), FS * seconds_limit)

      for start in range(0, limit - window_size, step):
        seg = signal[start : start + window_size, :]
        seg = (seg - np.mean(seg)) / (np.std(seg) + 1e-6)
        X_list.append(
            seg.T
        )  # (14, 128) — (channels, time), ίδιο με το αρχικό EEGDINO input
        y_list.append(label)
        rows.append({
            "window_id": wid,
            "subject_id": sub,
            "trial_id": trial,
            "session_id": sub,
            "dataset": "DREAMER",
            "label": label,
            "start_sample": start,
        })
        wid += 1

  X = np.array(X_list, dtype=np.float32)
  y = np.array(y_list, dtype=np.int64)
  manifest = pd.DataFrame(rows)
  return torch.tensor(X), torch.tensor(y), manifest


import scipy.io as sio

X_data, y_data, manifest = load_dreamer_with_manifest(mat_path)
manifest.to_csv("outputs/manifest.csv", index=False)
print(
    f"Σύνολο παραθύρων: {len(y_data)} | Θετικά (High Valence):"
    f" {y_data.sum().item()} ({y_data.float().mean().item():.1%})"
)
display(manifest.head())


def classify_channel(ch):
  name = ch.upper()
  if name.endswith("Z"):
    prefix, hemi = name[:-1], "midline"
  else:
    m = re.match(r"^([A-Z]+)(\d+)$", name)
    if m:
      prefix, num = m.group(1), int(m.group(2))
      hemi = "left" if num % 2 == 1 else "right"
    else:
      prefix, hemi = name, "midline"

  region_map = {
      "FP": ("frontal", "anterior"),
      "AF": ("frontal", "anterior"),
      "F": ("frontal", "anterior"),
      "FC": ("central", "anterior"),
      "FT": ("temporal", "anterior"),
      "C": ("central", "midline"),
      "T": ("temporal", "midline"),
      "TP": ("temporal", "posterior"),
      "CP": ("parietal", "anterior"),
      "P": ("parietal", "posterior"),
      "PO": ("occipital", "posterior"),
      "O": ("occipital", "posterior"),
      "CB": ("occipital", "posterior"),
  }
  region, axis = region_map.get(prefix, ("unknown", "unknown"))
  return region, hemi, axis


CHANNEL_REGION = {ch: classify_channel(ch) for ch in CHANNELS}


def build_montage_info(channels=CHANNELS):
  montage = mne.channels.make_standard_montage("standard_1005")
  positions = montage.get_positions()["ch_pos"]
  plottable = [ch for ch in channels if ch in positions]
  missing = [ch for ch in channels if ch not in positions]
  if missing:
    print(
        f"Σημείωση: {len(missing)} κανάλια χωρίς πρότυπη θέση 10-20/10-10 — δεν"
        f" εμφανίζονται στα scalp maps: {missing}"
    )
  info = mne.create_info(ch_names=plottable, sfreq=FS, ch_types="eeg")
  info.set_montage(montage)
  return info, plottable


MNE_INFO, PLOTTABLE_CHANNELS = build_montage_info()
PLOTTABLE_IDX = [CHANNELS.index(ch) for ch in PLOTTABLE_CHANNELS]


def plot_scalp_map(values_full, title, ax=None, cmap="RdBu_r"):
  values = np.asarray(values_full)[PLOTTABLE_IDX]
  fig = None
  if ax is None:
    fig, ax = plt.subplots(figsize=(4, 4))
  mne.viz.plot_topomap(
      values, MNE_INFO, axes=ax, show=False, cmap=cmap, contours=4
  )
  ax.set_title(title, fontsize=10)
  if fig is not None:
    plt.show()
  return ax


def channel_to_region_table(values, channels=CHANNELS):
  rows = []
  for ch, v in zip(channels, values):
    region, hemi, axis = CHANNEL_REGION[ch]
    rows.append({
        "channel": ch,
        "value": float(v),
        "region": region,
        "hemisphere": hemi,
        "axis": axis,
    })
  df = pd.DataFrame(rows)
  region_table = df.groupby("region")["value"].sum().sort_values(ascending=False)
  return df, region_table


def frontal_asymmetry(values, channels=CHANNELS):
  candidate_pairs = [
      ("AF3", "AF4"),
      ("F3", "F4"),
      ("F7", "F8"),
      ("FC5", "FC6"),
      ("Fp1", "Fp2"),
  ]
  idx = {ch: i for i, ch in enumerate(channels)}
  out = {}
  for l, r in candidate_pairs:
    lu, ru = l.upper(), r.upper()
    idx_upper = {k.upper(): v for k, v in idx.items()}
    if lu in idx_upper and ru in idx_upper:
      out[f"{l}-{r}"] = float(values[idx_upper[ru]] - values[idx_upper[lu]])
  return out


def s1_window_split(manifest, seed):
  idx = manifest.index.values
  train_idx, test_idx = train_test_split(
      idx, test_size=0.2, random_state=seed, stratify=manifest["label"].values
  )
  yield train_idx, test_idx, "S1_full"


def s2_trial_split(manifest, seed, n_splits=5):
  groups = (
      manifest["subject_id"].astype(str) + "_" + manifest["trial_id"].astype(str)
  )
  gkf = GroupKFold(n_splits=n_splits)
  idx = manifest.index.values
  for fold, (tr, te) in enumerate(
      gkf.split(idx, manifest["label"].values, groups=groups)
  ):
    yield idx[tr], idx[te], f"S2_fold{fold}"


def s3_subject_split(
    manifest, seed, run_full=RUN_FULL_LOSO, max_subjects=LOSO_MAX_SUBJECTS
):
  logo = LeaveOneGroupOut()
  idx = manifest.index.values
  groups = manifest["subject_id"].values
  subjects = sorted(manifest["subject_id"].unique())
  if not run_full:
    rng = np.random.RandomState(seed)
    subjects = list(
        rng.choice(
            subjects, size=min(max_subjects, len(subjects)), replace=False
        )
    )
  for tr, te in logo.split(idx, manifest["label"].values, groups=groups):
    held_out_subject = manifest.iloc[te]["subject_id"].iloc[0]
    if held_out_subject not in subjects:
      continue
    yield idx[tr], idx[te], f"S3_subject{held_out_subject}"


def build_eegdino():
  """Φρέσκο, προεκπαιδευμένο EEGDINO σε κάθε fold (ξεκινάει πάντα από τα ίδια
  pretrained βάρη — ΟΧΙ συνέχιση εκπαίδευσης από το προηγούμενο fold)."""
  return EEGDINO.from_pretrained(
      "braindecode/eegdino-small-pretrained",
      n_outputs=2,
      n_chans=N_CHANNELS,
      sfreq=FS,
  ).to(device)


def full_metrics(y_true, y_pred, y_prob_pos):
  prec, rec, f1c, _ = precision_recall_fscore_support(
      y_true, y_pred, labels=[0, 1], zero_division=0
  )
  vals, counts = np.unique(y_true, return_counts=True)
  majority_baseline = counts.max() / counts.sum()
  out = {
      "majority_baseline": majority_baseline,
      "balanced_accuracy": balanced_accuracy_score(y_true, y_pred),
      "macro_f1": f1_score(y_true, y_pred, average="macro", zero_division=0),
      "precision_class0": prec[0],
      "recall_class0": rec[0],
      "precision_class1": prec[1],
      "recall_class1": rec[1],
      "cohen_kappa": cohen_kappa_score(y_true, y_pred),
      "mcc": matthews_corrcoef(y_true, y_pred),
  }
  try:
    out["roc_auc"] = roc_auc_score(y_true, y_prob_pos)
  except ValueError:
    out["roc_auc"] = float("nan")
  out["learned_anything"] = (
      out["balanced_accuracy"] > out["majority_baseline"] + 0.02
  )
  return out


def train_one_split(train_idx, test_idx, seed, n_epochs=N_EPOCHS, verbose=False):
  torch.manual_seed(seed)
  np.random.seed(seed)

  train_ds = TensorDataset(X_data[train_idx], y_data[train_idx])
  test_ds = TensorDataset(X_data[test_idx], y_data[test_idx])
  train_loader = DataLoader(train_ds, batch_size=BATCH_SIZE, shuffle=True)
  test_loader = DataLoader(test_ds, batch_size=BATCH_SIZE, shuffle=False)

  model = build_eegdino()
  optimizer = optim.AdamW(
      model.parameters(), lr=LR, weight_decay=1e-2
  )  # ίδιο LR/optimizer με το αρχικό
  scheduler = optim.lr_scheduler.ReduceLROnPlateau(
      optimizer, "max", patience=2, factor=0.5
  )
  criterion = nn.CrossEntropyLoss()

  for epoch in range(n_epochs):
    model.train()
    for xb, yb in train_loader:
      xb, yb = xb.to(device), yb.to(device)
      optimizer.zero_grad()
      loss = criterion(model(xb), yb)
      loss.backward()
      optimizer.step()

    model.eval()
    val_correct, val_n = 0, 0
    with torch.no_grad():
      for xb, yb in test_loader:
        xb, yb = xb.to(device), yb.to(device)
        val_correct += (model(xb).argmax(1) == yb).sum().item()
        val_n += len(yb)
    val_acc = val_correct / max(val_n, 1)
    scheduler.step(val_acc)
    if verbose:
      print(f"  epoch {epoch+1:02d} | val_acc {val_acc:.2%}")

  model.eval()
  all_true, all_pred, all_prob = [], [], []
  with torch.no_grad():
    for xb, yb in test_loader:
      xb = xb.to(device)
      out = torch.softmax(model(xb), dim=1).cpu().numpy()
      all_true.append(yb.numpy())
      all_pred.append(out.argmax(1))
      all_prob.append(out[:, 1])
  y_true = np.concatenate(all_true)
  y_pred = np.concatenate(all_pred)
  y_prob = np.concatenate(all_prob)
  metrics = full_metrics(y_true, y_pred, y_prob)
  return model, y_true, y_pred, y_prob, metrics


def plot_confusion(y_true, y_pred, title, ax=None):
  cm = confusion_matrix(y_true, y_pred, labels=[0, 1])
  fig = None
  if ax is None:
    fig, ax = plt.subplots(figsize=(3.5, 3))
  sns.heatmap(
      cm,
      annot=True,
      fmt="d",
      cmap="Blues",
      ax=ax,
      xticklabels=["Low", "High"],
      yticklabels=["Low", "High"],
  )
  ax.set_xlabel("Predicted")
  ax.set_ylabel("True")
  ax.set_title(title, fontsize=10)
  if fig is not None:
    plt.show()
  return cm


def expected_calibration_error(y_true, y_prob_pos, n_bins=10):
  bins = np.linspace(0, 1, n_bins + 1)
  ece = 0.0
  n = len(y_true)
  for i in range(n_bins):
    mask = (y_prob_pos >= bins[i]) & (
        y_prob_pos < bins[i + 1] if i < n_bins - 1 else y_prob_pos <= bins[i + 1]
    )
    if mask.sum() == 0:
      continue
    conf = y_prob_pos[mask].mean()
    acc = (y_true[mask] == 1).mean()
    ece += (mask.sum() / n) * abs(acc - conf)
  return ece


def reliability_diagram(
    y_true, y_prob_pos, n_bins=10, title="Reliability diagram"
):
  bins = np.linspace(0, 1, n_bins + 1)
  accs, confs = [], []
  for i in range(n_bins):
    lo, hi = bins[i], bins[i + 1]
    mask = (y_prob_pos >= lo) & (
        y_prob_pos < hi if i < n_bins - 1 else y_prob_pos <= hi
    )
    if mask.sum() == 0:
      continue
    accs.append((y_true[mask] == 1).mean())
    confs.append(y_prob_pos[mask].mean())
  plt.figure(figsize=(4, 4))
  plt.plot([0, 1], [0, 1], "k--", label="ideal")
  plt.plot(confs, accs, "o-", label="model")
  plt.xlabel("Confidence")
  plt.ylabel("Empirical accuracy")
  plt.title(title)
  plt.legend()
  plt.show()
  ece = expected_calibration_error(y_true, y_prob_pos, n_bins)
  brier = brier_score_loss((y_true == 1).astype(int), y_prob_pos)
  print(f"ECE = {ece:.4f} | Brier = {brier:.4f}")
  return ece, brier


def per_subject_accuracy(y_true, y_pred, subject_ids):
  df = pd.DataFrame({"y": y_true, "p": y_pred, "subject": subject_ids})
  return df.groupby("subject").apply(lambda g: (g["y"] == g["p"]).mean())


results = []
saved_models = {}

for seed in SEEDS:
  print(f"\n=== Seed {seed} ===")

  for train_idx, test_idx, fold_name in s1_window_split(manifest, seed):
    print(f"-> Εκπαίδευση S1 (fold: {fold_name}) - Υπολογισμός εποχών...")
    model, y_true, y_pred, y_prob, m = train_one_split(
        train_idx, test_idx, seed, verbose=True
    )
    m.update({
        "model": "EEGDINO",
        "dataset": DATASET_NAME,
        "split_level": "S1",
        "fold": fold_name,
        "seed": seed,
    })
    results.append(m)
    saved_models[("S1", seed)] = (model, test_idx, y_true, y_pred, y_prob)

  for train_idx, test_idx, fold_name in s2_trial_split(
      manifest, seed, n_splits=5
  ):
    print(f"-> Εκπαίδευση S2 (fold: {fold_name}) - Υπολογισμός εποχών...")
    model, y_true, y_pred, y_prob, m = train_one_split(
        train_idx, test_idx, seed, verbose=True
    )
    m.update({
        "model": "EEGDINO",
        "dataset": DATASET_NAME,
        "split_level": "S2",
        "fold": fold_name,
        "seed": seed,
    })
    results.append(m)
    saved_models[("S2", seed, fold_name)] = (
        model,
        test_idx,
        y_true,
        y_pred,
        y_prob,
    )

  for train_idx, test_idx, fold_name in s3_subject_split(manifest, seed):
    print(f"-> Εκπαίδευση S3 (fold: {fold_name}) - Υπολογισμός εποχών...")
    model, y_true, y_pred, y_prob, m = train_one_split(
        train_idx, test_idx, seed, verbose=True
    )
    m.update({
        "model": "EEGDINO",
        "dataset": DATASET_NAME,
        "split_level": "S3",
        "fold": fold_name,
        "seed": seed,
    })
    results.append(m)
    saved_models[("S3", seed, fold_name)] = (
        model,
        test_idx,
        y_true,
        y_pred,
        y_prob,
    )

results_df = pd.DataFrame(results)
results_df.to_csv("outputs/results_S1_S2_S3.csv", index=False)
display(results_df.head())

summary = (
    results_df.groupby("split_level")
    .agg(
        balanced_accuracy_mean=("balanced_accuracy", "mean"),
        balanced_accuracy_std=("balanced_accuracy", "std"),
        macro_f1_mean=("macro_f1", "mean"),
        macro_f1_std=("macro_f1", "std"),
        majority_baseline_mean=("majority_baseline", "mean"),
    )
    .reset_index()
)

s1_acc = summary.loc[
    summary.split_level == "S1", "balanced_accuracy_mean"
].values
s3_acc = summary.loc[
    summary.split_level == "S3", "balanced_accuracy_mean"
].values
drop = (s1_acc[0] - s3_acc[0]) if len(s1_acc) and len(s3_acc) else float("nan")
print(f"Πτώση S1→S3 (balanced accuracy): {drop:.2%}")
display(summary)


def bootstrap_ci(
    y_true, y_pred, metric_fn, n_boot=N_BOOTSTRAP, alpha=CI_ALPHA, seed=0
):
  rng = np.random.RandomState(seed)
  n = len(y_true)
  stats = []
  for _ in range(n_boot):
    idx = rng.randint(0, n, n)
    stats.append(metric_fn(y_true[idx], y_pred[idx]))
  lo, hi = np.percentile(stats, [100 * alpha / 2, 100 * (1 - alpha / 2)])
  return np.mean(stats), lo, hi


any_s3_key = next(k for k in saved_models if k[0] == "S3")
_, _, y_true_s3, y_pred_s3, _ = saved_models[any_s3_key]
mean_ba, lo, hi = bootstrap_ci(y_true_s3, y_pred_s3, balanced_accuracy_score)
print(f"S3 balanced accuracy: {mean_ba:.2%} (95% CI [{lo:.2%}, {hi:.2%}])")

s1_key = next(k for k in saved_models if k[0] == "S1")
_, s1_test_idx, s1_true, s1_pred, _ = saved_models[s1_key]
s1_subjects = manifest.loc[s1_test_idx, "subject_id"].values
s1_per_subj = per_subject_accuracy(s1_true, s1_pred, s1_subjects)

s3_keys = [k for k in saved_models if k[0] == "S3"]
s3_per_subj = {}
for k in s3_keys:
  _, test_idx, y_true, y_pred, _ = saved_models[k]
  subj = manifest.loc[test_idx, "subject_id"].iloc[0]
  s3_per_subj[subj] = (y_true == y_pred).mean()
s3_per_subj = pd.Series(s3_per_subj)

common = s1_per_subj.index.intersection(s3_per_subj.index)
if len(common) >= 5:
  stat, p = wilcoxon(s1_per_subj.loc[common], s3_per_subj.loc[common])
  reject, p_adj, _, _ = multipletests([p], method="holm")
  print(
      f"Wilcoxon S1 vs S3 (n={len(common)} εθελοντές): stat={stat:.3f},"
      f" p={p:.4f}, p_holm={p_adj[0]:.4f}, σημαντικό={reject[0]}"
  )
else:
  print(
      "Πολύ λίγοι κοινοί εθελοντές μεταξύ S1-test-set και S3-folds για Wilcoxon"
      " (αύξησε LOSO_MAX_SUBJECTS ή RUN_FULL_LOSO)."
  )

fig, axes = plt.subplots(1, 3, figsize=(10.5, 3))
for ax, level in zip(axes, ["S1", "S2", "S3"]):
  key = next((k for k in saved_models if k[0] == level), None)
  if key is None:
    continue
  _, _, y_true, y_pred, _ = saved_models[key]
  plot_confusion(y_true, y_pred, title=f"{level}", ax=ax)
plt.tight_layout()
plt.show()

_, _, y_true_s3, y_pred_s3, y_prob_s3 = saved_models[any_s3_key]
reliability_diagram(y_true_s3, y_prob_s3, title="Reliability — S3 (LOSO)")
subj_balance = manifest.groupby("subject_id")["label"].mean().sort_values()
plt.figure(figsize=(9, 3))
sns.barplot(
    x=subj_balance.index.astype(str), y=subj_balance.values, color="steelblue"
)
plt.axhline(0.5, color="red", linestyle="--", linewidth=1)
plt.ylabel("% High Valence")
plt.xlabel("subject_id")
plt.title(f"Ισορροπία κλάσεων ανά εθελοντή ({DATASET_NAME})")
plt.xticks(rotation=90)
plt.show()
outlier_subjects = subj_balance[(subj_balance < 0.15) | (subj_balance > 0.85)]
print("Πιθανοί degenerate-risk subjects (πολύ ανισόρροποι):")
print(outlier_subjects)

tsne_model, tsne_test_idx, _, _, _ = saved_models[any_s3_key]
tsne_loader = DataLoader(
    TensorDataset(X_data[tsne_test_idx], y_data[tsne_test_idx]),
    batch_size=64,
    shuffle=False,
)
tsne_model.eval()
f_list, l_list = [], []
with torch.no_grad():
  for xb, yb in tsne_loader:
    # Χρησιμοποιούμε την τελευταία κρυφή αναπαράσταση πριν τον classifier head, αν
    # εκτίθεται ως model.forward_features(); αλλιώς κρατάμε τα logits ως προσέγγιση.
    if hasattr(tsne_model, "forward_features"):
      feat = tsne_model.forward_features(xb.to(device))
      if feat.dim() > 2:
        feat = feat.mean(dim=1)
    else:
      feat = tsne_model(xb.to(device))
    f_list.append(feat.cpu().numpy())
    l_list.append(yb.numpy())
X_tsne = TSNE(n_components=2, init="pca", random_state=0).fit_transform(
    np.concatenate(f_list)
)
plt.figure(figsize=(7, 5))
sns.scatterplot(
    x=X_tsne[:, 0],
    y=X_tsne[:, 1],
    hue=np.concatenate(l_list),
    palette="coolwarm",
)
plt.title("t-SNE Clusters (S3 / LOSO model)")
plt.show()

# =====================================================================
# ΠΟΣΟΤΙΚΑ ΜΕΤΡΑ ΔΙΑΧΩΡΙΣΙΜΟΤΗΤΑΣ
# Το t-SNE είναι μόνο οπτική ένδειξη και μπορεί να "δείξει" ομάδες που δεν
# αντιστοιχούν σε πραγματική δομή, ανάλογα με το perplexity. Τα παρακάτω μέτρα
# τεκμηριώνουν αριθμητικά τον βαθμό διαχωρισμού των δύο κλάσεων.
# =====================================================================
from sklearn.metrics import silhouette_score, davies_bouldin_score

features_s3 = np.concatenate(f_list)
labels_s3 = np.concatenate(l_list)


def fisher_ratio(features, labels):
  """Λόγος διασποράς μεταξύ κλάσεων προς εντός κλάσεων (μεγαλύτερο = πιο διαχωρίσιμο)."""
  classes = np.unique(labels)
  overall_mean = features.mean(axis=0)
  between, within = 0.0, 0.0
  for c in classes:
    fc = features[labels == c]
    between += len(fc) * np.sum((fc.mean(axis=0) - overall_mean) ** 2)
    within += np.sum((fc - fc.mean(axis=0)) ** 2)
  return float(between / within) if within > 0 else float("nan")


def class_distances(features, labels):
  """Μέση ενδο-κλασική και δια-κλασική απόσταση."""
  classes = np.unique(labels)
  centroids = {c: features[labels == c].mean(axis=0) for c in classes}
  intra = np.mean([
      np.linalg.norm(features[labels == c] - centroids[c], axis=1).mean()
      for c in classes
  ])
  inter = np.mean([
      np.linalg.norm(centroids[a] - centroids[b])
      for i, a in enumerate(classes) for b in classes[i + 1:]
  ])
  return float(intra), float(inter)


sil = silhouette_score(features_s3, labels_s3)
db = davies_bouldin_score(features_s3, labels_s3)
fr = fisher_ratio(features_s3, labels_s3)
intra, inter = class_distances(features_s3, labels_s3)

print("\n===== Ποσοτικά μέτρα διαχωρισιμότητας (EEGDINO / DREAMER, S3) =====")
print(f"Αναπαραστάσεις: {features_s3.shape} | Κατανομή κλάσεων: {np.bincount(labels_s3)}")
print(f"Silhouette score      : {sil:+.4f}   (→ +1 άριστος διαχωρισμός, ~0 επικάλυψη)")
print(f"Davies-Bouldin index  : {db:.4f}    (μικρότερο = καλύτερο)")
print(f"Fisher discriminant   : {fr:.4f}    (μεγαλύτερο = καλύτερο)")
print(f"Μέση ενδο-κλασική απόσταση : {intra:.4f}")
print(f"Μέση δια-κλασική απόσταση  : {inter:.4f}")
print(f"Λόγος inter/intra          : {inter/intra:.4f}  (>1 υποδηλώνει διαχωρίσιμες κλάσεις)")

separability_df = pd.DataFrame([{
    "model": "EEGDINO", "dataset": DATASET_NAME, "split_level": "S3",
    "silhouette": sil, "davies_bouldin": db, "fisher_ratio": fr,
    "intra_class_dist": intra, "inter_class_dist": inter,
    "inter_over_intra": inter / intra,
}])
separability_df.to_csv("outputs/separability_metrics.csv", index=False)
display(separability_df)

xai_model, xai_test_idx, _, _, _ = saved_models[any_s3_key]
xai_model.eval()

xai_loader = DataLoader(
    TensorDataset(X_data[xai_test_idx], y_data[xai_test_idx]),
    batch_size=32,
    shuffle=False,
)
xai_batch_x, xai_batch_y = next(iter(xai_loader))
background = xai_batch_x[: min(30, len(xai_batch_x))].to(device)
explain_samples = xai_batch_x[:N_EXPLAIN_SAMPLES].to(device)
explain_labels = xai_batch_y[:N_EXPLAIN_SAMPLES]


def _to_map(attr_tensor):
  return (
      attr_tensor.squeeze().detach().cpu().numpy()
  )  # (N_CHANNELS, WINDOW_SIZE)


def explain_integrated_gradients(sample, target=1):
  ig = IntegratedGradients(xai_model)
  attr = ig.attribute(sample, target=target, n_steps=32)
  return _to_map(attr)


def explain_saliency(sample, target=1):
  sal = Saliency(xai_model)
  attr = sal.attribute(sample.requires_grad_(True), target=target)
  return _to_map(attr)


def explain_occlusion(sample, target=1):
  occ = Occlusion(xai_model)
  attr = occ.attribute(
      sample, target=target, sliding_window_shapes=(1, 10), strides=(1, 5)
  )
  return _to_map(attr)


def explain_deeplift(sample, target=1):
  dl = DeepLift(xai_model)
  baseline = torch.zeros_like(sample)
  attr = dl.attribute(sample, baselines=baseline, target=target)
  return _to_map(attr)


def explain_feature_ablation_channel(sample, target=1):
  ablator = FeatureAblation(xai_model)
  feature_mask = (
      torch.arange(N_CHANNELS)
      .reshape(1, N_CHANNELS, 1)
      .expand(1, N_CHANNELS, WINDOW_SIZE)
      .to(device)
  )
  attr = ablator.attribute(sample, target=target, feature_mask=feature_mask)
  return _to_map(attr)


def explain_lrp(sample, target=1):
  try:
    lrp = LRP(xai_model)
    attr = lrp.attribute(sample.requires_grad_(True), target=target)
    return _to_map(attr)
  except Exception as e:
    print(
        f"LRP απέτυχε ({e}); επιστρέφεται 0-map. (Αναμενόμενο — το EEGDINO είναι"
        " transformer/DINO αρχιτεκτονική, το captum LRP στοχεύει κυρίως CNN/MLP"
        " layers.)"
    )
    return np.zeros((N_CHANNELS, WINDOW_SIZE))


def explain_shap_gradient(background, sample, target=1):
  explainer = shap.GradientExplainer(xai_model, background)
  sv = explainer.shap_values(sample)

  # ΔΙΟΡΘΩΣΗ: νεότερες εκδόσεις του shap επιστρέφουν ΕΠΙΠΛΕΟΝ άξονα κλάσεων στο τέλος,
  # π.χ. (1, 14, 128, 2). Το προηγούμενο np.squeeze(arr)[0] έπαιρνε το πρώτο ΚΑΝΑΛΙ
  # αντί για την κλάση-στόχο, δίνοντας σχήμα (128, 2) και τελικά 128 τιμές αντί για 14.
  if isinstance(sv, list):
    arr = np.array(sv[target] if len(sv) > target else sv[0])
  else:
    arr = np.array(sv)
  arr = np.squeeze(arr)

  if arr.ndim == 3:
    if arr.shape[-1] == 2:        # (14, 128, n_classes)
      arr = arr[..., target]
    elif arr.shape[0] == 2:       # (n_classes, 14, 128)
      arr = arr[target]
    else:
      arr = arr[0]
  return arr


def explain_lime(all_X_flat, sample_flat, target=1):
  def predict_p(x):
    x_t = (
        torch.tensor(x, dtype=torch.float32)
        .reshape(-1, N_CHANNELS, WINDOW_SIZE)
        .to(device)
    )
    xai_model.eval()
    with torch.no_grad():
      return torch.softmax(xai_model(x_t), dim=1).cpu().numpy()

  expl = lime_tabular.LimeTabularExplainer(
      all_X_flat,
      mode="classification",
      class_names=["Low Valence", "High Valence"],
  )
  exp = expl.explain_instance(sample_flat, predict_p, num_features=40)
  out = np.zeros(N_CHANNELS * WINDOW_SIZE)
  for feat_idx, w in exp.as_map()[1]:
    out[feat_idx] = w
  return out.reshape(N_CHANNELS, WINDOW_SIZE)


def explain_kernelshap(all_X_flat, sample_flat, target=1, nsamples=100):
  def predict_p1(x):
    x_t = (
        torch.tensor(x, dtype=torch.float32)
        .reshape(-1, N_CHANNELS, WINDOW_SIZE)
        .to(device)
    )
    xai_model.eval()
    with torch.no_grad():
      return torch.softmax(xai_model(x_t), dim=1).cpu().numpy()[:, target]

  bg = shap.sample(all_X_flat, 30)
  explainer = shap.KernelExplainer(predict_p1, bg)
  sv = explainer.shap_values(sample_flat, nsamples=nsamples, silent=True)
  return np.array(sv).reshape(N_CHANNELS, WINDOW_SIZE)


# Σημείωση: το GradCAM (LayerGradCam) παραλείπεται εδώ — χρειάζεται να δείξει σε
# συγκεκριμένο convolutional layer, αλλά το EEGDINO είναι transformer (DINO) αρχιτεκτονική
# χωρίς γνωστά conv layers στη δημόσια τεκμηρίωση· δεν το μαντεύω.
EXPLAINERS_CHANNELTIME = {
    "IntegratedGradients": lambda s: explain_integrated_gradients(s),
    "Saliency": lambda s: explain_saliency(s),
    "Occlusion": lambda s: explain_occlusion(s),
    "DeepLift": lambda s: explain_deeplift(s),
    "FeatureAblation": lambda s: explain_feature_ablation_channel(s),
    "LRP": lambda s: explain_lrp(s),
}

print(
    "Explainers έτοιμοι:",
    list(EXPLAINERS_CHANNELTIME.keys())
    + ["SHAP(Gradient)", "LIME", "KernelSHAP"],
)
sample0 = explain_samples[0:1]
target0 = int(explain_labels[0].item())

ig_map = explain_integrated_gradients(
    sample0, target=target0
)  # (N_CHANNELS, WINDOW_SIZE)
channel_importance = np.abs(ig_map).sum(axis=1)

region_df, region_table = channel_to_region_table(channel_importance)
print(region_df.sort_values("value", ascending=False))
print("\nRelevance ανά περιοχή:\n", region_table)

plot_scalp_map(channel_importance, title=f"IG relevance — target class {target0}")

asym = frontal_asymmetry(channel_importance)
print("\nΜετωπιαία ασυμμετρία (δεξιά - αριστερή, >0 => δεξιά ισχυρότερη):")
for k, v in asym.items():
  print(f"  {k}: {v:+.4f}")


# =====================================================================
# GRAD-CAM: εντοπισμός συνελικτικού επιπέδου δυναμικά
# Το EEGDINO είναι transformer (DINO), αλλά το patch-embedding στάδιο υλοποιείται
# συνήθως με συνελικτικά επίπεδα. Τα εντοπίζουμε δυναμικά αντί να μαντέψουμε ονόματα·
# αν δεν βρεθεί κανένα, η μέθοδος παραλείπεται με σαφές μήνυμα αντί να σκάσει.
# =====================================================================
from captum.attr import LayerGradCam

conv_layers = [(n, m) for n, m in xai_model.named_modules()
               if isinstance(m, (nn.Conv1d, nn.Conv2d))]

gradcam_map = None
if conv_layers:
  layer_name, layer_mod = conv_layers[-1]   # το τελευταίο συνελικτικό επίπεδο
  print(f"\nGrad-CAM στο επίπεδο: {layer_name} ({type(layer_mod).__name__})")
  try:
    gradcam = LayerGradCam(xai_model, layer_mod)
    gc_attr = gradcam.attribute(sample0, target=target0, relu_attributions=True)
    gc = gc_attr.squeeze().detach().cpu().numpy()

    # Το Grad-CAM επιστρέφει χάρτη στις διαστάσεις του επιπέδου, όχι της εισόδου.
    # Τον φέρνουμε σε (N_CHANNELS, WINDOW_SIZE) με interpolation.
    gc_t = torch.tensor(gc, dtype=torch.float32)
    while gc_t.dim() < 3:
      gc_t = gc_t.unsqueeze(0)
    gradcam_map = torch.nn.functional.interpolate(
        gc_t.unsqueeze(0) if gc_t.dim() == 3 else gc_t,
        size=(N_CHANNELS, WINDOW_SIZE), mode="bilinear", align_corners=False
    ).squeeze().detach().cpu().numpy()
    if gradcam_map.ndim > 2:
      gradcam_map = gradcam_map.mean(axis=0)
    print(f"Grad-CAM χάρτης: {gradcam_map.shape}")
  except Exception as e:
    print(f"Το Grad-CAM απέτυχε ({e}); παραλείπεται.")
    gradcam_map = None
else:
  print("\nΔεν βρέθηκαν συνελικτικά επίπεδα στο EEGDINO -- το Grad-CAM παραλείπεται.")


# =====================================================================
# ΧΑΡΤΕΣ ΚΡΑΝΙΟΥ ΚΑΙ TOP-5 ΓΙΑ ΚΑΘΕ ΜΕΘΟΔΟ XAI
# Μέχρι εδώ είχε παραχθεί scalp map μόνο για το Integrated Gradients. Εδώ
# υπολογίζεται και απεικονίζεται ξεχωριστά κάθε μέθοδος, ώστε να είναι δυνατή
# η οπτική σύγκρισή τους.
# =====================================================================
all_maps = {}

for name, fn in EXPLAINERS_CHANNELTIME.items():
  try:
    all_maps[name] = fn(sample0)
  except Exception as e:
    print(f"{name}: απέτυχε ({e})")

# Οι τρεις μέθοδοι εκτός του EXPLAINERS_CHANNELTIME (δέχονται διαφορετικά ορίσματα)
sample0_flat = sample0.detach().reshape(1, -1).cpu().numpy()[0]
all_X_flat = background.detach().reshape(background.shape[0], -1).cpu().numpy()

try:
  all_maps["SHAP(Gradient)"] = explain_shap_gradient(background, sample0, target=target0)
except Exception as e:
  print(f"SHAP: απέτυχε ({e})")

try:
  all_maps["LIME"] = explain_lime(all_X_flat, sample0_flat, target=target0)
except Exception as e:
  print(f"LIME: απέτυχε ({e})")

try:
  all_maps["KernelSHAP"] = explain_kernelshap(all_X_flat, sample0_flat, target=target0)
except Exception as e:
  print(f"KernelSHAP: απέτυχε ({e})")

if gradcam_map is not None:
  all_maps["GradCAM"] = gradcam_map

# --- Έλεγχος και κανονικοποίηση σχήματος: (κανάλια, χρόνος) ---
# Κάποιες βιβλιοθήκες επιστρέφουν τον χάρτη ανάστροφα ή με επιπλέον άξονα. Χωρίς
# αυτόν τον έλεγχο, το plot_topomap δέχεται λάθος πλήθος τιμών και η ανάλυση διακόπτεται.
clean_maps = {}
for name, m in all_maps.items():
  arr = np.asarray(m)
  if arr.ndim != 2:
    print(f"  Το {name} έδωσε χάρτη σχήματος {arr.shape}· παραλείπεται.")
    continue
  if arr.shape[0] != N_CHANNELS and arr.shape[1] == N_CHANNELS:
    print(f"  Το {name} έδωσε ανάστροφο χάρτη {arr.shape}· διορθώνεται.")
    arr = arr.T
  if arr.shape[0] != N_CHANNELS:
    print(f"  Το {name} έδωσε ασύμβατο χάρτη {arr.shape}· παραλείπεται.")
    continue
  clean_maps[name] = arr
all_maps = clean_maps

print(f"\nΜέθοδοι με έγκυρο χάρτη: {list(all_maps.keys())}")

# --- Ξεχωριστός τοπογραφικός χάρτης ανά μέθοδο ---
channel_importances_all = {}
for name, m in all_maps.items():
  imp = np.abs(m).sum(axis=1)
  channel_importances_all[name] = imp
  plot_scalp_map(imp, title=f"{name} — target class {target0}")

  top5_idx = np.argsort(imp)[::-1][:5]
  print(f"\nTop-5 ηλεκτρόδια ({name}):")
  for i in top5_idx:
    ch = CHANNELS[i]
    region, hemi, _ = CHANNEL_REGION[ch]
    print(f"  {ch} ({region}-{hemi}): {imp[i]:.6f}")

# --- Συγκεντρωτικός πίνακας σημαντικότητας ανά ηλεκτρόδιο και μέθοδο ---
channel_scores_df = pd.DataFrame(channel_importances_all, index=CHANNELS)
channel_scores_df.index.name = "channel"
print("\n===== Σημαντικότητα ανά ηλεκτρόδιο και μέθοδο (EEGDINO / DREAMER) =====")
display(channel_scores_df.round(6))
channel_scores_df.to_csv("outputs/xai_channel_scores.csv")

# --- Heatmaps Καναλιών x Χρόνου ανά μέθοδο ---
n_show = len(all_maps)
fig, axes = plt.subplots(n_show, 1, figsize=(12, 2.6 * n_show))
if n_show == 1:
  axes = [axes]
for ax, (name, m) in zip(axes, all_maps.items()):
  im = ax.imshow(np.asarray(m), aspect="auto", cmap="RdBu_r",
                 vmin=-np.abs(m).max(), vmax=np.abs(m).max())
  ax.set_title(name, fontsize=10)
  ax.set_ylabel("Κανάλια")
  ax.set_yticks(range(0, N_CHANNELS, max(1, N_CHANNELS // 7)))
  ax.set_yticklabels([CHANNELS[i] for i in range(0, N_CHANNELS, max(1, N_CHANNELS // 7))],
                     fontsize=7)
  fig.colorbar(im, ax=ax, fraction=0.02)
axes[-1].set_xlabel("Χρόνος (δείγματα)")
plt.tight_layout()
plt.show()

K_VALUES = [5, 10, 20, 30, 50]


def predicted_prob(sample, target):
  with torch.no_grad():
    return torch.softmax(xai_model(sample), dim=1)[0, target].item()


def rank_flat_indices(attr_map, descending=True):
  flat = attr_map.flatten()
  order = np.argsort(-flat if descending else flat)
  return order


def deletion_insertion_curve(
    sample, target, attr_map, mode="deletion", k_values=K_VALUES
):
  order = rank_flat_indices(attr_map)
  n_total = order.size
  base = sample.clone()
  probs = []
  for k in k_values:
    n_k = int(n_total * k / 100)
    idx_to_touch = order[:n_k]
    flat = base.view(1, -1).clone()
    if mode == "deletion":
      flat[0, idx_to_touch] = 0.0
    else:
      new_flat = torch.zeros_like(flat)
      new_flat[0, idx_to_touch] = base.view(1, -1)[0, idx_to_touch]
      flat = new_flat
    perturbed_sample = flat.view_as(base)
    probs.append(predicted_prob(perturbed_sample, target))
  return np.array(probs)


def aopc(probs, base_prob):
  return float(np.mean(base_prob - probs)) if len(probs) else float("nan")


faithfulness_rows = []
for name, fn in EXPLAINERS_CHANNELTIME.items():
  for i in range(min(N_EXPLAIN_SAMPLES, len(explain_samples))):
    sample = explain_samples[i : i + 1]
    target = int(explain_labels[i].item())
    base_prob = predicted_prob(sample, target)
    attr_map = fn(sample)

    del_probs = deletion_insertion_curve(sample, target, attr_map, "deletion")
    ins_probs = deletion_insertion_curve(sample, target, attr_map, "insertion")

    rand_map = np.random.RandomState(i).rand(N_CHANNELS, WINDOW_SIZE)
    del_probs_rand = deletion_insertion_curve(
        sample, target, rand_map, "deletion"
    )
    ins_probs_rand = deletion_insertion_curve(
        sample, target, rand_map, "insertion"
    )

    faithfulness_rows.append({
        "explainer": name,
        "sample": i,
        "AOPC_deletion": aopc(del_probs, base_prob),
        # ΔΙΟΡΘΩΣΗ: η σειρά ορισμάτων ήταν ανάποδη -- aopc(probs, base_prob), όχι
        # aopc(base_prob, probs). Το base_prob είναι μονή τιμή (float), οπότε όταν
        # μπαίνει ΠΡΩΤΟ ως "probs", το len(probs) μέσα στο aopc() σκάει με
        # "TypeError: object of type 'float' has no len()".
        "AOPC_insertion": aopc(ins_probs, base_prob) * -1,
        "AOPC_deletion_random": aopc(del_probs_rand, base_prob),
        "AOPC_insertion_random": aopc(ins_probs_rand, base_prob) * -1,
    })

faithfulness_df = pd.DataFrame(faithfulness_rows)
faithfulness_summary = faithfulness_df.groupby("explainer").mean(
    numeric_only=True
)
faithfulness_summary["deletion_beats_random"] = (
    faithfulness_summary["AOPC_deletion"]
    > faithfulness_summary["AOPC_deletion_random"]
)
faithfulness_summary.to_csv("outputs/faithfulness_AOPC.csv")
display(faithfulness_summary)


def cascading_randomization_test(model, sample, target):
  def explain_with(m):
    sal = Saliency(m)
    attr = sal.attribute(sample.clone().requires_grad_(True), target=target)
    return attr.squeeze().detach().cpu().numpy().flatten()

  original_map = explain_with(model)
  # Τυχαιοποιούμε σταδιακά τα named_children από την έξοδο προς την είσοδο (γενικό,
  # δεν υποθέτει συγκεκριμένα ονόματα layer όπως στο ACTNN — το EEGDINO έχει άγνωστη
  # εσωτερική δομή σε επίπεδο δημόσιας τεκμηρίωσης).
  children = list(model.named_children())
  results = {"none_randomized": 1.0}
  m = copy.deepcopy(model)
  m_children = dict(m.named_children())
  for name, _ in reversed(children):
    layer = m_children[name]
    for p in layer.parameters():
      nn.init.normal_(p, mean=0.0, std=0.1)
    new_map = explain_with(m)
    rho, _ = spearmanr(original_map, new_map)
    results[f"after_randomizing_{name}"] = rho
  return results


rand_results = cascading_randomization_test(
    xai_model, explain_samples[0:1], int(explain_labels[0].item())
)
print("Model-randomization test (Spearman ρ με τον αρχικό χάρτη):")
for k, v in rand_results.items():
  print(f"  {k}: {v:+.3f}")

plt.figure(figsize=(7, 3))
plt.plot(list(rand_results.keys()), list(rand_results.values()), "o-")
plt.xticks(rotation=30, ha="right")
plt.ylabel("Spearman ρ με τον αρχικό χάρτη")
plt.title("Cascading model-randomization sanity check")
plt.axhline(0, color="gray", linestyle="--", linewidth=1)
plt.tight_layout()
plt.show()

shuffled_y = y_data.clone()
rng = np.random.RandomState(0)
perm = rng.permutation(len(shuffled_y))
shuffled_y = shuffled_y[perm]

_, s3_test_idx, _, _, _ = saved_models[any_s3_key]
s3_train_idx = np.setdiff1d(manifest.index.values, s3_test_idx)

shuffled_train_ds = TensorDataset(X_data[s3_train_idx], shuffled_y[s3_train_idx])
shuffled_loader = DataLoader(
    shuffled_train_ds, batch_size=BATCH_SIZE, shuffle=True
)

shuffled_model = build_eegdino()
opt = optim.AdamW(shuffled_model.parameters(), lr=LR, weight_decay=1e-2)
crit = nn.CrossEntropyLoss()
for epoch in range(N_EPOCHS):
  shuffled_model.train()
  for xb, yb in shuffled_loader:
    xb, yb = xb.to(device), yb.to(device)
    opt.zero_grad()
    loss = crit(shuffled_model(xb), yb)
    loss.backward()
    opt.step()

sal_normal = Saliency(xai_model)
sal_shuffled = Saliency(shuffled_model)
sample = explain_samples[0:1]
target = int(explain_labels[0].item())
map_normal = (
    sal_normal.attribute(sample.clone().requires_grad_(True), target=target)
    .squeeze()
    .detach()
    .cpu()
    .numpy()
)
map_shuffled = (
    sal_shuffled.attribute(sample.clone().requires_grad_(True), target=target)
    .squeeze()
    .detach()
    .cpu()
    .numpy()
)
rho_data, _ = spearmanr(map_normal.flatten(), map_shuffled.flatten())
print(
    "Data-randomization test: Spearman ρ (κανονικό vs shuffled-labels μοντέλο)"
    f" = {rho_data:+.3f}"
)
print(
    "Αναμενόμενο: χαμηλό ρ. Αν παραμένει υψηλό, ο explainer αντικατοπτρίζει"
    " κυρίως τη δομή του σήματος."
)


def rbo(list1, list2, p=0.9):
  s1, s2 = set(), set()
  score, weight = 0.0, 1.0
  for d in range(max(len(list1), len(list2))):
    if d < len(list1):
      s1.add(list1[d])
    if d < len(list2):
      s2.add(list2[d])
    overlap = len(s1 & s2)
    score += weight * (overlap / (d + 1))
    weight *= p
  return (1 - p) * score


def jaccard_topk(v1, v2, k):
  top1 = set(np.argsort(-np.abs(v1))[:k])
  top2 = set(np.argsort(-np.abs(v2))[:k])
  return len(top1 & top2) / len(top1 | top2)


channel_importances = {}
for name, fn in EXPLAINERS_CHANNELTIME.items():
  attr_map = fn(sample0)
  channel_importances[name] = np.abs(attr_map).sum(axis=1)

names = list(channel_importances.keys())
agree_spearman = pd.DataFrame(index=names, columns=names, dtype=float)
agree_jaccard5 = pd.DataFrame(index=names, columns=names, dtype=float)
agree_rbo = pd.DataFrame(index=names, columns=names, dtype=float)

for a, b in itertools.product(names, names):
  va, vb = channel_importances[a], channel_importances[b]
  agree_spearman.loc[a, b] = spearmanr(va, vb)[0]
  agree_jaccard5.loc[a, b] = jaccard_topk(va, vb, 5)
  order_a = list(np.argsort(-np.abs(va)))
  order_b = list(np.argsort(-np.abs(vb)))
  agree_rbo.loc[a, b] = rbo(order_a, order_b)

print("Spearman ρ (channel importance):")
display(agree_spearman.round(2))
print("\nJaccard top-5 κανάλια:")
display(agree_jaccard5.round(2))

fig, ax = plt.subplots(figsize=(6, 5))
sns.heatmap(
    agree_spearman.astype(float),
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    vmin=-1,
    vmax=1,
    ax=ax,
)
ax.set_title("Συμφωνία explainers — Spearman ρ (channel importance)")
plt.tight_layout()
plt.show()


def bandstop_filter(x, low, high, fs=FS, order=4):
  nyq = fs / 2
  low_n, high_n = max(low / nyq, 1e-3), min(high / nyq, 0.999)
  b, a = sps.butter(order, [low_n, high_n], btype="bandstop")
  return sps.filtfilt(b, a, x, axis=-1)


def bandpass_filter(x, low, high, fs=FS, order=4):
  nyq = fs / 2
  low_n, high_n = max(low / nyq, 1e-3), min(high / nyq, 0.999)
  b, a = sps.butter(order, [low_n, high_n], btype="bandpass")
  return sps.filtfilt(b, a, x, axis=-1)


_, s3_test_idx_full, y_true_full, y_pred_full, _ = saved_models[any_s3_key]
X_test_np = X_data[s3_test_idx_full].numpy()  # (N, N_CHANNELS, WINDOW_SIZE)
y_test_np = y_data[s3_test_idx_full].numpy()

base_macro_f1 = f1_score(
    y_true_full, y_pred_full, average="macro", zero_division=0
)
band_drop_rows = []
for band_name, (lo, hi) in BANDS.items():
  filtered = np.stack(
      [bandstop_filter(x, lo, hi) for x in X_test_np]
  ).astype(np.float32)
  filtered_t = torch.tensor(filtered).to(device)
  with torch.no_grad():
    preds = xai_model(filtered_t).argmax(1).cpu().numpy()
  band_f1 = f1_score(y_test_np, preds, average="macro", zero_division=0)
  band_drop_rows.append({
      "band": band_name,
      "range_hz": f"{lo}-{hi}",
      "macro_f1_after_removal": band_f1,
      "drop": base_macro_f1 - band_f1,
  })

band_drop_df = pd.DataFrame(band_drop_rows).sort_values("drop", ascending=False)
print(f"Baseline Macro F1 (χωρίς αφαίρεση ζώνης): {base_macro_f1:.3f}")
display(band_drop_df)

channels_x_bands = np.zeros((N_CHANNELS, len(BANDS)))
# ΔΙΟΡΘΩΣΗ: το sample0 κουβαλάει requires_grad=True από τις κλήσεις των explainers
# (Saliency/LRP καλούν .requires_grad_(True) στο ίδιο tensor), οπότε το .numpy() σκάει.
sample_np = sample0.detach().squeeze().cpu().numpy()  # (N_CHANNELS, WINDOW_SIZE)

for b_idx, (band_name, (lo, hi)) in enumerate(BANDS.items()):
  # ΔΙΟΡΘΩΣΗ: η filtfilt εκτελεί διέλευση και προς τις δύο κατευθύνσεις και επιστρέφει
  # πίνακα με ΑΡΝΗΤΙΚΑ strides, τα οποία το PyTorch δεν υποστηρίζει. Το ascontiguousarray
  # δημιουργεί συνεχόμενο αντίγραφο στη μνήμη, χωρίς να αλλάζει τις τιμές.
  band_signal = np.ascontiguousarray(bandpass_filter(sample_np, lo, hi))
  band_tensor = torch.tensor(
      band_signal[None, :, :], dtype=torch.float32, device=device
  )
  ig_band = explain_integrated_gradients(band_tensor, target=target0)
  channels_x_bands[:, b_idx] = np.abs(ig_band).sum(axis=1)

cxb_df = pd.DataFrame(
    channels_x_bands, index=CHANNELS, columns=list(BANDS.keys())
)
plt.figure(figsize=(6, 8))
sns.heatmap(cxb_df, annot=(N_CHANNELS <= 16), fmt=".2f", cmap="viridis")
plt.title(f"Channels × Bands relevance (IG, target class {target0})")
plt.tight_layout()
plt.show()

fig, axes = plt.subplots(1, len(BANDS), figsize=(4 * len(BANDS), 4))
for ax, band_name in zip(axes, BANDS.keys()):
  plot_scalp_map(cxb_df[band_name].values, title=band_name, ax=ax)
plt.tight_layout()
plt.show()

fa_frontal = frontal_asymmetry(cxb_df["alpha"].values)
print(
    "Frontal alpha asymmetry (δεξιά-αριστερή, ένδειξη valence στη βιβλιογραφία):"
)
for k, v in fa_frontal.items():
  print(f"  {k}: {v:+.4f}")


######################################################################
# ΑΝΑΛΥΣΗ ΕΣΩΤΕΡΙΚΩΝ ΑΝΑΠΑΡΑΣΤΑΣΕΩΝ: ΔΙΑΧΩΡΙΣΙΜΟΤΗΤΑ ΚΑΙ ΟΜΟΙΟΤΗΤΑ (CKA)
#
# Σε αντίθεση με τις υπόλοιπες αρχιτεκτονικές της εργασίας, η εσωτερική δομή του
# EEGDINO δεν τεκμηριώνεται δημόσια σε επίπεδο ονομάτων επιπέδων. Για τον λόγο αυτό
# τα στάδια προς σύγκριση ΑΝΑΚΑΛΥΠΤΟΝΤΑΙ δυναμικά: τοποθετούνται forward hooks σε
# όλα τα υπο-modules πρώτου επιπέδου και κρατούνται όσα επιστρέφουν πραγματικό
# tensor κατά τη διάρκεια μιας δοκιμαστικής πρόβλεψης.
######################################################################

CKA_SAMPLES = 400        # πλήθος παραθύρων για τη σύγκριση
CKA_MAX_DIMS = 4096      # άνω όριο διαστάσεων ανά στάδιο (για τη μνήμη)


def _center_gram(K):
    n = K.shape[0]
    H = np.eye(n) - np.ones((n, n)) / n
    return H @ K @ H


def linear_cka(X, Y):
    """Linear CKA μεταξύ δύο πινάκων (δείγματα x διαστάσεις). Τα δείγματα πρέπει
    να είναι τα ΙΔΙΑ και στην ΙΔΙΑ σειρά στους δύο πίνακες."""
    X = np.asarray(X, dtype=np.float64)
    Y = np.asarray(Y, dtype=np.float64)
    X = X - X.mean(axis=0, keepdims=True)
    Y = Y - Y.mean(axis=0, keepdims=True)
    Kx = _center_gram(X @ X.T)
    Ky = _center_gram(Y @ Y.T)
    hsic = float((Kx * Ky).sum())
    norm = float(np.sqrt((Kx * Kx).sum()) * np.sqrt((Ky * Ky).sum()))
    return hsic / norm if norm > 0 else float('nan')


def discover_stages(model, probe_batch):
    """Εντοπίζει ποια υπο-modules πρώτου επιπέδου παράγουν αξιοποιήσιμο tensor."""
    seen = {}

    def mk(name):
        def hook(_m, _i, out):
            t = out[0] if isinstance(out, (tuple, list)) and len(out) else out
            if torch.is_tensor(t) and t.dim() >= 2:
                seen[name] = True
        return hook

    handles = [m.register_forward_hook(mk(n)) for n, m in model.named_children()]
    model.eval()
    try:
        with torch.no_grad():
            _ = model(probe_batch)
    finally:
        for h in handles:
            h.remove()
    return list(seen.keys())


def get_layer_reps(model, X_tensor, stage_names, batch=32):
    """Ενεργοποιήσεις των σταδίων, ισοπεδωμένες σε δύο διαστάσεις. Αν ένα στάδιο
    παράγει πολύ μεγάλο διάνυσμα, γίνεται μέσος όρος στον άξονα των tokens/χρόνου
    ώστε να παραμείνει διαχειρίσιμο."""
    store, reps = {}, {n: [] for n in stage_names}

    def mk(name):
        def hook(_m, _i, out):
            t = out[0] if isinstance(out, (tuple, list)) and len(out) else out
            if torch.is_tensor(t):
                store[name] = t.detach().cpu().numpy()
        return hook

    handles = [m.register_forward_hook(mk(n))
               for n, m in model.named_children() if n in stage_names]
    model.eval()
    try:
        with torch.no_grad():
            for i in range(0, len(X_tensor), batch):
                xb = X_tensor[i:i + batch].to(device)
                _ = model(xb)
                for n in stage_names:
                    if n not in store:
                        continue
                    a = store[n]
                    flat = a.reshape(len(xb), -1)
                    if flat.shape[1] > CKA_MAX_DIMS and a.ndim >= 3:
                        flat = a.mean(axis=1).reshape(len(xb), -1)   # μέσος όρος tokens
                    reps[n].append(flat[:, :CKA_MAX_DIMS])
    finally:
        for h in handles:
            h.remove()
    return {n: np.concatenate(v) for n, v in reps.items() if v}


# --- Κοινό υποσύνολο παραθύρων, ίδιο και στην ίδια σειρά για κάθε μοντέλο ---
rng_cka = np.random.RandomState(0)
cka_idx = rng_cka.choice(xai_test_idx, size=min(CKA_SAMPLES, len(xai_test_idx)),
                         replace=False)
X_cka = X_data[cka_idx]
labels_cka = y_data[cka_idx].numpy() if torch.is_tensor(y_data) else np.asarray(y_data)[cka_idx]

stage_names = discover_stages(xai_model, X_cka[:4].to(device))
print(f"Στάδια που εντοπίστηκαν: {stage_names}")

reps_s3 = get_layer_reps(xai_model, X_cka, stage_names)
for k, v in reps_s3.items():
    print(f"Αναπαράσταση '{k}': {v.shape}")

# --- Μέτρα διαχωρισιμότητας ανά στάδιο ---
sep_rows = []
for name, feats in reps_s3.items():
    if np.allclose(feats.std(axis=0), 0):
        print(f"  Το στάδιο '{name}' έχει μηδενική διασπορά· παραλείπεται.")
        continue
    intra, inter = class_distances(feats, labels_cka)
    sep_rows.append({
        'stage': name,
        'silhouette': silhouette_score(feats, labels_cka),
        'davies_bouldin': davies_bouldin_score(feats, labels_cka),
        'fisher_ratio': fisher_ratio(feats, labels_cka),
        'intra_class_dist': intra,
        'inter_class_dist': inter,
        'inter_over_intra': inter / intra if intra > 0 else float('nan'),
    })

if sep_rows:
    separability_df = pd.DataFrame(sep_rows).set_index('stage')
    print("\n===== Μέτρα διαχωρισιμότητας ανά στάδιο (μοντέλο S3) =====")
    print(f"Κατανομή κλάσεων: {np.bincount(labels_cka)}")
    display(separability_df.round(4))
    print("Silhouette: +1 άριστος διαχωρισμός, ~0 επικάλυψη, <0 λάθος ανάθεση.")
    separability_df.to_csv("outputs/separability_per_stage.csv")

    fig, axes = plt.subplots(1, 2, figsize=(11, 3.2))
    separability_df['silhouette'].plot(kind='bar', ax=axes[0], color='#4C72B0')
    axes[0].axhline(0, color='gray', lw=1); axes[0].set_title('Silhouette ανά στάδιο')
    axes[0].tick_params(axis='x', rotation=30)
    separability_df['inter_over_intra'].plot(kind='bar', ax=axes[1], color='#55A868')
    axes[1].axhline(1, color='red', ls='--', lw=1)
    axes[1].set_title('Λόγος δια-/ενδο-κλασικής απόστασης')
    axes[1].tick_params(axis='x', rotation=30)
    plt.tight_layout(); plt.show()

# --- (α) Ομοιότητα μεταξύ σταδίων του μοντέλου S3 ---
names = list(reps_s3.keys())
cka_layers = pd.DataFrame(index=names, columns=names, dtype=float)
for a, b in itertools.product(names, names):
    cka_layers.loc[a, b] = linear_cka(reps_s3[a], reps_s3[b])

print("\n===== CKA μεταξύ σταδίων (μοντέλο S3) =====")
display(cka_layers.round(3))
cka_layers.to_csv("outputs/cka_between_layers.csv")

fig, ax = plt.subplots(figsize=(1.2 * len(names) + 2, len(names) + 1))
sns.heatmap(cka_layers.astype(float), annot=True, fmt='.2f', cmap='viridis',
            vmin=0, vmax=1, ax=ax)
ax.set_title('Linear CKA μεταξύ σταδίων (S3)')
plt.tight_layout(); plt.show()

# --- (β) Ομοιότητα ίδιου σταδίου μεταξύ των μοντέλων S1 και S3 ---
s1_key = next((k for k in saved_models if k[0] == "S1"), None)
model_s1 = saved_models[s1_key][0] if s1_key else None

if model_s1 is None:
    print("\nΤο μοντέλο S1 δεν είναι διαθέσιμο· η σύγκριση S1 vs S3 παραλείπεται.")
else:
    reps_s1 = get_layer_reps(model_s1, X_cka, stage_names)
    common = [n for n in names if n in reps_s1 and reps_s1[n].shape == reps_s3[n].shape]
    cka_cross = pd.DataFrame(
        [{'stage': n, 'CKA_S1_vs_S3': linear_cka(reps_s1[n], reps_s3[n])} for n in common]
    ).set_index('stage')

    print("\n===== CKA ίδιου σταδίου: μοντέλο S1 έναντι μοντέλου S3 =====")
    display(cka_cross.round(3))
    print("Χαμηλές τιμές σημαίνουν ότι τα δύο πρωτόκολλα οδηγούν σε ποιοτικά")
    print("διαφορετικές αναπαραστάσεις, παρότι η αρχιτεκτονική είναι ίδια.")
    cka_cross.to_csv("outputs/cka_s1_vs_s3.csv")

    fig, ax = plt.subplots(figsize=(1.2 * len(common) + 2, 3.2))
    cka_cross['CKA_S1_vs_S3'].plot(kind='bar', ax=ax, color='#4C72B0')
    ax.set_ylim(0, 1); ax.set_ylabel('Linear CKA')
    ax.set_title('Ομοιότητα αναπαραστάσεων S1 vs S3')
    plt.xticks(rotation=30); plt.tight_layout(); plt.show()

    
# =====================================================================
# LAYER-WISE PROBING
# Γραμμικοί ταξινομητές πάνω στις παγωμένες αναπαραστάσεις ΚΑΘΕ σταδίου,
# ώστε να φανεί σε ποιο σημείο του δικτύου η κάθε πληροφορία γίνεται
# γραμμικά αποκωδικοποιήσιμη. Συγκρίνεται η συναισθηματική ετικέτα με
# μεταβλητές που το μοντέλο ΔΕΝ θα έπρεπε να κωδικοποιεί (ταυτότητα
# εθελοντή, δοκιμασία), οπότε υψηλή ακρίβεια σε αυτές αποτελεί ένδειξη
# συντόμευσης μέσω Διαρροής Ταυτότητας Υποκειμένου.
# =====================================================================
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split as _tts

META = manifest            # <-- στα δύο CCNN notebooks άλλαξέ το σε: META = info

probe_targets = {'label (συναίσθημα)': 'label',
                 'subject_id (ταυτότητα)': 'subject_id',
                 'trial_id (δοκιμασία)': 'trial_id'}
if 'session_id' in META.columns and META['session_id'].nunique() > 1:
    probe_targets['session_id (συνεδρία)'] = 'session_id'

probe_rows = []
for stage, feats in reps_s3.items():
    if np.allclose(feats.std(axis=0), 0):
        print(f"  Το στάδιο '{stage}' έχει μηδενική διασπορά· παραλείπεται.")
        continue
    for label_name, col in probe_targets.items():
        y = META.loc[cka_idx, col].values
        classes, counts = np.unique(y, return_counts=True)
        if len(classes) < 2 or counts.min() < 2:
            continue
        strat = y if counts.min() >= 2 else None
        Xtr, Xte, ytr, yte = _tts(feats, y, test_size=0.3, random_state=0, stratify=strat)
        clf = LogisticRegression(max_iter=1000).fit(Xtr, ytr)
        acc = clf.score(Xte, yte)
        chance = counts.max() / counts.sum()      # πλειοψηφική κλάση ως γραμμή βάσης
        probe_rows.append({'stage': stage, 'target': label_name,
                           'probe_accuracy': acc, 'chance_level': chance,
                           'ratio': acc / chance if chance > 0 else float('nan')})

probe_df = pd.DataFrame(probe_rows)
if not probe_df.empty:
    table = probe_df.pivot(index='stage', columns='target', values='probe_accuracy')
    print("\n===== Layer-wise probing: ακρίβεια γραμμικού ταξινομητή ανά στάδιο =====")
    print(table.round(4))
    print("\nΛόγος προς τη γραμμή βάσης (τιμές >>1 υποδηλώνουν αποκωδικοποιήσιμη πληροφορία):")
    print(probe_df.pivot(index='stage', columns='target', values='ratio').round(2))
    probe_df.to_csv('layerwise_probing.csv', index=False)

    fig, ax = plt.subplots(figsize=(1.6 * len(table) + 3, 3.4))
    table.plot(kind='bar', ax=ax)
    ax.set_ylabel('Ακρίβεια probe'); ax.set_ylim(0, 1)
    ax.set_title('Layer-wise probing ανά στάδιο')
    plt.xticks(rotation=0); plt.legend(fontsize=8); plt.tight_layout(); plt.show()
```
